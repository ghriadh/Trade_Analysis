# A. Introduction & Framework
</div>

**Direction 1** : *"Which economic indicators explain trade performance?"*

**Design principles :**
- All 16 indicators included — no a priori exclusion
- J (RF) selects top indicators → guides E/F/G/H automatically
- Population used directly (not derived from GDP/GDP_pc)
- FDI available and used in consistency analysis
- All figures : trade × indicator only

**Structure :**
```
C  — Data Loading (all dataframes + complete mapping)
D  — Combined Metrics (Trade Intensity, 6-situation Elasticity, Per Capita)
J  — RF + OLS (selects top indicators per trade variable)
I.1/I.2 — Spearman confirmation
E  — Export × top indicators
F  — Import × top indicators
G  — Total Trade × top indicators
H  — Balance & Coverage × top indicators
I.3→I.7 — Synthesis (regression, PCA, temporal, groups, concentration)
N  — Save
O  — Conclusions
```

## 2. Style & Helpers
</div>

````python
C = {"blue":"#3498db","green":"#2ecc71","red":"#e74c3c","orange":"#f39c12", "purple":"#9b59b6","grey":"#95a5a6","dark":"#1a1a2e","teal":"#1abc9c"}

SIZE_COLORS      = {"Leader":C["green"],"Major":C["blue"],"Mid":C["orange"],"Small":C["red"]}
BALANCE_COLORS   = {"Strong Surplus":C["green"],"Slight Surplus":C["teal"], "Slight Deficit":C["orange"],"Structural Deficit":C["red"]}
GROWTH_COLORS    = {"Strong Grower":C["green"],"Moderate Grower":C["blue"], "Weak Grower":C["orange"],"Declining":C["red"]}
STABILITY_COLORS = {"Stable":C["green"],"Moderate":C["blue"], "Volatile":C["orange"],"Very Volatile":C["red"]}
CONSIST_COLORS   = {"Very Consistent":C["green"],"Consistent":C["blue"], "Inconsistent":C["orange"],"Very Inconsistent":C["red"]}

def styled_figure(figsize=(12,6)):
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    ax.tick_params(colors="white"); ax.xaxis.label.set_color("white")
    ax.yaxis.label.set_color("white"); ax.title.set_color("white")
    for sp in ax.spines.values(): sp.set_edgecolor("#444")
    ax.grid(alpha=0.2,color="#555"); return fig,ax

def styled_subplots(nrows=1,ncols=2,figsize=(18,6)):
    fig,axes = plt.subplots(nrows,ncols,figsize=figsize)
    fig.patch.set_facecolor(C["dark"])
    for ax in np.array(axes).flatten():
        ax.set_facecolor("#16213e"); ax.tick_params(colors="white")
        ax.xaxis.label.set_color("white"); ax.yaxis.label.set_color("white")
        ax.title.set_color("white")
        for sp in ax.spines.values(): sp.set_edgecolor("#444")
        ax.grid(alpha=0.2,color="#555")
    return fig,axes

def normalize_01(s):
    mn,mx = s.min(),s.max(); return (s-mn)/(mx-mn+1e-9)

def spearman_matrix(df, trade_vars, ind_vars, min_obs=25):
    rows=[]
    for tv in trade_vars:
        for iv in ind_vars:
            sub = df[[tv,iv]].dropna()
            if len(sub)<min_obs: continue
            r,p = spearmanr(sub[tv],sub[iv])
            rows.append({"trade_var":tv,"indicator":iv, "rho":round(r,3),"pvalue":round(p,4), "sig":p<0.05,"abs_rho":abs(r)})
    return pd.DataFrame(rows)

def heatmap_corr(corr_df, title, trade_label_map=None, ind_label_map=None, min_abs_rho=0.15, figsize=(12,7), fname=None):
    """Plot Spearman heatmap with readable labels."""
    strong = corr_df[(corr_df["sig"])&(corr_df["abs_rho"]>=min_abs_rho)].copy()
    if strong.empty: print(f"No sig. correlations above |ρ|={min_abs_rho}"); return
    if trade_label_map:  strong["trade_var"] = strong["trade_var"].map(trade_label_map).fillna(strong["trade_var"])
    if ind_label_map:    strong["indicator"] = strong["indicator"].map(ind_label_map).fillna(strong["indicator"])
    pivot = strong.pivot_table(index="indicator",columns="trade_var",values="rho",aggfunc="first")
    pivot = pivot.sort_values(pivot.columns[0],ascending=False)
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1, annot=True,fmt=".2f",annot_kws={"size":8},
                linewidths=0.5,linecolor="#333",cbar_kws={"label":"Spearman ρ"})
    ax.set_title(f"{title}\n(sig |ρ|≥{min_abs_rho})",color="white",pad=10)
    ax.tick_params(colors="white",labelsize=8)
    plt.xticks(rotation=30,ha="right"); plt.tight_layout()
    f = fname or title.lower().replace(" ","_").replace("×","x")[:40]
    plt.savefig(f"../Figures/link/{f}.png",dpi=150,bbox_inches="tight"); plt.show()

def scatter_ind(df, x_col, y_col, color_col=None, color_map=None, title="", xlabel="", ylabel="", countries=None, xlog=False, clip_x=None, fname=None):
    """Standard scatter trade × indicator."""
    fig,ax = styled_figure(figsize=(13,8))
    ax.set_title(title,color="white")
    cols = [x_col,y_col] + ([color_col] if color_col else []) + (["country_name"] if "country_name" in df.columns else [])
    df_p = df[cols].dropna()
    if clip_x: df_p = df_p.copy(); df_p[x_col] = df_p[x_col].clip(*clip_x)
    if color_col and color_map:
        for grp,col in color_map.items():
            sub = df_p[df_p[color_col]==grp]
            ax.scatter(sub[x_col],sub[y_col],color=col,alpha=0.6,s=50,label=grp)
        ax.legend(facecolor="#16213e",labelcolor="white",fontsize=8)
    else: ax.scatter(df_p[x_col],df_p[y_col],color=C["blue"],alpha=0.6,s=50)
    if countries and "country_name" in df_p.columns:
        for _,row in df_p[df_p["country_name"].isin(countries)].iterrows():
            ax.annotate(row["country_name"],(row[x_col],row[y_col]), fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    if xlog: ax.set_xscale("log")
    ax.set_xlabel(xlabel,color="white"); ax.set_ylabel(ylabel,color="white")
    sub2 = df_p[[x_col,y_col]].dropna()
    r,p = spearmanr(sub2[x_col],sub2[y_col])
    ax.text(0.05,0.95,f"Spearman ρ={r:.2f} (p={p:.3f})", transform=ax.transAxes,color="white",fontsize=9)
    plt.tight_layout()
    if fname: plt.savefig(f"Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

def boxplot_quartile(df, value_col, group_col, title="", ylabel="", fname=None):
    """Boxplot by quartile — clearer than scatter for weak ρ."""
    df_q = df[[value_col,group_col]].dropna().copy()
    labels = ["Q1(Low)","Q2","Q3","Q4(High)"]
    df_q["quartile"] = pd.qcut(df_q[group_col],4,labels=labels,duplicates="drop")
    fig,ax = styled_figure(figsize=(10,6))
    ax.set_title(title,color="white")
    data_box = [df_q[df_q["quartile"]==q][value_col].dropna().values for q in labels]
    bp = ax.boxplot(data_box,patch_artist=True,medianprops=dict(color="white",linewidth=2))
    for patch,col in zip(bp["boxes"],[C["red"],C["orange"],C["blue"],C["green"]]):
        patch.set_facecolor(col); patch.set_alpha(0.7)
    ax.set_xticklabels(labels,color="white")
    ax.set_ylabel(ylabel,color="white")
    plt.tight_layout()
    if fname: plt.savefig(f"Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

countries = ["USA","China","Germany","France","India","Brazil","Japan","Canada"]
print("✅ Style & helpers OK")
````

````python
# indicator list
all_indicators_full = sorted(df[df["entity_type"]=="COUNTRY"]["indicator_name"].unique())
ind_cols = [i for i in all_indicators_full if i in df["indicator_name"].unique()]

# IND_LABELS — full name → short label (for figures)
IND_LABELS = {
                "Current account balance (% of GDP)"                                   : "Current Account % GDP",
                "Current account balance (BoP, current US$)"                           : "Current Account (US$)",
                "Expense (% of GDP)"                                                   : "Expense % GDP",
                "Exports of goods and services (% of GDP)"                             : "Exports % GDP",
                "Exports of goods and services (current US$)"                          : "Exports (US$)",
                "External debt stocks, total (DOD, current US$)"                       : "External Debt (US$)",
                "Foreign direct investment, net inflows (BoP, current US$)"            : "FDI Inflows (US$)",
                "GDP (current US$)"                                                    : "GDP (US$)",
                "GDP per capita (current US$)"                                         : "GDP per Capita (US$)",
                "Imports of goods and services (% of GDP)"                             : "Imports % GDP",
                "Imports of goods and services (current US$)"                          : "Imports (US$)",
                "Inflation, consumer prices (annual %)"                                : "Inflation % (CPI)",
                "Population, total"                                                    : "Population",
                "Real interest rate (%)"                                               : "Real Interest Rate %",
                "Trade (% of GDP)"                                                     : "Trade % GDP",
                "Unemployment, total (% of total labor force) (modeled ILO estimate)"  : "Unemployment %",
                }

# IND_PCT — indicators en % clip  —  % GDP indicators : directly comparable across countries
IND_PCT = {
            "Inflation, consumer prices (annual %)"                                : (-5,   50),  
            "Real interest rate (%)"                                               : (-20,  45),  
            "Unemployment, total (% of total labor force) (modeled ILO estimate)"  : (0,    40),
            "Current account balance (% of GDP)"                                   : (-30,  30),
            "Expense (% of GDP)"                                                   : (0,   100),  
            "Exports of goods and services (% of GDP)"                             : (0,   250),  
            "Imports of goods and services (% of GDP)"                             : (0,   250),  
            "Trade (% of GDP)"                                                     : (0,   400),
            }

# US$ absolute indicators : need normalization by GDP or Population for cross-country comparison
IND_ABS = {
            "GDP (current US$)",
            "GDP per capita (current US$)",
            "Population, total",
            "Exports of goods and services (current US$)",
            "Imports of goods and services (current US$)",
            "External debt stocks, total (DOD, current US$)",
            "Foreign direct investment, net inflows (BoP, current US$)",
            "Current account balance (BoP, current US$)"
            }
````

````python
def get_ind_params(ind):
    """    Return display parameters for a given indicator (full name).    Returns : label, clip, use_log, needs_norm
    Usage in scatter :    label, clip, use_log, needs_norm = get_ind_params(ind)
                          x = np.log(df[ind].clip(lower=1)) if use_log else df[ind].clip(*clip) if clip else df[ind]
                          ax.set_xlabel(label)    """
    
    label      = IND_LABELS.get(ind, ind)
    clip       = IND_PCT.get(ind, None)      # None if not in IND_PCT
    use_log    = ind in IND_ABS              # log scaling if US$ absolute
    needs_norm = ind in IND_ABS and ind not in { "GDP per capita (current US$)",  "Population, total"   }
    
    return label, clip, use_log, needs_norm

# Test
for ind in list(IND_PCT.keys())[:2] + list(IND_ABS)[:2]:
    label, clip, use_log, needs_norm = get_ind_params(ind)
    print(f"  {label:<35} clip={clip}  log={use_log}  norm={needs_norm}")
````

````python
# df_structural is the backbone
df_master = df_structural.copy()

# Merge structural indicators 
df_master = df_master.merge(df_ind_struct, on="country_name", how="left")

# Merge GDP CAGR & VOL
df_master = df_master.merge(gdp_cagr, on="country_name", how="left")

# Merge combined scores
df_master = df_master.merge(df_combined[["country_name","combined_score","f_norm_rank", "g_norm_rank","combined_profile"]], 
                            on="country_name", how="left")

# Merge economic scores (multi_df)
econ_score_cols = [c for c in multi_df.columns if "_value_norm" in c or "_growth_norm" in c]
df_master = df_master.merge(multi_df[["country_name"]+econ_score_cols], on="country_name", how="left")

print(f"✅ df_master : {df_master.shape}")
print(f"   Countries : {df_master['country_name'].nunique()}")
print(f"   Columns   : {len(df_master.columns)}")
df_master.sample()
````

## C8. Derived Columns (Per Capita, Trade Ratios, Elasticity)
</div>

````python
# Per capita trade 
if pop_col in df_master.columns:
    mask = df_master[pop_col].notna() & (df_master[pop_col]>0)
    for flow_col,pc_col in [("total_export","export_pc"),("total_import","import_pc")]:
        if flow_col in df_master.columns:
            df_master.loc[mask,pc_col] = (df_master.loc[mask,flow_col] / df_master.loc[mask,pop_col])
    print(f"✅ Per capita trade computed from Population directly")

# Trade Intensity & ratios (using GDP directly)
if gdp_col in df_master.columns:
    mask = df_master[gdp_col].notna() & (df_master[gdp_col]>0)
    
    df_master.loc[mask,"trade_intensity"] = ((df_master.loc[mask,"total_export"]+df_master.loc[mask,"total_import"]) / df_master.loc[mask,gdp_col])
    df_master.loc[mask,"export_dep"] = df_master.loc[mask,"total_export"]/df_master.loc[mask,gdp_col]
    df_master.loc[mask,"import_dep"] = df_master.loc[mask,"total_import"]/df_master.loc[mask,gdp_col]
    df_master.loc[mask,"balance_dep"] = ((df_master.loc[mask,"total_export"]-df_master.loc[mask,"total_import"]) /df_master.loc[mask,gdp_col])

n_ti = df_master["trade_intensity"].notna().sum()
print(f"✅ Trade ratios/GDP : {n_ti} countries")

# Trade Elasticity = CAGR_export / CAGR_GDP
mask_e = df_master["cagr_gdp"].notna() & (df_master["cagr_gdp"]!=0)
df_master.loc[mask_e,"trade_elasticity_x"] = (df_master.loc[mask_e,"cagr_export"]/df_master.loc[mask_e,"cagr_gdp"])
df_master.loc[mask_e,"trade_elasticity_m"] = (df_master.loc[mask_e,"cagr_import"]/df_master.loc[mask_e,"cagr_gdp"])

n_el = df_master["trade_elasticity_x"].notna().sum()
print(f"✅ Trade elasticity  : {n_el} countries")

print(f"\n✅ df_master : {df_master.shape} | {df_master['country_name'].nunique()} countries")
print(f"   Indicator columns available : {len([c for c in ind_cols if c in df_master.columns])} / {len(ind_cols)}")

df_master.sample()
````

````python
# final summary
print(f"\n{'='*55}")
print(f"  df_master SUMMARY")
print(f"{'='*55}")
print(f"  Shape     : {df_master.shape}")
print(f"  Countries : {df_master['country_name'].nunique()}")
print(f"\n  Trade variables (from df_structural) :")
trade_cols = [c for c in df_master.columns if c in df_structural.columns and c != "country_name"]
print(f"    {len(trade_cols)} columns")
print(f"\n  Indicator columns (original full names) :")
ind_in_master = [c for c in ind_cols_available if c in df_master.columns]
print(f"    {len(ind_in_master)} / {len(ind_cols_available)}")
print(f"\n  Derived columns :")
derived = ["export_pc","import_pc","trade_intensity","export_dep", "import_dep","balance_dep","trade_elasticity_x","trade_elasticity_m", 
           "cagr_gdp","vol_gdp"]
for c in derived:
    n = df_master[c].notna().sum() if c in df_master.columns else 0
    print(f"    {c:<30} : {n} countries")
````

````python
# Normalize large-scale indicators by GDP or Population 
for ind in ["External debt stocks, total (DOD, current US$)", "Foreign direct investment, net inflows (BoP, current US$)"]:
    if ind in df_master.columns and gdp_col in df_master.columns:
        mask = df_master[gdp_col].notna() & (df_master[gdp_col] > 0)
        short = IND_LABELS.get(ind, "").replace(" ", "_").replace("(", "").replace(")", "").replace("$", "").strip()
        new_col = f"norm_{short}"
        df_master.loc[mask, new_col] = (df_master.loc[mask, ind] / df_master.loc[mask, gdp_col])
        print(f"✅ {new_col:<45} : {df_master[new_col].notna().sum()} countries")
````

## C9. Define Trade Variable Groups

````python
trade_cols_structural = [c for c in df_structural.columns if c != "country_name"]
trade_cols_derived    = ["export_pc","import_pc","trade_intensity",
                         "export_dep","import_dep","balance_dep",
                         "trade_elasticity_x","trade_elasticity_m"]
all_trade_cols = [c for c in trade_cols_structural + trade_cols_derived if c in df_master.columns]
````

Preliminary analysis linking trade structural metrics to the two most fundamental
economic indicators : **GDP** and **GDP per Capita**.

|---|---|---|---|

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### D1.1 — Does higher income lead to more trade openness?
</div>

````python
gdp_pc_col = "GDP per capita (current US$)"
df_d1 = df_master[["country_name","trade_intensity",gdp_pc_col,"size_profile"]].dropna()

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Trade Intensity (X+M)/GDP × GDP per Capita",
             fontsize=14, color="white", y=1.01)

# ── Left : Scatter Trade Intensity vs GDP_pc ──────────────────────────────────
for sp, col in SIZE_COLORS.items():
    sub = df_d1[df_d1["size_profile"] == sp]
    axes[0].scatter(sub[gdp_pc_col], sub["trade_intensity"],
                    color=col, alpha=0.6, s=50, label=sp)

# Regression line
from scipy.stats import linregress
log_gdp = np.log(df_d1[gdp_pc_col].clip(lower=1))
slope, intercept, r, p, _ = linregress(log_gdp, df_d1["trade_intensity"])
x_line = np.linspace(log_gdp.min(), log_gdp.max(), 100)
axes[0].plot(np.exp(x_line), intercept + slope * x_line,
             color="white", linewidth=1.5, linestyle="--", alpha=0.6,
             label=f"Trend (ρ={r:.2f})")

# Annotate selected countries
for _, row in df_d1[df_d1["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row[gdp_pc_col], row["trade_intensity"]),
                     fontsize=8, color="white",
                     xytext=(5, 3), textcoords="offset points")

axes[0].set_xscale("log")
axes[0].set_xlabel("GDP per Capita (log scale)", color="white")
axes[0].set_ylabel("Trade Intensity (X+M)/GDP", color="white")
axes[0].set_title("Trade Intensity vs GDP per Capita\n"
                  "(scatter + trend)", color="white")
axes[0].legend(facecolor="#16213e", labelcolor="white", fontsize=8)
r_sp, p_sp = spearmanr(log_gdp, df_d1["trade_intensity"])
axes[0].text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})",
             transform=axes[0].transAxes, color="white", fontsize=9)

# ── Right : Boxplot by GDP_pc Quartile ───────────────────────────────────────
df_d1["gdp_pc_q"] = pd.qcut(df_d1[gdp_pc_col], 4,
                              labels=["Q1\n(Poorest)", "Q2", "Q3", "Q4\n(Richest)"])
labels_q  = ["Q1\n(Poorest)", "Q2", "Q3", "Q4\n(Richest)"]
data_box  = [df_d1[df_d1["gdp_pc_q"] == q]["trade_intensity"].dropna().values
             for q in labels_q]
bp = axes[1].boxplot(data_box, patch_artist=True,
                     medianprops=dict(color="white", linewidth=2))
for patch, col in zip(bp["boxes"], [C["red"], C["orange"], C["blue"], C["green"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)
axes[1].set_xticklabels(labels_q, color="white")
axes[1].set_title("Trade Intensity by GDP_pc Quartile\n"
                  "(does the trend hold across groups?)", color="white")
axes[1].set_ylabel("Trade Intensity (X+M)/GDP", color="white")

medians = []
for i, q in enumerate(labels_q):
    vals = df_d1[df_d1["gdp_pc_q"] == q]["trade_intensity"].dropna()
    med  = vals.median()
    medians.append(med)
    axes[1].text(i+1, med + 0.3, f"{med:.1f}",
                 ha="center", color="white", fontsize=9, fontweight="bold")

plt.tight_layout()
plt.savefig("../Figures/link/D1_1_trade_intensity_gdppc.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Insight ───────────────────────────────────────────────────────────────────
print("=" * 60)
print("  D1.1 — INSIGHT")
print("=" * 60)
print(f"  Spearman ρ = {r_sp:.2f} → Weak relationship")
print(f"  Medians by quartile : {[f'{m:.1f}' for m in medians]}")
if medians[-1] > medians[0]:
    print(f"  → Richer countries are slightly more open ({medians[0]:.1f} → {medians[-1]:.1f})")
else:
    print(f"  → No clear monotonic trend — wealth does not drive openness")
print()
print("  Key anomalies visible in scatter :")
print("  · USA, Japan : high GDP_pc but very LOW trade intensity")
print("    → Large domestic market absorbs most production")
print("  · Mid/Small outliers above trend : small open economies")
print("    (transit hubs, resource exporters)")
print()
print("  Answer : GDP per capita is a WEAK predictor of trade openness (ρ=0.29)")
print("  → Trade openness is more a structural policy choice than a wealth effect")
````

## D2. Trade Elasticity — 6 Situations
</div>

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### D2.1 — Classification of trade-GDP growth relationship
</div>

**Trade Elasticity** = CAGR_export / CAGR_GDP

Classification based first on the **sign** of each CAGR, then on the **magnitude** :

| # | CAGR_export | CAGR_GDP | Reading |
|---|---|---|---|
| 1 | + | + | > 1 — Trade-led : trade grows faster → opening |
| 2 | + | + | 0→1 — Endogenous : economy grows faster than trade |
| 3 | + | − | < 0 — Trade Resistant : trade grows despite GDP decline |
| 4 | − | + | < 0 — Decoupling : GDP grows but trade declines |
| 5 | − | − | > 1 — Double Decline : trade collapses faster than GDP |
| 6 | − | − | 0→1 — Double Decline : GDP more affected than trade |

**Warning** : situations 5 and 6 both produce a **positive elasticity ratio** from (-/-).
Sign-based classification must come **before** using the ratio value.


````python
def classify_6sit(row):
    xe = row.get("cagr_export", np.nan)
    xg = row.get("cagr_gdp",    np.nan)
    if pd.isna(xe) or pd.isna(xg): return "Unknown"
    if xe >= 0 and xg >= 0:
        return "1-Trade-led"   if xe > xg else "2-Endogenous"
    elif xe >= 0 and xg < 0:
        return "3-Trade Resistant"
    elif xe < 0 and xg >= 0:
        return "4-Decoupling"
    else:
        return "5-Double Decline (trade faster)" if abs(xe) > abs(xg)                else "6-Double Decline (GDP faster)"

SIT_COLORS = {
    "1-Trade-led"                          : C["green"],
    "2-Endogenous"                         : C["teal"],
    "3-Trade Resistant"                    : C["blue"],
    "4-Decoupling"                         : C["orange"],
    "5-Double Decline (trade faster)"      : C["red"],
    "6-Double Decline (GDP faster)"        : C["purple"],
    "Unknown"                              : C["grey"],
}
SIT_LABELS = {k: k.split("-", 1)[1] if "-" in k else k for k in SIT_COLORS}

df_d2 = df_master[["country_name","cagr_export","cagr_gdp",
                    "trade_elasticity_x","size_profile"]].dropna(
    subset=["cagr_export","cagr_gdp"])
df_d2["situation"] = df_d2.apply(classify_6sit, axis=1)

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Trade Elasticity — 6 Situations",
             fontsize=14, color="white", y=1.01)

# ── Left : Scatter colored by situation ──────────────────────────────────────
for sit, col in SIT_COLORS.items():
    sub = df_d2[df_d2["situation"] == sit]
    if sub.empty: continue
    axes[0].scatter(sub["cagr_gdp"] * 100, sub["cagr_export"] * 100,
                    color=col, alpha=0.7, s=60, label=SIT_LABELS[sit])

lim = (df_d2[["cagr_export","cagr_gdp"]].abs().max().max()) * 100 * 1.1
axes[0].plot([-lim, lim], [-lim, lim], color="white",
             linestyle="--", alpha=0.4, label="Elasticity = 1")
axes[0].axhline(0, color=C["grey"], linestyle=":", alpha=0.4)
axes[0].axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_d2[df_d2["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row["cagr_gdp"]*100, row["cagr_export"]*100),
                     fontsize=8, color="white",
                     xytext=(5, 3), textcoords="offset points")

axes[0].set_xlabel("CAGR GDP (%)", color="white")
axes[0].set_ylabel("CAGR Export (%)", color="white")
axes[0].set_title("CAGR Export vs CAGR GDP\n(color = situation)",
                  color="white")
axes[0].legend(facecolor="#16213e", labelcolor="white", fontsize=7)

# ── Right : Country count by situation ───────────────────────────────────────
sit_counts = df_d2["situation"].value_counts()
axes[1].barh([SIT_LABELS.get(s, s) for s in sit_counts.index],
             sit_counts.values,
             color=[SIT_COLORS.get(s, C["grey"]) for s in sit_counts.index],
             alpha=0.85)
axes[1].set_title("Country Count by Situation", color="white")
axes[1].set_xlabel("Number of countries", color="white")
for i, v in enumerate(sit_counts.values):
    axes[1].text(v + 0.3, i, str(v), va="center", color="white", fontsize=9)

plt.tight_layout()
plt.savefig("../Figures/link/D2_1_elasticity_6sit.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Insight ───────────────────────────────────────────────────────────────────
print("=" * 60)
print("  D2.1 — INSIGHT")
print("=" * 60)
n_total = len(df_d2)
for sit, cnt in sit_counts.items():
    print(f"  {SIT_LABELS.get(sit,sit):<40} : {cnt:>3} ({cnt/n_total*100:.1f}%)")
print()
print("  Key findings :")
sit1 = sit_counts.get("1-Trade-led", 0)
sit2 = sit_counts.get("2-Endogenous", 0)
sit4 = sit_counts.get("4-Decoupling", 0)
print(f"  · {sit1} countries Trade-led → trade growing faster than GDP")
print(f"    → Most common situation : global trade integration deepening")
print(f"  · {sit2} countries Endogenous → economies growing faster than trade")
print(f"    → Domestic demand driving growth more than external trade")
print(f"  · {sit4} countries Decoupling → GDP grows but trade declines")
print(f"    → Structural loss of competitiveness or trade reorientation")
print()
print("  Selected countries situations :")
print(df_d2[df_d2["country_name"].isin(countries)][
    ["country_name","cagr_export","cagr_gdp","trade_elasticity_x","situation"]
].sort_values("trade_elasticity_x", ascending=False).to_string(index=False))
````

````python
# ── Trade Elasticity value vs GDP CAGR ───────────────────────────────────────
# Clip extreme elasticity values for readability (keep (+/-) situations only)
df_d22 = df_d2[df_d2["situation"].isin(["1-Trade-led","2-Endogenous"])].copy()
df_d22["elasticity_clipped"] = df_d22["trade_elasticity_x"].clip(0, 5)

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Trade Elasticity Value × GDP CAGR\n"
             "(Positive situations only : Trade-led & Endogenous)",
             fontsize=13, color="white", y=1.01)

# ── Left : Scatter Elasticity vs CAGR_GDP ────────────────────────────────────
for sit in ["1-Trade-led","2-Endogenous"]:
    sub = df_d22[df_d22["situation"] == sit]
    axes[0].scatter(sub["cagr_gdp"]*100, sub["elasticity_clipped"],
                    color=SIT_COLORS[sit], alpha=0.7, s=60,
                    label=SIT_LABELS[sit])

axes[0].axhline(1, color="white", linestyle="--", alpha=0.5,
                label="Elasticity = 1")
axes[0].axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_d22[df_d22["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row["cagr_gdp"]*100, row["elasticity_clipped"]),
                     fontsize=8, color="white",
                     xytext=(5, 3), textcoords="offset points")

axes[0].set_xlabel("CAGR GDP (%)", color="white")
axes[0].set_ylabel("Trade Elasticity (clipped 0-5)", color="white")
axes[0].set_title("Trade Elasticity vs GDP CAGR\n"
                  "(above 1 = trade-led, below 1 = endogenous)",
                  color="white")
axes[0].legend(facecolor="#16213e", labelcolor="white", fontsize=8)
r, p = spearmanr(df_d22["cagr_gdp"], df_d22["elasticity_clipped"])
axes[0].text(0.05, 0.95, f"Spearman ρ={r:.2f} (p={p:.3f})",
             transform=axes[0].transAxes, color="white", fontsize=9)

# ── Right : Boxplot Elasticity by CAGR_GDP Quartile ──────────────────────────
df_d22["gdp_cagr_q"] = pd.qcut(df_d22["cagr_gdp"], 4,
                                 labels=["Q1\n(Slowest)", "Q2",
                                         "Q3", "Q4\n(Fastest)"])
labels_q = ["Q1\n(Slowest)","Q2","Q3","Q4\n(Fastest)"]
data_box  = [df_d22[df_d22["gdp_cagr_q"]==q]["elasticity_clipped"].dropna().values
             for q in labels_q]
bp = axes[1].boxplot(data_box, patch_artist=True,
                     medianprops=dict(color="white", linewidth=2))
for patch, col in zip(bp["boxes"],
                      [C["red"], C["orange"], C["blue"], C["green"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)
axes[1].axhline(1, color="white", linestyle="--", alpha=0.5,
                label="Elasticity = 1")
axes[1].set_xticklabels(labels_q, color="white")
axes[1].set_title("Trade Elasticity by GDP CAGR Quartile\n"
                  "(do faster economies open more?)", color="white")
axes[1].set_ylabel("Trade Elasticity", color="white")
axes[1].legend(facecolor="#16213e", labelcolor="white", fontsize=8)

for i, q in enumerate(labels_q):
    vals = df_d22[df_d22["gdp_cagr_q"]==q]["elasticity_clipped"].dropna()
    if not vals.empty:
        axes[1].text(i+1, vals.median()+0.05, f"{vals.median():.2f}",
                     ha="center", color="white", fontsize=9, fontweight="bold")

plt.tight_layout()
plt.savefig("../Figures/link/D2_2_elasticity_vs_gdpcagr.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Insight ───────────────────────────────────────────────────────────────────
print("=" * 60)
print("  D2.2 — INSIGHT")
print("=" * 60)
print(f"  Spearman ρ = {r:.2f} between Trade Elasticity and GDP CAGR")
print()
print("  Medians by GDP CAGR quartile :")
for q in labels_q:
    vals = df_d22[df_d22["gdp_cagr_q"]==q]["elasticity_clipped"].dropna()
    print(f"  {q.replace(chr(10),' '):<20} median elasticity = {vals.median():.2f}")
print()
print("  Key findings :")
print("  · Elasticity does NOT systematically increase with GDP growth rate")
print("  · Fast-growing economies can be either trade-led OR endogenous")
print("  · The relationship between growth speed and trade openness is weak")
print("  → GDP growth rate alone does not predict trade elasticity")
````

## D3. Per Capita Trade × GDP per Capita
</div>

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### D3.1 — Do richer countries trade more per person?
</div>

````python
gdp_pc_col = "GDP per capita (current US$)"
df_d3 = df_master[["country_name","export_pc","import_pc",
                    gdp_pc_col,"size_profile"]].dropna()

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Per Capita Trade × GDP per Capita",
             fontsize=14, color="white", y=1.01)

# ── Left : Scatter Export_pc vs GDP_pc (log-log) ─────────────────────────────
for sp, col in SIZE_COLORS.items():
    sub = df_d3[df_d3["size_profile"] == sp]
    axes[0].scatter(sub[gdp_pc_col], sub["export_pc"],
                    color=col, alpha=0.5, s=40, label=f"Export {sp}")
    axes[0].scatter(sub[gdp_pc_col], sub["import_pc"],
                    color=col, alpha=0.25, s=25, marker="^")

# Regression line Export_pc
log_x  = np.log(df_d3[gdp_pc_col].clip(lower=1))
log_ye = np.log(df_d3["export_pc"].clip(lower=1))
log_ym = np.log(df_d3["import_pc"].clip(lower=1))
slope_e, int_e, re, _, _ = linregress(log_x, log_ye)
slope_m, int_m, rm, _, _ = linregress(log_x, log_ym)
x_line = np.linspace(log_x.min(), log_x.max(), 100)
axes[0].plot(np.exp(x_line), np.exp(int_e + slope_e * x_line),
             color=C["blue"], linewidth=2, linestyle="--",
             label=f"Export trend (ρ={re:.2f})")
axes[0].plot(np.exp(x_line), np.exp(int_m + slope_m * x_line),
             color=C["orange"], linewidth=2, linestyle=":",
             label=f"Import trend (ρ={rm:.2f})")

for _, row in df_d3[df_d3["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row[gdp_pc_col], row["export_pc"]),
                     fontsize=7, color="white",
                     xytext=(4, 2), textcoords="offset points")

axes[0].set_xscale("log"); axes[0].set_yscale("log")
axes[0].set_xlabel("GDP per Capita (log)", color="white")
axes[0].set_ylabel("Trade per Capita (log)", color="white")
axes[0].set_title("Export & Import per Capita vs GDP per Capita\n"
                  "(circles=export, triangles=import)", color="white")
axes[0].legend(facecolor="#16213e", labelcolor="white", fontsize=7)

# ── Right : Grouped Boxplot by GDP_pc Quartile ───────────────────────────────
df_d3["gdp_pc_q"] = pd.qcut(df_d3[gdp_pc_col], 4,
                              labels=["Q1\n(Poorest)","Q2","Q3","Q4\n(Richest)"])
labels_q = ["Q1\n(Poorest)","Q2","Q3","Q4\n(Richest)"]
x_pos    = np.arange(len(labels_q))
width    = 0.35

data_x = [df_d3[df_d3["gdp_pc_q"]==q]["export_pc"].dropna().values for q in labels_q]
data_m = [df_d3[df_d3["gdp_pc_q"]==q]["import_pc"].dropna().values for q in labels_q]

bp_x = axes[1].boxplot(data_x, positions=x_pos-width/2, widths=width,
                        patch_artist=True,
                        medianprops=dict(color="white", linewidth=2))
bp_m = axes[1].boxplot(data_m, positions=x_pos+width/2, widths=width,
                        patch_artist=True,
                        medianprops=dict(color="white", linewidth=2))
for patch in bp_x["boxes"]:
    patch.set_facecolor(C["blue"]); patch.set_alpha(0.7)
for patch in bp_m["boxes"]:
    patch.set_facecolor(C["orange"]); patch.set_alpha(0.7)

axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(labels_q, color="white")
axes[1].set_yscale("log")
axes[1].set_ylabel("Trade per Capita US$ (log)", color="white")
axes[1].set_title("Export & Import per Capita by GDP_pc Quartile",
                  color="white")
patches = [mpatches.Patch(color=C["blue"],   alpha=0.7, label="Export per Capita"),
           mpatches.Patch(color=C["orange"], alpha=0.7, label="Import per Capita")]
axes[1].legend(handles=patches, facecolor="#16213e", labelcolor="white", fontsize=9)

# Annotate medians
for i, q in enumerate(labels_q):
    me = df_d3[df_d3["gdp_pc_q"]==q]["export_pc"].median()
    mm = df_d3[df_d3["gdp_pc_q"]==q]["import_pc"].median()
    axes[1].text(i-width/2, me*1.3, f"{me:,.0f}",
                 ha="center", color="white", fontsize=7)
    axes[1].text(i+width/2, mm*1.3, f"{mm:,.0f}",
                 ha="center", color="white", fontsize=7)

r_x, _ = spearmanr(log_x, log_ye)
r_m, _ = spearmanr(log_x, log_ym)
axes[1].text(0.05, 0.95,
             f"Export ρ={r_x:.2f}  Import ρ={r_m:.2f}",
             transform=axes[1].transAxes, color="white", fontsize=9)

plt.tight_layout()
plt.savefig("../Figures/link/D3_1_percapita_gdppc.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Insight ───────────────────────────────────────────────────────────────────
print("=" * 60)
print("  D3.1 — INSIGHT")
print("=" * 60)
print(f"  Spearman ρ Export_pc vs GDP_pc = {r_x:.2f}")
print(f"  Spearman ρ Import_pc vs GDP_pc = {r_m:.2f}")
print()
print("  Key findings :")
print("  · STRONG positive relationship — much stronger than Trade Intensity (ρ=0.29)")
print("  · Import_pc correlates even more strongly than Export_pc with GDP_pc")
print("    → Richer populations consume more imported goods")
print("  · Log-log linearity → constant elasticity relationship :")
print(f"    Export: 1% increase in GDP_pc → {slope_e:.2f}% increase in Export_pc")
print(f"    Import: 1% increase in GDP_pc → {slope_m:.2f}% increase in Import_pc")
print()
print("  Contrast with D1 :")
print("  · GDP_pc strongly predicts trade PER PERSON (ρ≈0.82-0.93)")
print("  · GDP_pc weakly predicts trade AS SHARE OF GDP (ρ=0.29)")
print("  → Richer countries trade more in absolute terms per person,")
print("    but their economies grow proportionally even faster")
````

# D. Combined Metrics
</div>

Three structural dimensions linking trade data to economic fundamentals :

| Section | Metric | Formula | Question |
|---|---|---|---|
| **D1** | Trade Intensity | (X+M) / GDP | How open is the economy? |
| **D2** | Trade Elasticity | CAGR_export / CAGR_GDP | Does trade grow faster than the economy? |
| **D3** | Per Capita Trade | X / Population, M / Population | How much trade per person? |

All metrics derived from `df_master`. GDP and Population from the indicators dataset directly.

## D2. Trade Elasticity — 6 Situations
</div>

````python
def classify_6sit(row):
    xe = row.get("cagr_export", np.nan)
    xg = row.get("cagr_gdp",    np.nan)
    if pd.isna(xe) or pd.isna(xg):
        return "Unknown"
    if xe >= 0 and xg >= 0:
        return "1 — Trade-led"      if xe > xg else "2 — Endogenous"
    elif xe >= 0 and xg < 0:
        return "3 — Trade Resistant"
    elif xe < 0 and xg >= 0:
        return "4 — Decoupling"
    else:
        return "5 — Double Decline (trade faster)" if abs(xe) > abs(xg)                else "6 — Double Decline (GDP faster)"

SIT_COLORS = {
    "1 — Trade-led"                          : C["green"],
    "2 — Endogenous"                         : C["teal"],
    "3 — Trade Resistant"                    : C["blue"],
    "4 — Decoupling"                         : C["orange"],
    "5 — Double Decline (trade faster)"      : C["red"],
    "6 — Double Decline (GDP faster)"        : C["purple"],
    "Unknown"                                : C["grey"],
}

df_d2 = df_master[["country_name", "cagr_export", "cagr_gdp",
                    "trade_elasticity_x", "size_profile"]].dropna(
    subset=["cagr_export", "cagr_gdp"])
df_d2["situation"] = df_d2.apply(classify_6sit, axis=1)

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Trade Elasticity — 6 Situations (CAGR Export vs CAGR GDP)",
             fontsize=14, color="white", y=1.01)

# ── Left : Scatter colored by situation ──────────────────────────────────────
for sit, col in SIT_COLORS.items():
    sub = df_d2[df_d2["situation"] == sit]
    if sub.empty: continue
    axes[0].scatter(sub["cagr_gdp"] * 100, sub["cagr_export"] * 100,
                    color=col, alpha=0.7, s=60,
                    label=sit.split(" — ", 1)[1] if " — " in sit else sit)

lim = (df_d2[["cagr_export", "cagr_gdp"]].abs().max().max()) * 100 * 1.1
axes[0].plot([-lim, lim], [-lim, lim], color="white", linestyle="--",
             alpha=0.4, label="Elasticity = 1")
axes[0].axhline(0, color=C["grey"], linestyle=":", alpha=0.4)
axes[0].axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_d2[df_d2["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row["cagr_gdp"] * 100, row["cagr_export"] * 100),
                     fontsize=8, color="white",
                     xytext=(5, 3), textcoords="offset points")

axes[0].set_xlabel("CAGR GDP (%)", color="white")
axes[0].set_ylabel("CAGR Export (%)", color="white")
axes[0].set_title("CAGR Export vs CAGR GDP\n(color = situation)", color="white")
axes[0].legend(facecolor="#16213e", labelcolor="white", fontsize=7)

# ── Right : Country count by situation ───────────────────────────────────────
sit_counts = df_d2["situation"].value_counts()
sit_labels = [s.split(" — ", 1)[1] if " — " in s else s for s in sit_counts.index]
sit_clrs   = [SIT_COLORS.get(s, C["grey"]) for s in sit_counts.index]
axes[1].barh(sit_labels, sit_counts.values, color=sit_clrs, alpha=0.85)
axes[1].set_title("Country Count by Situation", color="white")
axes[1].set_xlabel("Number of countries", color="white")
for i, (v, lbl) in enumerate(zip(sit_counts.values, sit_labels)):
    axes[1].text(v + 0.5, i, str(v), va="center", color="white", fontsize=9)

plt.tight_layout()
plt.savefig("../Figures/link/D2_elasticity_6sit.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Summary ───────────────────────────────────────────────────────────────────
print("Situation Distribution:")
for sit, count in sit_counts.items():
    pct = count / len(df_d2) * 100
    print(f"  {sit:<45} : {count:>3} ({pct:.1f}%)")
print(f"\nSelected countries:")
print(df_d2[df_d2["country_name"].isin(countries)][
    ["country_name", "cagr_export", "cagr_gdp", "trade_elasticity_x", "situation"]
].sort_values("trade_elasticity_x", ascending=False).to_string(index=False))
````

## D2. Trade Elasticity — 6 Situations
</div>

**Trade Elasticity** = CAGR_export / CAGR_GDP

Classifies each country into 6 structural situations based on the **sign** of each CAGR,  
then the **magnitude** of the ratio when both are same-signed.

| # | CAGR_export | CAGR_GDP | Elasticity | Reading |
|---|---|---|---|---|
| 1 | + | + | > 1 | Trade-led : trade grows faster → opening |
| 2 | + | + | 0→1 | Endogenous : economy grows faster than trade |
| 3 | + | − | < 0 | Trade Resistant : trade grows despite GDP decline |
| 4 | − | + | < 0 | Decoupling : GDP grows but trade declines |
| 5 | − | − | > 1 | Double Decline : trade collapses faster than GDP |
| 6 | − | − | 0→1 | Double Decline : GDP more affected than trade |

**Key distinction** : situations 5 and 6 both have negative elasticity from (-/-) — the ratio alone  
is misleading. Sign-based classification is required before using the elasticity value.


````python
def classify_6sit(row):
    xe,xg = row.get("cagr_export",np.nan),row.get("cagr_gdp",np.nan)
    if pd.isna(xe) or pd.isna(xg): return "Unknown"
    if xe>=0 and xg>=0: return "1-Trade-led" if xe>xg else "2-Endogenous"
    elif xe>=0 and xg<0: return "3-Trade Resistant"
    elif xe<0 and xg>=0: return "4-Decoupling"
    else: return "5-Double Decline (trade faster)" if abs(xe)>abs(xg) else "6-Double Decline (econ faster)"

SIT_COLORS = {"1-Trade-led":C["green"],"2-Endogenous":C["teal"],
              "3-Trade Resistant":C["blue"],"4-Decoupling":C["orange"],
              "5-Double Decline (trade faster)":C["red"],
              "6-Double Decline (econ faster)":C["purple"]}

df_d2 = df_master[["country_name","cagr_export","cagr_gdp","size_profile"]].dropna()
df_d2["situation"] = df_d2.apply(classify_6sit,axis=1)

fig,axes = styled_subplots(1,2,figsize=(18,7))
fig.suptitle("Trade Elasticity — 6 Situations",fontsize=14,color="white",y=1.01)

for sit,col in SIT_COLORS.items():
    sub = df_d2[df_d2["situation"]==sit]
    axes[0].scatter(sub["cagr_gdp"]*100,sub["cagr_export"]*100,
                    color=col,alpha=0.7,s=60,label=sit.split("-",1)[1])
lim = (df_d2[["cagr_export","cagr_gdp"]].abs().max().max())*100*1.1
axes[0].plot([-lim,lim],[-lim,lim],color="white",linestyle="--",alpha=0.4)
axes[0].axhline(0,color=C["grey"],linestyle=":",alpha=0.4)
axes[0].axvline(0,color=C["grey"],linestyle=":",alpha=0.4)
for _,row in df_d2[df_d2["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],(row["cagr_gdp"]*100,row["cagr_export"]*100),
                     fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
axes[0].set_xlabel("CAGR GDP (%)",color="white")
axes[0].set_ylabel("CAGR Export (%)",color="white")
axes[0].set_title("CAGR Export vs CAGR GDP (6 situations)",color="white")
axes[0].legend(facecolor="#16213e",labelcolor="white",fontsize=7)

sit_counts = df_d2["situation"].value_counts()
axes[1].barh([s.split("-",1)[1] for s in sit_counts.index],
             sit_counts.values,
             color=[SIT_COLORS.get(s,C["grey"]) for s in sit_counts.index],alpha=0.8)
axes[1].set_title("Country Distribution by Situation",color="white")
axes[1].set_xlabel("Number of countries",color="white")
plt.tight_layout()
plt.savefig("../Figures/link/D2_elasticity_6sit.png",dpi=150,bbox_inches="tight")
plt.show()
print("\nSituation distribution:")
print(df_d2["situation"].value_counts().to_string())
````

## D3. Export & Import per Capita
</div>

**Export per Capita** and **Import per Capita** are part of the Trade Intensity family —  
they normalize trade volumes by population rather than GDP.

They answer a different question from trade_intensity (which normalizes by GDP) :  
*"How much does each person in this country trade?"* rather than  
*"How large is trade relative to the economy?"*

**Key insight** : per capita normalization **reclassifies** countries vs absolute volume ranking.  
A country like China ranks #1 in export volume but much lower in export per capita.  
Small, wealthy, open economies (Luxembourg, Singapore) dominate the per capita ranking.


````python
df_d3 = df_master[[
    "country_name", "total_export", "total_import",
    "export_pc", "import_pc", gdp_pc_col, "size_profile"
]].dropna(subset=["export_pc", "import_pc"])

# Rank by volume vs per capita
df_d3["rank_x_vol"] = df_d3["total_export"].rank(ascending=False).astype(int)
df_d3["rank_x_pc"]  = df_d3["export_pc"].rank(ascending=False).astype(int)
df_d3["rank_m_vol"] = df_d3["total_import"].rank(ascending=False).astype(int)
df_d3["rank_m_pc"]  = df_d3["import_pc"].rank(ascending=False).astype(int)
df_d3["rank_delta_x"] = df_d3["rank_x_vol"] - df_d3["rank_x_pc"]  # + = climbs in pc ranking
df_d3["rank_delta_m"] = df_d3["rank_m_vol"] - df_d3["rank_m_pc"]

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Per Capita Trade — Reclassification vs Volume Ranking",
             fontsize=14, color="white", y=1.01)

# ── Left : Export per Capita by GDP_pc Quartile (boxplot) ────────────────────
df_d3["gdp_pc_q"] = pd.qcut(
    df_d3[gdp_pc_col], 4,
    labels=["Q1\n(Poorest)", "Q2", "Q3", "Q4\n(Richest)"]
)
labels_q = ["Q1\n(Poorest)", "Q2", "Q3", "Q4\n(Richest)"]

data_box_x = [df_d3[df_d3["gdp_pc_q"] == q]["export_pc"].dropna().values for q in labels_q]
data_box_m = [df_d3[df_d3["gdp_pc_q"] == q]["import_pc"].dropna().values for q in labels_q]

x_pos = np.arange(len(labels_q))
width = 0.35

# Export pc
bp_x = axes[0].boxplot(data_box_x, positions=x_pos - width/2, widths=width,
                        patch_artist=True, medianprops=dict(color="white", linewidth=2))
for patch in bp_x["boxes"]:
    patch.set_facecolor(C["blue"]); patch.set_alpha(0.7)

# Import pc
bp_m = axes[0].boxplot(data_box_m, positions=x_pos + width/2, widths=width,
                        patch_artist=True, medianprops=dict(color="white", linewidth=2))
for patch in bp_m["boxes"]:
    patch.set_facecolor(C["orange"]); patch.set_alpha(0.7)

axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(labels_q, color="white")
axes[0].set_title("Export & Import per Capita by GDP_pc Quartile\n"
                  "(richer countries trade more per person)", color="white")
axes[0].set_ylabel("Trade per Capita (US$)", color="white")
axes[0].set_yscale("log")
patches = [mpatches.Patch(color=C["blue"], alpha=0.7, label="Export per Capita"),
           mpatches.Patch(color=C["orange"], alpha=0.7, label="Import per Capita")]
axes[0].legend(handles=patches, facecolor="#16213e", labelcolor="white", fontsize=9)

r_x, p_x = spearmanr(np.log(df_d3[gdp_pc_col].clip(lower=1)),
                       np.log(df_d3["export_pc"].clip(lower=1)))
r_m, p_m = spearmanr(np.log(df_d3[gdp_pc_col].clip(lower=1)),
                       np.log(df_d3["import_pc"].clip(lower=1)))
axes[0].text(0.05, 0.95,
             f"Export ρ={r_x:.2f}  Import ρ={r_m:.2f}",
             transform=axes[0].transAxes, color="white", fontsize=9)

# ── Right : Top rank movers Export Volume → Per Capita ───────────────────────
top_movers = (
    df_d3.reindex(df_d3["rank_delta_x"].abs().sort_values(ascending=False).index)
    .head(25)
)
colors_mv = [C["green"] if v > 0 else C["red"] for v in top_movers["rank_delta_x"]]
axes[1].barh(top_movers["country_name"], top_movers["rank_delta_x"],
             color=colors_mv, alpha=0.8)
axes[1].axvline(0, color="white", linewidth=1)
axes[1].set_title("Rank Movers : Export Volume → Per Capita\n"
                  "(green = climbs, red = drops)", color="white")
axes[1].set_xlabel("Rank Change (positions)", color="white")

plt.tight_layout()
plt.savefig("../Figures/link/D3_per_capita_trade.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Key movers ────────────────────────────────────────────────────────────────
print("Selected countries — Volume rank vs Per Capita rank (Export):")
print(df_d3[df_d3["country_name"].isin(countries)][
    ["country_name", "rank_x_vol", "rank_x_pc", "rank_delta_x"]
].sort_values("rank_delta_x").to_string(index=False))
````

# J. RF + OLS — All Indicators × All Trade Variables
</div>

RF and OLS run on **all 16 indicators simultaneously**.  
Results are stored in `top_indicators_per_var` — used automatically in E/F/G/H.  
No manual indicator selection.


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J1. Random Forest — All Trade Variables × All Indicators
</div>

````python
all_rf = {}

print(f"Running RF for {len(all_vars)} trade targets × {len(ind_vars)} indicators")
print("="*60)

for target in all_vars:
    df_rf = df_master[[target]+ind_vars].dropna()
    if len(df_rf)<30:
        print(f"  ⚠️  {ALL_TRADE_LABELS.get(target,target):<35} n={len(df_rf)} — skip")
        continue
    X = df_rf[ind_vars].values; y = df_rf[target].values
    rf = RandomForestRegressor(n_estimators=300,random_state=42,max_features="sqrt")
    rf.fit(X,y)
    fi = pd.DataFrame({"indicator":ind_vars,"importance":rf.feature_importances_})
    fi = fi.sort_values("importance",ascending=False)
    r2 = rf.score(X,y)
    all_rf[target] = {"fi":fi,"r2":r2,"label":ALL_TRADE_LABELS.get(target,target),"n":len(df_rf)}
    top = fi.iloc[0]
    print(f"  ✅ {ALL_TRADE_LABELS.get(target,target):<35} R²={r2:.3f}  n={len(df_rf)}  Top: {IND_LABELS.get(top['indicator'],top['indicator'])}")

# ── Extract top 3 indicators per trade variable ───────────────────────────────
top_indicators_per_var = {
    target: results["fi"].head(3)["indicator"].tolist()
    for target,results in all_rf.items()
}

print(f"\n✅ top_indicators_per_var built — guides E/F/G/H automatically")
````

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J3. OLS — Direction & Magnitude (top 5 per target)
</div>

````python
print("OLS Regression — Standardized Coefficients (▲ positive / ▼ negative)")
print("="*65)

ols_summary = []
for target,results in all_rf.items():
    top5 = results["fi"].head(5)["indicator"].tolist()
    df_ols = df_master[[target]+top5].dropna()
    if len(df_ols)<25: continue
    X_s = StandardScaler().fit_transform(df_ols[top5].values)
    y_s = df_ols[target].values
    lr = LinearRegression().fit(X_s,y_s)
    r2 = lr.score(X_s,y_s)
    print(f"\n  {results['label']} (RF R²={results['r2']:.3f} | OLS R²={r2:.3f})")
    print(f"  {'─'*55}")
    for ind_name,coef in sorted(zip(top5,lr.coef_),key=lambda x:abs(x[1]),reverse=True):
        direction = "▲" if coef>0 else "▼"
        label = IND_LABELS.get(ind_name,ind_name)
        print(f"    {direction} {coef:+.4f}  {label}")
    ols_summary.append({"target":results["label"],"r2_rf":results["r2"],"r2_ols":r2})
````

## I3. Multiple Regression — All Trade Vars ~ All 16 Indicators
</div>

````python
coef_matrix = []
print("Multiple Regression (OLS) — All trade vars ~ all 16 indicators")
print("="*65)

for target in all_vars:
    df_reg = df_master[[target]+ind_vars].dropna()
    if len(df_reg)<30: continue
    X_s = StandardScaler().fit_transform(df_reg[ind_vars].values)
    y_s = df_reg[target].values
    lr = LinearRegression().fit(X_s,y_s)
    r2 = lr.score(X_s,y_s)
    label = ALL_TRADE_LABELS.get(target,target)
    coef_matrix.append(dict(zip(["target"]+[IND_LABELS.get(i,i) for i in ind_vars],
                                 [label]+list(lr.coef_))))
    top_coef = sorted(zip(ind_vars,lr.coef_),key=lambda x:abs(x[1]),reverse=True)[0]
    d = "▲" if top_coef[1]>0 else "▼"
    print(f"  {label:<35} R²={r2:.3f}  Top: {d} {IND_LABELS.get(top_coef[0],top_coef[0])}")

df_coef = pd.DataFrame(coef_matrix).set_index("target")
fig,ax = plt.subplots(figsize=(max(12,len(df_coef.columns)*1.3),max(6,len(df_coef)*0.6)))
fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
sns.heatmap(df_coef,ax=ax,cmap="RdYlGn",center=0,annot=True,fmt=".2f",
            annot_kws={"size":7},linewidths=0.5,linecolor="#333")
ax.set_title("Multiple Regression Coefficients (standardized)\nAll Trade Vars ~ All 16 Indicators",
             color="white",pad=10)
ax.tick_params(colors="white",labelsize=8)
plt.xticks(rotation=35,ha="right"); plt.tight_layout()
plt.savefig("Figures/I3_multiple_regression.png",dpi=150,bbox_inches="tight")
plt.show()
````

## I4. PCA Croisé — Latent Dimensions Trade + Economy
</div>

````python
pca_all_cols = [v for v in all_vars if v in df_master.columns] + ind_vars
df_pca = df_master[["country_name"]+pca_all_cols].dropna()
X_pca  = StandardScaler().fit_transform(df_pca[pca_all_cols].values)

pca = PCA(n_components=min(6,len(pca_all_cols)))
pca.fit(X_pca)
expl = pca.explained_variance_ratio_

print(f"✅ PCA : {len(df_pca)} countries × {len(pca_all_cols)} variables")
print(f"   PC1={expl[0]:.1%}  PC2={expl[1]:.1%}  PC3={expl[2]:.1%}  Cumul(3)={expl[:3].sum():.1%}")

# Loadings
loadings = pd.DataFrame(pca.components_[:2].T,index=pca_all_cols,columns=["PC1","PC2"])
trade_set = set(all_vars)
loadings["category"] = ["trade" if v in trade_set else "indicator" for v in pca_all_cols]
loadings.index = [ALL_TRADE_LABELS.get(v,IND_LABELS.get(v,v)) for v in pca_all_cols]
loadings = loadings.sort_values("PC1",ascending=False)

fig,axes = styled_subplots(1,2,figsize=(18,max(7,len(loadings)*0.35)))
fig.suptitle(f"PCA Loadings — PC1 ({expl[0]:.1%}) & PC2 ({expl[1]:.1%})",
             fontsize=13,color="white",y=1.01)
for i,pc in enumerate(["PC1","PC2"]):
    cols_bar = [C["blue"] if c=="trade" else C["orange"] for c in loadings["category"]]
    axes[i].barh(loadings.index,loadings[pc],color=cols_bar,alpha=0.8)
    axes[i].axvline(0,color="white",linewidth=1)
    axes[i].set_title(f"{pc}\n(blue=trade, orange=indicator)",color="white")
    axes[i].set_xlabel("Loading",color="white")
    axes[i].tick_params(colors="white",labelsize=7)
    for sp in axes[i].spines.values(): sp.set_edgecolor("#444")
    axes[i].grid(alpha=0.2,color="#555")
plt.tight_layout()
plt.savefig("Figures/I4_pca_loadings.png",dpi=150,bbox_inches="tight")
plt.show()

# Country scatter PC1 vs PC2
scores_pca = pd.DataFrame(pca.transform(X_pca)[:,:2],columns=["PC1","PC2"])
scores_pca["country_name"] = df_pca["country_name"].values
scores_pca = scores_pca.merge(df_master[["country_name","size_profile"]],on="country_name",how="left")

fig,ax = styled_figure(figsize=(13,8))
ax.set_title(f"Country Positioning — PC1 vs PC2",color="white")
for sp,col in SIZE_COLORS.items():
    sub = scores_pca[scores_pca["size_profile"]==sp]
    ax.scatter(sub["PC1"],sub["PC2"],color=col,alpha=0.5,s=40,label=sp)
for _,row in scores_pca[scores_pca["country_name"].isin(countries)].iterrows():
    ax.scatter(row["PC1"],row["PC2"],color=C["orange"],s=120,zorder=5,edgecolors="white",linewidths=1)
    ax.annotate(row["country_name"],(row["PC1"],row["PC2"]),fontsize=9,color="white",xytext=(5,3),textcoords="offset points")
ax.axhline(0,color=C["grey"],linestyle="--",alpha=0.4)
ax.axvline(0,color=C["grey"],linestyle="--",alpha=0.4)
ax.legend(facecolor="#16213e",labelcolor="white",fontsize=8)
plt.tight_layout()
plt.savefig("Figures/I4_pca_scatter.png",dpi=150,bbox_inches="tight")
plt.show()
````

## I5. Temporal Stability of Correlations (3 Sub-Periods)
</div>

````python
PERIODS = {"2015–2019":(2015,2019),"2020–2021":(2020,2021),"2022–2024":(2022,2024)}
df_annual_full = df_trade.merge(df_ind_annual,on=["country_name","year"],how="inner")

# Key pairs: use top RF pairs
KEY_PAIRS = [(tv,iv)
             for tv,top in top_indicators_per_var.items()
             for iv in top[:1]  # top 1 indicator per trade var
             if tv in df_annual_full.columns and iv in df_annual_full.columns][:12]

stability_rows = []
for period,(y1,y2) in PERIODS.items():
    df_p = df_annual_full[(df_annual_full["year"]>=y1)&(df_annual_full["year"]<=y2)]
    for tv,iv in KEY_PAIRS:
        sub = df_p[[tv,iv]].dropna()
        if len(sub)<10: continue
        r,p = spearmanr(sub[tv],sub[iv])
        stability_rows.append({"period":period,"trade_var":tv,"indicator":iv,"rho":round(r,3),"sig":p<0.05})

df_stab = pd.DataFrame(stability_rows)
if not df_stab.empty:
    df_stab["pair"] = (df_stab["trade_var"].map(ALL_TRADE_LABELS).fillna(df_stab["trade_var"])
                       +" × "+df_stab["indicator"].map(IND_LABELS).fillna(df_stab["indicator"]))
    pivot_stab = df_stab.pivot_table(index="pair",columns="period",values="rho",aggfunc="first")
    pivot_stab = pivot_stab[list(PERIODS.keys())]
    fig,ax = plt.subplots(figsize=(10,max(4,len(pivot_stab)*0.55)))
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot_stab,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1,
                annot=True,fmt=".2f",annot_kws={"size":9},linewidths=0.5,linecolor="#333")
    ax.set_title("Temporal Stability — Spearman ρ by Sub-Period",color="white",pad=10)
    ax.tick_params(colors="white",labelsize=8)
    plt.tight_layout()
    plt.savefig("Figures/I5_temporal_stability.png",dpi=150,bbox_inches="tight")
    plt.show()
    print("\n💡 Stable ρ across periods = structural relationship")
    print("   Volatile ρ = conjunctural (COVID or post-COVID driven)")
````

## I6. Heterogeneity by Size Group
</div>

````python
group_rows = []
for group in ["Leader","Major","Mid","Small"]:
    df_g = df_master[df_master["size_profile"]==group]
    if len(df_g)<10: continue
    for tv in data_vars:
        for iv in ind_vars:
            sub = df_g[[tv,iv]].dropna()
            if len(sub)<12: continue
            r,p = spearmanr(sub[tv],sub[iv])
            group_rows.append({"group":group,"trade_var":tv,"indicator":iv,
                                "rho":round(r,3),"sig":p<0.05,"n":len(sub)})

df_group = pd.DataFrame(group_rows)

# Key export vars × all indicators, compare across groups
if not df_group.empty:
    for tv in ["export_share_structural","cagr_export","volatility_export","negfreq_export"]:
        sub = df_group[df_group["trade_var"]==tv].copy()
        if sub.empty: continue
        sub["ind_label"] = sub["indicator"].map(IND_LABELS).fillna(sub["indicator"])
        pivot_g = sub.pivot_table(index="ind_label",columns="group",values="rho",aggfunc="first")
        if pivot_g.empty: continue
        fig,ax = plt.subplots(figsize=(9,max(4,len(pivot_g)*0.5)))
        fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
        sns.heatmap(pivot_g,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1,
                    annot=True,fmt=".2f",annot_kws={"size":8},linewidths=0.5,linecolor="#333")
        ax.set_title(f"Heterogeneity — {ALL_TRADE_LABELS.get(tv,tv)}\nSpearman ρ by Size Group",
                     color="white",pad=10)
        ax.tick_params(colors="white",labelsize=8)
        plt.tight_layout()
        plt.savefig(f"Figures/I6_{tv[:15]}.png",dpi=150,bbox_inches="tight")
        plt.show()
````

# O. Conclusions — Direction 1
</div>

### Design Principles Applied

| Rule | Implementation |
|---|---|
| All 16 indicators | Dynamic loading from `df` — no exclusion |
| No manual selection | J (RF) → `top_indicators_per_var` → guides E/F/G/H |
| Complete mapping | `IND_LABELS` : all 16 full names → short readable labels |
| Clip rules | `IND_CLIP` : per-indicator clip for scatter readability |
| Log scale | `IND_LOG` : set of indicators needing log x-axis |
| Population direct | From `Population, total` indicator — not GDP/GDP_pc derivation |
| FDI included | Used in E4 consistency analysis |
| No trade × trade | All figures : trade variable × economic indicator only |
| Boxplots for weak ρ | H section uses quartile boxplots instead of dispersed scatters |

### Structure Delivered
C → D → J → I.1/I.2 → E/F/G/H → I.3→I.7 → O

> *"Which trade variables predict future economic performance?"*  
> Lagged correlations, export-led growth testing, Granger causality.
