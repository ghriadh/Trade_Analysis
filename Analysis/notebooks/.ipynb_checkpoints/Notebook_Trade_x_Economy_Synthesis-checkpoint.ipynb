## 3. Extract GDP Annual Series

```python
# ── Extract GDP (current US$) from long indicators dataframe ──────────────────
GDP_INDICATOR = "GDP (current US$)"

df_gdp = (
    df_indicators[
        (df_indicators["entity_type"]   == "COUNTRY") &
        (df_indicators["indicator_name"] == GDP_INDICATOR)
    ][["country_name", "country_iso3", "year", "value"]]
    .rename(columns={"value": "gdp"})
    .copy()
)

print(f"✅ GDP series : {df_gdp.shape}")
print(f"   Countries  : {df_gdp['country_name'].nunique()}")
print(f"   Period     : {df_gdp['year'].min()} – {df_gdp['year'].max()}")
print(f"   Missing    : {df_gdp['gdp'].isna().sum()} / {len(df_gdp)}")
df_gdp.sample(5)
```

## 4. Quality-Controlled Merge (Trade × GDP)
</div>

```python
# ── Strategy : conservative merge with data quality flagging ─────────────────
#
# Rules:
#   R1 — country kept only if ≥6/10 valid years on GDP, X, M
#   R2 — internal gaps ≤2 consecutive years → linear interpolation
#   R3 — edge gaps ≤1 year → forward/backward fill
#   R4 — combined metric only computed where BOTH sources available same year
#   R5 — data_quality_flag : complete / interpolated / partial

def flag_quality(s, threshold=6):
    """Assign quality flag based on non-null count in a series of 10 values."""
    n_valid = s.notna().sum()
    n_total = len(s)
    if n_valid == n_total:
        return "complete"
    elif n_valid >= threshold:
        return "interpolated"
    else:
        return "partial"

def conservative_fill(s, max_gap=2, edge_fill=1):
    """Linear interpolation for internal gaps ≤ max_gap, ffill/bfill for edge ≤ edge_fill."""
    s_interp = s.interpolate(method="linear", limit=max_gap, limit_direction="both")
    s_interp = s_interp.ffill(limit=edge_fill).bfill(limit=edge_fill)
    return s_interp

# ── Merge trade annual + GDP ──────────────────────────────────────────────────
df_panel = df_trade.merge(
    df_gdp[["country_name", "year", "gdp"]],
    on=["country_name", "year"],
    how="left"
)

# ── Apply interpolation per country ──────────────────────────────────────────
df_panel = df_panel.sort_values(["country_name", "year"])
df_panel["gdp_raw"] = df_panel["gdp"].copy()

df_panel["gdp"] = (
    df_panel.groupby("country_name")["gdp"]
    .transform(conservative_fill)
)

# ── Quality flag ──────────────────────────────────────────────────────────────
quality_gdp = (
    df_panel.groupby("country_name")["gdp_raw"]
    .apply(flag_quality)
    .reset_index()
    .rename(columns={"gdp_raw": "data_quality_flag"})
)
quality_x = (
    df_panel.groupby("country_name")["X"]
    .apply(flag_quality)
    .reset_index()
    .rename(columns={"X": "data_quality_x"})
)

df_panel = df_panel.merge(quality_gdp, on="country_name", how="left")
df_panel = df_panel.merge(quality_x, on="country_name", how="left")

# ── Filter : keep only countries with ≥6 years valid on both GDP and X ────────
valid_countries = quality_gdp[quality_gdp["data_quality_flag"] != "partial"]["country_name"].tolist()
valid_x         = quality_x  [quality_x  ["data_quality_x"]   != "partial"]["country_name"].tolist()
valid_both      = set(valid_countries) & set(valid_x)

df_panel = df_panel[df_panel["country_name"].isin(valid_both)].copy()

print(f"✅ Panel shape     : {df_panel.shape}")
print(f"   Countries kept  : {df_panel['country_name'].nunique()} / {df_trade['country_name'].nunique()}")
print(f"   Quality flags   :")
print(df_panel.drop_duplicates("country_name")["data_quality_flag"].value_counts())
```

# D. Combined Metrics
</div>

Four structural ratios that anchor trade data to GDP :

| Metric | Formula | Interpretation |
|---|---|---|
| **Trade Intensity** | (X + M) / GDP | Degree of trade openness |
| **Export Dependency** | X / GDP | Export reliance relative to economy size |
| **Import Coverage** | M / GDP | Import weight relative to economy size |
| **Trade Balance Ratio** | (X − M) / GDP | Structural surplus/deficit as % of GDP |


<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 1. Compute Annual Ratios
</div>

```python
# ── Annual combined metrics ───────────────────────────────────────────────────
# Only computed where BOTH trade and GDP are non-null (Rule R4)

mask = df_panel["gdp"].notna() & (df_panel["gdp"] > 0) & df_panel["X"].notna() & df_panel["M"].notna()

df_panel.loc[mask, "trade_intensity"]    = (df_panel.loc[mask, "X"] + df_panel.loc[mask, "M"]) / df_panel.loc[mask, "gdp"]
df_panel.loc[mask, "export_dep"]         = df_panel.loc[mask, "X"] / df_panel.loc[mask, "gdp"]
df_panel.loc[mask, "import_dep"]         = df_panel.loc[mask, "M"] / df_panel.loc[mask, "gdp"]
df_panel.loc[mask, "trade_balance_gdp"]  = (df_panel.loc[mask, "X"] - df_panel.loc[mask, "M"]) / df_panel.loc[mask, "gdp"]

print("✅ Annual combined metrics computed")
print(f"   Coverage trade_intensity : {df_panel['trade_intensity'].notna().sum()} / {len(df_panel)} rows")
df_panel[["country_name", "year", "X", "M", "gdp", "trade_intensity", "export_dep", "trade_balance_gdp"]].sample(5)
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 2. Country-Level Aggregation
</div>

```python
# ── Aggregate to country level (median — robust to COVID outlier 2020) ────────
df_combined_metrics = (
    df_panel.groupby("country_name").agg(
        trade_intensity_med   = ("trade_intensity",   "median"),
        export_dep_med        = ("export_dep",         "median"),
        import_dep_med        = ("import_dep",         "median"),
        trade_balance_gdp_med = ("trade_balance_gdp",  "median"),
        trade_intensity_std   = ("trade_intensity",   "std"),
        trade_intensity_2024  = ("trade_intensity",   "last"),
        data_quality_flag     = ("data_quality_flag", "first"),
        n_years_gdp           = ("gdp",               lambda x: x.notna().sum()),
    ).reset_index()
)

# ── Z-score normalization ──────────────────────────────────────────────────────
for col in ["trade_intensity_med", "export_dep_med", "import_dep_med"]:
    df_combined_metrics[f"z_{col}"] = zscore(df_combined_metrics[col].fillna(df_combined_metrics[col].median()))

print(f"✅ Country-level combined metrics : {df_combined_metrics.shape}")
df_combined_metrics.sample(5)
```

# E. Explicative Analysis — Which Indicators Explain Trade?
</div>

**Objective** : Identify which economic indicators (from the indicators notebook)  
best explain trade outcomes — volume, openness, vulnerability, growth.

**Approach** :
1. Pivot indicators to wide format (country × indicator, median over period)
2. Merge with trade scores
3. Correlation heatmap
4. Random Forest feature importance (non-linear relationships)
5. OLS regression summary


<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 1. Build Wide Indicators Table
</div>

```python
# ── Pivot indicators : country × indicator → median value ────────────────────
df_ind_wide = (
    df_indicators[df_indicators["entity_type"] == "COUNTRY"]
    .groupby(["country_name", "indicator_name"])["value"]
    .median()
    .unstack("indicator_name")
    .reset_index()
)

print(f"✅ Wide indicators : {df_ind_wide.shape}")
print(f"   Indicators : {df_ind_wide.shape[1] - 1}")
df_ind_wide.sample(3)
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 2. Merge with Trade Scores
</div>

```python
# ── Merge trade scores with indicators wide table ─────────────────────────────
trade_cols = [
    "country_name",
    "structural_score",       # F : Trade Power (global structure)
    "combined_score",         # H : Combined trade score
    "f_norm_rank",            # F normalized rank
    "g_norm_rank",            # G normalized rank (bilateral resilience)
]
df_trade_scores = df_combined[trade_cols].copy()

# Add Trade Intensity from combined metrics
df_trade_scores = df_trade_scores.merge(
    df_combined_metrics[["country_name", "trade_intensity_med", "export_dep_med", "trade_balance_gdp_med"]],
    on="country_name", how="left"
)

# Final merge
df_explainer = df_trade_scores.merge(df_ind_wide, on="country_name", how="inner")

print(f"✅ Explainer table : {df_explainer.shape}")
print(f"   Countries : {df_explainer['country_name'].nunique()}")
print(f"   Trade targets available : structural_score, combined_score, f_norm_rank, g_norm_rank, trade_intensity_med")
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 3. Correlation Heatmap — Indicators vs Trade Metrics
</div>

```python
# ── Correlation between all indicators and trade metrics ──────────────────────
trade_targets = ["structural_score", "combined_score", "trade_intensity_med", "export_dep_med", "g_norm_rank"]
indicator_cols = [c for c in df_explainer.columns if c not in trade_targets + ["country_name"]]

# Compute Spearman correlations
corr_results = []
for ind in indicator_cols:
    for target in trade_targets:
        sub = df_explainer[[ind, target]].dropna()
        if len(sub) < 20:
            continue
        r, p = spearmanr(sub[ind], sub[target])
        corr_results.append({"indicator": ind, "target": target, "rho": r, "pvalue": p, "significant": p < 0.05})

df_corr = pd.DataFrame(corr_results)

# Pivot for heatmap
corr_pivot = df_corr.pivot(index="indicator", columns="target", values="rho")

# Filter : keep only indicators with at least one significant |rho| ≥ 0.25
strong = df_corr[(df_corr["significant"]) & (df_corr["rho"].abs() >= 0.25)]["indicator"].unique()
corr_pivot_strong = corr_pivot.loc[corr_pivot.index.isin(strong)].dropna(how="all")
corr_pivot_strong = corr_pivot_strong.sort_values("combined_score", ascending=False)

fig, ax = plt.subplots(figsize=(12, max(6, len(corr_pivot_strong) * 0.45)))
fig.patch.set_facecolor(C["dark"])
ax.set_facecolor("#16213e")

sns.heatmap(
    corr_pivot_strong,
    ax=ax, cmap="RdYlGn", center=0, vmin=-1, vmax=1,
    annot=True, fmt=".2f", annot_kws={"size": 8},
    linewidths=0.5, linecolor="#333",
    cbar_kws={"label": "Spearman ρ"}
)
ax.set_title("Spearman Correlation — Indicators × Trade Metrics\n(significant |ρ| ≥ 0.25)", 
             color="white", fontsize=13, pad=12)
ax.set_xlabel("Trade Target", color="white")
ax.set_ylabel("Economic Indicator", color="white")
ax.tick_params(colors="white", labelsize=8)
plt.tight_layout()
plt.savefig("Figures/E3_correlation_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()

print(f"\n📊 Top correlations with combined_score:")
top_corr = df_corr[(df_corr["target"] == "combined_score") & (df_corr["significant"])]
print(top_corr.sort_values("rho", ascending=False, key=abs).head(10)[["indicator", "rho", "pvalue"]].to_string(index=False))
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 4. Random Forest — Feature Importance
</div>

```python
# ── Target : combined_score (trade performance overall) ─────────────────────
TARGET = "combined_score"

df_rf = df_explainer[["country_name", TARGET] + indicator_cols].dropna(subset=[TARGET])

# Keep indicators with ≥ 50 valid values
valid_inds = [c for c in indicator_cols if df_rf[c].notna().sum() >= 50]
df_rf_clean = df_rf[["country_name", TARGET] + valid_inds].dropna()

X_rf = df_rf_clean[valid_inds].values
y_rf = df_rf_clean[TARGET].values

print(f"✅ RF dataset : {df_rf_clean.shape} | {len(valid_inds)} indicators | {len(y_rf)} countries")

# Train
rf = RandomForestRegressor(n_estimators=300, random_state=42, max_features="sqrt")
rf.fit(X_rf, y_rf)

# Feature importance
fi = pd.DataFrame({
    "indicator"  : valid_inds,
    "importance" : rf.feature_importances_
}).sort_values("importance", ascending=False)

print(f"\n🌲 R² (in-sample) : {rf.score(X_rf, y_rf):.3f}")
print(f"\n📊 Top 15 indicators explaining Trade Combined Score:")
print(fi.head(15).to_string(index=False))

# Plot
fig, ax = styled_figure(figsize=(13, 7))
top15 = fi.head(15).sort_values("importance")
bars = ax.barh(top15["indicator"], top15["importance"], color=C["blue"], alpha=0.8)
ax.set_title(f"Random Forest Feature Importance → {TARGET}\n(Top 15 Economic Indicators)", color="white")
ax.set_xlabel("Importance Score", color="white")
for bar in bars:
    w = bar.get_width()
    ax.text(w + 0.001, bar.get_y() + bar.get_height()/2, f"{w:.3f}", va="center", ha="left", color="white", fontsize=8)
plt.tight_layout()
plt.savefig("Figures/E4_rf_importance.png", dpi=150, bbox_inches="tight")
plt.show()
```

# F. Combined Clustering — Trade + Economy
</div>

**Objective** : Create unified country profiles combining both trade and economic dimensions.  
Merge trade scores (F, G, H) with economic scores (Health, Performance, Competitiveness)  
and run a joint clustering.


<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 1. Build Combined Feature Matrix
</div>

```python
# ── Features from Trade (normalized ranks 0→1) ───────────────────────────────
trade_feat = df_combined[["country_name", "f_norm_rank", "g_norm_rank", "combined_score"]].copy()

# ── Features from Indicators (normalized scores 0→100) ───────────────────────
econ_feat_cols = [c for c in multi_df.columns if "_value_norm" in c or "_growth_norm" in c]
econ_feat = multi_df[["country_name"] + econ_feat_cols].copy()

# Normalize econ scores to 0→1
for col in econ_feat_cols:
    econ_feat[col] = econ_feat[col] / 100.0

# ── Merge ─────────────────────────────────────────────────────────────────────
df_cluster_input = trade_feat.merge(econ_feat, on="country_name", how="inner")
df_cluster_input = df_cluster_input.merge(
    df_combined_metrics[["country_name", "trade_intensity_med", "export_dep_med"]],
    on="country_name", how="left"
)

# Normalize trade_intensity and export_dep to 0→1
for col in ["trade_intensity_med", "export_dep_med"]:
    mn, mx = df_cluster_input[col].min(), df_cluster_input[col].max()
    df_cluster_input[f"{col}_norm"] = (df_cluster_input[col] - mn) / (mx - mn + 1e-9)

# ── Final feature set ─────────────────────────────────────────────────────────
FEATURE_COLS = (
    ["f_norm_rank", "g_norm_rank"] +          # Trade structure + bilateral resilience
    econ_feat_cols +                           # Economic scores
    ["trade_intensity_med_norm", "export_dep_med_norm"]  # Combined metrics
)
FEATURE_COLS = [c for c in FEATURE_COLS if c in df_cluster_input.columns]

df_clust = df_cluster_input.dropna(subset=FEATURE_COLS).copy()
X_clust  = df_clust[FEATURE_COLS].values

print(f"✅ Clustering dataset : {df_clust.shape}")
print(f"   Countries : {len(df_clust)}")
print(f"   Features  : {len(FEATURE_COLS)}")
print(f"   Features  : {FEATURE_COLS}")
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 2. Optimal K Selection
</div>

```python
# ── Elbow + Silhouette ────────────────────────────────────────────────────────
scaler   = RobustScaler()
X_scaled = scaler.fit_transform(X_clust)

k_range    = range(2, 9)
inertias   = []
silhouettes = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled, labels))

fig, axes = styled_subplots(1, 2, figsize=(14, 5))
fig.suptitle("Optimal K — Elbow & Silhouette", fontsize=13, color="white", y=1.01)

axes[0].plot(list(k_range), inertias, marker="o", color=C["blue"])
axes[0].set_title("Elbow Curve (Inertia)", color="white")
axes[0].set_xlabel("K", color="white")
axes[0].set_ylabel("Inertia", color="white")

axes[1].plot(list(k_range), silhouettes, marker="o", color=C["green"])
best_k = list(k_range)[np.argmax(silhouettes)]
axes[1].axvline(best_k, color=C["orange"], linestyle="--", label=f"Best K = {best_k}")
axes[1].set_title("Silhouette Score", color="white")
axes[1].set_xlabel("K", color="white")
axes[1].set_ylabel("Silhouette", color="white")
axes[1].legend(facecolor="#16213e", labelcolor="white")

plt.tight_layout()
plt.show()
print(f"\n🎯 Best K (max silhouette) : {best_k}")
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 3. KMeans — Combined Clustering
</div>

```python
# ── Parameter : set K ──────────────────────────────────────────────────────
K = best_k   # Or override: K = 4
# ──────────────────────────────────────────────────────────────────────────

km_final = KMeans(n_clusters=K, random_state=42, n_init=20)
df_clust["combined_cluster"] = km_final.fit_predict(X_scaled)
df_clust["combined_cluster"] = df_clust["combined_cluster"].astype(str)

print(f"✅ K = {K} | Silhouette = {silhouette_score(X_scaled, km_final.labels_):.3f}")
print("\nCluster distribution:")
print(df_clust["combined_cluster"].value_counts().sort_index())
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

# G. Global Resilience Score
</div>

**Formula** :

$$\text{Resilience\_Score} = w_1 \cdot \text{g\_norm\_rank} + w_2 \cdot \text{(1 - vol\_export\_norm)} + w_3 \cdot \text{trade\_intensity\_norm} + w_4 \cdot \text{gdp\_growth\_score\_norm}$$

| Component | Proxy | Weight | Direction |
|---|---|---|---|
| Bilateral resilience | `g_norm_rank` (Section G) | 30% | Higher = better |
| Trade stability | `1 - volatility_export_norm` | 25% | Lower vol = better |
| Trade openness | `trade_intensity_med_norm` | 20% | Higher = better |
| Economic momentum | `Performance_growth_norm` | 25% | Higher = better |

All components normalized 0→1. Final score ∈ [0, 1].


<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## 1. Compute Resilience Score
</div>

```python
# ── Prepare components ────────────────────────────────────────────────────────
df_resil = df_combined[["country_name", "g_norm_rank"]].copy()

# Volatility export (from df_trade_dynamic) — normalize
vol_col = "volatility_export"
if vol_col in df_trade_dynamic.columns:
    vol_data = df_trade_dynamic[["country_name", vol_col]].copy()
    mn, mx   = vol_data[vol_col].min(), vol_data[vol_col].max()
    vol_data["vol_export_norm"] = (vol_data[vol_col] - mn) / (mx - mn + 1e-9)
    df_resil = df_resil.merge(vol_data[["country_name", "vol_export_norm"]], on="country_name", how="left")
    df_resil["stability_score"] = 1 - df_resil["vol_export_norm"]
else:
    df_resil["stability_score"] = 0.5  # neutral fallback

# Trade intensity (from combined metrics)
df_resil = df_resil.merge(
    df_combined_metrics[["country_name", "trade_intensity_med_norm"]],
    on="country_name", how="left"
)

# Economic momentum — Performance growth normalized
perf_growth_col = [c for c in multi_df.columns if "Performance_growth_norm" in c]
if perf_growth_col:
    perf_data = multi_df[["country_name", perf_growth_col[0]]].copy()
    perf_data[perf_growth_col[0]] = perf_data[perf_growth_col[0]] / 100.0
    df_resil = df_resil.merge(perf_data, on="country_name", how="left")
    df_resil["econ_momentum"] = df_resil[perf_growth_col[0]]
else:
    df_resil["econ_momentum"] = 0.5

# ── Weights ───────────────────────────────────────────────────────────────────
W_BILATERAL  = 0.30
W_STABILITY  = 0.25
W_OPENNESS   = 0.20
W_MOMENTUM   = 0.25

df_resil = df_resil.dropna(subset=["g_norm_rank", "stability_score", "trade_intensity_med_norm", "econ_momentum"])

df_resil["resilience_score"] = (
    W_BILATERAL * df_resil["g_norm_rank"] +
    W_STABILITY * df_resil["stability_score"] +
    W_OPENNESS  * df_resil["trade_intensity_med_norm"] +
    W_MOMENTUM  * df_resil["econ_momentum"]
)

# Classification
def classify_resilience(s):
    if   s >= 0.70: return "🟢 High Resilience"
    elif s >= 0.50: return "🟡 Moderate Resilience"
    elif s >= 0.30: return "🟠 Low Resilience"
    else:           return "🔴 Fragile"

df_resil["resilience_profile"] = df_resil["resilience_score"].apply(classify_resilience)

print(f"✅ Resilience Score computed : {len(df_resil)} countries")
print("\nProfile distribution:")
print(df_resil["resilience_profile"].value_counts())
print(f"\nSelected countries:")
print(df_resil[df_resil["country_name"].isin(countries)][
    ["country_name", "resilience_score", "resilience_profile"]
].sort_values("resilience_score", ascending=False).to_string(index=False))
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

# H. Unified Country Profile Card
</div>

**Objective** : For any selected country, synthesize all dimensions on a single card :  
Trade structure (F/G/H) × Economic scores (Health/Performance/Competitiveness) × Resilience


```python
def plot_unified_profile_card(country_name):
    """
    Unified Country Profile Card : 6 panels
    1. Trade metrics (latest year)
    2. Economic Scores radar
    3. Trade + Resilience positioning
    4. Combined metrics timeline
    5. Score summary table
    6. Profile badges
    """
    # ── Data retrieval ────────────────────────────────────────────────────────
    trade_row = df_combined[df_combined["country_name"] == country_name]
    resil_row = df_resil[df_resil["country_name"] == country_name]
    econ_row  = multi_df[multi_df["country_name"] == country_name]
    metrics_row = df_combined_metrics[df_combined_metrics["country_name"] == country_name]
    annual_data = df_panel[df_panel["country_name"] == country_name].sort_values("year")

    if trade_row.empty:
        print(f"⚠️  {country_name} not found in trade data")
        return
    
    tr = trade_row.iloc[0]
    rs = resil_row.iloc[0]  if not resil_row.empty  else None
    ec = econ_row.iloc[0]   if not econ_row.empty   else None
    mt = metrics_row.iloc[0] if not metrics_row.empty else None

    # ── Layout ────────────────────────────────────────────────────────────────
    fig = plt.figure(figsize=(20, 14))
    fig.patch.set_facecolor(C["dark"])
    
    gs = gridspec.GridSpec(3, 3, figure=fig, hspace=0.5, wspace=0.4)
    
    ax_header  = fig.add_subplot(gs[0, :])
    ax_trade   = fig.add_subplot(gs[1, 0])
    ax_radar   = fig.add_subplot(gs[1, 1], polar=True)
    ax_pos     = fig.add_subplot(gs[1, 2])
    ax_timeline = fig.add_subplot(gs[2, 0])
    ax_scores  = fig.add_subplot(gs[2, 1])
    ax_badges  = fig.add_subplot(gs[2, 2])

    for ax in [ax_header, ax_trade, ax_pos, ax_timeline, ax_scores, ax_badges]:
        ax.set_facecolor("#16213e")
        for spine in ax.spines.values():
            spine.set_edgecolor("#444")
    ax_radar.set_facecolor("#16213e")

    # ── Header ────────────────────────────────────────────────────────────────
    ax_header.axis("off")
    combined_profile = tr.get("combined_profile", "N/A")
    header_color = PROFILE_COLORS.get(combined_profile, C["grey"])
    ax_header.text(0.5, 0.7, f"🌍  {country_name}  —  Unified Trade × Economy Profile",
                   ha="center", va="center", fontsize=18, fontweight="bold", color="white", transform=ax_header.transAxes)
    ax_header.text(0.5, 0.25, f"Trade Profile: {combined_profile}   |   Resilience: {rs['resilience_profile'] if rs is not None else 'N/A'}",
                   ha="center", va="center", fontsize=13, color=header_color, transform=ax_header.transAxes)

    # ── Panel 1 : Trade Metrics ───────────────────────────────────────────────
    ax_trade.axis("off")
    ax_trade.set_title("Trade Scores", color="white", fontsize=11, pad=8)
    
    rows_trade = [
        ("Combined Score (H)", f"{tr.get('combined_score', 0):.3f}"),
        ("Trade Power (F)",    f"{tr.get('structural_score', 0):.3f}"),
        ("Bilateral Resil. (G)", f"{tr.get('g_norm_rank', 0):.3f}"),
        ("Trade Intensity",   f"{mt['trade_intensity_med']:.1%}" if mt is not None and not pd.isna(mt.get('trade_intensity_med',np.nan)) else "N/A"),
        ("Export Dep./GDP",   f"{mt['export_dep_med']:.1%}" if mt is not None and not pd.isna(mt.get('export_dep_med',np.nan)) else "N/A"),
        ("Trade Balance/GDP", f"{mt['trade_balance_gdp_med']:+.1%}" if mt is not None and not pd.isna(mt.get('trade_balance_gdp_med',np.nan)) else "N/A"),
    ]
    y_pos = 0.85
    for label, val in rows_trade:
        ax_trade.text(0.05, y_pos, label, transform=ax_trade.transAxes, color=C["grey"], fontsize=9)
        ax_trade.text(0.95, y_pos, val,   transform=ax_trade.transAxes, color="white",  fontsize=9, ha="right", fontweight="bold")
        y_pos -= 0.14

    # ── Panel 2 : Economic Scores Radar ──────────────────────────────────────
    if ec is not None:
        score_names = ["Economic Health", "Performance", "Competitiveness"]
        val_cols    = [f"{s}_value_norm" for s in score_names]
        grow_cols   = [f"{s}_growth_norm" for s in score_names]
        
        val_vals  = [ec.get(c, 50) / 100 for c in val_cols]
        grow_vals = [ec.get(c, 50) / 100 for c in grow_cols]
        
        angles = np.linspace(0, 2 * np.pi, len(score_names), endpoint=False).tolist()
        angles += angles[:1]
        val_vals  += val_vals[:1]
        grow_vals += grow_vals[:1]

        ax_radar.set_facecolor("#16213e")
        ax_radar.plot(angles, val_vals,  color=C["blue"],   linewidth=2, label="Value")
        ax_radar.fill(angles, val_vals,  color=C["blue"],   alpha=0.15)
        ax_radar.plot(angles, grow_vals, color=C["orange"], linewidth=2, label="Growth", linestyle="--")
        ax_radar.fill(angles, grow_vals, color=C["orange"], alpha=0.10)
        ax_radar.set_xticks(angles[:-1])
        ax_radar.set_xticklabels(score_names, color="white", size=8)
        ax_radar.set_yticks([0.25, 0.5, 0.75, 1.0])
        ax_radar.set_yticklabels(["25", "50", "75", "100"], color="#888", size=7)
        ax_radar.set_title("Economic Scores", color="white", fontsize=11, pad=15)
        ax_radar.legend(loc="upper right", bbox_to_anchor=(1.35, 1.1), facecolor="#16213e", labelcolor="white", fontsize=7)
        ax_radar.tick_params(colors="white")
    
    # ── Panel 3 : Trade vs Resilience Positioning ─────────────────────────────
    ax_pos.scatter(df_pos["combined_score"], df_pos["resilience_score"], color=C["grey"], alpha=0.3, s=20)
    if rs is not None:
        ax_pos.scatter(tr.get("combined_score", 0), rs["resilience_score"],
                       color=PROFILE_COLORS.get(combined_profile, C["orange"]), s=180, zorder=5, edgecolors="white", linewidths=1.5)
        ax_pos.annotate(country_name, (tr.get("combined_score", 0), rs["resilience_score"]),
                        fontsize=9, color="white", xytext=(6, 4), textcoords="offset points")
    ax_pos.axvline(df_pos["combined_score"].median(), color=C["grey"], linestyle="--", alpha=0.4)
    ax_pos.axhline(df_pos["resilience_score"].median(), color=C["grey"], linestyle="--", alpha=0.4)
    ax_pos.set_title("Trade Power vs Resilience", color="white", fontsize=11)
    ax_pos.set_xlabel("Combined Score (H)", color="white", fontsize=8)
    ax_pos.set_ylabel("Resilience Score",   color="white", fontsize=8)
    ax_pos.tick_params(colors="white", labelsize=7)

    # ── Panel 4 : Trade Intensity Timeline ────────────────────────────────────
    if not annual_data.empty and "trade_intensity" in annual_data.columns:
        ti = annual_data.dropna(subset=["trade_intensity"])
        ax_timeline.plot(ti["year"], ti["trade_intensity"], marker="o", color=C["blue"], linewidth=2)
        ax_timeline.fill_between(ti["year"], ti["trade_intensity"], alpha=0.2, color=C["blue"])
        ax_timeline.set_title("Trade Intensity (X+M)/GDP", color="white", fontsize=11)
        ax_timeline.set_xlabel("Year", color="white", fontsize=8)
        ax_timeline.yaxis.set_major_formatter(mpl.ticker.PercentFormatter(xmax=1, decimals=0))
    ax_timeline.tick_params(colors="white", labelsize=7)

    # ── Panel 5 : Score Summary Bars ──────────────────────────────────────────
    ax_scores.axis("off")
    ax_scores.set_title("Score Summary", color="white", fontsize=11, pad=8)
    
    score_items = [
        ("Trade Power",    tr.get("f_norm_rank", 0), C["blue"]),
        ("Bilateral Res.", tr.get("g_norm_rank", 0), C["green"]),
        ("Combined (H)",   tr.get("combined_score", 0), C["orange"]),
        ("Resilience",     rs["resilience_score"] if rs is not None else 0, C["purple"]),
    ]
    if ec is not None:
        val_cols_check = [c for c in [f"Economic Health_value_norm", "Performance_value_norm", "Competitiveness_value_norm"] if c in ec.index]
        for col in val_cols_check:
            short_name = col.replace("_value_norm","").replace("Economic Health","Econ. Health")
            score_items.append((short_name, ec.get(col, 0)/100, C["blue"]))

    bar_y = np.arange(len(score_items))
    bar_h = [s[1] for s in score_items]
    bar_c = [s[2] for s in score_items]
    bar_labels = [s[0] for s in score_items]
    
    ax_scores.axis("on")
    ax_scores.set_facecolor("#16213e")
    bars = ax_scores.barh(bar_y, bar_h, color=bar_c, alpha=0.8, height=0.6)
    ax_scores.set_yticks(bar_y)
    ax_scores.set_yticklabels(bar_labels, color="white", fontsize=8)
    ax_scores.set_xlim(0, 1)
    ax_scores.axvline(0.5, color=C["grey"], linestyle="--", alpha=0.4)
    ax_scores.tick_params(colors="white", labelsize=7)
    for bar, val in zip(bars, bar_h):
        ax_scores.text(min(val + 0.02, 0.97), bar.get_y() + bar.get_height()/2,
                       f"{val:.2f}", va="center", color="white", fontsize=7)
    for spine in ax_scores.spines.values():
        spine.set_edgecolor("#444")
    ax_scores.grid(alpha=0.2, color="#555")

    # ── Panel 6 : Profile Badges ──────────────────────────────────────────────
    ax_badges.axis("off")
    ax_badges.set_title("Profiles", color="white", fontsize=11, pad=8)
    
    badges = [
        ("Trade Profile",     tr.get("structural_profile", "N/A"),  0.80),
        ("Bilateral Profile", tr.get("profile_full_periode", "N/A"), 0.60),
        ("H Combined",        combined_profile,                       0.40),
        ("Resilience",        rs["resilience_profile"] if rs is not None else "N/A", 0.20),
    ]
    for label, badge, y in badges:
        ax_badges.text(0.05, y + 0.07, label, transform=ax_badges.transAxes, color=C["grey"], fontsize=8)
        badge_color = PROFILE_COLORS.get(badge, C["orange"])
        ax_badges.text(0.05, y - 0.02, badge, transform=ax_badges.transAxes,
                       color=badge_color, fontsize=10, fontweight="bold")

    plt.suptitle(f"Unified Profile — {country_name}", fontsize=16, color="white", y=1.01, fontweight="bold")
    plt.savefig(f"Figures/H_profile_{country_name.replace(' ','_')}.png", dpi=150, bbox_inches="tight")
    plt.show()

print("✅ plot_unified_profile_card() ready")
```

## K2. HHI-Weighted Vulnerability Metric
</div>

**Rationale** : HHI alone measures partner concentration. Import dependency alone measures trade weight vs GDP.  
Their **product** captures both dimensions simultaneously :

$$\text{HHI\_Weighted\_Vuln} = \text{hhi\_import\_norm} \times \text{import\_dep}$$

A country with high HHI *and* high import/GDP is doubly exposed — partner shock = macro shock.  
A country with high HHI but low import/GDP is concentrated but insulated.


```python
# ── Compute HHI-Weighted Vulnerability (country-level, median) ────────────────
hhi_col = "hhi_import_norm" if "hhi_import_norm" in df_bilateral_annual.columns else "hhi_norm"

df_hhi_weighted = (
    df_bilateral_annual
    .dropna(subset=[hhi_col, "import_dep"])
    .groupby("country_name")
    .agg(
        hhi_import_norm_med  = (hhi_col,      "median"),
        import_dep_med       = ("import_dep",  "median"),
        score_global_med     = ("score_global","median") if "score_global" in df_bilateral_annual.columns else ("import_dep","median"),
        bdis_med             = ("bdis",        "median") if "bdis" in df_bilateral_annual.columns else ("import_dep","median"),
    )
    .reset_index()
)

# Compute combined metric
df_hhi_weighted["hhi_weighted_vuln"] = (
    df_hhi_weighted["hhi_import_norm_med"] * df_hhi_weighted["import_dep_med"]
)

# Normalize 0→1
mn = df_hhi_weighted["hhi_weighted_vuln"].min()
mx = df_hhi_weighted["hhi_weighted_vuln"].max()
df_hhi_weighted["hhi_weighted_vuln_norm"] = (df_hhi_weighted["hhi_weighted_vuln"] - mn) / (mx - mn + 1e-9)

print(f"✅ HHI-Weighted Vulnerability : {df_hhi_weighted.shape}")
print(f"\nTop 10 most exposed (HHI × Import/GDP):")
print(df_hhi_weighted.nlargest(10, "hhi_weighted_vuln")[
    ["country_name", "hhi_import_norm_med", "import_dep_med", "hhi_weighted_vuln"]
].to_string(index=False))
```

```python
# ── Scatter : HHI Import vs Import/GDP — colored by combined metric ───────────
fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("HHI-Weighted Vulnerability — Partner Concentration × Import/GDP", 
             fontsize=14, color="white", y=1.01)

# Left : scatter
sc = axes[0].scatter(
    df_hhi_weighted["hhi_import_norm_med"],
    df_hhi_weighted["import_dep_med"],
    c=df_hhi_weighted["hhi_weighted_vuln_norm"],
    cmap="YlOrRd", alpha=0.6, s=50
)
axes[0].axhline(df_hhi_weighted["import_dep_med"].median(),    color=C["grey"], linestyle="--", alpha=0.5)
axes[0].axvline(df_hhi_weighted["hhi_import_norm_med"].median(),color=C["grey"], linestyle="--", alpha=0.5)
axes[0].set_xlabel("HHI Import (normalized)", color="white")
axes[0].set_ylabel("Import / GDP", color="white")
axes[0].set_title("HHI × Import Dependency\n(color = combined vulnerability)", color="white")
plt.colorbar(sc, ax=axes[0], label="HHI-Weighted Vuln.")

# Annotate selected countries
for _, row in df_hhi_weighted[df_hhi_weighted["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row["hhi_import_norm_med"], row["import_dep_med"]),
                     fontsize=8, color="white", xytext=(5, 3), textcoords="offset points")

# Right : top 20 ranking
top20 = df_hhi_weighted.nlargest(20, "hhi_weighted_vuln").sort_values("hhi_weighted_vuln")
colors_bar = [C["red"] if c in countries else C["orange"] for c in top20["country_name"]]
axes[1].barh(top20["country_name"], top20["hhi_weighted_vuln"], color=colors_bar, alpha=0.8)
axes[1].set_title("Top 20 — HHI-Weighted Vulnerability", color="white")
axes[1].set_xlabel("HHI Import × Import/GDP", color="white")

plt.tight_layout()
plt.savefig("Figures/K2_hhi_weighted_vulnerability.png", dpi=150, bbox_inches="tight")
plt.show()
```

## K3. Annual Bilateral Profile Stability
</div>

Track how each country's bilateral profile (`profile_score_global`) evolved over 2015–2024.  
**A stable profile** → structural, not cyclical.  
**A degrading profile** → vulnerability is worsening → priority alert.  
**An improving profile** → resilience-building trajectory.


```python
# ── Profile encoding for trend analysis ──────────────────────────────────────
PROFILE_ENCODE = {
    "🟢 Robust"                : 4,
    "🟡 Stable but Concentrated": 3,
    "🟠 Vulnerable"            : 2,
    "🔴 Critical Dependency"   : 1,
}

if "profile_score_global" in df_bilateral_annual.columns:
    df_bilateral_annual["profile_score_num"] = (
        df_bilateral_annual["profile_score_global"].map(PROFILE_ENCODE)
    )

    # ── Stability metrics per country ─────────────────────────────────────────
    stability_df = (
        df_bilateral_annual
        .dropna(subset=["profile_score_num"])
        .groupby("country_name")
        .agg(
            profile_mean   = ("profile_score_num", "mean"),
            profile_std    = ("profile_score_num", "std"),
            profile_first  = ("profile_score_num", "first"),
            profile_last   = ("profile_score_num", "last"),
            n_years        = ("profile_score_num", "count"),
        )
        .reset_index()
    )
    stability_df["profile_drift"] = stability_df["profile_last"] - stability_df["profile_first"]
    stability_df["profile_stable"] = stability_df["profile_std"] < 0.5

    print(f"✅ Profile stability : {stability_df.shape}")
    print(f"\nSelected countries profile drift (+= improving, -= degrading):")
    print(
        stability_df[stability_df["country_name"].isin(countries)]
        [["country_name", "profile_mean", "profile_std", "profile_drift", "profile_stable"]]
        .sort_values("profile_drift", ascending=False)
        .to_string(index=False)
    )
else:
    print("⚠️  profile_score_global not found in df_bilateral_annual — check column names")
    stability_df = pd.DataFrame()
```

# L. Partner Economic Health
</div>

**Concept** : A country's trade exposure depends not only on HOW MUCH it trades with a partner,  
but also on HOW HEALTHY that partner's economy is.

$$\text{Partner\_Health\_Score}_i = \sum_j w_{ij} \times \text{EconHealth}_j$$

where $w_{ij}$ = share of country $i$'s trade with partner $j$ (from `df_bi_trade`).

**Interpretation** :
- High Partner Health → trade network concentrated on economically strong countries → lower systemic risk
- Low Partner Health → trade network dominated by fragile economies → amplified vulnerability

**Three sub-analyses** :
- **L1** — Extract top partners per country
- **L2** — Score each partner using `multi_df`
- **L3** — Compute weighted Partner Health Score + cross with trade vulnerability


<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## L1. Extract Top Partners per Country
</div>

```python
# ── From df_bi_trade : top 5 partners per country per flow (latest year) ──────
LATEST_YEAR = df_bi_trade["year"].max()
TOP_N_PARTNERS = 5

# Use share column if available, otherwise compute from value
if "share" in df_bi_trade.columns:
    share_col = "share"
else:
    # Compute share
    df_bi_trade = df_bi_trade.copy()
    total = df_bi_trade.groupby(["country_name", "year", "flow_code"])["value"].transform("sum")
    df_bi_trade["share"] = df_bi_trade["value"] / total.replace(0, np.nan)
    share_col = "share"

df_top_partners = (
    df_bi_trade[df_bi_trade["year"] == LATEST_YEAR]
    .dropna(subset=[share_col])
    .sort_values([share_col], ascending=False)
    .groupby(["country_name", "flow_code"])
    .head(TOP_N_PARTNERS)
    [["country_name", "partner_name", "flow_code", "year", share_col]]
    .copy()
)

print(f"✅ Top {TOP_N_PARTNERS} partners extracted ({LATEST_YEAR})")
print(f"   Countries : {df_top_partners['country_name'].nunique()}")
print(f"\nExample — USA top export partners:")
print(df_top_partners[
    (df_top_partners["country_name"] == "USA") & 
    (df_top_partners["flow_code"] == "X")
].to_string(index=False))
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## L2. Score Each Partner's Economic Health
</div>

```python
# ── Get Economic Health value score for each partner ─────────────────────────
econ_health_score_col = "Economic Health_value_norm"

if econ_health_score_col in multi_df.columns:
    partner_scores = multi_df[["country_name", econ_health_score_col]].rename(
        columns={"country_name": "partner_name", econ_health_score_col: "partner_econ_health"}
    )
    # Normalize to 0→1
    partner_scores["partner_econ_health"] = partner_scores["partner_econ_health"] / 100.0
else:
    # Fallback: use first available value score
    val_cols = [c for c in multi_df.columns if "_value_norm" in c]
    if val_cols:
        partner_scores = multi_df[["country_name", val_cols[0]]].rename(
            columns={"country_name": "partner_name", val_cols[0]: "partner_econ_health"}
        )
        partner_scores["partner_econ_health"] = partner_scores["partner_econ_health"] / 100.0
        print(f"ℹ️  Using {val_cols[0]} as partner health proxy")
    else:
        print("⚠️  No economic health score found in multi_df")
        partner_scores = pd.DataFrame(columns=["partner_name", "partner_econ_health"])

# ── Join partner scores to top partners table ────────────────────────────────
df_top_partners = df_top_partners.merge(partner_scores, on="partner_name", how="left")

coverage = df_top_partners["partner_econ_health"].notna().mean()
print(f"✅ Partner health score joined : {coverage:.1%} coverage")
print(f"\nExample — USA top partners with health score:")
print(df_top_partners[df_top_partners["country_name"] == "USA"].to_string(index=False))
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## L3. Weighted Partner Health Score
</div>

```python
# ── Compute weighted Partner Health Score per country × flow ─────────────────
def weighted_partner_health(group):
    """Weighted average of partner health, weighted by trade share."""
    valid = group.dropna(subset=["partner_econ_health"])
    if valid.empty or valid[share_col].sum() == 0:
        return np.nan
    return np.average(valid["partner_econ_health"], weights=valid[share_col])

df_partner_health = (
    df_top_partners
    .groupby(["country_name", "flow_code"])
    .apply(weighted_partner_health)
    .reset_index()
    .rename(columns={0: "partner_health_score"})
)

# Pivot import / export
df_partner_health_wide = df_partner_health.pivot(
    index="country_name", columns="flow_code", values="partner_health_score"
).reset_index()
df_partner_health_wide.columns.name = None

# Rename
col_map = {}
if "X" in df_partner_health_wide.columns: col_map["X"] = "partner_health_export"
if "M" in df_partner_health_wide.columns: col_map["M"] = "partner_health_import"
df_partner_health_wide = df_partner_health_wide.rename(columns=col_map)

# Combined partner health (average of import + export if both available)
health_cols = [c for c in ["partner_health_export", "partner_health_import"] if c in df_partner_health_wide.columns]
df_partner_health_wide["partner_health_combined"] = df_partner_health_wide[health_cols].mean(axis=1)

print(f"✅ Partner Health Score : {df_partner_health_wide.shape}")
print(f"\nSelected countries Partner Health:")
print(df_partner_health_wide[df_partner_health_wide["country_name"].isin(countries)].to_string(index=False))
```

```python
# ── Visualisation : Partner Health vs Bilateral Vulnerability ─────────────────
df_ph_viz = df_partner_health_wide.merge(
    df_combined[["country_name", "combined_score", "combined_profile"]],
    on="country_name", how="inner"
).merge(
    df_hhi_weighted[["country_name", "hhi_weighted_vuln_norm"]],
    on="country_name", how="left"
)

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Partner Economic Health — Network Quality Analysis", 
             fontsize=14, color="white", y=1.01)

# Left : Partner Health vs Trade Combined Score
sc = axes[0].scatter(
    df_ph_viz["combined_score"],
    df_ph_viz["partner_health_combined"],
    c=df_ph_viz["hhi_weighted_vuln_norm"],
    cmap="RdYlGn_r", alpha=0.6, s=50
)
axes[0].axhline(df_ph_viz["partner_health_combined"].median(), color=C["grey"], linestyle="--", alpha=0.5)
axes[0].axvline(df_ph_viz["combined_score"].median(),          color=C["grey"], linestyle="--", alpha=0.5)
axes[0].set_xlabel("Trade Combined Score (H)", color="white")
axes[0].set_ylabel("Weighted Partner Health", color="white")
axes[0].set_title("Trade Performance vs Partner Quality\n(color = HHI-Weighted Vulnerability)", color="white")
plt.colorbar(sc, ax=axes[0], label="HHI-Weighted Vuln.")

# Annotate selected countries
for _, row in df_ph_viz[df_ph_viz["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row["combined_score"], row["partner_health_combined"]),
                     fontsize=8, color="white", xytext=(5,3), textcoords="offset points")

# Right : Partner Health ranking (top/bottom 20)
n = 20
df_sorted = df_ph_viz.dropna(subset=["partner_health_combined"]).sort_values("partner_health_combined")
df_bot = df_sorted.head(n)
df_top = df_sorted.tail(n).sort_values("partner_health_combined")

for ax, data, title in [(axes[1], pd.concat([df_bot, df_top]).drop_duplicates(),
                          f"Top & Bottom {n} — Weighted Partner Health")]:
    midpoint = len(df_bot)
    colors_bar = [C["red"]] * len(df_bot) + [C["green"]] * len(df_top)
    ax.barh(data["country_name"], data["partner_health_combined"], color=colors_bar, alpha=0.8)
    ax.set_title(title, color="white")
    ax.set_xlabel("Weighted Partner Economic Health [0–1]", color="white")
    ax.axvline(df_ph_viz["partner_health_combined"].median(), color="white", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("Figures/L3_partner_health_score.png", dpi=150, bbox_inches="tight")
plt.show()
```

# M. Enhanced Resilience Score (Annual)
</div>

**Upgrade from Section G** : the original Resilience Score used `g_norm_rank` (full-period static).  
This section builds a **richer, time-varying version** using annual data from `df_final` + partner health.

**New formula** :

| Component | Source | Weight |
|---|---|---|
| Bilateral resilience (annual) | `1 - score_global` from `df_final` | 25% |
| Trade stability | `1 - volatility_export_norm` | 20% |
| Trade openness | `trade_intensity_med_norm` | 15% |
| Economic momentum | `Performance_growth_norm` | 20% |
| Partner health | `partner_health_combined` (Section L) | 10% |
| HHI-weighted vuln. (inverted) | `1 - hhi_weighted_vuln_norm` | 10% |

6 components vs 4 in Section G — richer, more robust.


<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## M1. Build Enhanced Resilience Score
</div>

```python
# ── Assemble all components ───────────────────────────────────────────────────
df_resil_v2 = df_combined[["country_name"]].copy()

# 1. Bilateral resilience (annual median from df_final → inverted)
if "score_global" in df_bilateral_annual.columns:
    bil_annual = (
        df_bilateral_annual
        .groupby("country_name")["score_global"]
        .median()
        .reset_index()
        .rename(columns={"score_global": "score_global_med"})
    )
    mn, mx = bil_annual["score_global_med"].min(), bil_annual["score_global_med"].max()
    bil_annual["bilateral_resil_v2"] = 1 - (bil_annual["score_global_med"] - mn) / (mx - mn + 1e-9)
    df_resil_v2 = df_resil_v2.merge(bil_annual[["country_name","bilateral_resil_v2"]], on="country_name", how="left")
else:
    df_resil_v2["bilateral_resil_v2"] = df_resil[["country_name","g_norm_rank"]].set_index("country_name").reindex(df_resil_v2["country_name"]).values

# 2. Trade stability (from original resilience computation)
vol_col = "vol_export_norm"
if vol_col in df_resil.columns:
    df_resil_v2 = df_resil_v2.merge(df_resil[["country_name", vol_col]], on="country_name", how="left")
    df_resil_v2["stability_v2"] = 1 - df_resil_v2[vol_col]
else:
    df_resil_v2["stability_v2"] = 0.5

# 3. Trade openness
df_resil_v2 = df_resil_v2.merge(
    df_combined_metrics[["country_name", "trade_intensity_med_norm"]],
    on="country_name", how="left"
)

# 4. Economic momentum
perf_col = [c for c in multi_df.columns if "Performance_growth_norm" in c]
if perf_col:
    perf_df = multi_df[["country_name", perf_col[0]]].copy()
    perf_df[perf_col[0]] = perf_df[perf_col[0]] / 100.0
    df_resil_v2 = df_resil_v2.merge(perf_df, on="country_name", how="left")
    df_resil_v2["econ_momentum_v2"] = df_resil_v2[perf_col[0]]
else:
    df_resil_v2["econ_momentum_v2"] = 0.5

# 5. Partner health (Section L)
if "partner_health_combined" in df_partner_health_wide.columns:
    df_resil_v2 = df_resil_v2.merge(
        df_partner_health_wide[["country_name", "partner_health_combined"]],
        on="country_name", how="left"
    )
else:
    df_resil_v2["partner_health_combined"] = 0.5

# 6. HHI-weighted vulnerability (inverted)
if "hhi_weighted_vuln_norm" in df_hhi_weighted.columns:
    df_resil_v2 = df_resil_v2.merge(
        df_hhi_weighted[["country_name", "hhi_weighted_vuln_norm"]],
        on="country_name", how="left"
    )
    df_resil_v2["hhi_resil"] = 1 - df_resil_v2["hhi_weighted_vuln_norm"]
else:
    df_resil_v2["hhi_resil"] = 0.5

# ── Weights ───────────────────────────────────────────────────────────────────
W = {
    "bilateral_resil_v2"       : 0.25,
    "stability_v2"             : 0.20,
    "trade_intensity_med_norm" : 0.15,
    "econ_momentum_v2"         : 0.20,
    "partner_health_combined"  : 0.10,
    "hhi_resil"                : 0.10,
}

components = list(W.keys())
df_resil_v2 = df_resil_v2.dropna(subset=components)

df_resil_v2["resilience_score_v2"] = sum(
    W[c] * df_resil_v2[c] for c in components
)

df_resil_v2["resilience_profile_v2"] = df_resil_v2["resilience_score_v2"].apply(classify_resilience)

print(f"✅ Enhanced Resilience Score v2 : {len(df_resil_v2)} countries")
print("\nProfile distribution:")
print(df_resil_v2["resilience_profile_v2"].value_counts())
print(f"\nSelected countries:")
print(df_resil_v2[df_resil_v2["country_name"].isin(countries)][
    ["country_name", "resilience_score_v2", "resilience_profile_v2"]
].sort_values("resilience_score_v2", ascending=False).to_string(index=False))
```

<div style="background-color: #e9f7ef; padding-left: 20px; border: 1px solid; border-radius: 5px;">

## M2. V1 vs V2 Comparison

```python
# ── Compare original vs enhanced resilience score ────────────────────────────
df_compare = df_resil[["country_name", "resilience_score"]].merge(
    df_resil_v2[["country_name", "resilience_score_v2"]],
    on="country_name", how="inner"
)
df_compare["score_delta"] = df_compare["resilience_score_v2"] - df_compare["resilience_score"]
df_compare["rank_v1"] = df_compare["resilience_score"].rank(ascending=False).astype(int)
df_compare["rank_v2"] = df_compare["resilience_score_v2"].rank(ascending=False).astype(int)
df_compare["rank_change"] = df_compare["rank_v1"] - df_compare["rank_v2"]  # positive = climbed

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Resilience Score — V1 (Section G) vs V2 Enhanced (Section M)", 
             fontsize=14, color="white", y=1.01)

# Left : scatter V1 vs V2
axes[0].scatter(df_compare["resilience_score"], df_compare["resilience_score_v2"],
                color=C["blue"], alpha=0.5, s=40)
lims = [min(df_compare["resilience_score"].min(), df_compare["resilience_score_v2"].min()),
        max(df_compare["resilience_score"].max(), df_compare["resilience_score_v2"].max())]
axes[0].plot(lims, lims, color=C["grey"], linestyle="--", alpha=0.5, label="V1 = V2")

for _, row in df_compare[df_compare["country_name"].isin(countries)].iterrows():
    axes[0].scatter(row["resilience_score"], row["resilience_score_v2"],
                    color=C["orange"], s=120, zorder=5, edgecolors="white")
    axes[0].annotate(row["country_name"],
                     (row["resilience_score"], row["resilience_score_v2"]),
                     fontsize=8, color="white", xytext=(5, 3), textcoords="offset points")
axes[0].set_xlabel("Resilience Score V1", color="white")
axes[0].set_ylabel("Resilience Score V2", color="white")
axes[0].set_title("V1 vs V2 Correlation\n(above line = upgraded by partner/HHI enrichment)", color="white")
axes[0].legend(facecolor="#16213e", labelcolor="white")

r_corr, _ = spearmanr(df_compare["resilience_score"], df_compare["resilience_score_v2"])
axes[0].text(0.05, 0.95, f"Spearman ρ = {r_corr:.3f}", transform=axes[0].transAxes,
             color="white", fontsize=10)

# Right : biggest rank movers
top_movers = df_compare.reindex(df_compare["rank_change"].abs().sort_values(ascending=False).index).head(20)
colors_rank = [C["green"] if v > 0 else C["red"] for v in top_movers["rank_change"]]
axes[1].barh(top_movers["country_name"], top_movers["rank_change"], color=colors_rank, alpha=0.8)
axes[1].axvline(0, color="white", linewidth=1)
axes[1].set_title("Biggest Rank Movers V1 → V2\n(green = climbed, red = dropped)", color="white")
axes[1].set_xlabel("Rank Change (positions)", color="white")

plt.tight_layout()
plt.savefig("Figures/M2_resilience_v1_vs_v2.png", dpi=150, bbox_inches="tight")
plt.show()
```

| **E** | Explicative analysis : top indicators by Spearman ρ + RF importance + OLS |
| **F** | Combined clustering (Trade + Economy) with K-Means + Silhouette |
| **G** | Resilience Score = f(Bilateral resilience, Stability, Openness, Momentum) |
| **H** | Unified Profile Card per country (8 dimensions in one view) |

### Key Questions Answered

1. **Which indicators explain trade best?** → Section E (RF importance + correlation)
2. **Trade Intensity / Export Dependency by country?** → Section D
3. **Country profiles combining trade + economy?** → Section F (clustering)
4. **Which countries are globally resilient?** → Section G
5. **Full country synthesis?** → Section H (profile card)

### Limitations

- Import Coverage (Reserves/M) excluded — reserves data not in current indicators dataset
- RF importance is in-sample — no cross-validation (limited by ~130 country observations)
- Resilience score weights are analytical choices — sensitivity analysis recommended
- Interpolated values (flagged as `interpolated`) introduce mild uncertainty
