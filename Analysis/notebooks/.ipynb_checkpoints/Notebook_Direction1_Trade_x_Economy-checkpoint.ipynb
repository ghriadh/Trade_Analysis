# A. Introduction & Framework
</div>

**Direction 1 — Trade as Outcome**
> *"Given a country's economic structure, which indicators explain its trade performance?"*

**9 Trade Families covered** — all using `df_structural` as backbone :

| Section | Trade Family | Key Variables | Indicators Used |
|---|---|---|---|
| **D** | Combined Metrics | Trade Intensity, Export/Import Dep, Balance/GDP, Per Capita, Elasticity | GDP, GDP per capita |
| **E** | Export Analysis | size_profile, growth_profile, cagr_export, export_share, export_pc | GDP, GDP_pc, Inflation, Real rate, FDI proxy |
| **F** | Import Analysis | import_share, cagr_import, volatility_import, import_pc | GDP, GDP_pc, Unemployment, Inflation |
| **G** | Total Trade Volume | total_trade, CAGR_trade, volatility | GDP CAGR, Trade%GDP cross-check |
| **H** | Balance & Coverage | coverage_ratio, balance_profile, balance_risk | Current account, External debt, Inflation |
| **I** | Unified Correlation | All families × all indicators | Full matrix |
| **J** | Feature Importance | RF + OLS per family | All indicators |
| **K** | Combined Clustering | Trade profiles + Economic scores | multi_df |
| **L** | Resilience Score | Composite index | All |
| **M** | Profile Card | Unified synthesis | All |

**Key inputs** :
- `df_structural` → country-level trade backbone (profiles, scores, risks)
- `df_trade` → annual trade flows
- `df_trade_dynamic` → dynamic features (CAGR, volatility)
- `df` → long indicators (country × indicator × year)
- `multi_df` → economic scores (Health / Performance / Competitiveness)

## 2. Style
</div>

```python
C = {"blue":"#3498db","green":"#2ecc71","red":"#e74c3c","orange":"#f39c12",
     "purple":"#9b59b6","grey":"#95a5a6","dark":"#1a1a2e","teal":"#1abc9c"}

SIZE_COLORS      = {"Leader":C["green"],"Major":C["blue"],"Mid":C["orange"],"Small":C["red"]}
BALANCE_COLORS   = {"Strong Surplus":C["green"],"Slight Surplus":C["teal"],
                    "Slight Deficit":C["orange"],"Structural Deficit":C["red"]}
GROWTH_COLORS    = {"Strong Grower":C["green"],"Moderate Grower":C["blue"],
                    "Weak Grower":C["orange"],"Declining":C["red"]}
STABILITY_COLORS = {"Stable":C["green"],"Moderate":C["blue"],
                    "Volatile":C["orange"],"Very Volatile":C["red"]}
CONSIST_COLORS   = {"Very Consistent":C["green"],"Consistent":C["blue"],
                    "Inconsistent":C["orange"],"Very Inconsistent":C["red"]}
PROFILE_COLORS   = {"🏆 Global Champion":C["green"],"⚡ Competitive":C["blue"],
                    "⚠️ Vulnerable":C["orange"],"🔴 Fragile":C["red"]}

def styled_figure(figsize=(12,6)):
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    ax.tick_params(colors="white"); ax.xaxis.label.set_color("white")
    ax.yaxis.label.set_color("white"); ax.title.set_color("white")
    for sp in ax.spines.values(): sp.set_edgecolor("#444")
    ax.grid(alpha=0.2,color="#555"); return fig,ax

def styled_subplots(nrows=1,ncols=2,figsize=(18,6)):
    fig,axes = plt.subplots(nrows,ncols,figsize=figsize)
    fig.patch.set_facecolor(C["dark"])
    for ax in np.array(axes).flatten():
        ax.set_facecolor("#16213e"); ax.tick_params(colors="white")
        ax.xaxis.label.set_color("white"); ax.yaxis.label.set_color("white")
        ax.title.set_color("white")
        for sp in ax.spines.values(): sp.set_edgecolor("#444")
        ax.grid(alpha=0.2,color="#555")
    return fig,axes

def normalize_01(s):
    mn,mx = s.min(),s.max(); return (s-mn)/(mx-mn+1e-9)

def spearman_corr(df, trade_vars, ind_vars, min_obs=25):
    """Compute Spearman ρ table between trade vars and indicator vars."""
    rows=[]
    for tv in trade_vars:
        for iv in ind_vars:
            sub = df[[tv,iv]].dropna()
            if len(sub)<min_obs: continue
            r,p = spearmanr(sub[tv],sub[iv])
            rows.append({"trade_var":tv,"indicator":iv,"rho":round(r,3),
                         "pvalue":round(p,4),"sig":p<0.05,"abs_rho":abs(r)})
    return pd.DataFrame(rows)

def plot_corr_heatmap(corr_df, title, min_abs_rho=0.20, figsize=(12,7)):
    """Plot Spearman correlation heatmap filtered by significance and strength."""
    strong = corr_df[(corr_df["sig"]) & (corr_df["abs_rho"]>=min_abs_rho)]
    if strong.empty:
        print(f"No significant correlations above |ρ|={min_abs_rho}")
        return
    pivot = strong.pivot(index="indicator",columns="trade_var",values="rho")
    pivot = pivot.sort_values(pivot.columns[0], ascending=False)
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1,
                annot=True,fmt=".2f",annot_kws={"size":8},
                linewidths=0.5,linecolor="#333",cbar_kws={"label":"Spearman ρ"})
    ax.set_title(f"{title}\n(sig. |ρ| ≥ {min_abs_rho})",color="white",pad=10)
    ax.tick_params(colors="white",labelsize=8)
    plt.tight_layout()
    fname = title.lower().replace(" ","_").replace("×","x")[:40]
    plt.savefig(f"../Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

print("✅ Style helpers OK")
```

## C3. Build Indicator Tables
</div>

```python
# ── Indicator name mapping (short keys) ──────────────────────────────────────
IND = {
    "gdp"          : "GDP (current US$)",
    "gdp_pc"       : "GDP per capita (current US$)",
    "inflation"    : "Inflation, consumer prices (annual %)",
    "real_rate"    : "Real interest rate (%)",
    "unemployment" : "Unemployment, total (% of total labor force) (modeled ILO estimate)",
    "current_acct" : "Current account balance (% of GDP)",
    "trade_pct_gdp": "Trade (% of GDP)",
    "ext_debt"     : "External debt stocks, total (DOD, current US$)",
    "expense"      : "Expense (% of GDP)",
}
IND = {k:v for k,v in IND.items() if v in df["indicator_name"].unique()}
print(f"✅ Indicators confirmed : {list(IND.keys())}")

# ── Annual pivot : country × year × indicator ─────────────────────────────────
df_ind_annual = (df[df["entity_type"]=="COUNTRY"].pivot_table(index=["country_name","year"], columns="indicator_name",values="value").reset_index())
df_ind_annual.columns.name = None
df_ind_annual = df_ind_annual.rename(columns={v:k for k,v in IND.items()})
ind_cols = [k for k in IND.keys() if k in df_ind_annual.columns]

# ── Annual growth rates per indicator ─────────────────────────────────────────
df_ind_annual = df_ind_annual.sort_values(["country_name","year"])
for col in ind_cols:
    df_ind_annual[f"{col}_yoy"] = (df_ind_annual.groupby("country_name")[col].pct_change()*100)

# ── Country-level median (structural) ─────────────────────────────────────────
df_ind_struct = df_ind_annual.groupby("country_name")[ind_cols].median().reset_index()

print(f"✅ df_ind_annual : {df_ind_annual.shape}")
print(f"✅ df_ind_struct : {df_ind_struct.shape}")
```

## C4. Master Merge — df_structural × Indicators
</div>

```python
# ── Core merge : df_structural + df_ind_struct + df_combined ─────────────────

# df_structural is the backbone : all trade structural info already computed
df_master = df_structural.merge(df_ind_struct, on="country_name", how="left")

# Add combined score & profile (Section H)
df_master = df_master.merge(df_combined[["country_name","combined_score","f_norm_rank","g_norm_rank","combined_profile"]], on="country_name", how="left")

# Add economic scores (multi_df)
econ_score_cols = [c for c in multi_df.columns if "_value_norm" in c or "_growth_norm" in c]
df_master = df_master.merge(multi_df[["country_name"]+econ_score_cols], on="country_name", how="left")

df_master.sample()
```

```python
# ── Derived : Population from GDP / GDP_per_capita ───────────────────────────
if "gdp" in df_master.columns and "gdp_pc" in df_master.columns:
    mask = df_master["gdp_pc"].notna() & (df_master["gdp_pc"]>0)
    df_master.loc[mask,"population"] = (df_master.loc[mask,"gdp"] / df_master.loc[mask,"gdp_pc"])
df_master.sample().T
# (calcule la population)
```

```python
# ── Per capita trade ──────────────────────────────────────────────────────────
# total_export and total_import come from df_structural (via df_trade_dynamic)
for flow_col, pc_col in [("total_export","export_pc"),("total_import","import_pc")]:
    if flow_col in df_master.columns and "population" in df_master.columns:
        mask = df_master["population"].notna() & (df_master["population"]>0)
        df_master.loc[mask,pc_col] = (df_master.loc[mask,flow_col] / df_master.loc[mask,"population"])
df_master.sample().T
# (export_pc : part d'individu in export = total export/population) - de meme import_pc
```

```python
# ── Trade Intensity & related ratios ─────────────────────────────────────────
if "gdp" in df_master.columns and (df_master["gdp"]>0).any():
    mask = df_master["gdp"].notna() & (df_master["gdp"]>0)
    df_master.loc[mask,"trade_intensity"] = ((df_master.loc[mask,"total_export"] + df_master.loc[mask,"total_import"])/ df_master.loc[mask,"gdp"])
    df_master.loc[mask,"export_dep"]  = df_master.loc[mask,"total_export"] / df_master.loc[mask,"gdp"]
    df_master.loc[mask,"import_dep"]  = df_master.loc[mask,"total_import"] / df_master.loc[mask,"gdp"]
    df_master.loc[mask,"balance_dep"] = (df_master.loc[mask,"total_export"]-df_master.loc[mask,"total_import"]) / df_master.loc[mask,"gdp"]
df_master.sample().T
# trade_intensity = (total export + total import) / gdp)
# export_dep = total export  / gdp
# import_dep = total import  / gdp
# balance_dep = (total export - total import) / gdp)
```

```python
# ── Trade Elasticity = CAGR_export / CAGR_GDP ────────────────────────────────

if "gdp" in df_ind_annual.columns:
    # Compute GDP CAGR from annual data
    
    def compute_cagr(series):
        s = series.dropna()
        if len(s)<2 or s.iloc[0]<=0 or s.iloc[-1]<=0: return np.nan
        return (s.iloc[-1]/s.iloc[0])**(1/(len(s)-1))-1
    
    gdp_cagr = (df_ind_annual.groupby("country_name")["gdp"].apply(compute_cagr).reset_index().rename(columns={"gdp":"cagr_gdp"}))
    df_master = df_master.merge(gdp_cagr, on="country_name", how="left")
    
    mask = df_master["cagr_gdp"].notna() & (df_master["cagr_gdp"]!=0)
    
    df_master.loc[mask,"trade_elasticity_x"] = (df_master.loc[mask,"cagr_export"] / df_master.loc[mask,"cagr_gdp"])
    df_master.loc[mask,"trade_elasticity_m"] = (df_master.loc[mask,"cagr_import"] / df_master.loc[mask,"cagr_gdp"])

df_master.sample().T
# trade_elasticity_x = cagr_export / cagr_gdp
```

## D2. Trade Elasticity — Does Trade Grow Faster than Economy?
</div>

```python
# Trade Elasticity = CAGR_export / CAGR_GDP
# > 1 : trade grows faster → country opening
# < 1 : economy grows faster → domestic-led growth
# < 0 : trade declines while economy grows → structural alert

if "trade_elasticity_x" in df_master.columns:
    df_d2 = df_master[["country_name","trade_elasticity_x","trade_elasticity_m",
                        "cagr_export","cagr_gdp","size_profile","growth_profile"]].dropna()

    fig,axes = styled_subplots(1,2,figsize=(18,7))
    fig.suptitle("Trade Elasticity — CAGR_Trade / CAGR_GDP",fontsize=14,color="white",y=1.01)

    # Left: CAGR_export vs CAGR_GDP scatter
    for gp,col in GROWTH_COLORS.items():
        sub = df_d2[df_d2["growth_profile"]==gp]
        axes[0].scatter(sub["cagr_gdp"]*100,sub["cagr_export"]*100,
                        color=col,alpha=0.6,s=50,label=gp)
    # diagonal = elasticity 1
    lim_min = min(df_d2["cagr_gdp"].min(),df_d2["cagr_export"].min())*100*1.1
    lim_max = max(df_d2["cagr_gdp"].max(),df_d2["cagr_export"].max())*100*1.1
    axes[0].plot([lim_min,lim_max],[lim_min,lim_max],
                 color="white",linestyle="--",alpha=0.5,label="Elasticity = 1")
    axes[0].axhline(0,color=C["grey"],linestyle=":",alpha=0.4)
    axes[0].axvline(0,color=C["grey"],linestyle=":",alpha=0.4)
    for _,row in df_d2[df_d2["country_name"].isin(countries)].iterrows():
        axes[0].annotate(row["country_name"],
                         (row["cagr_gdp"]*100,row["cagr_export"]*100),
                         fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    axes[0].set_xlabel("CAGR GDP (%)",color="white")
    axes[0].set_ylabel("CAGR Export (%)",color="white")
    axes[0].set_title("Export CAGR vs GDP CAGR\n(above diagonal = trade-led growth)",color="white")
    axes[0].legend(facecolor="#16213e",labelcolor="white",fontsize=7)

    # Right: Trade Elasticity distribution
    elas = df_d2["trade_elasticity_x"].clip(-3,5)
    colors_bar = [C["green"] if v>=1 else C["orange"] if v>=0 else C["red"] for v in elas.sort_values()]
    axes[1].barh(range(len(elas.sort_values())),elas.sort_values().values,
                 color=colors_bar,alpha=0.7,height=1.0)
    axes[1].axvline(1,color="white",linestyle="--",alpha=0.7,label="Elasticity=1")
    axes[1].axvline(0,color=C["red"],linestyle=":",alpha=0.5)
    axes[1].set_title("Trade Elasticity Distribution\n(>1=opening, 0-1=partial, <0=alert)",
                      color="white")
    axes[1].set_xlabel("CAGR_Export / CAGR_GDP",color="white")
    axes[1].set_yticks([])
    axes[1].legend(facecolor="#16213e",labelcolor="white")

    plt.tight_layout()
    plt.savefig("../Figures/D2_trade_elasticity.png",dpi=150,bbox_inches="tight")
    plt.show()

    print("\nTrade Elasticity — Selected countries:")
    print(df_d2[df_d2["country_name"].isin(countries)][
        ["country_name","cagr_export","cagr_gdp","trade_elasticity_x","growth_profile"]
    ].sort_values("trade_elasticity_x",ascending=False).to_string(index=False))
else:
    print("⚠️  trade_elasticity_x not computed — check cagr_gdp availability")
```

## D3. Per Capita Trade — Reclassification vs Size Profile
</div>

```python
# Export per capita reclassifies countries vs absolute volume
# Small absolute exporters can be large per capita exporters (Luxembourg, Singapore)

if "export_pc" in df_master.columns:
    df_d3 = df_master[["country_name","total_export","export_pc",
                        "gdp_pc","size_profile"]].dropna()
    df_d3["rank_total_export"] = df_d3["total_export"].rank(ascending=False).astype(int)
    df_d3["rank_export_pc"]  = df_d3["export_pc"].rank(ascending=False).astype(int)
    df_d3["rank_delta"] = df_d3["rank_total_export"] - df_d3["rank_export_pc"]  # +: climbs in pc ranking
    display(df_d3.sample(3))
    fig,axes = styled_subplots(1,2,figsize=(18,7))
    fig.suptitle("Export Volume vs Export per Capita — Rank Reclassification", fontsize=14,color="white",y=1.01)

    # Left: scatter absolute vs per capita (log)
    for sp,col in SIZE_COLORS.items():
        sub = df_d3[df_d3["size_profile"]==sp]
        axes[0].scatter(sub["total_export"],sub["export_pc"],
                        color=col,alpha=0.6,s=50,label=sp)
    for _,row in df_d3[df_d3["country_name"].isin(countries)].iterrows():
        axes[0].annotate(row["country_name"],
                         (row["total_export"],row["export_pc"]),
                         fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    axes[0].set_xscale("log"); axes[0].set_yscale("log")
    axes[0].set_xlabel("Total Export (log)",color="white")
    axes[0].set_ylabel("Export per Capita (log)",color="white")
    axes[0].set_title("Volume vs Per Capita\n(color = Size Profile)",color="white")
    axes[0].legend(facecolor="#16213e",labelcolor="white",fontsize=8)

    # Right: top rank movers
    top_movers = df_d3.reindex(df_d3["rank_delta"].abs().sort_values(ascending=False).index).head(25)
    colors_mv = [C["green"] if v>0 else C["red"] for v in top_movers["rank_delta"]]
    axes[1].barh(top_movers["country_name"],top_movers["rank_delta"],
                 color=colors_mv,alpha=0.8)
    axes[1].axvline(0,color="white",linewidth=1)
    axes[1].set_title("Rank Movers: Volume → Per Capita\n(green=climbs, red=drops)",color="white")
    axes[1].set_xlabel("Rank Change (positions)",color="white")

    plt.tight_layout()
    plt.savefig("../Figures/D3_export_pc_reclassification.png",dpi=150,bbox_inches="tight")
    plt.show()

    # Export per capita vs GDP per capita deviation
    fig,ax = styled_figure(figsize=(13,8))
    ax.set_title("Export per Capita vs GDP per Capita\nDeviations reveal structural exporters & rentier economies",
                 color="white")
    sc = ax.scatter(df_d3["gdp_pc"],df_d3["export_pc"],
                    c=[list(SIZE_COLORS.keys()).index(sp) if sp in SIZE_COLORS else 0
                       for sp in df_d3["size_profile"]],
                    cmap="RdYlGn",alpha=0.6,s=50)
    for _,row in df_d3[df_d3["country_name"].isin(countries)].iterrows():
        ax.annotate(row["country_name"],(row["gdp_pc"],row["export_pc"]),
                    fontsize=9,color="white",xytext=(6,4),textcoords="offset points")
    ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlabel("GDP per Capita (log)",color="white")
    ax.set_ylabel("Export per Capita (log)",color="white")
    r,p = spearmanr(df_d3["gdp_pc"],df_d3["export_pc"])
    ax.text(0.05,0.95,f"Spearman ρ={r:.2f}",transform=ax.transAxes,
            color="white",fontsize=10)
    plt.tight_layout()
    plt.savefig("../Figures/D3_export_pc_vs_gdp_pc.png",dpi=150,bbox_inches="tight")
    plt.show()
```

## E4. Consistency Profile (negfreq) × Real Interest Rate & External Debt
</div>

```python
# consistency_profile (Very Consistent/Consistent/Inconsistent/Very Inconsistent)
# Key hypothesis: high real interest rate → costly financing → more frequent negative export years
# External debt → forced adjustments → trade inconsistency

df_e4 = df_master[["country_name","negfreq_export","consistency_profile", "consistency_risk","real_rate","ext_debt","current_acct", "inflation","gdp"]] \
                 .dropna(subset=["consistency_profile"])

# Normalize ext_debt by GDP for comparability
if "ext_debt" in df_e4.columns and "gdp" in df_e4.columns:
    mask = df_e4["gdp"].notna() & (df_e4["gdp"]>0)
    df_e4.loc[mask,"ext_debt_gdp"] = df_e4.loc[mask,"ext_debt"] / df_e4.loc[mask,"gdp"]

consist_vars = ["negfreq_export","consistency_risk"]
ind_vars_c = [c for c in ["real_rate","ext_debt_gdp","current_acct","inflation"] if c in df_e4.columns]
corr_e4 = spearman_corr(df_e4,consist_vars,ind_vars_c)
print("📊 Consistency × Indicators Correlations:")
print(corr_e4.sort_values("abs_rho",ascending=False).to_string(index=False))
print("\n💡 Interpretation:")
print("  Real rate ↑  →  negfreq ↑  : costlier financing → more negative export years")
print("  Ext. debt ↑  →  negfreq ↑  : debt burden → forced adjustments")

plot_corr_heatmap(corr_e4,"Consistency Profile × Indicators",min_abs_rho=0.15)

if "real_rate" in df_e4.columns:
    fig,axes = styled_subplots(1,2,figsize=(18,7))
    fig.suptitle("Consistency Profile × Real Interest Rate & External Debt", fontsize=14,color="white",y=1.01)

    # Left: negfreq vs real_rate
    df_e4_clean = df_e4.dropna(subset=["real_rate"])
    for cp,col in CONSIST_COLORS.items():
        sub = df_e4_clean[df_e4_clean["consistency_profile"]==cp]
        axes[0].scatter(sub["real_rate"].clip(-20,30),sub["negfreq_export"], color=col,alpha=0.6,s=50,label=cp)
    for _,row in df_e4_clean[df_e4_clean["country_name"].isin(countries)].iterrows():
        axes[0].annotate(row["country_name"], (np.clip(row["real_rate"],-20,30),row["negfreq_export"]),
                         fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    axes[0].set_xlabel("Real Interest Rate (%)",color="white")
    axes[0].set_ylabel("Negative Export Frequency [0–1]",color="white")
    axes[0].set_title("Export Inconsistency vs Real Interest Rate\n(color = Consistency Profile)", color="white")
    axes[0].legend(facecolor="#16213e",labelcolor="white",fontsize=7)
    r,p = spearmanr(df_e4_clean["real_rate"].clip(-20,30),df_e4_clean["negfreq_export"])
    axes[0].text(0.05,0.95,f"ρ={r:.2f} p={p:.3f}", transform=axes[0].transAxes,color="white",fontsize=9)

    # Right: box consistency_risk by consistency_profile
    c_profiles = ["Very Consistent","Consistent","Inconsistent","Very Inconsistent"]
    data_box = [df_e4[df_e4["consistency_profile"]==cp]["consistency_risk"].dropna().values
                for cp in c_profiles if cp in df_e4["consistency_profile"].values]
    labels_c = [cp for cp in c_profiles if cp in df_e4["consistency_profile"].values]
    bp = axes[1].boxplot(data_box,patch_artist=True, medianprops=dict(color="white",linewidth=2))
    for patch,cp in zip(bp["boxes"],labels_c):   patch.set_facecolor(CONSIST_COLORS.get(cp,C["grey"])); patch.set_alpha(0.7)
    axes[1].set_xticklabels([l.replace(" ","") for l in labels_c], color="white",fontsize=7)
    axes[1].set_title("Consistency Risk by Consistency Profile",color="white")
    axes[1].set_ylabel("Consistency Risk [0–1]",color="white")

    plt.tight_layout()
    plt.savefig("../Figures/E4_consistency_profile_indicators.png",dpi=150,bbox_inches="tight")
    plt.show()
```

## F2. Import Growth × GDP Growth & Unemployment
</div>

```python
# Import growth is driven by domestic demand:
# GDP growth ↑ → more imports (income effect)
# Unemployment ↑ → lower domestic demand → fewer imports

df_f2 = df_master[["country_name","cagr_import","volatility_import","cagr_gdp","gdp_pc","inflation","unemployment","current_acct"]] \
                 .dropna(subset=["cagr_import"])

import_vars = ["cagr_import","volatility_import"]
ind_vars_f = [c for c in ["cagr_gdp","gdp_pc","inflation","unemployment","current_acct"] if c in df_f2.columns]
corr_f2 = spearman_corr(df_f2,import_vars,ind_vars_f)
print("📊 Import Growth/Volatility × Indicators:")
print(corr_f2.sort_values("abs_rho",ascending=False).to_string(index=False))

plot_corr_heatmap(corr_f2,"Import Growth × Indicators",min_abs_rho=0.15)

if "cagr_gdp" in df_f2.columns:
    fig,ax = styled_figure(figsize=(13,8))
    ax.set_title("Import CAGR vs GDP CAGR — Domestic Demand Effect",color="white")
    sc = ax.scatter(df_f2["cagr_gdp"]*100,df_f2["cagr_import"]*100, c=df_f2.get("unemployment", pd.Series([5]*len(df_f2))), cmap="RdYlGn_r",alpha=0.6,s=60)
    for _,row in df_f2[df_f2["country_name"].isin(countries)].iterrows():
        ax.annotate(row["country_name"],(row["cagr_gdp"]*100,row["cagr_import"]*100), fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    lim_min = min(df_f2["cagr_gdp"].min(),df_f2["cagr_import"].min())*100*1.1
    lim_max = max(df_f2["cagr_gdp"].max(),df_f2["cagr_import"].max())*100*1.1
    ax.plot([lim_min,lim_max],[lim_min,lim_max], color="white",linestyle="--",alpha=0.5,label="Import=GDP growth")
    ax.axhline(0,color=C["grey"],linestyle=":",alpha=0.4)
    ax.axvline(0,color=C["grey"],linestyle=":",alpha=0.4)
    ax.set_xlabel("GDP CAGR (%)",color="white")
    ax.set_ylabel("Import CAGR (%)",color="white")
    ax.legend(facecolor="#16213e",labelcolor="white")
    plt.colorbar(sc,ax=ax,label="Unemployment (%)")
    r,p = spearmanr(df_f2["cagr_gdp"],df_f2["cagr_import"])
    ax.text(0.05,0.95,f"ρ={r:.2f} p={p:.3f}", transform=ax.transAxes,color="white",fontsize=10)
    plt.tight_layout()
    plt.savefig("../Figures/F2_import_growth_gdp.png",dpi=150,bbox_inches="tight")
    plt.show()
```

# J. Feature Importance — Which Indicators Explain Trade Best?
</div>

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J1. Random Forest — Feature Importance
</div>

```python
# For each trade target, compute RF importance across all indicators

rf_results = {}
targets = {
    "structural_score" : "Trade Power Score (F)",
    "combined_score"   : "Combined Trade Score (H)",
    "growth_risk"      : "Growth Risk",
    "stability_risk"   : "Stability Risk",
    "consistency_risk" : "Consistency Risk",
    "balance_risk"     : "Balance Risk",
}

for target, label in targets.items():
    if target not in df_master.columns: continue
    df_rf = df_master[[target]+ind_vars_all].dropna()
    if len(df_rf)<30: continue
    X = df_rf[ind_vars_all].values
    y = df_rf[target].values
    rf = RandomForestRegressor(n_estimators=300,random_state=42,max_features="sqrt")
    rf.fit(X,y)
    fi = pd.DataFrame({"indicator":ind_vars_all,"importance":rf.feature_importances_})
    fi = fi.sort_values("importance",ascending=False)
    rf_results[target] = {"fi":fi,"r2":rf.score(X,y),"label":label}
    print(f"✅ {label:<35} R²={rf.score(X,y):.3f} | Top: {fi.iloc[0]['indicator']}")

print(f"\n✅ RF computed for {len(rf_results)} targets")
```

## J3. OLS Regression — Top Indicators per Target
</div>

```python
from sklearn.preprocessing import StandardScaler

ols_summary = []
for target,results in rf_results.items():
    top5 = results["fi"].head(5)["indicator"].tolist()
    df_ols = df_master[[target]+top5].dropna()
    if len(df_ols)<25: continue
    X_s = StandardScaler().fit_transform(df_ols[top5].values)
    y_s = df_ols[target].values
    lr = LinearRegression().fit(X_s,y_s)
    r2 = lr.score(X_s,y_s)
    print(f"\n{'═'*55}")
    print(f"  OLS — {results['label']} (R²={r2:.3f})")
    print(f"{'─'*55}")
    for ind_name,coef in sorted(zip(top5,lr.coef_),key=lambda x:abs(x[1]),reverse=True):
        direction = "▲" if coef>0 else "▼"
        print(f"  {direction} {coef:+.4f}  {ind_name}")
    ols_summary.append({"target":results["label"],"r2":r2,"n":len(df_ols)})

print(f"\n{'═'*55}")
print("OLS Summary Table:")
for row in sorted(ols_summary,key=lambda x:x["r2"],reverse=True):
    print(f"  {row['target']:<35} R²={row['r2']:.3f}  n={row['n']}")
```

# K. Combined Clustering — Trade Profiles + Economic Scores
</div>

```python
# Features: trade structural scores + economic scores (multi_df)
econ_score_cols = [c for c in multi_df.columns if "_value_norm" in c]
trade_feat_cols = ["f_norm_rank","g_norm_rank","growth_risk", "stability_risk","consistency_risk","balance_risk"]
trade_feat_cols = [c for c in trade_feat_cols if c in df_master.columns]

df_clust = df_master[["country_name"]+trade_feat_cols].dropna()
df_clust = df_clust.merge(
    multi_df[["country_name"]+econ_score_cols],on="country_name",how="inner"
)
# Normalize econ scores 0→1
for col in econ_score_cols:
    df_clust[col] = df_clust[col]/100.0

feat_cols = trade_feat_cols + econ_score_cols
df_clust = df_clust.dropna(subset=feat_cols)
X_c = RobustScaler().fit_transform(df_clust[feat_cols].values)

# Optimal K
sils = []
for k in range(2,9):
    km = KMeans(n_clusters=k,random_state=42,n_init=10)
    sils.append(silhouette_score(X_c,km.fit_predict(X_c)))

best_k = list(range(2,9))[np.argmax(sils)]
km_final = KMeans(n_clusters=best_k,random_state=42,n_init=20)
df_clust["cluster"] = km_final.fit_predict(X_c).astype(str)

print(f"✅ K={best_k} | Silhouette={silhouette_score(X_c,km_final.labels_):.3f}")
print(f"\nCluster distribution:")
print(df_clust["cluster"].value_counts().sort_index())

# Cluster profiles
profile_table = df_clust.groupby("cluster")[feat_cols].mean().round(3)
print("\nCluster Profiles (mean):")
display(profile_table.T.style.background_gradient(cmap="Blues",axis=1))
```

# L. Global Resilience Score
</div>

```python
def classify_resilience(s):
    if   s>=0.70: return "🟢 High Resilience"
    elif s>=0.50: return "🟡 Moderate Resilience"
    elif s>=0.30: return "🟠 Low Resilience"
    else:         return "🔴 Fragile"

# Components
df_resil = df_master[["country_name","g_norm_rank"]].copy()

# 1. Trade stability (1 - stability_risk)
if "stability_risk" in df_master.columns:
    df_resil["stability_comp"] = 1 - df_master["stability_risk"]

# 2. Trade openness (trade_intensity normalized)
if "trade_intensity" in df_master.columns:
    df_resil["openness_comp"] = normalize_01(df_master["trade_intensity"].fillna(0))

# 3. Economic momentum (Performance_growth_norm)
perf_col = [c for c in multi_df.columns if "Performance_growth_norm" in c]
if perf_col:
    perf = multi_df[["country_name",perf_col[0]]].copy()
    perf[perf_col[0]] /= 100.0
    df_resil = df_resil.merge(perf,on="country_name",how="left")
    df_resil["momentum_comp"] = df_resil[perf_col[0]]
else:
    df_resil["momentum_comp"] = 0.5

# 4. Balance resilience (1 - balance_risk)
if "balance_risk" in df_master.columns:
    df_resil["balance_comp"] = 1 - df_master.set_index("country_name")["balance_risk"].reindex(df_resil["country_name"]).values

WEIGHTS = {"g_norm_rank":0.30,"stability_comp":0.25,
           "openness_comp":0.20,"momentum_comp":0.15,"balance_comp":0.10}
comps = list(WEIGHTS.keys())
df_resil = df_resil.dropna(subset=comps)
df_resil["resilience_score"] = sum(WEIGHTS[c]*df_resil[c] for c in comps)
df_resil["resilience_profile"] = df_resil["resilience_score"].apply(classify_resilience)

print(f"✅ Resilience Score : {len(df_resil)} countries")
print(df_resil["resilience_profile"].value_counts())
print(f"\nSelected countries:")
print(df_resil[df_resil["country_name"].isin(countries)][
    ["country_name","resilience_score","resilience_profile"]
].sort_values("resilience_score",ascending=False).to_string(index=False))
```

```python
def plot_unified_profile_card(country_name):
    trade_row = df_master[df_master["country_name"]==country_name]
    resil_row = df_resil[df_resil["country_name"]==country_name]
    ec_row    = multi_df[multi_df["country_name"]==country_name]
    if trade_row.empty: print(f"⚠️ {country_name} not found"); return

    tr = trade_row.iloc[0]
    rs = resil_row.iloc[0] if not resil_row.empty else None
    ec = ec_row.iloc[0]    if not ec_row.empty    else None

    fig = plt.figure(figsize=(20,14))
    fig.patch.set_facecolor(C["dark"])
    gs = gridspec.GridSpec(3,3,figure=fig,hspace=0.5,wspace=0.4)
    ax_hd = fig.add_subplot(gs[0,:])
    ax_t1 = fig.add_subplot(gs[1,0])
    ax_rd = fig.add_subplot(gs[1,1],polar=True)
    ax_sc = fig.add_subplot(gs[1,2])
    ax_g  = fig.add_subplot(gs[2,0])
    ax_b  = fig.add_subplot(gs[2,1])
    ax_bd = fig.add_subplot(gs[2,2])

    for ax in [ax_hd,ax_t1,ax_sc,ax_g,ax_b,ax_bd]:
        ax.set_facecolor("#16213e")
        for sp in ax.spines.values(): sp.set_edgecolor("#444")

    # Header
    ax_hd.axis("off")
    cp = tr.get("combined_profile","N/A")
    hc = PROFILE_COLORS.get(cp,C["orange"])
    ax_hd.text(0.5,0.7,f"🌍 {country_name} — Unified Trade × Economy Profile",
               ha="center",va="center",fontsize=18,fontweight="bold",color="white",
               transform=ax_hd.transAxes)
    ax_hd.text(0.5,0.25,f"Trade: {cp}  |  Structural: {tr.get('structural_profile','N/A')}  |  Resilience: {rs['resilience_profile'] if rs is not None else 'N/A'}",
               ha="center",va="center",fontsize=12,color=hc,transform=ax_hd.transAxes)

    # Panel 1: Key Trade Values
    ax_t1.axis("off"); ax_t1.set_title("Trade Structural Values",color="white",fontsize=10,pad=8)
    rows_kv = [
        ("Export Share",      f"{tr.get('export_share_structural',np.nan):.2f}%"),
        ("Coverage Ratio",    f"{tr.get('coverage_ratio_structural',np.nan):.2f}"),
        ("CAGR Export",       f"{tr.get('cagr_export',np.nan)*100:.1f}%"),
        ("Volatility Export", f"{tr.get('volatility_export',np.nan):.2f}"),
        ("Neg. Freq Export",  f"{tr.get('negfreq_export',np.nan):.2f}"),
        ("Trade Intensity",   f"{tr.get('trade_intensity',np.nan)*100:.1f}%" if "trade_intensity" in tr.index else "N/A"),
    ]
    y=0.85
    for lbl,val in rows_kv:
        ax_t1.text(0.05,y,lbl,transform=ax_t1.transAxes,color=C["grey"],fontsize=9)
        ax_t1.text(0.95,y,val,transform=ax_t1.transAxes,color="white",fontsize=9,
                   ha="right",fontweight="bold")
        y-=0.14

    # Panel 2: 5 Profiles Radar
    PROFILE_LABELS = ["Size","Balance","Growth","Stability","Consistency"]
    PROFILE_ENCODE_VIZ = {
        "size_profile"        :{"Leader":1.0,"Major":0.67,"Mid":0.33,"Small":0.0},
        "balance_profile"     :{"Strong Surplus":1.0,"Slight Surplus":0.67,"Slight Deficit":0.33,"Structural Deficit":0.0},
        "growth_profile"      :{"Strong Grower":1.0,"Moderate Grower":0.67,"Weak Grower":0.33,"Declining":0.0},
        "stability_profile"   :{"Stable":1.0,"Moderate":0.67,"Volatile":0.33,"Very Volatile":0.0},
        "consistency_profile" :{"Very Consistent":1.0,"Consistent":0.67,"Inconsistent":0.33,"Very Inconsistent":0.0},
    }
    vals_radar = [PROFILE_ENCODE_VIZ[p].get(tr.get(p,""),0.5)
                  for p in PROFILE_ENCODE_VIZ.keys()]
    angles = np.linspace(0,2*np.pi,5,endpoint=False).tolist()
    angles += angles[:1]; vals_radar += vals_radar[:1]
    ax_rd.set_facecolor("#16213e")
    ax_rd.plot(angles,vals_radar,color=C["blue"],linewidth=2)
    ax_rd.fill(angles,vals_radar,color=C["blue"],alpha=0.2)
    ax_rd.set_xticks(angles[:-1])
    ax_rd.set_xticklabels(PROFILE_LABELS,color="white",size=8)
    ax_rd.set_yticks([0.25,0.5,0.75,1.0])
    ax_rd.set_yticklabels(["","","",""],color="#888")
    ax_rd.set_title("5 Trade Profiles",color="white",fontsize=10,pad=12)
    ax_rd.tick_params(colors="white")

    # Panel 3: Score bars
    ax_sc.axis("off"); ax_sc.set_title("Scores",color="white",fontsize=10,pad=8)
    score_items = [
        ("Structural Score",tr.get("structural_score",0),C["blue"]),
        ("Combined Score",  tr.get("combined_score",0),  C["green"]),
        ("Resilience",      rs["resilience_score"] if rs is not None else 0,C["purple"]),
    ]
    if ec is not None:
        for col in [c for c in ["Economic Health_value_norm","Performance_value_norm",
                                 "Competitiveness_value_norm"] if c in (ec.index if hasattr(ec,"index") else [])]:
            score_items.append((col.replace("_value_norm","").replace("Economic Health","Eco. Health"),
                                ec.get(col,0)/100,C["orange"]))
    bar_y = np.arange(len(score_items))
    ax_sc.axis("on"); ax_sc.set_facecolor("#16213e")
    bars = ax_sc.barh(bar_y,[s[1] for s in score_items],
                      color=[s[2] for s in score_items],alpha=0.8,height=0.6)
    ax_sc.set_yticks(bar_y)
    ax_sc.set_yticklabels([s[0] for s in score_items],color="white",fontsize=8)
    ax_sc.set_xlim(0,1.1)
    ax_sc.axvline(0.5,color=C["grey"],linestyle="--",alpha=0.4)
    ax_sc.tick_params(colors="white",labelsize=7)
    for sp in ax_sc.spines.values(): sp.set_edgecolor("#444")
    ax_sc.grid(alpha=0.2,color="#555")

    # Panel 4: Risk radar
    risk_cols = ["growth_risk","stability_risk","consistency_risk","balance_risk"]
    risk_labels = ["Growth","Stability","Consistency","Balance"]
    risk_vals = [float(tr.get(c,0.5)) for c in risk_cols]
    angles2 = np.linspace(0,2*np.pi,4,endpoint=False).tolist()
    angles2 += angles2[:1]; risk_vals += risk_vals[:1]
    ax_g.remove(); ax_g = fig.add_subplot(gs[2,0],polar=True)
    ax_g.set_facecolor("#16213e")
    ax_g.plot(angles2,risk_vals,color=C["red"],linewidth=2)
    ax_g.fill(angles2,risk_vals,color=C["red"],alpha=0.2)
    ax_g.set_xticks(angles2[:-1])
    ax_g.set_xticklabels(risk_labels,color="white",size=8)
    ax_g.set_yticks([0.25,0.5,0.75,1.0])
    ax_g.set_yticklabels([],color="#888")
    ax_g.set_title("Risk Dimensions",color="white",fontsize=10,pad=12)
    ax_g.tick_params(colors="white")

    # Panel 5: Economic scores bar
    if ec is not None:
        econ_disp = {
            "Economic Health_value_norm" :"Health (Value)",
            "Economic Health_growth_norm":"Health (Growth)",
            "Performance_value_norm"     :"Perf. (Value)",
            "Performance_growth_norm"    :"Perf. (Growth)",
        }
        econ_available = {k:v for k,v in econ_disp.items()
                          if k in (ec.index if hasattr(ec,"index") else [])}
        ax_b.axis("on"); ax_b.set_facecolor("#16213e")
        if econ_available:
            y_pos = np.arange(len(econ_available))
            vals  = [ec.get(k,0)/100 for k in econ_available.keys()]
            ax_b.barh(y_pos,vals,color=[C["blue"],C["orange"],C["green"],C["purple"]][:len(vals)],alpha=0.8,height=0.6)
            ax_b.set_yticks(y_pos)
            ax_b.set_yticklabels(list(econ_available.values()),color="white",fontsize=8)
            ax_b.set_xlim(0,1.1)
            ax_b.axvline(0.5,color=C["grey"],linestyle="--",alpha=0.4)
        ax_b.set_title("Economic Scores",color="white",fontsize=10,pad=8)
        ax_b.tick_params(colors="white",labelsize=7)
        for sp in ax_b.spines.values(): sp.set_edgecolor("#444")
        ax_b.grid(alpha=0.2,color="#555")

    # Panel 6: Profile badges
    ax_bd.axis("off"); ax_bd.set_title("Profile Labels",color="white",fontsize=10,pad=8)
    badges = [
        ("Size Profile",        tr.get("size_profile","N/A"),       SIZE_COLORS),
        ("Balance Profile",     tr.get("balance_profile","N/A"),     BALANCE_COLORS),
        ("Growth Profile",      tr.get("growth_profile","N/A"),      GROWTH_COLORS),
        ("Stability Profile",   tr.get("stability_profile","N/A"),   STABILITY_COLORS),
        ("Consistency Profile", tr.get("consistency_profile","N/A"), CONSIST_COLORS),
        ("Trade Profile",       tr.get("structural_profile","N/A"),  {}),
    ]
    y=0.90
    for lbl,badge,cmap in badges:
        ax_bd.text(0.05,y,lbl,transform=ax_bd.transAxes,color=C["grey"],fontsize=8)
        bc = cmap.get(badge,C["orange"]) if cmap else C["blue"]
        ax_bd.text(0.05,y-0.06,badge,transform=ax_bd.transAxes,
                   color=bc,fontsize=9,fontweight="bold")
        y-=0.16

    plt.suptitle(f"Unified Profile — {country_name}",fontsize=16,
                 color="white",y=1.01,fontweight="bold")
    plt.savefig(f"Figures/M_profile_{country_name.replace(' ','_')}.png",
                dpi=150,bbox_inches="tight")
    plt.show()

print("✅ plot_unified_profile_card() ready")
```

# O. Conclusions — Direction 1
</div>

### Framework Delivered

| Section | Analysis | Key Output |
|---|---|---|
| **C** | Data fusion + Population derivation | `df_master` (trade × indicators) |
| **D** | Trade Intensity, Elasticity, Per Capita | Combined metrics per country |
| **E** | Export: Size, Growth, Stability, Consistency × Indicators | Spearman ρ per family |
| **F** | Import: Volume, Growth, Stability × Indicators | Symmetric analysis |
| **G** | Total trade: Elasticity, Volatility, Cross-validation | Trade%GDP coherence check |
| **H** | Balance & Coverage × Current account, Debt, Inflation | Balance risk drivers |
| **I** | Unified correlation matrix | Full ρ heatmap |
| **J** | RF importance + OLS per target | Top indicators by family |
| **K** | Combined clustering Trade + Economy | Country typology |
| **L** | Resilience Score (4 components) | Country ranking |
| **M** | Unified Profile Card | 6-panel per-country synthesis |

### Key Economic Hypotheses Tested

1. **Export Size ~ GDP** (structural) — expected strong ρ
2. **Export Growth ~ GDP Growth** (dynamic) — synchronisation check
3. **Export Volatility ~ Inflation Volatility** — macro instability driver
4. **Consistency ~ Real Interest Rate** — financing cost → negative growth frequency
5. **Balance ~ Current Account** — cross-dataset coherence
6. **Trade Elasticity** — trade-led vs domestic-led growth by country

### Limitations

- 17 indicators available — Population, FDI, Exchange rate volatility absent from dataset
- Population derived as GDP/GDP_pc — approximation (varies slightly from official figures)
- Structural median aggregation loses annual dynamics (covered by panel in Section G)
- In-sample RF — no cross-validation (n≈130 countries limits train/test split)

### Next Step — Direction 2
> *Which trade variables predict future economic indicators?*
> Lagged correlations, Granger causality, export-led growth testing.
