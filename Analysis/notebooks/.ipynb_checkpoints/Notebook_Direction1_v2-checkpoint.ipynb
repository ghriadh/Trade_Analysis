## 2. Style & Helpers
</div>

```python
C = {"blue":"#3498db","green":"#2ecc71","red":"#e74c3c","orange":"#f39c12",
     "purple":"#9b59b6","grey":"#95a5a6","dark":"#1a1a2e","teal":"#1abc9c"}

SIZE_COLORS      = {"Leader":C["green"],"Major":C["blue"],"Mid":C["orange"],"Small":C["red"]}
BALANCE_COLORS   = {"Strong Surplus":C["green"],"Slight Surplus":C["teal"],
                    "Slight Deficit":C["orange"],"Structural Deficit":C["red"]}
GROWTH_COLORS    = {"Strong Grower":C["green"],"Moderate Grower":C["blue"],
                    "Weak Grower":C["orange"],"Declining":C["red"]}
STABILITY_COLORS = {"Stable":C["green"],"Moderate":C["blue"],
                    "Volatile":C["orange"],"Very Volatile":C["red"]}
CONSIST_COLORS   = {"Very Consistent":C["green"],"Consistent":C["blue"],
                    "Inconsistent":C["orange"],"Very Inconsistent":C["red"]}

def styled_figure(figsize=(12,6)):
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    ax.tick_params(colors="white"); ax.xaxis.label.set_color("white")
    ax.yaxis.label.set_color("white"); ax.title.set_color("white")
    for sp in ax.spines.values(): sp.set_edgecolor("#444")
    ax.grid(alpha=0.2,color="#555"); return fig,ax

def styled_subplots(nrows=1,ncols=2,figsize=(18,6)):
    fig,axes = plt.subplots(nrows,ncols,figsize=figsize)
    fig.patch.set_facecolor(C["dark"])
    for ax in np.array(axes).flatten():
        ax.set_facecolor("#16213e"); ax.tick_params(colors="white")
        ax.xaxis.label.set_color("white"); ax.yaxis.label.set_color("white")
        ax.title.set_color("white")
        for sp in ax.spines.values(): sp.set_edgecolor("#444")
        ax.grid(alpha=0.2,color="#555")
    return fig,axes

def normalize_01(s):
    mn,mx = s.min(),s.max(); return (s-mn)/(mx-mn+1e-9)

def spearman_matrix(df, trade_vars, ind_vars, min_obs=25):
    """Compute full Spearman ρ matrix between trade vars and indicator vars."""
    rows=[]
    for tv in trade_vars:
        for iv in ind_vars:
            sub = df[[tv,iv]].dropna()
            if len(sub)<min_obs: continue
            r,p = spearmanr(sub[tv],sub[iv])
            rows.append({"trade_var":tv,"indicator":iv,
                         "rho":round(r,3),"pvalue":round(p,4),
                         "sig":p<0.05,"abs_rho":abs(r)})
    return pd.DataFrame(rows)

def heatmap_corr(corr_df, title, min_abs_rho=0.15, figsize=(12,7), fname=None):
    strong = corr_df[(corr_df["sig"])&(corr_df["abs_rho"]>=min_abs_rho)]
    if strong.empty: print(f"No sig. correlations above |ρ|={min_abs_rho}"); return
    pivot = strong.pivot(index="indicator",columns="trade_var",values="rho")
    pivot = pivot.sort_values(pivot.columns[0],ascending=False)
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1,
                annot=True,fmt=".2f",annot_kws={"size":8},
                linewidths=0.5,linecolor="#333",cbar_kws={"label":"Spearman ρ"})
    ax.set_title(f"{title}\n(sig |ρ|≥{min_abs_rho})",color="white",pad=10)
    ax.tick_params(colors="white",labelsize=8)
    ax.set_xlabel("Trade Variable",color="white")
    ax.set_ylabel("Economic Indicator",color="white")
    plt.xticks(rotation=30,ha="right"); plt.tight_layout()
    f = fname or title.lower().replace(" ","_").replace("×","x")[:40]
    plt.savefig(f"Figures/{f}.png",dpi=150,bbox_inches="tight"); plt.show()

def scatter_trade_ind(df, x_col, y_col, color_col=None, color_map=None,
                      title="", xlabel="", ylabel="", countries=None,
                      xlog=False, ylog=False, fname=None, clip_x=None, clip_y=None):
    """Standard scatter trade × indicator with optional color grouping."""
    fig,ax = styled_figure(figsize=(13,8))
    ax.set_title(title,color="white")
    df_plot = df[[x_col,y_col]+(([color_col] if color_col else []))+(["country_name"] if "country_name" in df.columns else [])].dropna()
    if clip_x: df_plot[x_col] = df_plot[x_col].clip(*clip_x)
    if clip_y: df_plot[y_col] = df_plot[y_col].clip(*clip_y)
    if color_col and color_map:
        for grp,col in color_map.items():
            sub = df_plot[df_plot[color_col]==grp]
            ax.scatter(sub[x_col],sub[y_col],color=col,alpha=0.6,s=50,label=grp)
        ax.legend(facecolor="#16213e",labelcolor="white",fontsize=8)
    else:
        ax.scatter(df_plot[x_col],df_plot[y_col],color=C["blue"],alpha=0.6,s=50)
    if countries and "country_name" in df_plot.columns:
        for _,row in df_plot[df_plot["country_name"].isin(countries)].iterrows():
            ax.annotate(row["country_name"],(row[x_col],row[y_col]),
                        fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    if xlog: ax.set_xscale("log")
    if ylog: ax.set_yscale("log")
    ax.set_xlabel(xlabel,color="white"); ax.set_ylabel(ylabel,color="white")
    r,p = spearmanr(df_plot[x_col],df_plot[y_col])
    ax.text(0.05,0.95,f"Spearman ρ={r:.2f} (p={p:.3f})",
            transform=ax.transAxes,color="white",fontsize=9)
    plt.tight_layout()
    if fname: plt.savefig(f"Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

def boxplot_by_quartile(df, value_col, group_col, title="", ylabel="",
                        n_quartiles=4, fname=None):
    """Boxplot of value_col by quartile of group_col — clearer than scatter for weak ρ."""
    df_q = df[[value_col,group_col]].dropna().copy()
    labels = [f"Q{i+1}" for i in range(n_quartiles)]
    df_q["quartile"] = pd.qcut(df_q[group_col],n_quartiles,labels=labels)
    fig,ax = styled_figure(figsize=(10,6))
    ax.set_title(title,color="white")
    data_box = [df_q[df_q["quartile"]==q][value_col].dropna().values for q in labels]
    bp = ax.boxplot(data_box,patch_artist=True,medianprops=dict(color="white",linewidth=2))
    colors_q = [C["red"],C["orange"],C["blue"],C["green"]]
    for patch,col in zip(bp["boxes"],colors_q): patch.set_facecolor(col); patch.set_alpha(0.7)
    ax.set_xticklabels([f"{l}\n{group_col}" for l in labels],color="white",fontsize=8)
    ax.set_ylabel(ylabel,color="white")
    plt.tight_layout()
    if fname: plt.savefig(f"Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

countries = ["USA","China","Germany","France","India","Brazil","Japan","Canada"]
print("✅ Style helpers OK")
```

```python
# ── Short key mapping for all available indicators ────────────────────────────
IND = {
    "gdp"          : "GDP (current US$)",
    "gdp_pc"       : "GDP per capita (current US$)",
    "inflation"    : "Inflation, consumer prices (annual %)",
    "real_rate"    : "Real interest rate (%)",
    "unemployment" : "Unemployment, total (% of total labor force) (modeled ILO estimate)",
    "current_acct" : "Current account balance (% of GDP)",
    "trade_pct_gdp": "Trade (% of GDP)",
    "ext_debt"     : "External debt stocks, total (DOD, current US$)",
    "expense"      : "Expense (% of GDP)",
}
# Keep only those present in the dataset
IND = {k:v for k,v in IND.items()
       if v in df["indicator_name"].unique()}
IND_REVERSE = {v:k for k,v in IND.items()}
print(f"✅ {len(IND)} indicators confirmed : {list(IND.keys())}")
```

## C4. Build Annual Panel & Structural Table
</div>

```python
# ── Annual pivot : country × year ─────────────────────────────────────────────
df_ind_annual = (
    df[df["entity_type"]=="COUNTRY"]
    .pivot_table(index=["country_name","year"],
                 columns="indicator_name",values="value")
    .reset_index()
)
df_ind_annual.columns.name = None
df_ind_annual = df_ind_annual.rename(columns=IND_REVERSE)
ind_cols = [k for k in IND.keys() if k in df_ind_annual.columns]

# YoY growth per indicator (annual)
df_ind_annual = df_ind_annual.sort_values(["country_name","year"])
for col in ind_cols:
    df_ind_annual[f"{col}_yoy"] = df_ind_annual.groupby("country_name")[col].pct_change()*100

# ── Structural : country-level median ─────────────────────────────────────────
df_ind_struct = df_ind_annual.groupby("country_name")[ind_cols].median().reset_index()

# ── GDP CAGR from growth_df (already computed in Indicators notebook) ──────────
gdp_ind_name = IND.get("gdp","GDP (current US$)")
gdp_growth = growth_df[
    (growth_df["indicator_name"]==gdp_ind_name) &
    (growth_df["entity_type"]=="COUNTRY")
][["country_name","cagr","volatility"]].rename(
    columns={"cagr":"cagr_gdp","volatility":"vol_gdp"}
)

print(f"✅ df_ind_annual : {df_ind_annual.shape}")
print(f"✅ df_ind_struct : {df_ind_struct.shape}")
print(f"✅ GDP CAGR      : {gdp_growth.shape} — from growth_df (no recomputation)")
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## C5. Build Gini & CV Tables (from Indicators Analysis functions)
</div>

```python
# ── Reuse functions from Indicators notebook ──────────────────────────────────
# compute_global_concentration() → Gini per indicator per year
# compute_stability_all_indicators() → CV per indicator

# Gini for all indicators (cross-country concentration per year)
gini_all = []
for ind_name in IND.values():
    try:
        g = compute_global_concentration(df, ind_name)
        gini_all.append(g)
    except: pass

df_gini_all = pd.concat(gini_all, ignore_index=True) if gini_all else pd.DataFrame()

# CV for all indicators (cross-country stability)
df_cv_all = compute_stability_all_indicators(df)

# Gini structural (median over period per indicator)
if not df_gini_all.empty:
    df_gini_struct = (
        df_gini_all.groupby("indicator_name")["gini"]
        .median().reset_index()
        .rename(columns={"gini":"gini_median"})
        .assign(indicator=lambda x: x["indicator_name"].map(IND_REVERSE))
    )

print(f"✅ df_gini_all  : {df_gini_all.shape}")
print(f"✅ df_cv_all    : {df_cv_all.shape}")
print(df_cv_all.to_string(index=False))
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## C6. Master Merge — df_master
</div>

```python
# ── df_structural is the backbone — merge everything onto it ─────────────────
df_master = df_structural.merge(df_ind_struct, on="country_name", how="left")
df_master = df_master.merge(gdp_growth, on="country_name", how="left")
df_master = df_master.merge(
    df_combined[["country_name","combined_score","f_norm_rank",
                 "g_norm_rank","combined_profile"]],
    on="country_name", how="left"
)
econ_score_cols = [c for c in multi_df.columns
                   if "_value_norm" in c or "_growth_norm" in c]
df_master = df_master.merge(
    multi_df[["country_name"]+econ_score_cols], on="country_name", how="left"
)

# ── Population = GDP / GDP_per_capita ─────────────────────────────────────────
if "gdp" in df_master.columns and "gdp_pc" in df_master.columns:
    mask = df_master["gdp_pc"].notna() & (df_master["gdp_pc"]>0)
    df_master.loc[mask,"population"] = (
        df_master.loc[mask,"gdp"] / df_master.loc[mask,"gdp_pc"]
    )

# ── Per capita trade ──────────────────────────────────────────────────────────
for flow_col,pc_col in [("total_export","export_pc"),("total_import","import_pc")]:
    if flow_col in df_master.columns:
        mask = df_master.get("population",pd.Series()).notna() & (df_master.get("population",pd.Series())>0)
        df_master.loc[mask,pc_col] = (
            df_master.loc[mask,flow_col] / df_master.loc[mask,"population"]
        )

# ── Trade Intensity, Elasticity ───────────────────────────────────────────────
if "gdp" in df_master.columns:
    mask = df_master["gdp"].notna() & (df_master["gdp"]>0)
    df_master.loc[mask,"trade_intensity"] = (
        (df_master.loc[mask,"total_export"]+df_master.loc[mask,"total_import"])
        / df_master.loc[mask,"gdp"]
    )
    df_master.loc[mask,"export_dep"]  = df_master.loc[mask,"total_export"]/df_master.loc[mask,"gdp"]
    df_master.loc[mask,"import_dep"]  = df_master.loc[mask,"total_import"]/df_master.loc[mask,"gdp"]
    df_master.loc[mask,"balance_dep"] = (df_master.loc[mask,"total_export"]-df_master.loc[mask,"total_import"])/df_master.loc[mask,"gdp"]

# Trade Elasticity = CAGR_export / CAGR_GDP
mask_e = df_master["cagr_gdp"].notna() & (df_master["cagr_gdp"]!=0)
df_master.loc[mask_e,"trade_elasticity_x"] = (
    df_master.loc[mask_e,"cagr_export"] / df_master.loc[mask_e,"cagr_gdp"]
)
df_master.loc[mask_e,"trade_elasticity_m"] = (
    df_master.loc[mask_e,"cagr_import"] / df_master.loc[mask_e,"cagr_gdp"]
)

# External debt / GDP
if "ext_debt" in df_master.columns and "gdp" in df_master.columns:
    mask = df_master["gdp"].notna() & (df_master["gdp"]>0)
    df_master.loc[mask,"ext_debt_gdp"] = (
        df_master.loc[mask,"ext_debt"] / df_master.loc[mask,"gdp"]
    )

print(f"✅ df_master : {df_master.shape} | {df_master['country_name'].nunique()} countries")
print(f"   Columns  : {len(df_master.columns)}")
```

## D2. Trade Elasticity — 6 Situations
</div>

Trade Elasticity = CAGR_export / CAGR_GDP

| # | CAGR_export | CAGR_GDP | Elasticity | Reading |
|---|---|---|---|---|
| 1 | + | + | > 1 | Trade-led : opening |
| 2 | + | + | 0→1 | Endogenous : economy grows faster |
| 3 | + | − | < 0 | 🟢 Trade resistant : grows despite recession |
| 4 | − | + | < 0 | 🟠 Decoupling : economy grows, trade declines |
| 5 | − | − | > 1 | 🔴 Double decline : trade collapses faster |
| 6 | − | − | 0→1 | 🟡 Double decline : economy more affected |


```python
def classify_elasticity_6(row):
    xe, xg = row["cagr_export"], row["cagr_gdp"]
    if pd.isna(xe) or pd.isna(xg): return "Unknown"
    if xe>=0 and xg>=0:
        return "1-Trade-led" if xe>xg else "2-Endogenous"
    elif xe>=0 and xg<0:
        return "3-Trade Resistant"
    elif xe<0 and xg>=0:
        return "4-Decoupling"
    elif xe<0 and xg<0:
        return "5-Double Decline (trade faster)" if abs(xe)>abs(xg) else "6-Double Decline (econ faster)"
    return "Unknown"

SIT_COLORS = {
    "1-Trade-led"                       : C["green"],
    "2-Endogenous"                       : C["teal"],
    "3-Trade Resistant"                  : C["blue"],
    "4-Decoupling"                       : C["orange"],
    "5-Double Decline (trade faster)"    : C["red"],
    "6-Double Decline (econ faster)"     : C["purple"],
}

df_d2 = df_master[["country_name","cagr_export","cagr_gdp",
                    "trade_elasticity_x","size_profile"]].dropna()
df_d2["situation"] = df_d2.apply(classify_elasticity_6,axis=1)

fig,axes = styled_subplots(1,2,figsize=(18,7))
fig.suptitle("Trade Elasticity — 6 Situations",fontsize=14,color="white",y=1.01)

# Left: CAGR_export vs CAGR_GDP scatter (6 situations)
for sit,col in SIT_COLORS.items():
    sub = df_d2[df_d2["situation"]==sit]
    axes[0].scatter(sub["cagr_gdp"]*100,sub["cagr_export"]*100,
                    color=col,alpha=0.7,s=60,label=sit.split("-",1)[1])
lim = max(df_d2[["cagr_export","cagr_gdp"]].abs().max().max())*100*1.1
axes[0].plot([-lim,lim],[-lim,lim],color="white",linestyle="--",alpha=0.4,label="Elasticity=1")
axes[0].axhline(0,color=C["grey"],linestyle=":",alpha=0.4)
axes[0].axvline(0,color=C["grey"],linestyle=":",alpha=0.4)
for _,row in df_d2[df_d2["country_name"].isin(countries)].iterrows():
    axes[0].annotate(row["country_name"],
                     (row["cagr_gdp"]*100,row["cagr_export"]*100),
                     fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
axes[0].set_xlabel("CAGR GDP (%)",color="white")
axes[0].set_ylabel("CAGR Export (%)",color="white")
axes[0].set_title("CAGR Export vs CAGR GDP\n(6 situations)",color="white")
axes[0].legend(facecolor="#16213e",labelcolor="white",fontsize=7)

# Right: Situation distribution
sit_counts = df_d2["situation"].value_counts()
colors_sit = [SIT_COLORS.get(s,C["grey"]) for s in sit_counts.index]
axes[1].barh([s.split("-",1)[1] for s in sit_counts.index],
             sit_counts.values,color=colors_sit,alpha=0.8)
axes[1].set_title("Country Distribution by Situation",color="white")
axes[1].set_xlabel("Number of countries",color="white")
plt.tight_layout()
plt.savefig("Figures/D2_trade_elasticity_6sit.png",dpi=150,bbox_inches="tight")
plt.show()

print("\nSituation distribution:")
print(df_d2["situation"].value_counts().to_string())
print("\nSelected countries:")
print(df_d2[df_d2["country_name"].isin(countries)][
    ["country_name","cagr_export","cagr_gdp","trade_elasticity_x","situation"]
].sort_values("trade_elasticity_x",ascending=False).to_string(index=False))
```

# J. RF + OLS — Top Indicators per Trade Variable
</div>

**RF** identifies the most important indicators (non-linear, multivariate).  
**OLS** confirms the direction and magnitude (linear, signed coefficients).  
Results guide sections E, F, G, H.


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J1. Define Trade Variable Groups
</div>

```python
# ── Trade variables : raw data ────────────────────────────────────────────────
TRADE_VARS_DATA = {
    # Export
    "export_share_structural" : "Export Share",
    "cagr_export"             : "CAGR Export",
    "volatility_export"       : "Volatility Export",
    "negfreq_export"          : "Neg.Freq Export",
    "export_dep"              : "Export/GDP",
    # Import
    "import_share_structural" : "Import Share",
    "cagr_import"             : "CAGR Import",
    "volatility_import"       : "Volatility Import",
    "import_dep"              : "Import/GDP",
    # Total
    "trade_intensity"         : "Trade Intensity",
    "trade_elasticity_x"      : "Trade Elasticity",
    # Balance
    "coverage_ratio_structural":"Coverage Ratio",
    "balance_dep"             : "Balance/GDP",
}

# ── Trade variables : scores & risks ─────────────────────────────────────────
TRADE_VARS_SCORES = {
    "structural_score"  : "Structural Score",
    "combined_score"    : "Combined Score",
    "growth_risk"       : "Growth Risk",
    "stability_risk"    : "Stability Risk",
    "consistency_risk"  : "Consistency Risk",
    "balance_risk"      : "Balance Risk",
}

data_vars   = [v for v in TRADE_VARS_DATA.keys()   if v in df_master.columns]
score_vars  = [v for v in TRADE_VARS_SCORES.keys() if v in df_master.columns]
all_vars    = data_vars + score_vars
ind_vars    = [c for c in ind_cols if c in df_master.columns]

print(f"✅ Trade data vars  : {len(data_vars)}")
print(f"✅ Trade score vars : {len(score_vars)}")
print(f"✅ Indicator vars   : {len(ind_vars)} → {ind_vars}")
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J2. Random Forest — Trade Data Variables × All Indicators
</div>

```python
rf_results_data = {}

for target in data_vars:
    df_rf = df_master[[target]+ind_vars].dropna()
    if len(df_rf)<30: continue
    X = df_rf[ind_vars].values; y = df_rf[target].values
    rf = RandomForestRegressor(n_estimators=300,random_state=42,max_features="sqrt")
    rf.fit(X,y)
    fi = pd.DataFrame({"indicator":ind_vars,"importance":rf.feature_importances_})
    fi = fi.sort_values("importance",ascending=False)
    rf_results_data[target] = {"fi":fi,"r2":rf.score(X,y),
                                "label":TRADE_VARS_DATA.get(target,target)}

print("RF Results — Trade Data Variables:")
print(f"{'Target':<30} {'R²':>6}  Top Indicator")
print("─"*60)
for t,r in sorted(rf_results_data.items(),key=lambda x:-x[1]["r2"]):
    top = r["fi"].iloc[0]
    print(f"{r['label']:<30} {r['r2']:>6.3f}  {top['indicator']} ({top['importance']:.3f})")
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J3. Random Forest — Trade Scores × All Indicators
</div>

```python
rf_results_scores = {}

for target in score_vars:
    df_rf = df_master[[target]+ind_vars].dropna()
    if len(df_rf)<30: continue
    X = df_rf[ind_vars].values; y = df_rf[target].values
    rf = RandomForestRegressor(n_estimators=300,random_state=42,max_features="sqrt")
    rf.fit(X,y)
    fi = pd.DataFrame({"indicator":ind_vars,"importance":rf.feature_importances_})
    fi = fi.sort_values("importance",ascending=False)
    rf_results_scores[target] = {"fi":fi,"r2":rf.score(X,y),
                                  "label":TRADE_VARS_SCORES.get(target,target)}

print("RF Results — Trade Score Variables:")
print(f"{'Target':<30} {'R²':>6}  Top Indicator")
print("─"*60)
for t,r in sorted(rf_results_scores.items(),key=lambda x:-x[1]["r2"]):
    top = r["fi"].iloc[0]
    print(f"{r['label']:<30} {r['r2']:>6.3f}  {top['indicator']} ({top['importance']:.3f})")

# ── Combined importance heatmap ───────────────────────────────────────────────
all_rf = {**rf_results_data,**rf_results_scores}
if all_rf:
    fi_matrix = pd.DataFrame({
        r["label"]:r["fi"].set_index("indicator")["importance"]
        for t,r in all_rf.items()
    }).fillna(0)

    fig,ax = plt.subplots(figsize=(max(14,len(fi_matrix.columns)*1.5),
                                   max(5,len(fi_matrix)*0.6)))
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(fi_matrix,ax=ax,cmap="YlOrRd",annot=True,fmt=".3f",
                annot_kws={"size":7},linewidths=0.5,linecolor="#333",
                cbar_kws={"label":"RF Importance"})
    ax.set_title("RF Feature Importance — All Trade Variables × All Indicators",
                 color="white",pad=10)
    ax.tick_params(colors="white",labelsize=8)
    plt.xticks(rotation=30,ha="right"); plt.tight_layout()
    plt.savefig("Figures/J3_rf_importance_heatmap.png",dpi=150,bbox_inches="tight")
    plt.show()
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## J4. OLS — Direction & Magnitude
</div>

```python
print("OLS Regression — Standardized Coefficients")
print("(▲ = higher indicator → higher trade var, ▼ = inverse)")
print("="*65)

ols_summary = []
for target in all_vars:
    res = all_rf.get(target)
    if not res: continue
    top5 = res["fi"].head(5)["indicator"].tolist()
    df_ols = df_master[[target]+top5].dropna()
    if len(df_ols)<25: continue
    X_s = StandardScaler().fit_transform(df_ols[top5].values)
    y_s = df_ols[target].values
    lr = LinearRegression().fit(X_s,y_s)
    r2 = lr.score(X_s,y_s)
    print(f"\n  {res['label']} (R²={r2:.3f})")
    print(f"  {'─'*50}")
    for ind_name,coef in sorted(zip(top5,lr.coef_),key=lambda x:abs(x[1]),reverse=True):
        direction = "▲" if coef>0 else "▼"
        print(f"    {direction} {coef:+.4f}  {ind_name}")
    ols_summary.append({"target":res["label"],"r2_rf":res["r2"],"r2_ols":r2})

print("\n" + "="*65)
print("Summary — R² comparison RF vs OLS:")
for row in sorted(ols_summary,key=lambda x:-x["r2_rf"]):
    print(f"  {row['target']:<35} RF={row['r2_rf']:.3f}  OLS={row['r2_ols']:.3f}")
```

<div style="background-color:#eaf2f8;padding-left:20px;border:1px solid;border-radius:5px;">

## H1. Coverage Ratio × Current Account, Ext.Debt, Real Rate, Inflation
</div>

```python
df_h1 = df_master[["country_name","coverage_ratio_structural","balance_profile",
                    "current_acct","ext_debt_gdp","real_rate","inflation"]].dropna(subset=["coverage_ratio_structural"])

cov_ind_pairs = [
    ("current_acct", "Current Account % GDP", C["green"]),
    ("ext_debt_gdp", "External Debt / GDP",   C["orange"]),
    ("real_rate",    "Real Interest Rate",     C["blue"]),
    ("inflation",    "Inflation CPI (%)",      C["red"]),
]

fig,axes = styled_subplots(2,2,figsize=(18,14))
fig.suptitle("Coverage Ratio (X/M) × Economic Indicators",fontsize=14,color="white",y=1.01)
axes_flat = axes.flatten()

for i,(ind,label,col) in enumerate(cov_ind_pairs):
    df_sub = df_h1[["country_name","coverage_ratio_structural","balance_profile",ind]].dropna()
    for bp_,bc in BALANCE_COLORS.items():
        sub = df_sub[df_sub["balance_profile"]==bp_]
        x_vals = sub[ind].clip(-5,50) if ind=="inflation" else sub[ind].clip(-20,30) if ind=="real_rate" else sub[ind].clip(0,3) if "debt" in ind else sub[ind]
        axes_flat[i].scatter(x_vals,sub["coverage_ratio_structural"],
                             color=bc,alpha=0.6,s=50,label=bp_)
    for _,row in df_sub[df_sub["country_name"].isin(countries)].iterrows():
        xv = row[ind]
        if ind=="inflation": xv=np.clip(xv,-5,50)
        elif ind=="real_rate": xv=np.clip(xv,-20,30)
        elif "debt" in ind: xv=min(xv,3)
        axes_flat[i].annotate(row["country_name"],(xv,row["coverage_ratio_structural"]),
                              fontsize=7,color="white",xytext=(4,2),textcoords="offset points")
    axes_flat[i].axhline(1.0,color="white",linestyle="--",alpha=0.5,label="Coverage=1")
    axes_flat[i].axvline(0,color=C["grey"],linestyle=":",alpha=0.3)
    axes_flat[i].set_xlabel(label,color="white")
    axes_flat[i].set_ylabel("Coverage Ratio (X/M)",color="white")
    axes_flat[i].set_title(f"Coverage Ratio vs {label}",color="white")
    axes_flat[i].legend(facecolor="#16213e",labelcolor="white",fontsize=6)
    r,p = spearmanr(df_sub[ind].clip(-99,99),df_sub["coverage_ratio_structural"])
    axes_flat[i].text(0.05,0.95,f"ρ={r:.2f} p={p:.3f}",
                      transform=axes_flat[i].transAxes,color="white",fontsize=9)

plt.tight_layout()
plt.savefig("Figures/H1_coverage_ratio_indicators.png",dpi=150,bbox_inches="tight")
plt.show()
```

```python
# Boxplot by quartile — more readable than scatter for weak/moderate ρ
fig,axes = styled_subplots(1,2,figsize=(18,7))
fig.suptitle("Coverage Ratio by Indicator Quartile",fontsize=14,color="white",y=1.01)

for i,(ind,label) in enumerate([("current_acct","Current Account % GDP"),
                                  ("ext_debt_gdp","External Debt / GDP")]):
    df_sub = df_h1[["coverage_ratio_structural",ind]].dropna()
    df_sub["quartile"] = pd.qcut(df_sub[ind],4,labels=["Q1
(Low)","Q2","Q3","Q4
(High)"])
    data_box = [df_sub[df_sub["quartile"]==q]["coverage_ratio_structural"].dropna().values
                for q in ["Q1
(Low)","Q2","Q3","Q4
(High)"]]
    bp = axes[i].boxplot(data_box,patch_artist=True,
                         medianprops=dict(color="white",linewidth=2))
    qcolors = [C["red"],C["orange"],C["blue"],C["green"]]
    for patch,col in zip(bp["boxes"],qcolors):
        patch.set_facecolor(col); patch.set_alpha(0.7)
    axes[i].set_xticklabels(["Q1
(Low)","Q2","Q3","Q4
(High)"],color="white")
    axes[i].axhline(1.0,color="white",linestyle="--",alpha=0.5,label="Coverage=1")
    axes[i].set_title(f"Coverage Ratio by {label} Quartile",color="white")
    axes[i].set_ylabel("Coverage Ratio (X/M)",color="white")
    axes[i].legend(facecolor="#16213e",labelcolor="white")

plt.tight_layout()
plt.savefig("Figures/H2_coverage_by_quartile.png",dpi=150,bbox_inches="tight")
plt.show()
```

## I3. Multiple Regression — 1 Trade Var ~ All Indicators
</div>

```python
print("Multiple Regression (OLS) — All trade vars ~ all indicators simultaneously")
print("Standardized coefficients — effect net of all other indicators")
print("="*65)

multi_reg_results = []
for target in all_vars:
    df_reg = df_master[[target]+ind_vars].dropna()
    if len(df_reg)<30: continue
    X_s = StandardScaler().fit_transform(df_reg[ind_vars].values)
    y_s = df_reg[target].values
    lr = LinearRegression().fit(X_s,y_s)
    r2 = lr.score(X_s,y_s)
    coefs = pd.DataFrame({"indicator":ind_vars,"coef":lr.coef_,"abs_coef":np.abs(lr.coef_)})
    top = coefs.nlargest(3,"abs_coef")
    label = {**TRADE_VARS_DATA,**TRADE_VARS_SCORES}.get(target,target)
    multi_reg_results.append({"target":label,"r2":r2,"n":len(df_reg),
                               "top_indicator":top.iloc[0]["indicator"],
                               "top_coef":top.iloc[0]["coef"]})

df_multi_reg = pd.DataFrame(multi_reg_results).sort_values("r2",ascending=False)
print(df_multi_reg.to_string(index=False))

# Heatmap of all coefficients
coef_matrix = []
for target in all_vars:
    df_reg = df_master[[target]+ind_vars].dropna()
    if len(df_reg)<30: continue
    X_s = StandardScaler().fit_transform(df_reg[ind_vars].values)
    y_s = df_reg[target].values
    lr = LinearRegression().fit(X_s,y_s)
    label = {**TRADE_VARS_DATA,**TRADE_VARS_SCORES}.get(target,target)
    coef_matrix.append(dict(zip(["target"]+ind_vars,[label]+list(lr.coef_))))

df_coef = pd.DataFrame(coef_matrix).set_index("target")

fig,ax = plt.subplots(figsize=(max(12,len(df_coef.columns)*1.5),max(6,len(df_coef)*0.6)))
fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
sns.heatmap(df_coef,ax=ax,cmap="RdYlGn",center=0,annot=True,fmt=".2f",
            annot_kws={"size":7},linewidths=0.5,linecolor="#333")
ax.set_title("Multiple Regression Coefficients\nAll Trade Vars ~ All Indicators (standardized)",
             color="white",pad=10)
ax.tick_params(colors="white",labelsize=8)
plt.xticks(rotation=30,ha="right"); plt.tight_layout()
plt.savefig("Figures/I3_multiple_regression.png",dpi=150,bbox_inches="tight")
plt.show()
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## I4. PCA Croisé — Latent Dimensions Trade + Economy
</div>

```python
# PCA on combined trade + economic features
# Reveals latent axes that simultaneously summarize both dimensions

pca_trade_vars  = [v for v in all_vars  if v in df_master.columns]
pca_ind_vars    = [v for v in ind_vars  if v in df_master.columns]
pca_all_vars    = pca_trade_vars + pca_ind_vars

df_pca = df_master[["country_name"]+pca_all_vars].dropna()
X_pca  = StandardScaler().fit_transform(df_pca[pca_all_vars].values)

pca = PCA(n_components=min(6,len(pca_all_vars)))
pca.fit(X_pca)
explained = pca.explained_variance_ratio_

print(f"✅ PCA : {len(df_pca)} countries × {len(pca_all_vars)} variables")
print(f"   PC1: {explained[0]:.1%}  PC2: {explained[1]:.1%}  PC3: {explained[2]:.1%}")
print(f"   Cumulative (3 PCs): {explained[:3].sum():.1%}")

# Loadings heatmap PC1 & PC2
loadings = pd.DataFrame(pca.components_[:2].T,
                         index=pca_all_vars,
                         columns=["PC1","PC2"])
loadings["category"] = ["trade" if v in pca_trade_vars else "indicator" for v in pca_all_vars]
loadings = loadings.sort_values("PC1",ascending=False)

fig,axes = styled_subplots(1,2,figsize=(18,max(7,len(loadings)*0.35)))
fig.suptitle(f"PCA Loadings — PC1 ({explained[0]:.1%}) & PC2 ({explained[1]:.1%})",
             fontsize=13,color="white",y=1.01)

for i,pc in enumerate(["PC1","PC2"]):
    cols = [C["blue"] if c=="trade" else C["orange"] for c in loadings["category"]]
    axes[i].barh(loadings.index,loadings[pc],color=cols,alpha=0.8)
    axes[i].axvline(0,color="white",linewidth=1)
    axes[i].set_title(f"{pc} Loadings\n(blue=trade, orange=indicator)",color="white")
    axes[i].set_xlabel("Loading",color="white")
    axes[i].tick_params(colors="white",labelsize=7)
    for sp in axes[i].spines.values(): sp.set_edgecolor("#444")
    axes[i].grid(alpha=0.2,color="#555")

plt.tight_layout()
plt.savefig("Figures/I4_pca_loadings.png",dpi=150,bbox_inches="tight")
plt.show()

# Scatter PC1 vs PC2 countries
pca_scores = pd.DataFrame(pca.transform(X_pca)[:,:2],
                            columns=["PC1","PC2"])
pca_scores["country_name"] = df_pca["country_name"].values
pca_scores = pca_scores.merge(
    df_master[["country_name","size_profile"]],on="country_name",how="left"
)

fig,ax = styled_figure(figsize=(13,8))
ax.set_title(f"Country Positioning — PC1 ({explained[0]:.1%}) vs PC2 ({explained[1]:.1%})",color="white")
for sp,col in SIZE_COLORS.items():
    sub = pca_scores[pca_scores["size_profile"]==sp]
    ax.scatter(sub["PC1"],sub["PC2"],color=col,alpha=0.5,s=40,label=sp)
for _,row in pca_scores[pca_scores["country_name"].isin(countries)].iterrows():
    ax.scatter(row["PC1"],row["PC2"],color=C["orange"],s=120,zorder=5,
               edgecolors="white",linewidths=1)
    ax.annotate(row["country_name"],(row["PC1"],row["PC2"]),
                fontsize=9,color="white",xytext=(5,3),textcoords="offset points")
ax.axhline(0,color=C["grey"],linestyle="--",alpha=0.4)
ax.axvline(0,color=C["grey"],linestyle="--",alpha=0.4)
ax.legend(facecolor="#16213e",labelcolor="white",fontsize=8)
plt.tight_layout()
plt.savefig("Figures/I4_pca_scatter.png",dpi=150,bbox_inches="tight")
plt.show()
```

## I5. Temporal Stability of Correlations (3 Sub-Periods)
</div>

```python
# Sub-periods: pre-COVID, COVID, post-COVID
PERIODS = {
    "2015–2019": (2015,2019),
    "2020–2021": (2020,2021),
    "2022–2024": (2022,2024),
}

# Join annual trade + annual indicators
df_annual_full = df_trade.merge(df_ind_annual,on=["country_name","year"],how="inner")

# Key trade-indicator pairs to track
KEY_PAIRS = [
    ("export_growth","inflation"),
    ("export_growth","real_rate"),
    ("import_growth","gdp"),
    ("coverage_ratio","current_acct"),
]
# Keep only pairs where both columns exist
KEY_PAIRS = [(a,b) for a,b in KEY_PAIRS
             if a in df_annual_full.columns and b in df_annual_full.columns]

stability_results = []
for period,(y1,y2) in PERIODS.items():
    df_p = df_annual_full[(df_annual_full["year"]>=y1)&(df_annual_full["year"]<=y2)]
    for tv,iv in KEY_PAIRS:
        sub = df_p[[tv,iv]].dropna()
        if len(sub)<15: continue
        r,p = spearmanr(sub[tv],sub[iv])
        stability_results.append({"period":period,"trade_var":tv,
                                   "indicator":iv,"rho":round(r,3),"sig":p<0.05})

df_stability = pd.DataFrame(stability_results)

if not df_stability.empty:
    # Heatmap: rows = pairs, cols = periods
    df_stability["pair"] = df_stability["trade_var"]+" × "+df_stability["indicator"]
    pivot_stab = df_stability.pivot(index="pair",columns="period",values="rho")
    pivot_stab = pivot_stab[list(PERIODS.keys())]

    fig,ax = plt.subplots(figsize=(10,max(4,len(pivot_stab)*0.6)))
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot_stab,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1,
                annot=True,fmt=".2f",annot_kws={"size":9},
                linewidths=0.5,linecolor="#333")
    ax.set_title("Temporal Stability of Correlations\n(Spearman ρ by sub-period)",
                 color="white",pad=10)
    ax.tick_params(colors="white",labelsize=9)
    plt.tight_layout()
    plt.savefig("Figures/I5_temporal_stability.png",dpi=150,bbox_inches="tight")
    plt.show()

    print("\n💡 Key: stable ρ across periods = structural relationship")
    print("        volatile ρ = conjunctural (COVID-driven or cyclical)")
```

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## I6. Heterogeneity by Group (Size Profile)
</div>

```python
# Does the trade × indicator relationship differ by size profile?
# Spearman per group — RF only if n ≥ 40

group_corr_results = []
for group in ["Leader","Major","Mid","Small"]:
    df_g = df_master[df_master["size_profile"]==group]
    if len(df_g)<15: continue
    for tv in data_vars:
        for iv in ind_vars:
            sub = df_g[[tv,iv]].dropna()
            if len(sub)<15: continue
            r,p = spearmanr(sub[tv],sub[iv])
            group_corr_results.append({
                "group":group,"trade_var":tv,"indicator":iv,
                "rho":round(r,3),"sig":p<0.05,"abs_rho":abs(r),"n":len(sub)
            })

df_group_corr = pd.DataFrame(group_corr_results)

# Focus on key pairs — compare ρ across groups
if not df_group_corr.empty:
    key_tvs = ["export_share_structural","cagr_export","volatility_export","negfreq_export"]
    key_tvs = [v for v in key_tvs if v in df_group_corr["trade_var"].values]

    for tv in key_tvs:
        sub = df_group_corr[df_group_corr["trade_var"]==tv]
        if sub.empty: continue
        pivot_g = sub.pivot(index="indicator",columns="group",values="rho")

        fig,ax = plt.subplots(figsize=(10,max(4,len(pivot_g)*0.5)))
        fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
        sns.heatmap(pivot_g,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1,
                    annot=True,fmt=".2f",annot_kws={"size":8},
                    linewidths=0.5,linecolor="#333")
        ax.set_title(f"Heterogeneity — {TRADE_VARS_DATA.get(tv,tv)} × Indicators by Size Group",
                     color="white",pad=10)
        ax.tick_params(colors="white",labelsize=8)
        plt.tight_layout()
        plt.savefig(f"Figures/I6_{tv}_by_group.png",dpi=150,bbox_inches="tight")
        plt.show()
```

### I7.2 Trade Stability × Economic Stability (CV)
</div>

```python
# Trade stability: volatility_export (country-level, structural)
# Economic stability: CV per indicator (df_cv_all) — but this is a global scalar
# Cross at country level: volatility_export vs CV of each indicator for that country

# Country-level CV per indicator
country_cv = (
    df_ind_annual.groupby("country_name")[ind_cols]
    .std()
    .divide(df_ind_annual.groupby("country_name")[ind_cols].mean().abs()+1e-9)
    .reset_index()
)
country_cv.columns = ["country_name"] + [f"cv_{c}" for c in ind_cols]
cv_cols = [f"cv_{c}" for c in ind_cols]

df_stab_cross = df_master[["country_name","volatility_export","volatility_import",
                             "stability_profile"]].merge(country_cv,on="country_name",how="left")

# Correlations: trade volatility vs country-level CV of indicators
stab_vars = ["volatility_export","volatility_import"]
corr_stab = spearman_matrix(df_stab_cross,stab_vars,
                             [c for c in cv_cols if c in df_stab_cross.columns])

if not corr_stab.empty:
    heatmap_corr(corr_stab,"Trade Stability × Economic Stability (CV)",
                 min_abs_rho=0.15,figsize=(10,6),fname="I7_2_stability_cross")
    print("\n💡 High CV_inflation + High volatility_export → macro instability drives trade instability")
```

### I7.3 Convergence Croisée — Trade Shares × Economic Convergence
</div>

```python
# Sigma convergence: std of export_share over time (are countries converging in trade?)
# Compare with sigma convergence of GDP per capita (from indicators)

# Trade sigma convergence: std of export_share per year
trade_sigma = (
    df_trade.groupby("year")["export_share"]
    .std().reset_index().rename(columns={"export_share":"sigma_trade"})
)

# GDP sigma convergence from indicators (recompute quickly)
gdp_sigma = (
    df_ind_annual.groupby("year")["gdp_pc"]
    .std()
    .divide(df_ind_annual.groupby("year")["gdp_pc"].mean())
    .reset_index().rename(columns={"gdp_pc":"sigma_gdp_pc"})
)

df_sigma_cross = trade_sigma.merge(gdp_sigma,on="year",how="inner")

fig,axes = styled_subplots(1,2,figsize=(18,7))
fig.suptitle("Convergence Croisée — Trade Shares vs Economic Convergence",
             fontsize=14,color="white",y=1.01)

axes[0].plot(df_sigma_cross["year"],df_sigma_cross["sigma_trade"],
             color=C["blue"],marker="o",label="σ Trade (export share)",linewidth=2)
ax2 = axes[0].twinx(); ax2.set_facecolor("#16213e"); ax2.tick_params(colors="white")
ax2.plot(df_sigma_cross["year"],df_sigma_cross["sigma_gdp_pc"],
         color=C["orange"],marker="s",label="σ GDP/capita (CV)",linewidth=2,linestyle="--")
axes[0].set_xlabel("Year",color="white")
axes[0].set_ylabel("σ Export Share",color=C["blue"])
ax2.set_ylabel("CV GDP per Capita",color=C["orange"])
axes[0].set_title("Sigma Convergence — Trade vs Economy",color="white")
lines1,labels1 = axes[0].get_legend_handles_labels()
lines2,labels2 = ax2.get_legend_handles_labels()
axes[0].legend(lines1+lines2,labels1+labels2,facecolor="#16213e",labelcolor="white",fontsize=8)

# Scatter sigma_trade vs sigma_gdp_pc per year
axes[1].scatter(df_sigma_cross["sigma_gdp_pc"],df_sigma_cross["sigma_trade"],
                color=C["blue"],s=80,alpha=0.8)
for _,row in df_sigma_cross.iterrows():
    axes[1].annotate(str(int(row["year"])),(row["sigma_gdp_pc"],row["sigma_trade"]),
                     fontsize=8,color="white",xytext=(3,2),textcoords="offset points")
axes[1].set_xlabel("σ GDP per Capita (economic dispersion)",color="white")
axes[1].set_ylabel("σ Export Share (trade dispersion)",color="white")
axes[1].set_title("Economic vs Trade Dispersion per Year",color="white")
r,p = spearmanr(df_sigma_cross["sigma_gdp_pc"],df_sigma_cross["sigma_trade"])
axes[1].text(0.05,0.95,f"ρ={r:.2f} p={p:.3f}",transform=axes[1].transAxes,color="white",fontsize=9)

plt.tight_layout()
plt.savefig("Figures/I7_3_convergence_cross.png",dpi=150,bbox_inches="tight")
plt.show()
```
