Signal intéressant : India = Global Leader par taille, Volatile Emerging par structure
Cross-reference G.1 QD : India trade efficiency = 0.46 (sous-performer). G.6 Q5 : India below trend sur Inflation et Real Rate. H.3 : import_pc très bas. I.1 : coverage = 0.67.
→ Signal fort : tous les indicateurs individuels (G/H/I) montrent India comme structurellement faible, mais J.6 la classe Global Leader par taille. C'est la tension entre échelle (nombre absolu d'exports) et efficience (qualité de la structure). Le méta-profil capture l'échelle, pas l'efficience.

**Insight — J.1 :**

**Internal score validation :**
The score distribution confirms monotone consistency : medians progress
regularly from Weak (−0.42) → Moderate (−0.22) → Good (−0.05) → Strong (+0.73).
The "Strong" group is the only one with a clearly positive median. Its high

India at only +0.52 despite its "Strong" classification.

**KW — GDP per Capita (H=27.3, p<0.0001) :**

**KW — Current Account % GDP (H=20.4, p=0.0001) :**

````python
# [S] indicators : boxplot by quartile
# Larger countries have higher export share — but how does the distribution look?
INDS_ES_BOXPLOT = [
    ("GDP (current US$)", "GDP (US$)"),
    ("External debt stocks, total (DOD, current US$)", "External Debt / GDP"),
    ("Foreign direct investment, net inflows (BoP, current US$)", "FDI Inflows / GDP")]

gdp_col = "GDP (current US$)"

for ind_full, label_override in INDS_ES_BOXPLOT:
    if ind_full not in df_master.columns: continue
    flag = get_ind_flag(ind_full)
    _, clip, use_log, _ = get_ind_params(ind_full)
    fname = f"G1_es_boxplot_{label_override.replace(' ','_').replace('/','per')[:25]}"

    df_plot = df_master[["country_name","export_share_structural",
                          ind_full, gdp_col]].dropna().copy()

    # Normalize [S] abs by GDP
    if ind_full in IND_ABS and ind_full != gdp_col:
        mask = df_plot[gdp_col] > 0
        df_plot.loc[mask, ind_full] = (df_plot.loc[mask, ind_full] / df_plot.loc[mask, gdp_col])
        use_log = False; clip = None

    if clip: df_plot[ind_full] = df_plot[ind_full].clip(*clip)
    x_vals = np.log(df_plot[ind_full].clip(lower=1e-9)) if use_log else df_plot[ind_full]

    try:
        df_plot["quartile"] = pd.qcut(x_vals, 4,
                                       labels=["Q1(Low)","Q2","Q3","Q4(High)"],
                                       duplicates="drop")
    except ValueError: continue

    labels_q = ["Q1(Low)","Q2","Q3","Q4(High)"]
    data_box  = [df_plot[df_plot["quartile"]==q]["export_share_structural"].dropna().values
                 for q in labels_q]

    fig, ax = styled_figure(figsize=(11, 7))
    ax.set_title(f"Export Share by {label_override} Quartile  {flag}",
                 color="white")
    bp = ax.boxplot(data_box, patch_artist=True,
                    medianprops=dict(color="white", linewidth=2))
    for patch, col in zip(bp["boxes"],
                          [C["red"],C["orange"],C["blue"],C["green"]]):
        patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{q.split(chr(10))[0]}{label_override}(n={len(d)})"
               for q, d in zip(labels_q, data_box)]
    ax.set_xticklabels(xlabels, color="white", fontsize=8)
    ax.set_ylabel("Export Share (% world)", color="white")

    for i, vals in enumerate(data_box):
        if len(vals) == 0: continue
        ax.text(i+1, np.median(vals), f"{np.median(vals):.2f}",
                ha="center", va="bottom",
                color="white", fontsize=8, fontweight="bold")

    # Annotate selected countries
    for q_idx, q in enumerate(labels_q):
        sub = df_plot[(df_plot["quartile"]==q) &
                      (df_plot["country_name"].isin(countries))]
        for _, row in sub.iterrows():
            ax.scatter(q_idx+1, row["export_share_structural"],
                       color=C["orange"], s=100, zorder=6,
                       edgecolors="white", linewidths=1.2)
            ax.annotate(row["country_name"],
                        (q_idx+1, row["export_share_structural"]),
                        fontsize=7, color="white", fontweight="bold",
                        xytext=(10, 0), textcoords="offset points",
                        bbox=dict(boxstyle="round,pad=0.2",
                                  facecolor=C["dark"], alpha=0.7,
                                  edgecolor="white", linewidth=0.5))

    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()
````

````python
# [E] indicators : scatter + boxplot (causal signal)
INDS_ES_SCATTER = [
    ("Current account balance (% of GDP)", "Current Account % GDP",
     (-30, 30)),
    ("Real interest rate (%)",             "Real Interest Rate %",
     (-20, 45)),
]

for ind_full, label_ind, clip in INDS_ES_SCATTER:
    if ind_full not in df_master.columns: continue
    flag = get_ind_flag(ind_full)
    fname_base = f"G1_es_{label_ind.replace(' ','_').replace('%','pct')[:25]}"

    df_plot = df_master[["country_name","export_share_structural",ind_full]].dropna().copy()
    df_plot[ind_full] = df_plot[ind_full].clip(*clip)

    from scipy.stats import linregress
    r_sp, p_sp = spearmanr(df_plot[ind_full], df_plot["export_share_structural"])

    # ── Scatter ───────────────────────────────────────────────────────────────
    fig, ax = styled_figure(figsize=(13, 8))
    ax.set_title(f"Export Share × {label_ind}  {flag}"
                 f"(scatter — causal signal)", color="white")
    ax.scatter(df_plot[ind_full], df_plot["export_share_structural"],
               color=C["blue"], alpha=0.4, s=35)

    slope, intercept, r, _, _ = linregress(df_plot[ind_full],
                                            df_plot["export_share_structural"])
    x_line = np.linspace(df_plot[ind_full].min(), df_plot[ind_full].max(), 100)
    ax.plot(x_line, intercept + slope * x_line,
            color="white", linewidth=1.5, linestyle="--",
            alpha=0.6, label=f"Trend (r={r:.2f})")

    # Annotate countries
    for _, row in df_plot[df_plot["country_name"].isin(countries)].iterrows():
        ax.scatter(row[ind_full], row["export_share_structural"],
                   color=C["orange"], s=150, zorder=6,
                   edgecolors="white", linewidths=1.5)
        ax.annotate(row["country_name"],
                    (row[ind_full], row["export_share_structural"]),
                    fontsize=9, color="white", fontweight="bold",
                    xytext=(8, 4), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2",
                              facecolor=C["dark"], alpha=0.7,
                              edgecolor="white", linewidth=0.5))

    ax.set_xlabel(label_ind, color="white")
    ax.set_ylabel("Export Share (% world)", color="white")
    ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})",
            transform=ax.transAxes, color="white", fontsize=9)
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname_base}_scatter.png",
                dpi=150, bbox_inches="tight")
    plt.show()

    # ── Boxplot ───────────────────────────────────────────────────────────────
    labels_q = ["Q1(Low)","Q2","Q3","Q4(High)"]
    df_plot["quartile"] = pd.qcut(df_plot[ind_full], 4,
                                   labels=labels_q, duplicates="drop")
    data_box = [df_plot[df_plot["quartile"]==q]["export_share_structural"].dropna().values
                for q in labels_q]

    fig, ax = styled_figure(figsize=(11, 7))
    ax.set_title(f"Export Share by {label_ind} Quartile  {flag}",
                 color="white")
    bp = ax.boxplot(data_box, patch_artist=True,
                    medianprops=dict(color="white", linewidth=2))
    for patch, col in zip(bp["boxes"],
                          [C["red"],C["orange"],C["blue"],C["green"]]):
        patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{q.split(chr(10))[0]}{label_ind}(n={len(d)})"
               for q, d in zip(labels_q, data_box)]
    ax.set_xticklabels(xlabels, color="white", fontsize=8)
    ax.set_ylabel("Export Share (% world)", color="white")

    for i, vals in enumerate(data_box):
        if len(vals) == 0: continue
        ax.text(i+1, np.median(vals), f"{np.median(vals):.2f}",
                ha="center", va="bottom",
                color="white", fontsize=8, fontweight="bold")

    for q_idx, q in enumerate(labels_q):
        sub = df_plot[(df_plot["quartile"]==q) &
                      (df_plot["country_name"].isin(countries))]
        for _, row in sub.iterrows():
            ax.scatter(q_idx+1, row["export_share_structural"],
                       color=C["orange"], s=100, zorder=6,
                       edgecolors="white", linewidths=1.2)
            ax.annotate(row["country_name"],
                        (q_idx+1, row["export_share_structural"]),
                        fontsize=7, color="white", fontweight="bold",
                        xytext=(10, 0), textcoords="offset points",
                        bbox=dict(boxstyle="round,pad=0.2",
                                  facecolor=C["dark"], alpha=0.7,
                                  edgecolor="white", linewidth=0.5))
    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname_base}_boxplot.png",
                dpi=150, bbox_inches="tight")
    plt.show()
````

#### Question D — Among same-GDP countries, who captures more export share? a supprimer
</div>

````python
from scipy.stats import linregress

gdp_col = "GDP (current US$)"
target  = "export_share_structural"

df_qd = df_master[["country_name", target, gdp_col, "size_profile"]].dropna().copy()

# Compute residuals from log-linear trend (same as Question A)
df_qd["log_gdp"]   = np.log10(df_qd[gdp_col].clip(lower=1))
slope, intercept, _, _, _ = linregress(df_qd["log_gdp"], df_qd[target])
df_qd["predicted"] = intercept + slope * df_qd["log_gdp"]
df_qd["residual"]  = df_qd[target] - df_qd["predicted"]

# ── Figure : Residuals vs GDP ─────────────────────────────────────────────────
fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Question D — Among same-GDP countries, who captures more export share?"
             "Residuals from GDP trend (above = over-performer, below = under-performer)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_qd[df_qd["size_profile"] == sp]
    ax.scatter(sub[gdp_col], sub["residual"], color=col, alpha=0.5, s=40, label=sp)

ax.axhline(0, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Expected (on trend)")
ax.axhline(0, color="white", linewidth=1.5, linestyle="--", alpha=0.6)

# Annotate selected countries
for _, row in df_qd[df_qd["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row["residual"], color=C["orange"], s=150,zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row["residual"]), fontsize=9, color="white", fontweight="bold",
                xytext=(8, 4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

# Color zones
ax.axhspan(0, df_qd["residual"].max()*1.1, alpha=0.05, color=C["green"], label="Over-performers")
ax.axhspan(df_qd["residual"].min()*1.1, 0, alpha=0.05, color=C["red"], label="Under-performers")

ax.set_xscale("log")
ax.set_xlabel("GDP (US$) — log scale", color="white")
ax.set_ylabel("Residual (actual − predicted export share)", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G1_QD_residuals.png", dpi=150, bbox_inches="tight")
plt.show()

# Top over and under-performers
print("Top 10 over-performers (export share >> GDP would predict) :")
print(df_qd.nlargest(10, "residual")[["country_name","residual",target,"size_profile"]].to_string(index=False))
print()
print("Top 10 under-performers (export share << GDP would predict) :")
print(df_qd.nsmallest(10, "residual")[["country_name","residual",target,"size_profile"]].to_string(index=False))
````

**Insight — Question D :**

**Answer :** At any given GDP level, export share varies significantly — 
trade orientation determines who captures more than their economic size would predict.

**Top over-performers (above trend) :**
- **China** : largest positive residual — export share far exceeds GDP prediction
  → deliberate export-led growth model, manufacturing scale, competitive pricing
- **Germany** : strong over-performer among high-GDP countries
  → specialized industrial exports, high value-added products, global supply chain integration

**Top under-performers (below trend) :**
- **USA** : largest negative residual among Leaders — GDP is the world's largest
  but export share is below what the trend predicts
  → service-dominated economy (finance, tech, healthcare), large domestic market
- **Japan** : similar pattern to USA — strong GDP but export share below trend
  → large domestic consumption, mature economy with high internal absorption
- **India, Brazil** : moderate GDP but export share well below prediction
  → domestically oriented, commodity exports, limited manufacturing integration

**Key finding :** GDP explains *where* a country can be, but trade orientation
determines *how much* of the opportunity it captures. The residual is essentially
a measure of **commercial efficiency** — how effectively a country converts its
economic mass into world export share.

### G.1.3 — Size Profile × Indicators (KW Boxplots)

````python
# KW boxplots : indicator distribution by Size Profile groups
INDS_SIZE_PROFILE = [
    ("GDP (current US$)",                   "GDP (US$)",             None),
    ("GDP per capita (current US$)",         "GDP per Capita (US$)",  None),
    ("Current account balance (% of GDP)",   "Current Account % GDP", (-30,30)),
    ("Real interest rate (%)",               "Real Interest Rate %",  (-20,45)),
]

SIZE_PROFILE_MAPPING = {"Leader":4,"Major":3,"Mid":2,"Small":1}
groups_ordered = ["Leader","Major","Mid","Small"]
group_colors   = [C["green"],C["blue"],C["orange"],C["red"]]
gdp_col        = "GDP (current US$)"

for ind_full, label_ind, clip in INDS_SIZE_PROFILE:
    if ind_full not in df_master.columns: continue
    flag = get_ind_flag(ind_full)
    fname = f"G1_size_profile_{label_ind.replace(' ','_').replace('%','pct').replace('/','per')[:25]}"

    # KW stats
    kw_row = df_kw[(df_kw["profile_var"]=="size_profile") &
                   (df_kw["indicator"]==ind_full)]
    eta2 = kw_row["eta2"].values[0]   if not kw_row.empty else np.nan
    p_kw = kw_row["pvalue"].values[0] if not kw_row.empty else np.nan

    # Spearman on _num
    sp_row = df_spearman[(df_spearman["trade_var"]=="size_profile_num") &
                         (df_spearman["indicator"]==ind_full)]
    rho = sp_row["rho"].values[0] if not sp_row.empty else np.nan

    df_sub = df_master[["country_name","size_profile",ind_full]].dropna().copy()

    # Normalize [S] abs by GDP
    _, _, use_log, _ = get_ind_params(ind_full)
    if ind_full in IND_ABS and ind_full != gdp_col:
        df_gdp = df_master[["country_name",gdp_col]].dropna()
        df_sub  = df_sub.merge(df_gdp, on="country_name", how="left")
        mask    = df_sub[gdp_col] > 0
        df_sub.loc[mask, ind_full] = (
            df_sub.loc[mask, ind_full] / df_sub.loc[mask, gdp_col]
        )
        label_ind_plot = label_ind.replace("(US$)","/ GDP")
        use_log = False; clip = None
    else:
        label_ind_plot = label_ind

    if clip: df_sub[ind_full] = df_sub[ind_full].clip(*clip)
    data_box = [df_sub[df_sub["size_profile"]==g][ind_full].dropna().values
                for g in groups_ordered]

    fig, ax = styled_figure(figsize=(12, 7))
    ax.set_title(
        f"Size Profile × {label_ind_plot}  {flag}"
        f"KW η²={eta2:.3f} (p={p_kw:.4f}) | Spearman ρ={rho:+.3f}",
        color="white"
    )
    bp = ax.boxplot(data_box, patch_artist=True,
                    medianprops=dict(color="white", linewidth=2))
    for patch, col in zip(bp["boxes"], group_colors):
        patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{g}(n={len(d)})" for g, d in zip(groups_ordered, data_box)]
    ax.set_xticklabels(xlabels, color="white", fontsize=9)
    ax.set_ylabel(label_ind_plot, color="white")
    if use_log: ax.set_yscale("log")

    for i, vals in enumerate(data_box):
        if len(vals) == 0: continue
        ax.text(i+1, np.median(vals), f"{np.median(vals):.2f}",
                ha="center", va="bottom",
                color="white", fontsize=8, fontweight="bold")

    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()
````

````python
# Current Account → Export Share (boxplot by CA quartile)
ca_col = "Current account balance (% of GDP)"
df_inv = df_master[["country_name", ca_col,
                     "size_profile","export_share_structural"]].dropna()
df_inv[ca_col] = df_inv[ca_col].clip(-30, 30)

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Inverse Direction — Economic Indicator → Export Profile",
             fontsize=13, color="white", y=1.01)

# Left : Export Share by CA quartile
df_inv["ca_q"] = pd.qcut(df_inv[ca_col], 4,
                          labels=["Q1(Deficit)","Q2","Q3","Q4(Surplus)"])
labels_ca = ["Q1(Deficit)","Q2","Q3","Q4(Surplus)"]
data_box  = [df_inv[df_inv["ca_q"]==q]["export_share_structural"].dropna().values
             for q in labels_ca]
bp = axes[0].boxplot(data_box, patch_artist=True,
                     medianprops=dict(color="white", linewidth=2))
for patch, col in zip(bp["boxes"],
                      [C["red"],C["orange"],C["blue"],C["green"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)
for i, (q, vals) in enumerate(zip(labels_ca, data_box)):
    if len(vals) == 0: continue
    axes[0].text(i+1, np.median(vals), f"{np.median(vals):.2f}",
                 ha="center", va="bottom",
                 color="white", fontsize=8, fontweight="bold")
for q_idx, q in enumerate(labels_ca):
    sub = df_inv[(df_inv["ca_q"]==q) & (df_inv["country_name"].isin(countries))]
    for _, row in sub.iterrows():
        axes[0].scatter(q_idx+1, row["export_share_structural"],
                        color=C["orange"], s=100, zorder=6,
                        edgecolors="white", linewidths=1.2)
        axes[0].annotate(row["country_name"],
                         (q_idx+1, row["export_share_structural"]),
                         fontsize=7, color="white", fontweight="bold",
                         xytext=(10,0), textcoords="offset points",
                         bbox=dict(boxstyle="round,pad=0.2",
                                   facecolor=C["dark"], alpha=0.7,
                                   edgecolor="white", linewidth=0.5))
axes[0].set_xticklabels(labels_ca, color="white")
axes[0].set_title("Export Share by Current Account Quartile"
                  "(surplus countries export more?)", color="white")
axes[0].set_ylabel("Export Share (% world)", color="white")
r, p = spearmanr(df_inv[ca_col], df_inv["export_share_structural"])
axes[0].text(0.05, 0.95, f"ρ={r:.2f} (p={p:.3f})",
             transform=axes[0].transAxes, color="white", fontsize=9)

# Right : Size Profile distribution by CA quartile
ct = pd.crosstab(df_inv["ca_q"], df_inv["size_profile"],
                 normalize="index") * 100
profiles_order = [p for p in ["Leader","Major","Mid","Small"] if p in ct.columns]
ct[profiles_order].plot(kind="bar", ax=axes[1],
                         color=[SIZE_COLORS[p] for p in profiles_order],
                         alpha=0.8, width=0.7)
axes[1].set_title("Size Profile Distribution by Current Account Quartile",
                  color="white")
axes[1].set_xlabel("", color="white")
axes[1].set_ylabel("Share (%)", color="white")
axes[1].set_xticklabels(["Q1(Deficit)","Q2","Q3","Q4(Surplus)"],
                         rotation=0, color="white")
axes[1].legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G1_inverse_ca_export.png",
            dpi=150, bbox_inches="tight")
plt.show()
````

**G.1 Key Insights :**

**Export Share :**
- Size effect dominates — GDP (US$) is the primary determinant (ρ=0.937)
- Current Account % GDP is the only actionable [E] signal (ρ=0.484)
  → surplus countries structurally export more
- Real Interest Rate inversely related (ρ=-0.357) → high rates constrain export capacity
- **Answer** : Export dominance is primarily a size effect.
  Beyond size, maintaining a current account surplus is the key economic lever.

**Size Profile :**
- All [E] indicators : KW and Spearman agree → robust monotone relationships
- GDP per Capita (η²=0.246) : Leaders are not necessarily the richest (China effect)
- Real Interest Rate inversely discriminates (η²=0.176, ρ=-0.433)
  → Leaders tend to have lower real rates → easier financing of export capacity
- **Answer** : Size profile reflects absolute GDP mass, not wealth.
  Financial conditions (real rate) provide the most actionable [E] signal.

````python
# ── Heatmap : G1 vars × all indicators ───────────────────────────────────────
g1_vars  = ["export_share_structural", "size_profile_num"]
g1_vars  = [v for v in g1_vars if v in df_master.columns]
g1_inds  = [i for i in ind_cols if i in df_master.columns]

rows_g1 = []
for tv in g1_vars:
    for iv in g1_inds:
        sub = df_master[[tv, iv]].dropna()
        if len(sub) < 20: continue
        r, p = spearmanr(sub[tv], sub[iv])
        rows_g1.append({
            "trade_label" : ALL_TRADE_LABELS.get(tv, tv),
            "ind_label"   : IND_LABELS.get(iv, iv),
            "ind_flag"    : get_ind_flag(iv),
            "rho"         : round(r, 3),
            "sig"         : p < 0.05,
            "abs_rho"     : abs(r),
        })

df_g1_corr = pd.DataFrame(rows_g1)
sig_g1     = df_g1_corr[(df_g1_corr["sig"]) & (df_g1_corr["abs_rho"] >= 0.15)]

if not sig_g1.empty:
    pivot_g1 = sig_g1.pivot_table(index="ind_label", columns="trade_label", values="rho", aggfunc="first"
    ).sort_values(sig_g1.groupby("ind_label")["abs_rho"].mean().sort_values(ascending=False).index[::-1])

    flag_color = {"[T]": C["orange"], "[S]": C["purple"], "[E]": C["green"]}
    ind_flag_map = df_g1_corr[["ind_label","ind_flag"]].drop_duplicates().set_index("ind_label")["ind_flag"].to_dict()
    row_colors_g1 = [flag_color.get(ind_flag_map.get(i,"[E]"),"white") for i in pivot_g1.index]

    fig, ax = plt.subplots(figsize=(8, max(5, len(pivot_g1)*0.5)))
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot_g1, ax=ax, cmap="RdYlGn", center=0, vmin=-1, vmax=1,
                annot=True, fmt=".2f", annot_kws={"size": 9},
                linewidths=0.4, linecolor="#333",
                cbar_kws={"label": "Spearman ρ"})
    ax.set_title("G1 — Export Share & Size Profile × All Indicators"
                 "(sig |ρ|≥0.15 | 🟠[T] 🟣[S] 🟢[E])", color="white", pad=10)
    ax.tick_params(colors="white", labelsize=9)
    for i, lbl in enumerate(ax.get_yticklabels()):
        lbl.set_color(row_colors_g1[i])
    plt.xticks(rotation=15, ha="right")
    plt.tight_layout()
    plt.savefig("../Figures/Link/G1_3_heatmap.png", dpi=150, bbox_inches="tight")
    plt.show()
````

````python
# ── GDP_pc quartile → Size Profile distribution ──────────────────────────────
gdp_pc_col = "GDP per capita (current US$)"
df_g14 = df_master[["country_name", gdp_pc_col, "size_profile", "export_share_structural"]].dropna()

df_g14["gdp_pc_q"] = pd.qcut(df_g14[gdp_pc_col], 4, labels=["Q1\n(Poorest)","Q2","Q3","Q4\n(Richest)"])

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("G1.4 — GDP per Capita → Export Share & Size Profile", fontsize=14, color="white", y=1.01)

# Left : Export Share distribution by GDP_pc quartile
labels_q  = ["Q1\n(Poorest)","Q2","Q3","Q4\n(Richest)"]
data_box  = [df_g14[df_g14["gdp_pc_q"]==q]["export_share_structural"].dropna().values for q in labels_q]
bp = axes[0].boxplot(data_box, patch_artist=True,
                     medianprops=dict(color="white", linewidth=2))
for patch, col in zip(bp["boxes"], [C["red"],C["orange"],C["blue"],C["green"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)
axes[0].set_xticklabels(labels_q, color="white")
axes[0].set_title("Export Share by GDP per Capita Quartile", color="white")
axes[0].set_ylabel("Export Share (% world)", color="white")

for i, q in enumerate(labels_q):
    vals = df_g14[df_g14["gdp_pc_q"]==q]["export_share_structural"].dropna()
    if not vals.empty:
        axes[0].text(i+1, vals.median()+0.05, f"{vals.median():.2f}", ha="center", color="white", fontsize=9, fontweight="bold")

r, p = spearmanr(np.log(df_g14[gdp_pc_col].clip(lower=1)),df_g14["export_share_structural"])
axes[0].text(0.05, 0.95, f"Spearman ρ={r:.2f} (p={p:.3f})", transform=axes[0].transAxes, color="white", fontsize=9)

# Right : Size Profile distribution by GDP_pc quartile
ct = pd.crosstab(df_g14["gdp_pc_q"], df_g14["size_profile"], normalize="index")*100
profiles_order = [p for p in ["Leader","Major","Mid","Small"] if p in ct.columns]
ct[profiles_order].plot(kind="bar", ax=axes[1], color=[SIZE_COLORS[p] for p in profiles_order], alpha=0.8, width=0.7)
axes[1].set_title("Size Profile Distribution by GDP per Capita Quartile", color="white")
axes[1].set_xlabel("", color="white")
axes[1].set_ylabel("Share (%)", color="white")
axes[1].set_xticklabels(labels_q, rotation=0, color="white")
axes[1].legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G1_4_gdppc_to_export.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight :**
- Richer countries (Q4) do **not** have higher export shares — export share is flat across GDP_pc quartiles
- This confirms D1 finding : wealth does not drive export dominance
- **Leaders appear in all GDP_pc quartiles** — being a world export leader is not a function of income level
- The Size Profile is driven by **absolute GDP** (total economic mass), not **GDP per capita** (wealth)
- **Answer** : Export dominance is a function of economic size, not wealth level.
  China (middle income) and Germany (high income) are both Leaders for different reasons.

### G.2.3 — Finding
</div>

**G.2 — Finding : Export Growth is Weakly Explained by Macro Indicators**

**Statistical summary :**
- CAGR Export : RF R²=0.860, OLS R²=0.079, Gap=0.781 🔴
- Avg Growth Export : RF R²=0.863, OLS R²=0.162, Gap=0.700 🔴

**Spearman correlations :**
- CAGR Export : 3 weak significant correlations
  — External Debt [S] ρ=+0.19, FDI [S] ρ=+0.18, Current Account US$ [T] ρ=-0.21
  — All [S] or [T], no significant [E] indicator
- Avg Growth Export : **no significant correlation** with any indicator

**CAGR vs Avg Growth — not identical :**
CAGR captures compound growth rate (geometric), Avg Growth is arithmetic mean.
Their RF profiles differ slightly — Expense % GDP leads Avg Growth (0.127)
while External Debt leads CAGR (0.110). Both confirm weak macro signal.

**Possible explanations :**
1. **Period effect (2015–2024)** : COVID collapse (2020) and rebound (2021)
   dominate variance and mask structural relationships
2. **n=65** : limited statistical power
3. **Missing drivers** : exchange rates, trade agreements, commodity prices,
   sector competitiveness — absent from this dataset
4. **Large gap RF/OLS** : relationship is non-linear and complex —
   no single macro indicator drives export growth monotonically

**Reference figure :**
CAGR Export vs CAGR GDP (ρ=0.29) — weak but the strongest available link.
Most countries cluster around 0–10% GDP growth with scattered export growth.
China is a notable case — positive GDP growth but negative export CAGR
(decoupling pattern confirmed from D2).

**Conclusion :** Export growth is largely idiosyncratic in this dataset.
Standard macro fundamentals do not predict it. This is analytically meaningful —
export growth dynamics require sector-level or policy-level data to explain.

**Context :**

meaningful links exist between the two dimensions.

Do trade structures reflect — or influence — macroeconomic outcomes?

**Approach :**

3. Analyze the significant relationships — direction, strength, anomalies
4. Test whether the links hold across profiles, quadrants and composite scores

The analysis explores both directions — macro conditions as predictors of
trade structure, and trade structure as predictor of macro outcomes — without
treating them as two separate studies. The first direction is analyzed more
deeply given the stronger statistical signal found.

## 2. Style & Helpers
</div>

````python
import os
os.makedirs("../Figures/Link", exist_ok=True)
os.makedirs("../Analysis_files/Link", exist_ok=True)
````

````python
C = {"blue":"#3498db","green":"#2ecc71","red":"#e74c3c","orange":"#f39c12", "purple":"#9b59b6","grey":"#95a5a6","dark":"#1a1a2e","teal":"#1abc9c"}

SIZE_COLORS      = {"Leader":C["green"],"Major":C["blue"],"Mid":C["orange"],"Small":C["red"]}
BALANCE_COLORS   = {"Strong Surplus":C["green"],"Slight Surplus":C["teal"], "Slight Deficit":C["orange"],"Structural Deficit":C["red"]}
GROWTH_COLORS    = {"Strong Grower":C["green"],"Moderate Grower":C["blue"], "Weak Grower":C["orange"],"Declining":C["red"]}
STABILITY_COLORS = {"Stable":C["green"],"Moderate":C["blue"], "Volatile":C["orange"],"Very Volatile":C["red"]}
CONSIST_COLORS   = {"Very Consistent":C["green"],"Consistent":C["blue"], "Inconsistent":C["orange"],"Very Inconsistent":C["red"]}

def styled_figure(figsize=(12,6)):
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    ax.tick_params(colors="white"); ax.xaxis.label.set_color("white")
    ax.yaxis.label.set_color("white"); ax.title.set_color("white")
    for sp in ax.spines.values(): sp.set_edgecolor("#444")
    ax.grid(alpha=0.2,color="#555"); return fig,ax

def styled_subplots(nrows=1,ncols=2,figsize=(18,6)):
    fig,axes = plt.subplots(nrows,ncols,figsize=figsize)
    fig.patch.set_facecolor(C["dark"])
    for ax in np.array(axes).flatten():
        ax.set_facecolor("#16213e"); ax.tick_params(colors="white")
        ax.xaxis.label.set_color("white"); ax.yaxis.label.set_color("white")
        ax.title.set_color("white")
        for sp in ax.spines.values(): sp.set_edgecolor("#444")
        ax.grid(alpha=0.2,color="#555")
    return fig,axes

def normalize_01(s):
    mn,mx = s.min(),s.max(); return (s-mn)/(mx-mn+1e-9)

def spearman_matrix(df, trade_vars, ind_vars, min_obs=25):
    rows=[]
    for tv in trade_vars:
        for iv in ind_vars:
            sub = df[[tv,iv]].dropna()
            if len(sub)<min_obs: continue
            r,p = spearmanr(sub[tv],sub[iv])
            rows.append({"trade_var":tv,"indicator":iv, "rho":round(r,3),"pvalue":round(p,4), "sig":p<0.05,"abs_rho":abs(r)})
    return pd.DataFrame(rows)

def heatmap_corr(corr_df, title, trade_label_map=None, ind_label_map=None, min_abs_rho=0.15, figsize=(12,7), fname=None):
    """Plot Spearman heatmap with readable labels."""
    strong = corr_df[(corr_df["sig"])&(corr_df["abs_rho"]>=min_abs_rho)].copy()
    if strong.empty: print(f"No sig. correlations above |ρ|={min_abs_rho}"); return
    if trade_label_map:  strong["trade_var"] = strong["trade_var"].map(trade_label_map).fillna(strong["trade_var"])
    if ind_label_map:    strong["indicator"] = strong["indicator"].map(ind_label_map).fillna(strong["indicator"])
    pivot = strong.pivot_table(index="indicator",columns="trade_var",values="rho",aggfunc="first")
    pivot = pivot.sort_values(pivot.columns[0],ascending=False)
    fig,ax = plt.subplots(figsize=figsize)
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot,ax=ax,cmap="RdYlGn",center=0,vmin=-1,vmax=1, annot=True,fmt=".2f",annot_kws={"size":8},
                linewidths=0.5,linecolor="#333",cbar_kws={"label":"Spearman ρ"})
    ax.set_title(f"{title}\n(sig |ρ|≥{min_abs_rho})",color="white",pad=10)
    ax.tick_params(colors="white",labelsize=8)
    plt.xticks(rotation=30,ha="right"); plt.tight_layout()
    f = fname or title.lower().replace(" ","_").replace("×","x")[:40]
    plt.savefig(f"../Figures/link/{f}.png",dpi=150,bbox_inches="tight"); plt.show()

def scatter_ind(df, x_col, y_col, color_col=None, color_map=None, title="", xlabel="", ylabel="", countries=None, xlog=False, clip_x=None, fname=None):
    """Standard scatter trade × indicator."""
    fig,ax = styled_figure(figsize=(13,8))
    ax.set_title(title,color="white")
    cols = [x_col,y_col] + ([color_col] if color_col else []) + (["country_name"] if "country_name" in df.columns else [])
    df_p = df[cols].dropna()
    if clip_x: df_p = df_p.copy(); df_p[x_col] = df_p[x_col].clip(*clip_x)
    if color_col and color_map:
        for grp,col in color_map.items():
            sub = df_p[df_p[color_col]==grp]
            ax.scatter(sub[x_col],sub[y_col],color=col,alpha=0.6,s=50,label=grp)
        ax.legend(facecolor="#16213e",labelcolor="white",fontsize=8)
    else: ax.scatter(df_p[x_col],df_p[y_col],color=C["blue"],alpha=0.6,s=50)
    if countries and "country_name" in df_p.columns:
        for _,row in df_p[df_p["country_name"].isin(countries)].iterrows():
            ax.annotate(row["country_name"],(row[x_col],row[y_col]), fontsize=8,color="white",xytext=(5,3),textcoords="offset points")
    if xlog: ax.set_xscale("log")
    ax.set_xlabel(xlabel,color="white"); ax.set_ylabel(ylabel,color="white")
    sub2 = df_p[[x_col,y_col]].dropna()
    r,p = spearmanr(sub2[x_col],sub2[y_col])
    ax.text(0.05,0.95,f"Spearman ρ={r:.2f} (p={p:.3f})", transform=ax.transAxes,color="white",fontsize=9)
    plt.tight_layout()
    if fname: plt.savefig(f"Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

def boxplot_quartile(df, value_col, group_col, title="", ylabel="", fname=None):
    """Boxplot by quartile — clearer than scatter for weak ρ."""
    df_q = df[[value_col,group_col]].dropna().copy()
    labels = ["Q1(Low)","Q2","Q3","Q4(High)"]
    df_q["quartile"] = pd.qcut(df_q[group_col],4,labels=labels,duplicates="drop")
    fig,ax = styled_figure(figsize=(10,6))
    ax.set_title(title,color="white")
    data_box = [df_q[df_q["quartile"]==q][value_col].dropna().values for q in labels]
    bp = ax.boxplot(data_box,patch_artist=True,medianprops=dict(color="white",linewidth=2))
    for patch,col in zip(bp["boxes"],[C["red"],C["orange"],C["blue"],C["green"]]):
        patch.set_facecolor(col); patch.set_alpha(0.7)
    ax.set_xticklabels(labels,color="white")
    ax.set_ylabel(ylabel,color="white")
    plt.tight_layout()
    if fname: plt.savefig(f"Figures/{fname}.png",dpi=150,bbox_inches="tight")
    plt.show()

def annotate_countries(ax, df, x_col, y_col, countries,
                       fontsize=9, xytext=(8,4)):
    """Annotate selected countries on a scatter plot."""
    for _, row in df[df["country_name"].isin(countries)].iterrows():
        ax.scatter(row[x_col], row[y_col], color=C["orange"], s=150, zorder=6, edgecolors="white", 
                   linewidths=1.5)
        ax.annotate(row["country_name"], (row[x_col], row[y_col]), fontsize=fontsize, color="white", 
                    fontweight="bold", xytext=xytext, textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

def kw_boxplot(ax, df, prof_col, ind_col, groups, colors, do_log=False, annotate_countries_list=None):
    """KW boxplot for a profile variable × indicator."""
    from scipy.stats import kruskal
    data = [df[df[prof_col]==g][ind_col].dropna().values for g in groups]
    avail = [(g,d,c) for g,d,c in zip(groups,data,colors) if len(d)>=3]
    if not avail: return None, None
    g_av, d_av, c_av = zip(*avail)
    stat, p = kruskal(*list(d_av))

    bp = ax.boxplot(list(d_av), patch_artist=True, medianprops=dict(color="white", linewidth=2))
    for patch, col in zip(bp["boxes"], c_av): patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{g}\n(n={len(d)})" for g,d in zip(g_av,d_av)]
    ax.set_xticklabels(xlabels, color="white", fontsize=9)
    if do_log: ax.set_yscale("log")

    for i, vals in enumerate(d_av):
        med = np.median(vals)
        label = f"${med/1e3:.0f}k" if do_log and med>1000 else f"{med:.2f}"
        ax.text(i+1, med, label, ha="center", va="bottom", color="white", fontsize=8, fontweight="bold")

    if annotate_countries_list is not None:
        for _, row in df[df["country_name"].isin(annotate_countries_list)].iterrows():
            if row[prof_col] not in g_av or pd.isna(row[ind_col]): continue
            g_idx = list(g_av).index(row[prof_col]) + 1
            ax.scatter(g_idx, row[ind_col], color=C["orange"], s=100, zorder=6, edgecolors="white", linewidths=1.2)
            ax.annotate(row["country_name"], (g_idx, row[ind_col]), fontsize=7, color="white", 
                        fontweight="bold", xytext=(10,0), textcoords="offset points",
                        bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))
    return stat, p

def spearman_scatter(ax, df, x_col, y_col, x_label, y_label, size_colors, do_log_x=False, do_log_y=False,
                     countries_list=None):
    """Scatter with Spearman annotation and optional country labels."""
    from scipy.stats import spearmanr
    from scipy.stats import linregress

    x_vals = np.log10(df[x_col].clip(lower=1)) if do_log_x else df[x_col]
    y_vals = np.log10(df[y_col].clip(lower=1)) if do_log_y else df[y_col]
    r_sp, p_sp = spearmanr(x_vals.values, y_vals.values)

    for sp, col in size_colors.items():
        sub = df[df["size_profile"]==sp]
        ax.scatter(sub[x_col], sub[y_col], color=col, alpha=0.5, s=40, label=sp)

    slope, intercept, _, _, _ = linregress(x_vals, y_vals)
    x_line = np.linspace(x_vals.min(), x_vals.max(), 100)
    ax.plot(10**x_line if do_log_x else x_line, 10**(intercept+slope*x_line) if do_log_y
            else intercept+slope*x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.6)

    if do_log_x: ax.set_xscale("log")
    if do_log_y: ax.set_yscale("log")
    ax.set_xlabel(x_label, color="white")
    ax.set_ylabel(y_label, color="white")
    ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

    if countries_list is not None: annotate_countries(ax, df, x_col, y_col, countries_list)

    return r_sp, p_sp

countries = ["USA","China","Germany","France","India","Brazil","Japan","Canada"]
print("✅ Style & helpers OK")
````

## C3. Indicator Mapping
</div>

**Strategy :**
- `df_master` columns → original full names (no renaming)
- New derived columns → short keys (`export_pc`, `debt_gdp`...)
- Figures → `IND_LABELS` used **only at plot time**
- `IND_PCT` → % indicators with clip range
- `IND_ABS` → US$ absolute indicators (log scale)
- `ind_cols` → all 16 indicators (no exclusion — filter results a posteriori)


````python
# ── Discover all indicators ───────────────────────────────────────────────────
all_indicators = sorted(df[df["entity_type"] == "COUNTRY"]["indicator_name"].unique())

# ── IND_LABELS : full name → short label (figures only) ──────────────────────
IND_LABELS = {
    "Current account balance (% of GDP)"                                   : "Current Account % GDP",
    "Current account balance (BoP, current US$)"                           : "Current Account (US$)",
    "Expense (% of GDP)"                                                   : "Expense % GDP",
    "Exports of goods and services (% of GDP)"                             : "Exports % GDP",
    "Exports of goods and services (current US$)"                          : "Exports (US$)",
    "External debt stocks, total (DOD, current US$)"                       : "External Debt (US$)",
    "Foreign direct investment, net inflows (BoP, current US$)"            : "FDI Inflows (US$)",
    "GDP (current US$)"                                                    : "GDP (US$)",
    "GDP per capita (current US$)"                                         : "GDP per Capita (US$)",
    "Imports of goods and services (% of GDP)"                             : "Imports % GDP",
    "Imports of goods and services (current US$)"                          : "Imports (US$)",
    "Inflation, consumer prices (annual %)"                                : "Inflation % (CPI)",
    "Population, total"                                                    : "Population",
    "Real interest rate (%)"                                               : "Real Interest Rate %",
    "Trade (% of GDP)"                                                     : "Trade % GDP",
    "Unemployment, total (% of total labor force) (modeled ILO estimate)"  : "Unemployment %",
}

# ── IND_PCT : % indicators with clip range ────────────────────────────────────
# Ranges validated against actual data maxima
IND_PCT = {
    "Inflation, consumer prices (annual %)"                                : (-5,   50),
    "Real interest rate (%)"                                               : (-20,  45),
    "Unemployment, total (% of total labor force) (modeled ILO estimate)"  : (0,    40),
    "Current account balance (% of GDP)"                                   : (-30,  30),
    "Expense (% of GDP)"                                                   : (0,   100),
    "Exports of goods and services (% of GDP)"                             : (0,   250),
    "Imports of goods and services (% of GDP)"                             : (0,   250),
    "Trade (% of GDP)"                                                     : (0,   400),
}

# ── IND_ABS : US$ absolute indicators (log scale + normalize if needed) ───────
IND_ABS = {
    "GDP (current US$)",
    "GDP per capita (current US$)",
    "Population, total",
    "Exports of goods and services (current US$)",
    "Imports of goods and services (current US$)",
    "External debt stocks, total (DOD, current US$)",
    "Foreign direct investment, net inflows (BoP, current US$)",
    "Current account balance (BoP, current US$)",
}

# ── ind_cols : all available indicators ──────────────────────────────────
ind_cols = [i for i in all_indicators if i in df["indicator_name"].unique()]

# ── Helper : get display parameters for any indicator ────────────────────────
def get_ind_params(ind):
    """
    Return display parameters for a given indicator (full name).
    label     : short readable label for axis/title
    clip      : (min, max) tuple for scatter, or None
    use_log   : True if log scale recommended
    needs_norm: True if US$ absolute and no % GDP version available
    """
    label      = IND_LABELS.get(ind, ind)
    clip       = IND_PCT.get(ind, None)
    use_log    = ind in IND_ABS
    needs_norm = ind in IND_ABS and ind not in {
        "GDP per capita (current US$)",
        "Population, total",
    }
    return label, clip, use_log, needs_norm

# ── Verify mapping ────────────────────────────────────────────────────────────
unmapped = [i for i in all_indicators if i not in IND_LABELS]
if unmapped:    print(f"⚠️  Unmapped indicators : {unmapped}")
else:  print(f"✅ {len(all_indicators)} indicators mapped | "  f"{len(IND_PCT)} with clip | {len(IND_ABS)} abs")
````

## C8. Derived Columns
</div>

````python
# Per Capita (Population from indicators dataset)
if pop_col in df_master.columns:
    mask = df_master[pop_col].notna() & (df_master[pop_col] > 0)
    for flow_col, pc_col in [("total_export","export_pc"), ("total_import","import_pc")]:
        if flow_col in df_master.columns:
            df_master.loc[mask, pc_col] = (df_master.loc[mask, flow_col] / df_master.loc[mask, pop_col])

# Trade Ratios vs GDP
if gdp_col in df_master.columns:
    mask = df_master[gdp_col].notna() & (df_master[gdp_col] > 0)
    df_master.loc[mask, "trade_intensity"] = (
        (df_master.loc[mask, "total_export"] + df_master.loc[mask, "total_import"]) / df_master.loc[mask, gdp_col])
    df_master.loc[mask, "export_dep"]  = (df_master.loc[mask, "total_export"] / df_master.loc[mask, gdp_col])
    df_master.loc[mask, "import_dep"]  = (df_master.loc[mask, "total_import"] / df_master.loc[mask, gdp_col])
    df_master.loc[mask, "balance_dep"] = (
        (df_master.loc[mask, "total_export"] - df_master.loc[mask, "total_import"])/ df_master.loc[mask, gdp_col])

# Trade Elasticity = CAGR_export / CAGR_GDP
mask_e = df_master["cagr_gdp"].notna() & (df_master["cagr_gdp"] != 0)
df_master.loc[mask_e, "trade_elasticity_x"] = (
                                        df_master.loc[mask_e, "cagr_export"] / df_master.loc[mask_e, "cagr_gdp"])
df_master.loc[mask_e, "trade_elasticity_m"] = (
                                        df_master.loc[mask_e, "cagr_import"] / df_master.loc[mask_e, "cagr_gdp"])

# Normalize US$ indicators by GDP (for cross-country comparability)
for ind in [
    "External debt stocks, total (DOD, current US$)",
    "Foreign direct investment, net inflows (BoP, current US$)"]:
    if ind in df_master.columns and gdp_col in df_master.columns:
        mask = df_master[gdp_col].notna() & (df_master[gdp_col] > 0)
        label = IND_LABELS.get(ind,"").replace(" ","_").replace("(","").replace(")","").replace("$","").strip()
        df_master.loc[mask, f"norm_{label}"] = (df_master.loc[mask, ind] / df_master.loc[mask, gdp_col])

print(f"✅ Derived columns added")
derived = ["export_pc","import_pc","trade_intensity","export_dep","import_dep","balance_dep",
           "trade_elasticity_x","trade_elasticity_m"]
for c in derived:
    n = df_master[c].notna().sum() if c in df_master.columns else 0
    print(f"  {c:<30} : {n} countries")


df_master.sample()
````

````python
PROFILE_ENCODINGS = {
"size_profile"         : {"Leader":4,"Major":3,"Mid":2,"Small":1},
"balance_profile"      : {"Strong Surplus":4,"Slight Surplus":3, "Slight Deficit":2,"Structural Deficit":1},
"growth_profile"       : {"Strong Grower":4,"Moderate Grower":3, "Weak Grower":2,"Declining":1},
"stability_profile"    : {"Stable":4,"Moderate":3, "Volatile":2,"Very Volatile":1},
"consistency_profile"  : {"Very Consistent":4,"Consistent":3,  "Inconsistent":2,"Very Inconsistent":1},
"structural_profile"   : {"Strong":4,"Good":3,"Moderate":2,"Weak":1},
"meta_profile" : {"Global Leader":6,"Established Power":5, "Rising Challenger":4,"Volatile Emerging":3, "Transitional":2, "Peripheral":1},
"profile_full_periode" : {"Robust":4, "Stable but Concentrated":3, "Vulnerable":2,"Critical Dependency":1},
"static_quadrant"      : {"Dominant":4,"Balanced":3, "Vulnerable":2,"Marginal":1},
"dynamic_quadrant"     : {"Star":4,"Risky Grower":3, "Stable Stagnant":2,"Declining":1},
"strategic_quadrant"   : {"Stars":4,"Cash Cow":3, "Emerging":2,"Marginal":1},
"combined_profile"     : {"Global Champion":4, "Competitive":3, "Vulnerable":2,"Fragile":1},
"strategic_quadrant_combined": {"Strong & Resilient":4, "Strong but Exposed":3, "Diversified Minor":2, "Weak & Vulnerable":1},
                        }
for col, mapping in PROFILE_ENCODINGS.items():
    if col in df_master.columns:
        df_master[f"{col}_num"] = df_master[col].map(mapping)
        print(f"✅ {col}_num : {df_master[f'{col}_num'].notna().sum()} countries")
````

## D.2. Trade Elasticity — 6 Situations

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### 2.1 — Classification of the trade-GDP growth relationship
</div>

**Trade Elasticity** = CAGR_export / CAGR_GDP

Classification is based first on the **sign** of each CAGR, then on the magnitude.

**Warning** : situations 5 and 6 both produce a positive elasticity ratio from (−/−).
Sign-based classification must come **before** using the ratio value.

| # | CAGR_export | CAGR_GDP | Reading |
|---|---|---|---|
| 1 | + | + | > 1 — Trade-led : trade grows faster → opening |
| 2 | + | + | 0→1 — Endogenous : economy grows faster than trade |
| 3 | + | − | Trade Resistant : trade grows despite GDP decline |
| 4 | − | + | Decoupling : GDP grows but trade declines |
| 5 | − | − | > 1 — Double Decline : trade collapses faster |
| 6 | − | − | 0→1 — Double Decline : GDP more affected |


````python
def classify_6sit(row):
    xe = row.get("cagr_export", np.nan)
    xg = row.get("cagr_gdp",    np.nan)
    if pd.isna(xe) or pd.isna(xg): return "Unknown"
    if xe >= 0 and xg >= 0:  return "1-Trade-led"   if xe > xg else "2-Endogenous"
    elif xe >= 0 and xg < 0: return "3-Trade Resistant"
    elif xe < 0 and xg >= 0: return "4-Decoupling"
    else:                    return "5-Double Decline (trade faster)" if abs(xe) > abs(xg) else "6-Double Decline (GDP faster)"

SIT_COLORS = {
                "1-Trade-led"                        : C["green"],
                "2-Endogenous"                       : C["teal"],
                "3-Trade Resistant"                  : C["blue"],
                "4-Decoupling"                       : C["orange"],
                "5-Double Decline (both negative)"   : C["red"],
                "6-Double Decline (GDP faster)"      : C["purple"],
                "Unknown"                            : C["grey"]
              }
SIT_LABELS = {k: k.split("-",1)[1] if "-" in k else k for k in SIT_COLORS}

df_d2 = df_master[["country_name","cagr_export","cagr_gdp", "trade_elasticity_x","size_profile"]].dropna(
    subset=["cagr_export","cagr_gdp"])
df_d2["situation"] = df_d2.apply(classify_6sit, axis=1)
````

````python
# ── Figure A : Scatter CAGR_export vs CAGR_GDP ───────────────────────────────
fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("CAGR Export vs CAGR GDP — 6 Situations", color="white")

for sit, col in SIT_COLORS.items():
    sub = df_d2[df_d2["situation"]==sit]
    if sub.empty: continue
    ax.scatter(sub["cagr_gdp"]*100, sub["cagr_export"]*100, color=col, alpha=0.7, s=60, label=SIT_LABELS[sit])

lim = (df_d2[["cagr_export","cagr_gdp"]].abs().max().max())*100*1.1
ax.plot([-lim,lim],[-lim,lim], color="white", linestyle="--", alpha=0.4, label="Elasticity = 1")
ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)
ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_d2[df_d2["country_name"].isin(countries)].iterrows():
    ax.annotate(row["country_name"], (row["cagr_gdp"]*100, row["cagr_export"]*100), fontsize=10, color="white",
                xytext=(5,3), fontweight="bold", textcoords="offset points")

ax.set_xlabel("CAGR GDP (%)", color="white")
ax.set_ylabel("CAGR Export (%)", color="white")
ax.set_xlim(-15, 25)
ax.set_ylim(-25, 40)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/D_2_1_scatter.png", dpi=150, bbox_inches="tight")
plt.show()
````

````python
# ── Figure B : Country count by situation ─────────────────────────────────────
sit_counts = df_d2["situation"].value_counts()
n_total    = len(df_d2)

fig, ax = styled_figure(figsize=(11, 6))
ax.set_title("Country Count by Situation", color="white")

ax.barh([SIT_LABELS.get(s,s) for s in sit_counts.index], sit_counts.values,
        color=[SIT_COLORS.get(s,C["grey"]) for s in sit_counts.index], alpha=0.85)
ax.set_xlabel("Number of countries", color="white")
for i, (s, v) in enumerate(sit_counts.items()):
    ax.text(v+0.3, i, f"{v} ({v/n_total*100:.0f}%)", va="center", color="white", fontsize=9)

plt.tight_layout()
plt.savefig("../Figures/Link/D_2_1_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight :**
- **94 countries Trade-led (51%)** → most common situation : global trade integration deepening
- **51 countries Endogenous (28%)** → economies growing faster than their trade
- **26 countries Decoupling (14%)** → GDP grows but trade declines → structural loss of competitiveness
- Only **4 countries Double Decline** → rare, typically fragile or sanctioned economies
- **Germany, USA** : Endogenous → their economies grow but trade grows more slowly
- **India, France, Brazil** : Trade-led → their exports grow faster than GDP —
  trade is structurally more dynamic than the domestic economy.
- **China** : Decoupling → economy still growing but trade volume declining
- **Answer** : Trade-led growth is the dominant pattern globally, but it is
  concentrated in mid-size economies. The largest economies tend toward
  endogenous growth where the domestic market absorbs most output.

### 2.2 — Trade Elasticity value × CAGR GDP
</div>

````python
# Positive situations only (1 & 2) — elasticity ratio is meaningful
df_d22 = df_d2[df_d2["situation"].isin(["1-Trade-led","2-Endogenous"])].copy()
df_d22["elasticity_clipped"] = df_d22["trade_elasticity_x"].clip(0, 5)

# ── Figure A : Scatter Elasticity vs CAGR_GDP ─────────────────────────────────
fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Trade Elasticity vs GDP CAGR\n" "(above 1 = trade-led, below 1 = endogenous)", color="white")

for sit in ["1-Trade-led","2-Endogenous"]:
    sub = df_d22[df_d22["situation"]==sit]
    ax.scatter(sub["cagr_gdp"]*100, sub["elasticity_clipped"], color=SIT_COLORS[sit], alpha=0.5, s=60,
               label=SIT_LABELS[sit])

ax.axhline(1, color="white", linestyle="--", alpha=0.5, label="Elasticity = 1")
ax.axvline(0, color=C["grey"],  linestyle=":",  alpha=0.4)

for _, row in df_d22[df_d22["country_name"].isin(countries)].iterrows():
    ax.annotate(row["country_name"], (row["cagr_gdp"]*100, row["elasticity_clipped"]), fontsize=9, fontweight="bold",
                color="white", xytext=(5,3), textcoords="offset points")

ax.set_xlabel("CAGR GDP (%)", color="white")
ax.set_ylabel("Trade Elasticity (clipped 0–5)", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
r, p = spearmanr(df_d22["cagr_gdp"], df_d22["elasticity_clipped"])
ax.text(0.6, 0.95, f"Spearman ρ={r:.2f} (p={p:.3f})", transform=ax.transAxes, color="white", fontsize=10)

plt.tight_layout()
plt.savefig("../Figures/link/D_2_2_scatter.png", dpi=150, bbox_inches="tight")
plt.show()
````

````python
# Figure B : Boxplot Elasticity by CAGR_GDP Quartile 
df_d22["gdp_cagr_q"] = pd.qcut(df_d22["cagr_gdp"], 4, labels=["Q1\n(Slowest)","Q2","Q3","Q4\n(Fastest)"])
labels_q = ["Q1\n(Slowest)","Q2","Q3","Q4\n(Fastest)"]
data_box  = [df_d22[df_d22["gdp_cagr_q"]==q]["elasticity_clipped"].dropna().values for q in labels_q]

fig, ax = styled_figure(figsize=(11, 7))
ax.set_title("Trade Elasticity by GDP CAGR Quartile\n" "(do faster-growing economies open more?)", color="white")

bp = ax.boxplot(data_box, patch_artist=True, 
                  medianprops  = dict(color="white", linewidth=2),
                  flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                  whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                  capprops     = dict(color="white", linewidth=1.5),
                  boxprops     = dict(linewidth=1.2))

for patch, col in zip(bp["boxes"], [C["red"],C["orange"],C["blue"],C["green"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)

ax.axhline(1, color="white", linestyle="--", alpha=0.5, label="Elasticity = 1")
ax.set_xticklabels(labels_q, color="white")
ax.set_ylabel("Trade Elasticity", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

for i, q in enumerate(labels_q):
    vals = df_d22[df_d22["gdp_cagr_q"]==q]["elasticity_clipped"].dropna()
    if not vals.empty:
        ax.text(i+1, vals.median()+0.05, f"{vals.median():.2f}", ha="center", color="white", fontsize=9, fontweight="bold")

plt.tight_layout()
plt.savefig("../Figures/Link/D_2_2_boxplot.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight :**
- Spearman ρ = −0.37 → **negative** relationship : slower-growing economies have **higher** elasticity
- Q1 (slowest GDP growth) median elasticity = 2.35 vs Q4 (fastest) = 1.05
- **Interpretation** : countries with slow GDP growth but positive export growth
  naturally produce high elasticity ratios — trade is their relative engine
- Fast-growing economies (emerging markets) grow across all sectors, so trade
  grows proportionally but not necessarily faster
- **Germany, USA** : elasticity < 1 → economy outpaces trade in relative terms
- **France, Brazil** : high elasticity despite moderate GDP growth —
  their export engines outpace their overall economic performance.
- **Caution** : elasticity is only meaningful when both CAGRs are positive
  (situations 1 & 2). Negative CAGRs produce misleading ratios — this is
  why sign-based classification precedes ratio computation.
- **Answer** : Faster-growing economies do NOT necessarily have higher trade elasticity.
  High elasticity often signals trade as a compensating engine in slow-growth contexts.

## D.3. Per Capita Trade × GDP per Capita
</div>

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### D3.1 — Scatter : Export & Import per Capita vs GDP per Capita
</div>

````python
df_d3 = df_master[["country_name","export_pc","import_pc", gdp_pc_col,"size_profile"]].dropna()
log_x  = np.log(df_d3[gdp_pc_col].clip(lower=1))
log_ye = np.log(df_d3["export_pc"].clip(lower=1))
log_ym = np.log(df_d3["import_pc"].clip(lower=1))
slope_e, int_e, _, _, _ = linregress(log_x, log_ye)
slope_m, int_m, _, _, _ = linregress(log_x, log_ym)
r_x, _ = spearmanr(log_x, log_ye)
r_m, _ = spearmanr(log_x, log_ym)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Export & Import per Capita vs GDP per Capita\n" "(circles = export, triangles = import)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_d3[df_d3["size_profile"]==sp]
    ax.scatter(sub[gdp_pc_col], sub["export_pc"], color=col, alpha=0.5, s=40, label=f"Export {sp}")
    ax.scatter(sub[gdp_pc_col], sub["import_pc"], color=col, alpha=0.25, s=25, marker="^")

x_line = np.linspace(log_x.min(), log_x.max(), 100)
ax.plot(np.exp(x_line), np.exp(int_e + slope_e * x_line), color=C["blue"], linewidth=2, linestyle="--",
        label=f"Export trend (ρ={r_x:.2f})")
ax.plot(np.exp(x_line), np.exp(int_m + slope_m * x_line), color=C["orange"], linewidth=2, linestyle=":",
        label=f"Import trend (ρ={r_m:.2f})")

for _, row in df_d3[df_d3["country_name"].isin(countries)].iterrows():
    ax.annotate(row["country_name"], (row[gdp_pc_col], row["export_pc"]), fontsize=9, fontweight="bold", 
                color="white", xytext=(4,2), textcoords="offset points")

ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("GDP per Capita (log)", color="white")
ax.set_ylabel("Trade per Capita (log)", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=7)

plt.tight_layout()
plt.savefig("../Figures/Link/D_3_1_scatter.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight :**
- Export ρ = 0.83 and Import ρ = 0.92 → **very strong** relationship (vs ρ=0.29 in D1.1)
- Log-log linearity → constant elasticity :
  a 1% increase in GDP_pc → ~0.83% increase in Export_pc and ~0.92% in Import_pc
- **Import_pc correlates more strongly** than Export_pc → richer populations
  consume more imported goods proportionally
- **Germany** : exports per person well above the trend → structural export surplus
- **India** : well below trend for both → low per capita trade despite large economy
- **USA** : below export trend for its income level → large domestic market
  reduces relative export dependence despite massive absolute volumes.
- **Answer** : Per capita GDP is a strong predictor of per capita trade —
  wealthier populations both export more and import more per person,
  with imports growing faster than exports as income rises.

**Question :** "Does macro indicator concentration follow world trade concentration over time?"

### 4.5 — Insights
</div>

**World trade is deconcentrating (2015–2019) :**
HHI drops from 765 to 405 — world exports become less dominated by top countries.
COVID 2020 causes a temporary reconcentration (504) before stabilizing at ~490.
The trend is toward a more multipolar trade structure.

**FDI and Exports (absolute) follow trade concentration (ρ=+0.72, +0.71) :**
Both are [S] size-driven indicators — when trade concentrates, capital flows
and absolute export volumes concentrate in the same countries. This confirms
size indicators track the trade system structure, not just individual country performance.

**Unemployment follows inversely (ρ=−0.65) — most unexpected signal :**
When world trade deconcentrates (more countries exporting), global unemployment
tends to be higher. Two readings :
1. Trade deconcentration coincides with economic dispersion — more countries
   participate but each with lower productivity and higher unemployment
2. COVID reversal : 2020 spike in unemployment + temporary HHI increase
   are structurally linked — global shocks simultaneously raise unemployment
   and reconcentrate trade toward dominant exporters
- Both are size-driven indicators — when trade concentrates, capital flows
  and absolute export volumes concentrate in the same countries. This confirms
  that FDI follows trade power, not trade openness.

**GDP volatility diverges from trade concentration (ρ = −0.59) :**
When macro instability rises (high CV), trade deconcentrates — shocks
force diversification of trade partners rather than concentration.
This is a counter-intuitive but robust finding across all 10 years.

**CA% diverges from trade (ρ=−0.38) :**
When current account positions concentrate (more extreme surpluses/deficits),
trade deconcentrates. Structural trade imbalances and geographic trade
concentration move in opposite directions — a country can run a large surplus
without dominating world trade volumes.

**No macro indicator fully tracks trade concentration :**
All correlations are moderate (|ρ| < 0.75) — trade concentration follows
its own structural logic, only partially explained by macro dynamics.

**Limit :** Only 10 data points (2015–2024) — Spearman on n=10 is
statistically fragile. Results are indicative, not confirmatory.
COVID 2020 is an influential point that may drive several correlations.

# E. RF + OLS — Which Indicators Explain Trade Variables?
</div>

**Objective** : Identify which economic indicators best explain each trade variable.

**Structure :**
- **E1** — Prepare RF dataset (encode categoricals, define variable groups)
- **E2** — Random Forest : general results on all 16 indicators
- **E3** — Annotate : flag trade-derived [T], size-driven [S], economic [E] indicators
- **E4** — Gap RF/OLS : identify non-linear or size-driven relationships
- **E5** — Decisions & Filtering : apply exclusions, filtered heatmap, filtered gap
- **E6** — Output : `top_indicators_per_var` → OLS direction & magnitude → helper `plot_trade_vs_ind()`

**Key design choices :**
- All 16 indicators included — no a priori exclusion based on economic theory.
  Results are filtered a posteriori based on [T]/[S]/[E] classification
  and RF/OLS gap analysis in E5.
- Results annotated a posteriori (trade-derived vs purely economic)
- `df_master` unchanged — encoding done in local `df_rf` copy
- OLS uses standardized coefficients → direction and magnitude readable


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## E1. Prepare RF Dataset
</div>

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E1.1 — Encode Categorical Variables
</div>

Categorical trade variables must be encoded numerically for RF/OLS.
Encoding done in a **local copy** `df_rf` — `df_master` is unchanged.


````python
# ── Local copy — df_master unchanged ─────────────────────────────────────────
df_rf = df_master.copy()

# ── Categorical encodings ─────────────────────────────────────────────────────
CATEGORICAL_ENCODE = {
"combined_profile" : {"Global Champion" : 4, "Competitive" : 3, "Vulnerable" : 2, "Fragile" : 1},
"strategic_quadrant_combined":{"Strong & Resilient": 4, "Strong but Exposed": 3, "Diversified Minor": 2, "Weak & Vulnerable": 1},
"size_profile" : {"Leader" : 4, "Major" : 3, "Mid" : 2, "Small" : 1},
"balance_profile" : {"Strong Surplus": 4, "Slight Surplus": 3, "Slight Deficit": 2, "Structural Deficit": 1 },
"growth_profile" : {"Strong Grower" : 4, "Moderate Grower" : 3, "Weak Grower" : 2, "Declining" : 1 },
"stability_profile" : {"Stable" : 4, "Moderate" : 3, "Volatile" : 2, "Very Volatile" : 1 },
"consistency_profile" : {"Very Consistent" : 4, "Consistent" : 3, "Inconsistent" : 2, "Very Inconsistent" : 1 },
"structural_profile" : { "Strong" : 4, "Good" : 3, "Moderate" : 2, "Weak" : 1 },
"profile_full_periode": {"Robust":4, "Stable but Concentrated":3, "Critical Dependency":2, "Vulnerable":1},
"static_quadrant" : {"Dominant"   : 4, "Balanced"   : 3, "Vulnerable" : 2, "Marginal"   : 1},
"dynamic_quadrant" : {"Star" : 4, "Risky Grower" : 3, "Stable Stagnant" : 2, "Declining"  : 1 },
"strategic_quadrant" : {"Stars" : 4, "Cash Cow" : 3, "Emerging" : 2, "Marginal"  : 1},  
"meta_profile" : {"Global Leader": 6, "Established Power": 5, "Rising Challenger": 4,
                  "Volatile Emerging": 3, "Peripheral": 2, "Transitional": 1}}

for col, mapping in CATEGORICAL_ENCODE.items():
    if col in df_rf.columns:
        df_rf[f"{col}_num"] = df_rf[col].map(mapping)

encoded_cols = [f"{c}_num" for c in CATEGORICAL_ENCODE if f"{c}_num" in df_rf.columns]
print(f"✅ Encoded : {encoded_cols}")
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E1.2 — Define Variable Groups
</div>

````python
# Trade target variables : All available — filter results a posteriori

# Numeric trade variables (directly usable in RF)
trade_vars_numeric = [c for c in all_trade_cols
                      if c in df_rf.columns
                      and df_rf[c].dtype in ["float64","int64"]
                      and df_rf[c].notna().sum() >= 30]

# Add encoded categoricals
trade_vars_all_rf = trade_vars_numeric + encoded_cols
trade_vars_all_rf = [c for c in trade_vars_all_rf if c in df_rf.columns]

# ── Indicator variables (all 16) ──────────────────────────────────────────────
ind_vars_rf = [c for c in ind_cols if c in df_rf.columns]

# ── Trade-derived indicators (for annotation in J3) ──────────────────────────
IND_TRADE_DERIVED = {
    "Exports of goods and services (current US$)",
    "Exports of goods and services (% of GDP)",
    "Imports of goods and services (current US$)",
    "Imports of goods and services (% of GDP)",
    "Trade (% of GDP)",
    "Current account balance (BoP, current US$)",  # US$ version
}

print(f"✅ Trade targets for RF : {len(trade_vars_all_rf)}")
print(f"✅ Indicator features   : {len(ind_vars_rf)}")
print(f"   of which trade-derived : {len([i for i in ind_vars_rf if i in IND_TRADE_DERIVED])} ")
````

<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## E2. Random Forest — General Results
</div>

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E2.1 — RF
</div>

RF run on **all 16 indicators** for each trade variable.
Results stored in `rf_results` dict. No exclusion at this stage.


````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

rf_results = {}

print(f"Running RF : {len(trade_vars_all_rf)} targets × {len(ind_vars_rf)} indicators")
print("=" * 50)

for target in trade_vars_all_rf:
    df_sub = df_rf[[target] + ind_vars_rf].dropna()
    if len(df_sub) < 30:        continue
    X = df_sub[ind_vars_rf].values
    y = df_sub[target].values

    rf = RandomForestRegressor(n_estimators=300, random_state=42, max_features="sqrt", n_jobs=-1)
    rf.fit(X, y)
    r2_rf = rf.score(X, y)

    fi = pd.DataFrame({"indicator"  : ind_vars_rf,
                       "importance" : rf.feature_importances_,}).sort_values("importance", ascending=False)

    # OLS on top 5 RF indicators
    top5 = fi.head(5)["indicator"].tolist()
    df_ols = df_sub[[target] + top5].dropna()
    r2_ols = np.nan
    ols_coefs = {}
    if len(df_ols) >= 25:
        X_s = StandardScaler().fit_transform(df_ols[top5].values)
        y_s = df_ols[target].values
        lr  = LinearRegression().fit(X_s, y_s)
        r2_ols = lr.score(X_s, y_s)
        ols_coefs = dict(zip(top5, lr.coef_))

    label = ALL_TRADE_LABELS.get(target, target)
    rf_results[target] = {
                            "label"     : label,
                            "fi"        : fi,
                            "r2_rf"     : r2_rf,
                            "r2_ols"    : r2_ols,
                            "ols_coefs" : ols_coefs,
                            "n"         : len(df_sub),
                        }

    top1     = fi.iloc[0]
    top1_lbl = IND_LABELS.get(top1["indicator"], top1["indicator"])
    print(f"  {label:<35} RF={r2_rf:.3f}  OLS={r2_ols:.3f}  "
          f"n={len(df_sub)}  Top: {top1_lbl} ({top1['importance']:.3f})")

print(f"\n✅ RF complete : {len(rf_results)} targets")
````

## E3. Annotation — Flag Trade-Derived Indicators
</div>

Some indicators in the dataset are **trade-derived** :
- Exports/Imports (`US$` and % GDP), Trade % GDP, Current Account (`US$`).
- When these appear as top RF indicators for a trade variable,
the relationship is **tautological** — not economically informative.
- This step flags them in the results without excluding them.

````python
# ── Flag trade-derived indicators in RF results ───────────────────────────────
print("RF Results — Top 3 indicators per trade variable")
print("  [T] = trade-derived indicator (tautological relationship)")
print("  [E] = purely economic indicator")
print("=" * 75)

for target, res in rf_results.items():
    print(f"\n  {res['label']} (RF R²={res['r2_rf']:.3f} | OLS R²={res['r2_ols']:.3f})")
    print(f"  {'─'*60}")
    for _, row in res["fi"].head(3).iterrows():
        ind   = row["indicator"]
        label = IND_LABELS.get(ind, ind)
        flag  = "[T]" if ind in IND_TRADE_DERIVED else "[E]"
        coef  = res["ols_coefs"].get(ind, np.nan)
        direction = "▲" if coef > 0 else "▼" if coef < 0 else " "
        print(f"    {flag} {direction} {row['importance']:.3f}  {label}")

# ── Summary : which targets are dominated by trade-derived indicators ──────────
print("\n" + "=" * 75)
print("  TAUTOLOGY CHECK — targets where top indicator is trade-derived :")
for target, res in rf_results.items():
    top1 = res["fi"].iloc[0]["indicator"]
    if top1 in IND_TRADE_DERIVED:
        print(f"  ⚠️  {res['label']:<35} → top = {IND_LABELS.get(top1, top1)}")
````

## E4. Gap RF/OLS — Non-Linear & Size-Driven Relationships
</div>

**Gap RF/OLS** = RF R² - OLS R²

A large gap signals that the relationship is :
- **Non-linear** : RF captures it, OLS misses it
- **Size-driven** : large absolute indicators (US$) capture country size, not economics
- **Not causally interpretable** : avoid scatter plots for these

| Gap | Interpretation | Action |
|---|---|---|
| < 0.10 | Linear relationship — OLS reliable | Use scatter + regression |
| 0.10 – 0.30 | Moderate non-linearity | Interpret with caution |
| > 0.30 | Strong non-linearity / size effect | Use boxplot — avoid scatter |


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E4.1 — Gap: Compute and classify
</div>

````python
gap_results = []
for target, res in rf_results.items():
    if np.isnan(res["r2_ols"]): continue
    gap = res["r2_rf"] - res["r2_ols"]
    if gap > 0.30:
        interpretation = "Strong non-linearity / size effect — avoid scatter"; flag = "🔴"
    elif gap > 0.10:
        interpretation = "Moderate non-linearity — interpret with caution";    flag = "🟡"
    else:
        interpretation = "Linear — OLS reliable"; flag = "🟢"

    gap_results.append({
        "target"         : res["label"],
        "r2_rf"          : round(res["r2_rf"],  3),
        "r2_ols"         : round(res["r2_ols"], 3),
        "gap"            : round(gap, 3),
        "flag"           : flag,
        "interpretation" : interpretation,
    })

df_gap = pd.DataFrame(gap_results).sort_values("gap", ascending=False)
print("RF/OLS Gap Analysis :")
df_gap[["target","r2_rf","r2_ols","gap","flag","interpretation"]]
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E4.2 — Gap Visualization
</div>

````python
fig, ax = styled_figure(figsize=(13, max(6, len(df_gap)*0.45)))
ax.set_title("RF R² vs OLS R² — Gap Indicates Non-Linearity\n"
             "Y-label color = gap severity  |  "  "Blue bars = RF R²  |  Orange bars = OLS R²  |  "
             "Dashed lines: gap thresholds 0.10 and 0.30", color="white", fontsize=12)

y_pos = np.arange(len(df_gap))
ax.barh(y_pos, df_gap["r2_rf"],  height=0.4, color=C["blue"],   alpha=0.8, label="RF R²")
ax.barh(y_pos - 0.4, df_gap["r2_ols"], height=0.4, color=C["orange"], alpha=0.8, label="OLS R²")

ax.set_yticks(y_pos - 0.2)
ax.set_yticklabels(df_gap["target"], color="white", fontsize=8)
ax.set_xlabel("R²", color="white")
ax.axvline(0.1, color=C["grey"],   linestyle="--", alpha=0.4, linewidth=2)
ax.axvline(0.3, color=C["orange"], linestyle="--", alpha=0.4, linewidth=2)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=9)


# Color y-labels by gap severity
for i, (_, row) in enumerate(df_gap.iterrows()):
    col = C["red"] if row["gap"]>0.30 else C["orange"] if row["gap"]>0.10 else C["green"]
    ax.get_yticklabels()[i].set_color(col)

plt.tight_layout()
plt.savefig("../Figures/Link/E_4_2_gap_rf_ols.png", dpi=150, bbox_inches="tight")
plt.show()
````

This section documents all filtering decisions made after reviewing J2/J3/J4 results.
Each decision is explicitly justified.

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E5.1 — Decision Log
</div>

**Target Variables — Exclusions**

| Variable | Decision | Reason |
|---|---|---|
| `z_cagr_export/import` | ❌ Exclude | Identical results to `cagr_export/import` — pure z-score transformation, no added value |
| `z_volatility_export/import` | ❌ Exclude | Identical to `volatility_export/import` |
| `z_negfreq_export/import` | ❌ Exclude | Identical to `negfreq_export/import` |
| `z_export_share` | ❌ Exclude | Identical to `export_share_structural` |
| `z_import_share` | ❌ Exclude | Identical to `import_share_structural` |
| `combined_profile_enc` | ❌ Exclude | Encoded version of combined_profile — redundant |
| `strategic_quadrant_enc` | ❌ Exclude | Encoded version of strategic_quadrant — redundant |
| `z_coverage_ratio` | ❌ Exclude | Identical to `coverage_ratio_structural` |
| `years_export/import` | ❌ Exclude | Technical variable (data coverage count) — not analytically meaningful |
| `total_export/import` | ✅ Keep + [S] | Size effect but direct trade volume — annotated as size-driven |
| `resilience_export/import` | ✅ Keep | Synthetic indicator (growth adjusted for volatility) — economically meaningful |
| `avg_growth_export/import` | ✅ Keep | Complementary to CAGR (arithmetic vs compound) |

**Indicator Variables — Flags**

| Indicator | Flag | Reason |
|---|---|---|
| `Population, total` | [S] Size-driven | Captures country size, not economic relationship |
| `GDP (current US$)` | [S] Size-driven | Absolute size — kept but annotated |
| `Exports/Imports (US$)` | [T] Trade-derived | Already flagged — tautological |
| `External Debt (US$)` | [S] Size-driven | Use `norm_External_Debt` (normalized by GDP) instead |
| `FDI Inflows (US$)` | [S] Size-driven | Use `norm_FDI` (normalized by GDP) instead |

**Visualization Type — Based on Gap RF/OLS**

| Gap | Flag | Visualization |
|---|---|---|
| > 0.30 | 🔴 | Boxplot by indicator quartile |
| 0.10 – 0.30 | 🟡 | Scatter + caution note |
| < 0.10 | 🟢 | Scatter + regression line (verify tautology) |

**Multicolinearity Note**

`Neg.Freq Export` vs `GDP per Capita` → OLS coefficient positive (counter-intuitive).
Likely multicolinearity artifact — flagged [M] in results.


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E5.2 — Apply Filters
</div>

````python
# ── Targets to exclude ───────────────────────────────────────────────────────
TARGETS_EXCLUDE = {
    # Z-scores : redundant with raw variables
    "z_cagr_export","z_cagr_import",
    "z_volatility_export","z_volatility_import",
    "z_negfreq_export","z_negfreq_import",
    "z_export_share","z_import_share",
    "z_coverage_ratio",
    # Technical variables : data coverage count
    "years_export","years_import",
}

# ── Indicator flags ───────────────────────────────────────────────────────────
IND_SIZE_DRIVEN = {
    "Population, total",   "GDP (current US$)", "External debt stocks, total (DOD, current US$)",
    "Foreign direct investment, net inflows (BoP, current US$)"}

# Use normalized versions where available
IND_USE_NORM = {"External debt stocks, total (DOD, current US$)" : "norm_External_Debt_US",
                "Foreign direct investment, net inflows (BoP, current US$)" : "norm_FDI_Inflows_US"}

# Multicolinearity suspects
IND_MULTICOL = {("negfreq_export", "GDP per capita (current US$)")}

def get_ind_flag(ind):
    """Return flag string for a given indicator."""
    if ind in IND_TRADE_DERIVED: return "[T]"
    if ind in IND_SIZE_DRIVEN:   return "[S]"
    return "[E]"

# ── Filtered RF results ───────────────────────────────────────────────────────
rf_results_filtered = {
    target: res
    for target, res in rf_results.items()
    if target not in TARGETS_EXCLUDE
}

print(f"✅ Targets before filtering : {len(rf_results)}")
print(f"✅ Targets after filtering  : {len(rf_results_filtered)}")
print(f"   Excluded ({len(TARGETS_EXCLUDE)}) : {sorted(TARGETS_EXCLUDE)}")
````

## E6. Output — top_indicators_per_var
</div>

Final output used by G/H/I/J sections.

**Selection logic :**
- Exclude only `[T]` trade-derived indicators (tautological)
- Keep `[S]` size-driven and `[E]` economic — both annotated, neither excluded
- Rank by RF importance (highest importance = most explanatory power)
- `viz_type` based on gap RF/OLS from J4

**Rationale for keeping [S] :**
  Excluding size-driven indicators produces absurd results —
  e.g. Unemployment % as top indicator for Export Share (RF importance=0.022)
  while External Debt (US$) captures 80% of variance through country size.
  [S] indicators are kept, flagged, and their size-driven nature is documented.
  Users of `top_indicators_per_var` should treat [S] results with caution.

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### E6.1 — Build top_indicators_per_var
</div>

````python
# ── Gap lookup ───────────────────────────────────────────────────────────────
gap_lookup = {row["target"]: row for _, row in df_gap_filtered.iterrows()}

top_indicators_per_var = {}

for target, res in rf_results_filtered.items():
    fi = res["fi"].copy()
    fi["label"] = fi["indicator"].map(IND_LABELS).fillna(fi["indicator"])
    fi["flag"]  = fi["indicator"].apply(get_ind_flag)

    # ── Exclude only [T] — keep [S] and [E] ranked by RF importance ──────────
    fi_non_T = fi[fi["flag"] != "[T]"]

    # ── Top 3 by importance — original names only (no norm replacement) ───────
    top3     = fi_non_T.head(3)["indicator"].tolist()
    top3     = [i for i in top3 if i in df_master.columns]
    best_imp = fi_non_T.iloc[0]["importance"] if not fi_non_T.empty else 0
    size_dominated = best_imp < 0.05

    # ── Visualization type from gap ───────────────────────────────────────────
    label   = res["label"]
    gap_row = gap_lookup.get(label, {})
    viz     = gap_row.get("viz_type", "boxplot")

    top_indicators_per_var[target] = {
        "indicators"     : top3,
        "viz_type"       : viz,
        "r2_rf"          : res["r2_rf"],
        "r2_ols"         : res["r2_ols"],
        "size_dominated" : size_dominated,
    }

# ── Summary ───────────────────────────────────────────────────────────────────
print("top_indicators_per_var — FINAL OUTPUT")
print("=" * 75)
for target, info in top_indicators_per_var.items():
    label    = ALL_TRADE_LABELS.get(target, target)
    inds     = [IND_LABELS.get(i, i) for i in info["indicators"]]
    flags    = [get_ind_flag(i) for i in info["indicators"]]
    viz_icon = {"boxplot":"📦","scatter_caution":"⚠️ 📈","scatter":"📈"}.get(info["viz_type"],"?")
    size_note = " 🔵[size-dominated]" if info["size_dominated"] else ""
    print(f"\n  {label:<35} {viz_icon}{size_note}")
    for ind, flag in zip(inds, flags):
        print(f"    {flag} {ind}")

print(f"\n✅ {len(top_indicators_per_var)} entries")
print(f"   📦 boxplot         : {sum(1 for v in top_indicators_per_var.values() if v['viz_type']=='boxplot')}")
print(f"   ⚠️  scatter_caution : {sum(1 for v in top_indicators_per_var.values() if v['viz_type']=='scatter_caution')}")
print(f"   📈 scatter         : {sum(1 for v in top_indicators_per_var.values() if v['viz_type']=='scatter')}")
print(f"   🔵 size-dominated  : {sum(1 for v in top_indicators_per_var.values() if v['size_dominated'])}")
````

### E6.3 — Helper : plot_trade_vs_ind()
</div>

Universal plotting function for G/H/I/J sections.

- Selects scatter or boxplot automatically based on `viz_type`
- Applies `get_ind_params()` for clip, log scale, label
- Annotates selected countries with **bbox** for readability
- Flags [T]/[S]/[E] shown in title


````python
def plot_trade_vs_ind(df, trade_var, ind, title=None, trade_label=None, countries=None, fname=None):
    """    Universal trade × indicator plot.
                - Selects scatter or boxplot based on viz_type
                - Normalizes [S] absolute indicators by GDP at plot time
                - Annotates selected countries with bbox    """
    info        = top_indicators_per_var.get(trade_var, {})
    viz_type    = info.get("viz_type", "boxplot") if isinstance(info, dict) else "boxplot"
    label_ind, clip, use_log, _ = get_ind_params(ind)
    label_trade = trade_label or ALL_TRADE_LABELS.get(trade_var, trade_var)
    flag_str    = get_ind_flag(ind)
    title_      = title or f"{label_trade} × {label_ind}  {flag_str}"

    # ── Prepare data ──────────────────────────────────────────────────────────
    gdp_col  = "GDP (current US$)"
    col_list = ["country_name", trade_var, ind]
    if ind in IND_ABS and gdp_col in df.columns:  col_list.append(gdp_col)

    df_plot = df[col_list].dropna().copy()

    # Normalize [S] absolute indicators by GDP at plot time
    if ind in IND_ABS and gdp_col in df_plot.columns:
        mask = df_plot[gdp_col] > 0
        df_plot.loc[mask, ind] = (df_plot.loc[mask, ind] / df_plot.loc[mask, gdp_col])
        label_ind = label_ind.replace("(US$)", "/ GDP")
        use_log   = False   # after normalization, log not needed
        clip      = None    # reset clip — normalized values are in [0,x]

    if clip: df_plot[ind] = df_plot[ind].clip(*clip)

    # ── BOXPLOT ───────────────────────────────────────────────────────────────
    if viz_type == "boxplot":
        fig, ax = styled_figure(figsize=(11, 7))
        ax.set_title(f"{title_}\n(boxplot — non-linear relationship)", color="white")

        x_vals = (np.log(df_plot[ind].clip(lower=1e-9)) if use_log else df_plot[ind])
        try:
            df_plot["quartile"] = pd.qcut(x_vals, 4, labels=["Q1\n(Low)","Q2","Q3","Q4\n(High)"], duplicates="drop")
        except ValueError:
            print(f"⚠️  Cannot create quartiles for {label_ind} — skipping")
            plt.close(); return

        labels_q = ["Q1\n(Low)","Q2","Q3","Q4\n(High)"]
        data_box = [df_plot[df_plot["quartile"]==q][trade_var].dropna().values for q in labels_q]

        bp = ax.boxplot(data_box, patch_artist=True, medianprops=dict(color="white", linewidth=2))
        for patch, col in zip(bp["boxes"], [C["red"],C["orange"],C["blue"],C["green"]]):
            patch.set_facecolor(col); patch.set_alpha(0.7)

        # Annotate medians
        for i, q in enumerate(labels_q):
            vals = df_plot[df_plot["quartile"]==q][trade_var].dropna()
            if not vals.empty:
                ax.text(i+1, vals.median(), f"{vals.median():.2f}", ha="center", va="bottom", color="white", 
                        fontsize=8, fontweight="bold")

        ax.set_xticklabels(labels_q, color="white")
        ax.set_ylabel(label_trade, color="white")
        ax.set_xlabel(f"{label_ind} Quartile{'  (log)' if use_log else ''}", color="white")

        # Annotate selected countries
        if countries:
            for q_idx, q in enumerate(labels_q):
                sub = df_plot[(df_plot["quartile"]==q) & (df_plot["country_name"].isin(countries))]
                for _, row in sub.iterrows():
                    ax.scatter(q_idx+1, row[trade_var], color=C["orange"], s=100, zorder=6, edgecolors="white", 
                               linewidths=1.2)
                    ax.annotate(
                        row["country_name"], (q_idx+1, row[trade_var]), fontsize=7, color="white", 
                        fontweight="bold", xytext=(10, 0), textcoords="offset points",
                        bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    # ── SCATTER ───────────────────────────────────────────────────────────────
    else:
        from scipy.stats import linregress
        fig, ax = styled_figure(figsize=(13, 8))
        caution = ("\n⚠️  Moderate non-linearity — interpret with caution" if viz_type == "scatter_caution" else "")
        ax.set_title(f"{title_}{caution}", color="white")

        x = (np.log(df_plot[ind].clip(lower=1e-9)) if use_log else df_plot[ind])

        ax.scatter(df_plot[ind] if not use_log else np.exp(x), df_plot[trade_var], color=C["blue"], alpha=0.4, s=35)

        slope, intercept, r, _, _ = linregress(x, df_plot[trade_var])
        x_line = np.linspace(x.min(), x.max(), 100)
        ax.plot(np.exp(x_line) if use_log else x_line, intercept + slope * x_line, color="white", linewidth=1.5, 
                linestyle="--", alpha=0.6, label=f"Trend (r={r:.2f})")

        if use_log: ax.set_xscale("log")
        ax.set_xlabel(label_ind, color="white")
        ax.set_ylabel(label_trade, color="white")

        r_sp, p_sp = spearmanr(x, df_plot[trade_var])
        ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", 
                fontsize=9)
        ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

        # Annotate selected countries with bbox
        if countries:
            for _, row in df_plot[df_plot["country_name"].isin(countries)].iterrows():
                ax.scatter(row[ind], row[trade_var], color=C["orange"], s=150, zorder=6, edgecolors="white", 
                           linewidths=1.5)
                ax.annotate(row["country_name"], (row[ind], row[trade_var]), fontsize=9, color="white", 
                            fontweight="bold", xytext=(8, 4), textcoords="offset points",
                            bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    plt.tight_layout()
    if fname:
        plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()

print("✅ plot_trade_vs_ind() corrected — normalizes [S] abs indicators at plot time")
````

# F. Spearman Correlation — Confirmation
</div>

**Objective** : Confirm the RF-selected indicator-trade pairs with Spearman rank correlation.

**Two complementary readings :**

| Section | Direction | Question |
|---|---|---|
| **F1** | Trade var → Indicators | For each trade variable, which indicators correlate most? |
| **F2** | Indicator → Trade vars | For each indicator, which trade variables does it affect? |


**Why Spearman after RF ?**
- RF captures non-linear importance but doesn't give direction (positive/negative)
- Spearman gives direction AND confirms the relationship is monotone
- Together : RF selects, Spearman confirms + signs
- If RF importance is high but Spearman ρ is low → likely threshold or interaction effect
  (not a monotone relationship) → boxplot required instead of scatter

**Key flags carried from E :**
- `[T]` trade-derived — tautological, visually separated
- `[S]` size-driven — captures country size
- `[E]` purely economic — actionable relationships
- `[M]` multicolinearity suspect


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## F.1. Compute Full Spearman Matrix
</div>

````python
from scipy.stats import spearmanr

MIN_OBS = 20  # lower than RF (65) — pairwise coverage

# ── Trade vars : filtered targets only ───────────────────────────────────────
trade_vars_F = [t for t in rf_results_filtered.keys() if t in df_master.columns]

# ── All 16 indicators ─────────────────────────────────────────────────────────
ind_vars_F = [i for i in ind_cols if i in df_master.columns]

# ── Compute matrix ────────────────────────────────────────────────────────────
rows = []
for tv in trade_vars_F:
    for iv in ind_vars_F:
        sub = df_master[[tv, iv]].dropna()
        if len(sub) < MIN_OBS:            continue
        r, p = spearmanr(sub[tv], sub[iv])
        rows.append({
            "trade_var"    : tv,
            "trade_label"  : ALL_TRADE_LABELS.get(tv, tv),
            "indicator"    : iv,
            "ind_label"    : IND_LABELS.get(iv, iv),
            "ind_flag"     : get_ind_flag(iv),
            "rho"          : round(r, 3),
            "pvalue"       : round(p, 4),
            "sig"          : p < 0.05,
            "abs_rho"      : abs(r),
            "n"            : len(sub),
        })

df_spearman = pd.DataFrame(rows)

print(f"✅ Spearman matrix : {df_spearman.shape}")
print(f"   Trade vars      : {df_spearman['trade_var'].nunique()}")
print(f"   Indicators      : {df_spearman['indicator'].nunique()}")
print(f"   Significant     : {df_spearman['sig'].sum()} / {len(df_spearman)}")
print(f"   |ρ| ≥ 0.40      : {(df_spearman['abs_rho'] >= 0.40).sum()}")
print(f"   |ρ| ≥ 0.20      : {(df_spearman['abs_rho'] >= 0.20).sum()}")
print(f"   n range         : {df_spearman['n'].min()} – {df_spearman['n'].max()}")
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F1.1 — Unified Heatmap : All Trade Vars × All Indicators
</div>

Heatmap shows Spearman ρ for all significant pairs (|ρ| ≥ 0.15).
Indicator rows colored by flag : 🟠[T] 🟣[S] 🟢[E].


````python
# ── Pivot : trade vars × indicators ──────────────────────────────────────────
MIN_ABS_RHO = 0.15

sig = df_spearman[(df_spearman["sig"]) & (df_spearman["abs_rho"] >= MIN_ABS_RHO)]

if sig.empty:   print(f"No significant correlations above |ρ|={MIN_ABS_RHO}")
else:
    pivot = sig.pivot_table(index="ind_label", columns="trade_label", values="rho", aggfunc="first")
    # Sort rows by mean absolute correlation
    pivot = pivot.loc[pivot.abs().mean(axis=1).sort_values(ascending=False).index]
    # keep top 20
    top20_cols = pivot.abs().mean(axis=0).nlargest(20).index
    pivot = pivot[top20_cols]
    # Row colors by flag
    ind_label_to_flag = (df_spearman[["ind_label","ind_flag"]].drop_duplicates().
                         set_index("ind_label")["ind_flag"].to_dict())
    flag_color = {"[T]": C["orange"], "[S]": C["purple"], "[E]": C["green"]}
    row_colors = [flag_color.get(ind_label_to_flag.get(i,"[E]"),"white") for i in pivot.index]

    fig, ax = plt.subplots(figsize=(max(14, len(pivot.columns)*1), max(6, len(pivot)*1)))
    fig.patch.set_facecolor(C["dark"]); 
    ax.set_facecolor("#16213e")
    sns.heatmap(pivot, ax=ax, cmap="RdYlGn", center=0, vmin=-1,vmax=1, annot=True, fmt=".2f", annot_kws={"size": 7},
                linewidths=0.4, linecolor="#333", cbar_kws={"label": "Spearman ρ"})
    ax.set_title(f"Spearman Correlation — All Trade Vars × All Indicators "
                 f"(sig |ρ|≥{MIN_ABS_RHO} | (T)=trade-derived  (S)=size-driven  (E)=economic", color="white", pad=10)
    ax.tick_params(colors="white", labelsize=7)

    for i, label in enumerate(ax.get_yticklabels()):  label.set_color(row_colors[i])

    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.savefig("../Figures/Link/F_1_1_spearman_heatmap.png", dpi=150, bbox_inches="tight")
    plt.show()

    print(f"\nTop 10 strongest correlations :")
    top10 = df_spearman[df_spearman["sig"]].nlargest(10, "abs_rho")
    display(top10[["trade_label","ind_label","ind_flag","rho","pvalue","n"]])
````

### F1.2 — Confirm RF Top Pairs

For each trade variable, compare :

**Key question** : Does the RF-selected indicator also show a significant monotonic relationship?
If RF R² is high but Spearman ρ is low → non-monotonic, likely threshold or interaction effect.


````python
print("RF Top Indicator Confirmation via Spearman")
print("=" * 70)

spearman_lookup = (df_spearman.set_index(["trade_var", "indicator"])[["rho","pvalue","sig","n"]].to_dict("index"))

for target, info in top_indicators_per_var.items():
    label      = ALL_TRADE_LABELS.get(target, target)
    indicators = info["indicators"]
    viz        = info["viz_type"]
    viz_icon   = {"boxplot":"📦","scatter_caution":"⚠️","scatter":"📈"}.get(viz,"?")

    print(f"\n  {viz_icon} {label} (RF R²={info['r2_rf']:.3f})")
    print(f"  {'─'*60}")

    for ind in indicators:
        # Look up original indicator name (undo norm replacement)
        orig_ind = {v:k for k,v in IND_USE_NORM.items()}.get(ind, ind)
        sp = spearman_lookup.get((target, orig_ind), {})
        rho = sp.get("rho", np.nan)
        p   = sp.get("pvalue", np.nan)
        n   = sp.get("n", 0)
        sig = "✅" if sp.get("sig", False) else "❌"
        flag = get_ind_flag(orig_ind)
        ind_label = IND_LABELS.get(orig_ind, orig_ind)

        # Multicolinearity check
        multicol = (target, orig_ind) in IND_MULTICOL or                    (orig_ind, target) in IND_MULTICOL
        mcol_note = " [M] ⚠️ multicolinearity suspect" if multicol else ""

        print(f"    {flag} {sig} ρ={rho:+.3f} (p={p:.3f}, n={n})  "  f"{ind_label}{mcol_note}")
````

## F.2 — Kruskal-Wallis : Categorical Trade Variables × Indicators
</div>

**Why Kruskal-Wallis ?**

Profile and quadrant variables are **ordinal categorical** — Spearman assumes continuous monotone relationships.
Kruskal-Wallis tests whether the distribution of an indicator differs significantly across profile groups,
without assuming linearity or normality.

**Interpretation :**
- H statistic + p-value : is there a significant difference between groups ?
- Significant result (p < 0.05) → at least two groups have different indicator distributions
- η² (eta-squared) : effect size — how much of the variance is explained by group membership
  - η² > 0.10 = strong effect
  - η² 0.05–0.10 = moderate effect
  - η² < 0.05 = weak effect

**Variables covered :**
All profile and quadrant variables with `_num` encoding in `df_master`.


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F.2.1 — Define Profile Variables & Run Kruskal-Wallis
</div>

````python
from scipy.stats import kruskal

# ── Profile variables (categorical ordinal) ───────────────────────────────────
PROFILE_VARS = {
"size_profile"                   : {"Leader":4,"Major":3,"Mid":2,"Small":1}, 
"balance_profile"             : {"Strong Surplus":4,"Slight Surplus":3, "Slight Deficit":2,"Structural Deficit":1},
"growth_profile"                 : {"Strong Grower":4,"Moderate Grower":3, "Weak Grower":2,"Declining":1},
"stability_profile"              : {"Stable":4,"Moderate":3, "Volatile":2,"Very Volatile":1},
"consistency_profile"            : {"Very Consistent":4,"Consistent":3, "Inconsistent":2,"Very Inconsistent":1},
"structural_profile"             : {"Strong":4,"Good":3,"Moderate":2,"Weak":1},
"static_quadrant"                : {"Dominant":4,"Balanced":3, "Vulnerable":2,"Marginal":1},
"dynamic_quadrant"               : {"Star":4,"Risky Grower":3, "Stable Stagnant":2,"Declining":1},
"strategic_quadrant"             : {"Stars":4,"Cash Cow":3, "Emerging":2,"Marginal":1},
"combined_profile"               : {"Global Champion":4,"Competitive":3, "Vulnerable":2,"Fragile":1},
"profile_full_periode"         : {"Robust":4, "Stable but Concentrated":3, "Critical Dependency":2, "Vulnerable":1},
"strategic_quadrant_combined": {"Strong & Resilient":4, "Strong but Exposed":3, "Diversified Minor":2, "Weak & Vulnerable":1},
}

# Keep only those available in df_master
profile_vars_available = {col: mapping for col, mapping in PROFILE_VARS.items() if col in df_master.columns}

print(f"✅ Profile variables available : {len(profile_vars_available)}")
for col in profile_vars_available:
    groups_found = df_master[col].dropna().unique()
    print(f"   · {col:<35} : {len(groups_found)} groups")

# ── Run Kruskal-Wallis ────────────────────────────────────────────────────────
kw_rows = []
for profile_col, mapping in profile_vars_available.items():
    groups_ordered = list(mapping.keys())  # ordered by encoding
    for ind in ind_cols:
        if ind not in df_master.columns:            continue
        # Build groups
        groups_data = []
        for grp in groups_ordered:
            vals = df_master[df_master[profile_col] == grp][ind].dropna().values
            if len(vals) >= 5: groups_data.append(vals)

        if len(groups_data) < 2: continue

        try:           stat, p = kruskal(*groups_data)
        except ValueError:            continue

        # Effect size : eta-squared approximation
        n_total = sum(len(g) for g in groups_data)
        eta2    = (stat - len(groups_data) + 1) / (n_total - len(groups_data))
        eta2    = max(0, eta2)  # clip to 0

        kw_rows.append({
            "profile_var"  : profile_col,
            "profile_label": profile_col.replace("_"," ").title(),
            "indicator"    : ind,
            "ind_label"    : IND_LABELS.get(ind, ind),
            "ind_flag"     : get_ind_flag(ind),
            "H_stat"       : round(stat, 3),
            "pvalue"       : round(p, 4),
            "sig"          : p < 0.05,
            "eta2"         : round(eta2, 3),
            "n_groups"     : len(groups_data),
        })

df_kw = pd.DataFrame(kw_rows)

print(f"\n✅ Kruskal-Wallis tests : {len(df_kw)}")
print(f"   Significant (p<0.05) : {df_kw['sig'].sum()}")
print(f"   Strong effect (η²>0.10) : {(df_kw['eta2']>0.10).sum()}")
print(f"   Moderate (η²>0.05)      : {(df_kw['eta2']>0.05).sum()}")
````

### F.2.2 — Heatmap : Profile Variables × Indicators
</div>

Heatmap shows **eta-squared (η²)** — effect size of the Kruskal-Wallis test.

Only significant results (p<0.05) shown.
Indicator rows colored by flag : 🟠[T] 🟣[S] 🟢[E].


````python
sig_kw = df_kw[df_kw["sig"]].copy()

if sig_kw.empty:  print("No significant KW results")
else: 
    pivot_kw = sig_kw.pivot_table(index="ind_label", columns="profile_label", values="eta2", aggfunc="first").fillna(0)
    # Sort rows by mean eta2
    pivot_kw = pivot_kw.loc[pivot_kw.mean(axis=1).sort_values(ascending=False).index]
    # Row colors by flag
    ind_flag_map = (df_kw[["ind_label","ind_flag"]].drop_duplicates().set_index("ind_label")["ind_flag"].to_dict())
    flag_color = {"[T]":C["orange"],"[S]":C["purple"],"[E]":C["green"]}
    row_colors = [flag_color.get(ind_flag_map.get(i,"[E]"),"white")  for i in pivot_kw.index]

    fig, ax = plt.subplots(figsize=(max(12, len(pivot_kw.columns)*1.2), max(6,  len(pivot_kw)*1)))
    fig.patch.set_facecolor(C["dark"]); ax.set_facecolor("#16213e")
    sns.heatmap(pivot_kw, ax=ax, cmap="YlOrRd", vmin=0, vmax=0.3, annot=True, fmt=".3f", annot_kws={"size":7}, 
                linewidths=0.4, linecolor="#333", cbar_kws={"label":"η² (effect size)"})
    ax.set_title("Kruskal-Wallis η² — Profile Variables × All Indicators"
                 "(significant p<0.05 | [T] [S] [E])", color="white", pad=10)
    ax.tick_params(colors="white", labelsize=7)
    for i, lbl in enumerate(ax.get_yticklabels()): lbl.set_color(row_colors[i])
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.savefig("../Figures/Link/F_2_2_kruskal_heatmap.png", dpi=150, bbox_inches="tight")
    plt.show()

    print("Top 15 strongest effects (η²) :")
    display(sig_kw[["profile_label","ind_label","ind_flag","H_stat","pvalue","eta2"]].nlargest(15,"eta2"))
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F.2.3 — Per Profile : Top Discriminating Indicators
</div>

````python
print("Top discriminating indicators per profile variable")
print("(Kruskal-Wallis significant, sorted by η²)")
print("=" * 70)

for profile_col in profile_vars_available:
    label = profile_col.replace("_"," ").title()
    sub   = df_kw[(df_kw["profile_var"]==profile_col) & (df_kw["sig"])].sort_values("eta2", ascending=False)

    if sub.empty:
        print(f"\n  {label} : no significant indicator")
        continue

    print(f"\n  {label} :")
    print(f"  {'─'*55}")
    for _, row in sub.head(5).iterrows():
        strength = ("🔴 Strong" if row["eta2"]>0.10 else "🟡 Moderate" if row["eta2"]>0.05  else "🟢 Weak")
        print(f"    {row['ind_flag']} {strength}  η²={row['eta2']:.3f}  "
              f"H={row['H_stat']:.1f}  {row['ind_label']}")
````

## F3. Indicator → Trade Variables
</div>

**Symmetric reading** : for each indicator, which trade variables does it explain?

Reveals the **transversal influence** of an indicator across the full trade structure.
An indicator that correlates with many trade variables is a **structural driver**.


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F3.1 — All Indicators
</div>

````python
# ── For each indicator : top correlated trade vars ────────────────────────────
print("Indicator → Trade Variables (|ρ| ≥ 0.20, p < 0.05)")
print("=" * 70)

for iv in ind_vars_F:
    ind_label = IND_LABELS.get(iv, iv)
    flag      = get_ind_flag(iv)
    sub = df_spearman[(df_spearman["indicator"] == iv) & (df_spearman["sig"]) & (df_spearman["abs_rho"] >= 0.20)] \
          .sort_values("abs_rho", ascending=False)

    if sub.empty:       continue

    print(f"\n  {flag} {ind_label} ({len(sub)} significant trade vars) :")
    for _, row in sub.head(5).iterrows():
        direction = "▲" if row["rho"] > 0 else "▼"
        print(f"    {direction} ρ={row['rho']:+.3f}  {row['trade_label']}")
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F3.2 — Economic Indicators Only (Transversal Influence)
</div>

````python
# ── Focus on [E] indicators : how broadly do they influence trade ? ────────────
econ_inds = [i for i in ind_vars_F if get_ind_flag(i) == "[E]"]

influence_scores = []
for iv in econ_inds:
    sub = df_spearman[(df_spearman["indicator"] == iv) & (df_spearman["sig"]) & (df_spearman["abs_rho"] >= 0.20)]
    influence_scores.append({
                                "indicator" : IND_LABELS.get(iv, iv),
                                "n_trade_vars" : len(sub),
                                "mean_abs_rho" : round(sub["abs_rho"].mean(), 3) if not sub.empty else 0,
                                "max_abs_rho"  : round(sub["abs_rho"].max(), 3) if not sub.empty else 0,
                            })

df_influence = (pd.DataFrame(influence_scores).sort_values("n_trade_vars", ascending=False))
fig, ax = styled_figure(figsize=(12, 7))
ax.set_title("Economic Indicators — Transversal Influence on Trade Variables"
             "(number of trade vars with |ρ| ≥ 0.20, p < 0.05)", color="white")

bars = ax.barh(df_influence["indicator"], df_influence["n_trade_vars"], color=C["green"], alpha=0.8)
ax.set_xlabel("Number of significantly correlated trade variables", color="white")

for bar, (_, row) in zip(bars, df_influence.iterrows()):
    ax.text(bar.get_width() + 0.1, bar.get_y() + bar.get_height()/2,
            f"n={int(row['n_trade_vars'])}  mean|ρ|={row['mean_abs_rho']:.2f}",
            va="center", color="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/F_3_2_indicator_influence.png", dpi=150, bbox_inches="tight")
plt.show()

print("\nTransversal Influence — [E] Indicators :")
df_influence
````

**Insight :**
Use this to identify which economic indicators are **structural drivers** of trade :
- High `n_trade_vars` → influences many dimensions of trade simultaneously
- High `mean_abs_rho` → strong monotonic relationship
- An indicator scoring high on both is the most analytically valuable for G/H/I/J sections

## F.4 — Synthesis : RF × Spearman × Kruskal-Wallis
</div>

**Objective** : Consolidate results from E (RF), F.1 (Spearman), F.2 (KW)
into a single reference table per trade variable.

This table guides the **manual selection** in F.4.

**Columns :**
- `RF_imp` : RF importance of top non-[T] indicator
- `SP_rho` : Spearman ρ of top non-[T] indicator (continuous vars)
- `KW_eta2` : KW η² of top non-[T] indicator (profile vars)
- `gap` : RF R² - OLS R² → viz_type
- `flag` : dominant flag ([E]/[S]/[T])
- `note` : automatic interpretation


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F.4.1 — Continuous Trade Variables (Spearman)
</div>

````python
# ── Continuous trade vars — use Spearman 
# Variables without _num suffix
continuous_vars = [t for t in rf_results_filtered.keys() if not t.endswith("_num") and t in df_master.columns]

rows_f3 = []
for target in continuous_vars:
    res   = rf_results_filtered.get(target, {})
    if not res: continue

    fi    = res["fi"].copy()
    fi["flag"] = fi["indicator"].apply(get_ind_flag)

    # Top non-[T] by RF importance
    top_rf = fi[fi["flag"] != "[T]"].iloc[0] if not fi[fi["flag"] != "[T]"].empty else None
    if top_rf is None: continue

    ind       = top_rf["indicator"]
    rf_imp    = round(top_rf["importance"], 3)
    ind_flag  = top_rf["flag"]
    ind_label = IND_LABELS.get(ind, ind)

    # Spearman ρ
    sp_row = df_spearman[(df_spearman["trade_var"]==target) & (df_spearman["indicator"]==ind)]
    sp_rho = sp_row["rho"].values[0] if not sp_row.empty else np.nan
    sp_sig = sp_row["sig"].values[0] if not sp_row.empty else False

    # Gap
    gap_row = df_gap_filtered[df_gap_filtered["target"]==ALL_TRADE_LABELS.get(target,target)]
    gap     = gap_row["gap"].values[0]      if not gap_row.empty else np.nan
    viz     = gap_row["viz_type"].values[0] if not gap_row.empty else "boxplot"
    flag_g  = gap_row["flag"].values[0]     if not gap_row.empty else "🔴"

    # Automatic note
    if ind_flag == "[T]":            note = "⚠️  Tautological — trade-derived top indicator"
    elif not sp_sig:                 note = "❌ No significant Spearman correlation"
    elif abs(sp_rho) >= 0.40:        note = f"✅ Strong monotone relationship (ρ={sp_rho:.2f})"
    elif abs(sp_rho) >= 0.20:        note = f"🟡 Moderate relationship (ρ={sp_rho:.2f})"
    else:                            note = f"🔴 Weak relationship (ρ={sp_rho:.2f})"

    rows_f3.append({
                    "trade_var"    : ALL_TRADE_LABELS.get(target, target),
                    "top_indicator": ind_label,
                    "ind_flag"     : ind_flag,
                    "RF_imp"       : rf_imp,
                    "SP_rho"       : round(sp_rho, 3) if not np.isnan(sp_rho) else np.nan,
                    "SP_sig"       : sp_sig,
                    "gap"          : round(gap, 3) if not np.isnan(gap) else np.nan,
                    "gap_flag"     : flag_g,
                    "viz_type"     : viz,
                    "note"         : note,
                   })

df_f3_continuous = pd.DataFrame(rows_f3).sort_values("SP_rho", ascending=False, key=abs)

print("Continuous Trade Variables — RF × Spearman × Gap")
print("=" * 90)
df_f3_continuous[['trade_var','ind_flag', 'RF_imp','SP_rho', 'gap', 'gap_flag', 'note']]  
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F.4.2 — Profile/Quadrant Variables (Kruskal-Wallis)
</div>

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### F.4.2.1 — Profile vars — use KW
</div>

````python
# ── Profile vars — use KW 
profile_num_vars = [t for t in rf_results_filtered.keys() if t.endswith("_num") and t in df_master.columns]

rows_f3_kw = []
for target in profile_num_vars:
    # Original profile col (remove _num)
    profile_col = target.replace("_num","")
    res = rf_results_filtered.get(target, {})
    if not res: continue

    fi = res["fi"].copy()
    fi["flag"] = fi["indicator"].apply(get_ind_flag)

    # Top non-[T] by RF importance
    top_rf = fi[fi["flag"] != "[T]"].iloc[0] if not fi[fi["flag"] != "[T]"].empty else None
    if top_rf is None: continue

    ind       = top_rf["indicator"]
    rf_imp    = round(top_rf["importance"], 3)
    ind_flag  = top_rf["flag"]
    ind_label = IND_LABELS.get(ind, ind)

    # KW eta2
    kw_row = df_kw[(df_kw["profile_var"]==profile_col) & (df_kw["indicator"]==ind)]
    kw_eta2 = kw_row["eta2"].values[0]   if not kw_row.empty else np.nan
    kw_sig  = kw_row["sig"].values[0]    if not kw_row.empty else False

    # Also check Spearman on _num
    sp_row = df_spearman[(df_spearman["trade_var"]==target) & (df_spearman["indicator"]==ind)]
    sp_rho = sp_row["rho"].values[0] if not sp_row.empty else np.nan
    sp_sig = sp_row["sig"].values[0] if not sp_row.empty else False

    # Gap
    gap_row = df_gap_filtered[df_gap_filtered["target"]==ALL_TRADE_LABELS.get(target,target)]
    gap    = gap_row["gap"].values[0]      if not gap_row.empty else np.nan
    viz    = gap_row["viz_type"].values[0] if not gap_row.empty else "boxplot"
    flag_g = gap_row["flag"].values[0]     if not gap_row.empty else "🔴"

    # Note
    if not kw_sig:               note = "❌ No significant KW result"
    elif kw_eta2 >= 0.10:        note = f"✅ Strong group discrimination (η²={kw_eta2:.3f})"
    elif kw_eta2 >= 0.05:        note = f"🟡 Moderate group discrimination (η²={kw_eta2:.3f})"
    else:                        note = f"🔴 Weak group discrimination (η²={kw_eta2:.3f})"

    if kw_sig and sp_sig:        note += " | ✅ KW+Spearman agree"
    elif kw_sig and not sp_sig:  note += " | ⚠️  Non-monotone"

    rows_f3_kw.append({
        "trade_var"    : ALL_TRADE_LABELS.get(target, target),
        "top_indicator": ind_label,
        "ind_flag"     : ind_flag,
        "RF_imp"       : rf_imp,
        "KW_eta2"      : round(kw_eta2, 3) if not np.isnan(kw_eta2) else np.nan,
        "SP_rho"       : round(sp_rho, 3)  if not np.isnan(sp_rho)  else np.nan,
        "gap"          : round(gap, 3)     if not np.isnan(gap)      else np.nan,
        "gap_flag"     : flag_g,
        "viz_type"     : viz,
        "note"         : note,
    })

df_f3_profiles = pd.DataFrame(rows_f3_kw).sort_values("KW_eta2", ascending=False)

print("Profile/Quadrant Variables — RF × KW × Spearman × Gap")
print("=" * 90)
df_f3_profiles[['trade_var','ind_flag', 'RF_imp','KW_eta2', 'SP_rho', 'gap', 'gap_flag', 'note']]
````

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### F.4.2.2 — KW vs Spearman Comparison
</div>

**Key insight** : compare Kruskal-Wallis results with Spearman for `_num` encoded variables.

- If both agree → robust relationship (monotone AND group-discriminating)
- If KW significant but Spearman weak → non-monotone group difference
  (e.g. Middle groups differ more than extremes)
- If Spearman significant but KW weak → continuous monotone but groups overlap a lot


````python
# ── Compare KW (profiles) with Spearman (_num encoded) ───────────────────────
print("KW vs Spearman comparison — profile variables")
print("=" * 70)

for profile_col in profile_vars_available:
    num_col = f"{profile_col}_num"
    if num_col not in df_master.columns: continue

    label = profile_col.replace("_"," ").title()
    print(f"\n  {label} :")
    print(f"  {'─'*55}")

    # Get KW top indicators
    kw_top = df_kw[(df_kw["profile_var"]==profile_col) & (df_kw["sig"]) & (~df_kw["ind_flag"].isin(["[T]"]))] \
            .sort_values("eta2",ascending=False).head(3)

    for _, kw_row in kw_top.iterrows():
        ind      = kw_row["indicator"]
        ind_lbl  = kw_row["ind_label"]
        eta2     = kw_row["eta2"]

        # Spearman for same pair using _num
        sub = df_master[[num_col, ind]].dropna()
        if len(sub) < 15:  sp_rho, sp_p = np.nan, np.nan
        else:              sp_rho, sp_p = spearmanr(sub[num_col], sub[ind])

        agreement = ""
        if not np.isnan(sp_rho):
            if kw_row["sig"] and sp_p < 0.05:             agreement = "✅ Both agree"
            elif kw_row["sig"] and sp_p >= 0.05:          agreement = "⚠️  KW sig, Spearman not → non-monotone"
            elif not kw_row["sig"] and sp_p < 0.05:       agreement = "⚠️  Spearman sig, KW not → continuous only"

        print(f"    {kw_row['ind_flag']} {ind_lbl:<30} " f"KW η²={eta2:.3f}  ρ={sp_rho:+.3f}  {agreement}")
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

### F.4.3 — Summary Figure
</div>

````python
# ── Profile/categorical variables to exclude from continuous bubble chart ───────
PROFILE_COLS_ORIGINAL = {
                            "size_profile","balance_profile","growth_profile",
                            "stability_profile","consistency_profile","structural_profile",
                            "static_quadrant","dynamic_quadrant","strategic_quadrant",
                            "combined_profile","strategic_quadrant_combined"
                        }

# ── Figure A : Continuous vars — RF importance × Spearman ρ ──────────────────
df_bubble = df_f3_continuous[~df_f3_continuous["trade_var"].isin([ALL_TRADE_LABELS.get(c,c) 
                                                                  for c in PROFILE_COLS_ORIGINAL])] \
                                                                  .dropna(subset=["RF_imp","SP_rho"])

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("RF Importance × Spearman ρ — Continuous Trade Variables\n"
             "(bubble size = |ρ|, color = gap severity)", color="white")

gap_colors = {"🔴": C["red"], "🟡": C["orange"], "🟢": C["green"]}

for _, row in df_bubble.iterrows():
    col  = gap_colors.get(row["gap_flag"], C["grey"])
    size = max(50, abs(row["SP_rho"]) * 500)
    ax.scatter(row["RF_imp"], row["SP_rho"], s=size, color=col, alpha=0.7, edgecolors="white", linewidths=0.8, 
               zorder=5)
    ax.annotate(row["trade_var"], (row["RF_imp"], row["SP_rho"]), fontsize=7, color="white", xytext=(6, 3), 
                textcoords="offset points", bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.6, edgecolor="none"))

ax.axhline(0,     color=C["grey"],   linestyle="--", alpha=0.4)
ax.axhline(0.40,  color=C["green"],  linestyle=":",  alpha=0.4)
ax.axhline(-0.40, color=C["green"],  linestyle=":",  alpha=0.4)
ax.axvline(0.05,  color=C["orange"], linestyle=":",  alpha=0.4)
ax.set_xlabel("RF Importance (top non-[T] indicator)", color="white")
ax.set_ylabel("Spearman ρ (top non-[T] indicator)",    color="white")

patches = [ mpatches.Patch(color=C["red"],    alpha=0.7, label=" gap>0.30 non-linear"),
            mpatches.Patch(color=C["orange"], alpha=0.7, label=" gap 0.10-0.30"),
            mpatches.Patch(color=C["green"],  alpha=0.7, label=" gap<0.10 linear")]
ax.legend(handles=patches, facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/F_3_bubble_continuous.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Figure B : Profile vars — RF importance × KW η² ──────────────────────────
df_bubble_kw = df_f3_profiles.dropna(subset=["RF_imp","KW_eta2"])

fig, ax = styled_figure(figsize=(13, 7))
ax.set_title("RF Importance × KW η² — Profile & Quadrant Variables\n"
             "(bubble size = η², color = gap severity)", color="white")

for _, row in df_bubble_kw.iterrows():
    col  = gap_colors.get(row["gap_flag"], C["grey"])
    size = max(50, row["KW_eta2"] * 1000)
    ax.scatter(row["RF_imp"], row["KW_eta2"], s=size, color=col, alpha=0.7, edgecolors="white", linewidths=0.8, 
               zorder=5)
    ax.annotate(row["trade_var"], (row["RF_imp"], row["KW_eta2"]), fontsize=7, color="white", xytext=(6, 3), 
                textcoords="offset points", bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.6, edgecolor="none"))

ax.axhline(0.10, color=C["green"],  linestyle=":", alpha=0.7, label="η²=0.10 (strong)")
ax.axhline(0.05, color=C["orange"], linestyle=":", alpha=0.7, label="η²=0.05 (moderate)")
ax.axvline(0.05, color=C["grey"],   linestyle=":", alpha=0.7, label="RF=0.05 (threshold)")
ax.set_xlabel("RF Importance (top non-[T] indicator)", color="white")
ax.set_ylabel("KW η² (effect size)",                   color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=10, loc = "upper left")

plt.tight_layout()
plt.savefig("../Figures/Link/F_3_bubble_profiles.png",  dpi=150, bbox_inches="tight")
plt.show()
````

````python
# ── Spearman bar chart — summary of all correlations for Export Share ─────────
target_label = "Export Share"
sp_sub = df_spearman[(df_spearman["trade_label"] == target_label) & (df_spearman["sig"]) &
                     (df_spearman["abs_rho"] >= 0.15)].sort_values("abs_rho", ascending=False)

flag_color = {"[T]": C["orange"], "[S]": C["purple"], "[E]": C["green"]}

fig, ax = styled_figure(figsize=(10, max(5, len(sp_sub)*0.5)))
ax.set_title(f"Export Share — Spearman ρ with All Indicators\n"
             f"(sig | |ρ|≥0.15 | [T] [S] [E])", color="white")

colors_bar = [flag_color.get(f, "white") for f in sp_sub["ind_flag"]]
bars = ax.barh(sp_sub["ind_label"], sp_sub["rho"], color=colors_bar, alpha=0.8)
ax.axvline(0,     color="white",     linewidth=1)
ax.axvline(0.40,  color=C["green"],  linestyle=":", alpha=0.4)
ax.axvline(-0.40, color=C["green"],  linestyle=":", alpha=0.4)
ax.set_xlabel("Spearman ρ", color="white")

for bar, (_, row) in zip(bars, sp_sub.iterrows()):
    offset = 0.005 if row["rho"] >= 0 else -0.005
    ha     = "left" if row["rho"] >= 0 else "right"
    ax.text(row["rho"] + offset, bar.get_y() + bar.get_height()/2, f"ρ={row['rho']:+.2f}",
            va="center", ha=ha, color="white", fontsize=9, fontweight= "bold")

for lbl, flag in zip(ax.get_yticklabels(), sp_sub["ind_flag"]):
    lbl.set_color(flag_color.get(flag, "white"))

patches = [mpatches.Patch(color=C["orange"], alpha=0.8, label="[T] Trade-derived"),
           mpatches.Patch(color=C["purple"], alpha=0.8, label="[S] Size-driven"),
           mpatches.Patch(color=C["green"],  alpha=0.8, label="[E] Economic")]
ax.legend(handles=patches, facecolor="#16213e", labelcolor="white", fontsize=8, loc="best")
plt.tight_layout()
plt.savefig("../Figures/Link/G_1_0_export_share_spearman_summary.png", dpi=150, bbox_inches="tight")
plt.show()
````

````python
# ── Data preparation ─────────────────────────────────────────────────────────
gdp_col    = "GDP (current US$)"
target     = "export_share_structural"

df_qa = df_master[["country_name", target, gdp_col, "size_profile"]].dropna()

# ── Scatter log-log Export Share vs GDP ───────────────────────────────────────
from scipy.stats import linregress

log_gdp = np.log10(df_qa[gdp_col].clip(lower=1))
r_sp, p_sp = spearmanr(log_gdp, df_qa[target])

# Regression line on log scale
slope, intercept, r_lin, _, _ = linregress(log_gdp, df_qa[target])
x_line = np.linspace(log_gdp.min(), log_gdp.max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Do larger economies dominate world exports?"
             "\n Export Share (% world) vs GDP (US$)", color="white")

# Background points colored by size profile
for sp, col in SIZE_COLORS.items():
    sub = df_qa[df_qa["size_profile"] == sp]
    ax.scatter(sub[gdp_col], sub[target], color=col, alpha=0.5, s=40, label=sp)

# Regression trend line
ax.plot(10**x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--",
        alpha=0.6, label=f"Trend (ρ={r_sp:.2f})")

# Annotate selected countries with bbox
for _, row in df_qa[df_qa["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row[target], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row[target]), fontsize=9, color="white", fontweight="bold",
                xytext=(8, 4), textcoords="offset points", 
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7,edgecolor="white",linewidth=0.5))

ax.set_xscale("log")
ax.set_xlabel("GDP (US$) — log scale", color="white")
ax.set_ylabel("Export Share (% world exports)", color="white")
ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_1_QA_export_share_vs_gdp.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Print anomalies ───────────────────────────────────────────────────────────
# Residuals from trend
df_qa = df_qa.copy()
df_qa["log_gdp"]   = np.log10(df_qa[gdp_col].clip(lower=1))
df_qa["predicted"] = intercept + slope * df_qa["log_gdp"]
df_qa["residual"]  = df_qa[target] - df_qa["predicted"]

print("Top 5 over-performers (export share >> GDP trend) :")
print(df_qa.nlargest(5,"residual")[["country_name","residual",target,gdp_col]].to_string(index=False))
print()
print("Top 5 under-performers (export share << GDP trend) :")
df_qa.nsmallest(5,"residual")[["country_name","residual",target,gdp_col]]
````

**Over-performers (above trend) :**

**Under-performers (below trend) :**

  reduces relative export orientation despite massive absolute volumes.
  → GDP driven by domestic consumption and services (finance, tech, healthcare),
  not by external trade 
- **Japan** : similar pattern to USA — large GDP but export share below trend
  → large domestic market absorbs significant production

**Explanation :** Export share reflects not just economic size but the **trade
orientation** of an economy. Countries with export-led growth models (China, Germany)
outperform their GDP rank. Countries with large domestic markets and service-dominated
economies (USA, Japan) underperform their GDP rank.

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question C — How different is GDP across the 4 size profiles?
</div>

````python
from scipy.stats import kruskal

gdp_col = "GDP (current US$)"
target  = "export_share_structural"

df_qc = df_master[["country_name", gdp_col, "size_profile"]].dropna()

groups_ordered = ["Leader", "Major", "Mid", "Small"]
group_colors   = [C["green"], C["blue"], C["orange"], C["red"]]

data_box = [df_qc[df_qc["size_profile"]==g][gdp_col].dropna().values for g in groups_ordered]

# KW test
stat, p = kruskal(*data_box)

fig, ax = styled_figure(figsize=(12, 7))
ax.set_title(f" How different is GDP across the 4 size profiles?"
             f"Kruskal-Wallis H={stat:.1f} (p={p:.4f})",  color="white")

bp = ax.boxplot(data_box, widths=width, patch_artist=True, 
                    medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)
                   )
for patch, col in zip(bp["boxes"], group_colors):
    patch.set_facecolor(col); patch.set_alpha(0.7)

# Labels with n
xlabels = [f"{g}(n={len(d)})" for g, d in zip(groups_ordered, data_box)]
ax.set_xticklabels(xlabels, color="white", fontsize=10)
ax.set_ylabel("GDP (US$)", color="white")
ax.set_yscale("log")

# Annotate medians in readable format
for i, (g, vals) in enumerate(zip(groups_ordered, data_box)):
    if len(vals) == 0: continue
    med = np.median(vals)
    if med >= 1e12:        label = f"${med/1e12:.1f}T"
    elif med >= 1e9:       label = f"${med/1e9:.0f}B"
    else:                  label = f"${med/1e6:.0f}M"
    ax.text(i+1, med, label, ha="center", va="bottom", color="white", fontsize=9, fontweight="bold")

# Annotate selected countries
for _, row in df_qc[df_qc["country_name"].isin(countries)].iterrows():
    g_idx = groups_ordered.index(row["size_profile"]) + 1
    ax.scatter(g_idx, row[gdp_col], color=SIZE_COLORS[row["size_profile"]], s=50, zorder=6, edgecolors="white", linewidths=1.2)
    ax.annotate(row["country_name"], (g_idx, row[gdp_col]), fontsize=7, color="white", fontweight="bold",
                xytext=(10, 0), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_1_QC_gdp_by_size_profile.png", dpi=150, bbox_inches="tight")
plt.show()

# Print ratio between groups
medians = [np.median(d) for d in data_box]
print("Median GDP by Size Profile :")
for g, med in zip(groups_ordered, medians): print(f"  {g:<8} : ${med/1e9:.0f}B")
print()
print(f"  Leader / Major ratio : {medians[0]/medians[1]:.1f}x")
print(f"  Leader / Mid ratio   : {medians[0]/medians[2]:.1f}x")
print(f"  Leader / Small ratio : {medians[0]/medians[3]:.1f}x")
````

**Insight — Question C :**

**Answer :** GDP differs dramatically across size profiles — statistically significant
(KW H=127.6, p<0.0001) and spanning several orders of magnitude.

**Key ratios (from figure) :**
- Leader median `$1.9T ≈ 4x` Major (\$503B)
- Leader median `$1.9T ≈ 10x` Mid (\$183B)
- Leader median `$1.9T ≈ 136x` Small (\$14B)

**Anomaly :** India and Brazil appear in the Major group despite GDP comparable
to some Leaders. Their large GDP does not translate into a large world export share
→ they are domestically oriented economies where internal consumption drives GDP
more than external trade.

**Wide variance within Small :** the group spans from micro-states (`$1B GDP`) to
moderate economies (`$100B+`) — all sharing the same characteristic of low export
share regardless of their economic size.

**Explanation :** Size profile is defined by **export share** (world trade weight),
not by GDP directly. A country can have significant economic mass but remain
classified as Small or Mid if its production is absorbed by the domestic market
rather than exported.


<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question D — Among same-GDP countries, who captures more export share?
</div>

**Approach :** Trade Efficiency Ratio = Export Share / GDP Share (world)

- Ratio > 1 → country exports **more** than its GDP weight in the world economy → over-performer
- Ratio = 1 → exports proportional to economic size → on trend  
- Ratio < 1 → country exports **less** than its GDP weight → under-performer

This avoids regression artifacts and directly measures commercial efficiency.


````python
gdp_col = "GDP (current US$)"
target  = "export_share_structural"

df_qd = df_master[["country_name", target, gdp_col, "size_profile"]].dropna().copy()

# ── Trade Efficiency Ratio = Export Share / GDP Share ────────────────────────
world_gdp = df_qd[gdp_col].sum()
df_qd["gdp_share"]   = df_qd[gdp_col] / world_gdp * 100  # % of world GDP
df_qd["trade_eff"]   = df_qd[target] / df_qd["gdp_share"].replace(0, np.nan)

# Remove extreme outliers (micro-states with near-zero GDP)
df_qd = df_qd[df_qd[gdp_col] > 1e9]  # GDP > $1B

# ── Figure : Trade Efficiency Ratio vs GDP ────────────────────────────────────
fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Among same-GDP countries, who captures more export share?"
             "\n Trade Efficiency Ratio = Export Share / GDP Share  (1 = proportional)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_qd[df_qd["size_profile"] == sp]
    ax.scatter(sub[gdp_col], sub["trade_eff"], color=col, alpha=0.5, s=40, label=sp)

# Reference line ratio = 1
ax.axhline(1, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Ratio = 1 (proportional)")

# Color zones
ymax = df_qd["trade_eff"].quantile(0.99)
ax.axhspan(1, ymax * 1.05, alpha=0.05, color=C["green"])
ax.axhspan(0, 1,           alpha=0.05, color=C["red"])

ax.text(df_qd[gdp_col].min()*1.5, ymax*0.8, "Over-performers", color=C["green"], fontsize=9, bbox= dict(facecolor="white"))
ax.text(df_qd[gdp_col].min()*1.5, 0.2, "Under-performers", color=C["red"], fontsize=9, bbox = dict(facecolor="white"))

# Annotate selected countries
for _, row in df_qd[df_qd["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row["trade_eff"], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row["trade_eff"]), fontsize=9, color="white", fontweight="bold",
        xytext=(8, 4), textcoords="offset points",
        bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xscale("log")
ax.set_xlabel("GDP (US$) — log scale", color="white")
ax.set_ylabel("Trade Efficiency Ratio (Export Share / GDP Share)", color="white")
ax.set_ylim(0, df_qd["trade_eff"].quantile(0.99) * 1)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_1_QD_trade_efficiency.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Top over and under-performers ─────────────────────────────────────────────
print("Top 10 over-performers (Trade Efficiency Ratio >> 1) :")
print(df_qd.nlargest(10, "trade_eff")[
    ["country_name","trade_eff","export_share_structural","size_profile"]].round(2).to_string(index=False))
print(50*"==")
print("Top 10 under-performers (Trade Efficiency Ratio << 1) :")
print(df_qd[df_qd[gdp_col] > 1e11].nsmallest(10, "trade_eff")[["country_name","trade_eff",
                            "export_share_structural","size_profile"]].round(3).to_string(index=False))
print(50*"==")
print("Selected countries Trade Efficiency :")
print(df_qd[df_qd["country_name"].isin(countries)][["country_name","trade_eff","export_share_structural",
                                "size_profile"]].sort_values("trade_eff", ascending=False).round(2))
````

**Insight — Question D :**

**Answer :** Trade Efficiency Ratio = Export Share / GDP Share directly measures
how well a country converts its economic mass into world export share.
Ratio > 1 = over-performer, Ratio < 1 = under-performer.

**Top over-performers :**
- **Germany** is the most efficient large exporter — Trade Efficiency Ratio ≈ 2.6 :
  it captures 2.6× more export share than its GDP share would predict.
  This reflects its manufacturing specialization and export-oriented industrial model.
  Highly specialized industrial model, high value-added exports, global supply chain integration.
- **Canada, France** : above 1 → export more than their economic weight

**Top under-performers :**
- **USA** has a Trade Efficiency Ratio < 1 (~0.32) — it captures significantly **less**
  export share than its GDP share predicts. The world's largest economy is also
  one of its least export-oriented — its massive domestic market absorbs most production.
  This is the clearest illustration that GDP size and export power are not equivalent.
- **China (0.75)** : surprisingly below 1 despite being the world's largest exporter
  in absolute terms. China's GDP (~17% of world) is growing faster than its exports (12.5%) in recent years, reflecting a gradual shift toward domestic consumption-led growth.
  The economy is driven by domestic investment (infrastructure, real estate) and
  growing internal consumption. China has been rebalancing away from export-led growth
  since 2010 — exports as % of GDP have been declining steadily.
- **India, Brazil (≈0.45)** : large GDP but export share well below economic weight
  → domestically oriented, commodity-dependent

**Key finding :** GDP determines the ceiling of potential export share.
Trade orientation determines how much of that ceiling is reached.
Germany is the most commercially efficient large economy.
China, despite dominating absolute export volumes, is an under-performer relative
to its economic size — a structural shift toward domestic consumption is underway.

### G.1.2 — Export per Capita × GDP per Capita
</div>

Same 4 questions as Part 1 (Export Share × GDP) but normalized by population.

**Key shift :** from absolute economic mass to per-person efficiency.
Divergences between Part 1 and Part 2 reveal structural differences
in how countries convert wealth into trade.


<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question A — Do richer countries export more per person?
</div>

````python
from scipy.stats import linregress

gdp_pc_col = "GDP per capita (current US$)"
target_pc  = "export_pc"

df_p2 = df_master[["country_name", target_pc, gdp_pc_col, "size_profile"]].dropna()

log_gdp_pc = np.log10(df_p2[gdp_pc_col].clip(lower=1))
log_exp_pc = np.log10(df_p2[target_pc].clip(lower=1))
r_sp, p_sp = spearmanr(log_gdp_pc, log_exp_pc)
slope, intercept, r_lin, _, _ = linregress(log_gdp_pc, log_exp_pc)
x_line = np.linspace(log_gdp_pc.min(), log_gdp_pc.max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Do richer countries export more per person?"
             "\n Export per Capita vs GDP per Capita (log-log)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_p2[df_p2["size_profile"] == sp]
    ax.scatter(sub[gdp_pc_col], sub[target_pc], color=col, alpha=0.5, s=40, label=sp)

ax.plot(10**x_line, 10**(intercept + slope * x_line),
        color="white", linewidth=1.5, linestyle="--", alpha=0.6, label=f"Trend (ρ={r_sp:.2f})")

for _, row in df_p2[df_p2["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_pc_col], row[target_pc], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", 
               linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_pc_col], row[target_pc]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8, 4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("GDP per Capita (US$) — log scale", color="white")
ax.set_ylabel("Export per Capita (US$) — log scale", color="white")
ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G_1_2_QA_export_pc_vs_gdp_pc.png", dpi=150, bbox_inches="tight")
plt.show()

print(f"Spearman ρ Part 1 (Export Share vs GDP)   : 0.94")
print(f"Spearman ρ Part 2 (Export_pc vs GDP_pc)   : {r_sp:.2f}")
print(f"Log-log slope : {slope:.2f} → 1% increase in GDP_pc → {slope:.2f}% increase in Export_pc")
````

**Insight — Question A :**

**Answer :** Yes — richer countries export significantly more per person (ρ=0.82).
The log-log slope of 1.22 means a 1% increase in GDP_pc leads to a 1.22% increase
in Export_pc — exports grow *faster* than income per person.

**Over-performers (above trend) :**
- **Germany** : exports far more per person than its GDP_pc predicts
  → industrial export model generates high value per person
- **Canada, France** : above trend → open economies with high export intensity

**Under-performers (below trend) :**
- **USA** : well below trend despite high GDP_pc → same pattern as Part 1.
  Service economy, large domestic market. GDP_pc is high but export_pc is modest.
- **Japan** : below trend → domestic consumption absorbs production
- **China, India, Brazil** : all below trend → low-to-middle GDP_pc
  AND export_pc below what their income level predicts

**Comparison with G.1.2 (Export Share vs GDP) :**
- ρ=0.82 here vs ρ=0.94 in G.1.2 → The relationship is slightly weaker per capita
- Absolute economic size explains export dominance better
  than individual wealth — a country can be rich per person but small in
  world exports (e.g. small wealthy nations)

**Confirmations from G.1.2 :**
- **Germany** : over-performer in both parts → industrial export model
  efficient at both the aggregate and per-person level
- **USA** : under-performer in both parts → service economy pattern
  is consistent whether measured by size or by wealth
- **Japan** : under-performer in both parts → large domestic market
  confirmed at both scales

**Contradiction vs G.1.2 :**
- **China** : over-performer in Part 1 (large absolute export share)

  not individual. 1.4B people dilute the per-capita figure despite
  massive aggregate exports. High GDP_pc countries export more per person
  but China's model is volume-based, not value-based per person.

### G.1.3 — Export Share × External Debt
</div>

**Indicator :** External Debt stocks, total (DOD, current US\$)

**Note :** External Debt (US\$) correlates strongly with GDP (ρ≈0.95 between the two).
Questions B and C would replicate GDP results → not repeated here.
Analysis uses the **normalized version (External Debt / GDP)** to isolate
the debt-specific signal from the pure size effect.

Questions A and D only.


<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question A — Do countries with more external debt (relative to GDP) export more?
</div>

````python
from scipy.stats import linregress

gdp_col  = "GDP (current US$)"
debt_col = "External debt stocks, total (DOD, current US$)"
target   = "export_share_structural"
df_debt = df_master[["country_name", target, debt_col, gdp_col, "size_profile"]].dropna().copy()

# Normalize by GDP
mask = df_debt[gdp_col] > 0
df_debt.loc[mask, "debt_gdp"] = (df_debt.loc[mask, debt_col] / df_debt.loc[mask, gdp_col])
df_debt = df_debt.dropna(subset=["debt_gdp"])

r_raw,  _ = spearmanr(np.log(df_debt[debt_col].clip(lower=1)), df_debt[target])
r_norm, p_norm = spearmanr(df_debt["debt_gdp"], df_debt[target])

print(f"Spearman ρ — External Debt (US$) vs Export Share      : {r_raw:.3f}")
print(f"Spearman ρ — External Debt / GDP vs Export Share      : {r_norm:.3f} (p={p_norm:.4f})")
print(f"→ After normalization by GDP, ρ drops from {r_raw:.2f} to {r_norm:.2f}")

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Do countries with more external debt/GDP export more?"
             "\n Export Share vs External Debt / GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_debt[df_debt["size_profile"]==sp]
    ax.scatter(sub["debt_gdp"], sub[target], color=col, alpha=0.5, s=40, label=sp)

# Trend line only if significant (p < 0.05)
if p_norm < 0.05:
    slope, intercept, _, _, _ = linregress(df_debt["debt_gdp"], df_debt[target])
    x_line = np.linspace(df_debt["debt_gdp"].min(), df_debt["debt_gdp"].max(), 100)
    ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.2, linestyle="--", alpha=0.5, label=f"Trend (ρ={r_norm:.2f})")
else:
    ax.text(0.98, 0.04, f"No significant relationship (ρ={r_norm:.2f}, p={p_norm:.3f})",
            transform=ax.transAxes, ha="right", va="bottom", fontsize=8, color=C["orange"],
            bbox=dict(boxstyle="round,pad=0.3", facecolor=C["dark"], alpha=0.8, edgecolor=C["orange"]))

for _, row in df_debt[df_debt["country_name"].isin(countries)].iterrows():
    ax.scatter(row["debt_gdp"], row[target], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row["debt_gdp"], row[target]), fontsize=9, color="white", fontweight="bold",
                xytext=(8, 4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("External Debt / GDP", color="white")
ax.set_ylabel("Export Share (% world)", color="white")
ax.set_xlim (-0.01, 2)
ax.text(0.5, 0.95, f"Spearman ρ={r_norm:.2f} (p={p_norm:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G_1_3_ExtDebt_QA.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question A :**

**Answer :** No significant relationship — ρ=−0.14 (p=0.134).
Countries with higher external debt relative to GDP do not systematically
export more or less. The relationship is indistinguishable from noise.

**Why :** External Debt / GDP measures financial leverage, not productive
capacity. Debt can finance consumption or investment equally — only
productive investment translates into export capacity, and this distinction
is not captured by the aggregate debt stock.

**Note :** The trend line has been removed from this figure — displaying
a regression line on a non-significant result would be misleading.

**Data coverage note :**
External Debt (DOD) is a World Bank indicator primarily collected
for developing and emerging economies — USA, Germany, France, Japan
have no data for this indicator in the World Bank dataset.
The analysis covers only countries with available data, which
introduces a selection bias toward emerging economies.
This partially explains the non-significant result — the sample
excludes the world's largest exporters.

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question D — Among same-debt/GDP countries, who captures more export share?
</div>

````python
# Trade Efficiency vs Debt = Export Share / (Debt Share of world)
world_debt = df_debt[debt_col].sum()
df_debt["debt_share"] = df_debt[debt_col] / world_debt * 100
df_debt["debt_eff"]   = df_debt[target] / df_debt["debt_share"].replace(0, np.nan)
df_debt = df_debt[df_debt[debt_col] > 1e8]  # remove micro-states

r_eff, p_eff = spearmanr(df_debt["debt_gdp"], df_debt["debt_eff"])

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Among same-debt/GDP countries, who exports more?"
             "\n Export Efficiency vs External Debt / GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_debt[df_debt["size_profile"]==sp]
    ax.scatter(sub["debt_gdp"], sub["debt_eff"], color=col, alpha=0.5, s=40, label=sp)

# Reference line ratio=1
ax.axhline(1, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Ratio = 1 (proportional)")

ymax = df_debt["debt_eff"].quantile(0.99)
ax.axhspan(1, ymax*1.05, alpha=0.05, color=C["green"])
ax.axhspan(0, 1,         alpha=0.05, color=C["red"])

for _, row in df_debt[df_debt["country_name"].isin(countries)].iterrows():
    ax.scatter(row["debt_gdp"], row["debt_eff"], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row["debt_gdp"], row["debt_eff"]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8, 4), textcoords="offset points",
                bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("External Debt / GDP", color="white")
ax.set_ylabel("Export Share / Debt Share (efficiency ratio)", color="white")
ax.set_ylim(0, 1.5)
ax.set_xlim (0, 1.5)
ax.text(0.5, 0.95, f"Spearman ρ={r_eff:.2f} (p={p_eff:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G_1_3_ExtDebt_QD.png", dpi=150, bbox_inches="tight")
plt.show()

print("Selected countries — Export/Debt efficiency :")
print(df_debt[df_debt["country_name"].isin(countries)][
    ["country_name","debt_gdp","debt_eff","size_profile"]
].sort_values("debt_eff", ascending=False).round(3).to_string(index=False))
````

**Insight — Question D :**

**Answer :** Among same-debt/GDP countries, the relationship between
debt leverage and export efficiency is not significant — confirming
that external debt structure does not discriminate between efficient
and inefficient exporters.

**G.1.3 overall conclusion :**
External debt (both absolute and normalized) carries no meaningful
signal for export performance. It is the weakest indicator in this
section — neither significant on share (QA) nor on efficiency (QD).
This is an important negative result: financial leverage is not
a prerequisite for or an obstacle to export success.

**Implication :** External Debt is confirmed as a [S] size-proxy
in absolute terms and carries no [E] signal in normalized form.
It is excluded from downstream analyses (H, I, J).

#### Question A — Does attracting more FDI translate into higher export share?
</div>

````python
from scipy.stats import linregress

gdp_col = "GDP (current US$)"
fdi_col = "Foreign direct investment, net inflows (BoP, current US$)"
target  = "export_share_structural"

df_fdi = df_master[["country_name", target, fdi_col, gdp_col, "size_profile"]].dropna().copy()

# Normalize by GDP
mask = df_fdi[gdp_col] > 0
df_fdi.loc[mask, "fdi_gdp"] = (df_fdi.loc[mask, fdi_col] / df_fdi.loc[mask, gdp_col])
df_fdi = df_fdi.dropna(subset=["fdi_gdp"])
# Clip extreme outliers
df_fdi["fdi_gdp"] = df_fdi["fdi_gdp"].clip(-0.5, 1.0)

r_raw,  _ = spearmanr(np.log(df_fdi[fdi_col].clip(lower=1)), df_fdi[target])
r_norm, p_norm = spearmanr(df_fdi["fdi_gdp"], df_fdi[target])

print(f"Spearman ρ — FDI (US$) vs Export Share      : {r_raw:.3f}")
print(f"Spearman ρ — FDI / GDP vs Export Share      : {r_norm:.3f} (p={p_norm:.4f})")
print(f"→ After normalization by GDP, ρ drops from {r_raw:.2f} to {r_norm:.2f}")

slope, intercept, r_lin, _, _ = linregress(df_fdi["fdi_gdp"], df_fdi[target])
x_line = np.linspace(df_fdi["fdi_gdp"].min(), df_fdi["fdi_gdp"].max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Does attracting more FDI/GDP translate into higher export share?"
             "\n Export Share vs FDI / GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_fdi[df_fdi["size_profile"]==sp]
    ax.scatter(sub["fdi_gdp"], sub[target], color=col, alpha=0.5, s=40, label=sp)

ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--",
        alpha=0.6, label=f"Trend (ρ={r_norm:.2f})")
ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_fdi[df_fdi["country_name"].isin(countries)].iterrows():
    ax.scatter(row["fdi_gdp"], row[target], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", 
               linewidths=1.5)
    ax.annotate(row["country_name"], (row["fdi_gdp"], row[target]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("FDI Inflows / GDP", color="white")
ax.set_xlim(-0.02, 0.08)
ax.set
ax.set_ylabel("Export Share (% world)", color="white")
ax.text(0.05, 0.95, f"Spearman ρ={r_norm:.2f} (p={p_norm:.3f})", transform=ax.transAxes, color="white", 
        fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G_1_4_FDI_QA.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Key finding :** ρ drops from 0.730 to -0.164 after normalizing by GDP.
Same pattern as External Debt — the original correlation was a **size effect**.
Large countries attract more absolute FDI AND export more, but the two are
not causally linked once size is removed.

**Answer :** No — countries with higher FDI/GDP do not export more (ρ=-0.16).
The slight negative trend suggests that very high FDI/GDP countries
(often small financial centers or tax havens) tend to have low export share.

**Implication :** FDI attraction does not translate into export capacity
at the country level when measured relative to economic size.


<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question D — Among same FDI/GDP countries, who converts FDI into exports most efficiently?
</div>

````python
# Export Efficiency = Export Share / FDI Share (world)
world_fdi = df_fdi[fdi_col].clip(lower=0).sum()
df_fdi["fdi_share"] = df_fdi[fdi_col].clip(lower=0) / world_fdi * 100
df_fdi["fdi_eff"]   = df_fdi[target] / df_fdi["fdi_share"].replace(0, np.nan)
df_fdi = df_fdi[df_fdi[fdi_col] > 0]  # keep only positive FDI

r_eff, p_eff = spearmanr(df_fdi["fdi_gdp"], df_fdi["fdi_eff"])

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Among same FDI/GDP countries, who converts FDI into exports?"
             "\n Export Efficiency = Export Share / FDI Share  vs  FDI / GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_fdi[df_fdi["size_profile"]==sp]
    ax.scatter(sub["fdi_gdp"], sub["fdi_eff"], color=col, alpha=0.5, s=40, label=sp)

ax.axhline(1, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Ratio = 1 (proportional)")

ymax = df_fdi["fdi_eff"].quantile(0.99)
ax.axhspan(1, ymax*1.05, alpha=0.05, color=C["green"])
ax.axhspan(0, 1,         alpha=0.05, color=C["red"])

for _, row in df_fdi[df_fdi["country_name"].isin(countries)].iterrows():
    ax.scatter(row["fdi_gdp"], row["fdi_eff"], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", 
               linewidths=1.5)
    ax.annotate(row["country_name"], (row["fdi_gdp"], row["fdi_eff"]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("FDI Inflows / GDP", color="white")
ax.set_ylabel("Export Share / FDI Share (efficiency ratio)", color="white")
ax.set_xlim(-0.01, 0.15)
ax.set_ylim(-0.1, 6)
ax.text(0.05, 0.95, f"Spearman ρ={r_eff:.2f} (p={p_eff:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_4_FDI_QD.png", dpi=150, bbox_inches="tight")
plt.show()

print("Selected countries — FDI conversion efficiency :")
df_fdi[df_fdi["country_name"].isin(countries)][["country_name","fdi_gdp","fdi_eff","size_profile"]] \
      .sort_values("fdi_eff", ascending=False).round(3)
````

**Insight — Question D :**

**Answer :** Strong negative relationship (ρ=-0.66) — countries with
**lower** FDI/GDP convert their FDI into exports more efficiently.
All selected countries have very low FDI/GDP (0.006–0.036).

**Ranking of FDI-to-export conversion efficiency :**
- Germany (2.79) → most efficient : low FDI/GDP, high export share
- Japan (1.83), France (1.75), Canada (1.52) → above ratio=1
- China (1.23) → slightly above median → FDI partially converted
- India (0.60), USA (0.35), Brazil (0.27) → below ratio=1

**Key observation :** USA attracts significant FDI (0.018 FDI/GDP)
but converts it poorly into exports (0.35) — FDI flows into the
domestic service economy rather than export-oriented sectors.

**Contrast with External Debt QD :** Both show negative efficiency
vs leverage ratio — high debt/GDP and high FDI/GDP are both associated
with lower export efficiency. Countries that rely more on external
financing (debt or equity) tend to export less relative to that financing.

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question A — Do surplus countries export more?

````python
from scipy.stats import linregress

ca_col = "Current account balance (% of GDP)"
target = "export_share_structural"

df_ca = df_master[["country_name", target, ca_col, "size_profile"]].dropna().copy()
df_ca[ca_col] = df_ca[ca_col].clip(-30, 30)

r_sp, p_sp = spearmanr(df_ca[ca_col], df_ca[target])
slope, intercept, r_lin, _, _ = linregress(df_ca[ca_col], df_ca[target])
x_line = np.linspace(df_ca[ca_col].min(), df_ca[ca_col].max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Do surplus countries export more?" "\n Export Share vs Current Account % GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_ca[df_ca["size_profile"]==sp]
    ax.scatter(sub[ca_col], sub[target], color=col, alpha=0.5, s=40, label=sp)

ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--",
        alpha=0.6, label=f"Trend (ρ={r_sp:.2f})")
ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_ca[df_ca["country_name"].isin(countries)].iterrows():
    ax.scatter(row[ca_col], row[target], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"],
                (row[ca_col], row[target]), fontsize=9, color="white", fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("Current Account % GDP", color="white")
ax.set_ylabel("Export Share (% world)", color="white")
ax.set_xlim (-15, 15)

ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G_1_5_CA_QA.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question A :**

**Answer :** Yes — surplus countries export more (ρ=0.48, p<0.001).
The relationship is moderate but significant and causal — a current account
surplus means exports exceed imports by definition, so the link is structural.

**Observation :** Most countries cluster near zero on both axes —
the relationship is driven by a small number of large exporters.
China and Germany are both surplus countries AND top exporters →
surplus and export dominance reinforce each other.

**Anomaly :** USA and France have current account **deficits** but remain
top exporters → large GDP overrides the CA signal. Being in deficit does
not prevent large export volumes when the economy is big enough.

**Key distinction from GDP analysis :** GDP (ρ=0.94) explains export share
much better than CA% GDP (ρ=0.48). CA is a genuine economic signal but
secondary to absolute size.

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question B — Is there a CA surplus threshold?
</div>

````python
df_ca["ca_q"] = pd.qcut(df_ca[ca_col], 4, labels=["Q1(Deficit)","Q2","Q3","Q4(Surplus)"])
Q_COLORS = {"Q1(Deficit)":C["red"],"Q2":C["orange"],"Q3":C["blue"],"Q4(Surplus)":C["green"]}

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Is there a CA surplus threshold?"
             "\n Export Share vs Current Account % GDP — colored by CA Quartile", color="white")

for q, col in Q_COLORS.items():
    sub = df_ca[df_ca["ca_q"]==q]
    ax.scatter(sub[ca_col], sub[target], color=col, alpha=0.6, s=50, label=q.replace(""," "))

ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.5)
ax.axvline(0, color="white", linestyle=":", alpha=0.4, linewidth=1)

# Quartile boundaries
q_bounds = df_ca[ca_col].quantile([0.25, 0.50, 0.75])
for q_val in q_bounds:
    ax.axvline(q_val, color=C["grey"], linestyle=":", alpha=0.5, linewidth=2)
    ax.text(q_val, df_ca[target].max()*0.9, f"{q_val:.1f}%", color=C["orange"], fontsize=9, fontweight = "bold", ha="center")

for _, row in df_ca[df_ca["country_name"].isin(countries)].iterrows():
    ax.scatter(row[ca_col], row[target], color=Q_COLORS[row["ca_q"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[ca_col], row[target]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("Current Account % GDP", color="white")
ax.set_ylabel("Export Share (% world)", color="white")
ax.set_xlim (-15, 15)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8, title="CA Quartile", title_fontsize=8)

print("Export Share by CA Quartile :")
for q in ["Q1(Deficit)","Q2","Q3","Q4(Surplus)"]:
    vals = df_ca[df_ca["ca_q"]==q][target]
    print(f"  {q.replace(chr(10),' '):<20} median={vals.median():.3f}  "
          f"mean={vals.mean():.3f}  n={len(vals)}")

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_5_CA_QB.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question B :**

**Answer :** No hard threshold — the relationship is progressive across
quartiles with no sudden jump.

**Medians by quartile :**
- Q1 (Deficit <-5.2%) : median = 0.004 → near zero
- Q2 (-5.2% to -2.2%) : median = 0.058 → modest
- Q3 (-2.2% to +1.5%) : median = 0.137 → moderate
- Q4 (Surplus >+1.5%) : median = 0.253 → highest

**Key observation :** Q4 mean (1.264) is much higher than median (0.253)
→ a few very large surplus exporters (China, Germany, Japan) pull the mean up.
Most surplus countries still export little.

**Contrast with GDP threshold (G.1 QB) :** GDP created a hard threshold
at \$259B — below that, exports were essentially zero. CA creates no such
threshold — even Q1 deficit countries can be significant exporters (USA, France).
CA position is a tendency, not a prerequisite.

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question D — At same CA level, who exports more than their position predicts?
</div>

````python
# Residuals from CA → Export Share trend
df_ca = df_ca.copy()
df_ca["predicted"] = intercept + slope * df_ca[ca_col]
df_ca["residual"]  = df_ca[target] - df_ca["predicted"]

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("At same CA level, who exports more than predicted?"
             "\n Residuals from CA trend (above = over-performer, below = under-performer)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_ca[df_ca["size_profile"]==sp]
    ax.scatter(sub[ca_col], sub["residual"], color=col, alpha=0.5, s=40, label=sp)

ax.axhline(0, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Expected (on trend)")
ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

ymax = df_ca["residual"].max()
ymin = df_ca["residual"].min()
ax.axhspan(0,    ymax*1.05, alpha=0.05, color=C["green"])
ax.axhspan(ymin*1.05, 0,   alpha=0.05, color=C["red"])

for _, row in df_ca[df_ca["country_name"].isin(countries)].iterrows():
    ax.scatter(row[ca_col], row["residual"], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", 
               linewidths=1.5)
    ax.annotate(row["country_name"], (row[ca_col], row["residual"]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("Current Account % GDP", color="white")
ax.set_ylabel("Residual (actual − predicted export share)", color="white")
ax.set_xlim (-15, 15)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_5_CA_QD.png", dpi=150, bbox_inches="tight")
plt.show()

print("Top 5 over-performers (export share >> CA prediction) :")
print(df_ca.nlargest(5,"residual")[["country_name","residual",target,ca_col,"size_profile"]
].round(3).to_string(index=False))
print(50*"==")

print("Top 5 under-performers :")
print(df_ca.nsmallest(5,"residual")[["country_name","residual",target,ca_col,"size_profile"]
].round(3).to_string(index=False))
print(50*"==")

print("Selected countries :")
print(df_ca[df_ca["country_name"].isin(countries)][["country_name","residual",target,ca_col,"size_profile"]
].sort_values("residual",ascending=False).round(3).to_string(index=False))
````

**Insight — Question D :**

**Answer :** All selected countries are over-performers — they export more
than their CA position would predict. This confirms that **GDP size dominates
over CA position** when explaining export share.

**Top over-performers :**
- **China (residual=11.77)** : small surplus (+1.67%) but massive export share
  → CA position barely explains China's dominance — it's driven by size and
  industrial model, not by surplus level
- **Germany (11.39)** : large surplus (+7.40%) AND large export share
  → both CA and size reinforce each other
- **USA (6.47)** : deficit country (-2.53%) but still 3rd largest over-performer
  → pure size effect overwhelms the CA signal completely

**Under-performers pattern :**
All top 5 under-performers have very high CA surpluses (13–29% GDP) but near-zero
export share — these are small financial centers (Macao, Bermuda, Andorra, Djibouti,
Kiribati) where the surplus comes from services/finance, not from goods exports.

**Key finding :** CA surplus is a tendency toward export performance, not a
guarantee. A country can have a large surplus with zero goods exports (financial
centers) or a deficit with large goods exports (USA, France). GDP size remains
the dominant factor — CA position is a secondary structural signal

<div style="background-color:#fef9e7;padding-left:20px;border:1px solid;border-radius:5px;">

#### Question B — Is there a rate threshold above which export performance deteriorates?
</div>

````python
df_rate["rate_q"] = pd.qcut(df_rate[rate_col], 4, labels=["Q1(Lowest)","Q2","Q3","Q4(Highest)"])
Q_COLORS = {"Q1(Lowest)":C["green"],"Q2":C["blue"], "Q3":C["orange"],"Q4(Highest)":C["red"]}

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Is there a rate threshold above which exports deteriorate?"
             "\n Export Share vs Real Interest Rate — colored by Rate Quartile", color="white")

for q, col in Q_COLORS.items():
    sub = df_rate[df_rate["rate_q"]==q]
    ax.scatter(sub[rate_col], sub[target], color=col, alpha=0.6, s=50, label=q.replace(""," "))

ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.5)
ax.axvline(0, color="white", linestyle=":", alpha=0.4, linewidth=1)

# Quartile boundaries
q_bounds = df_rate[rate_col].quantile([0.25, 0.50, 0.75])
for q_val in q_bounds:
    ax.axvline(q_val, color=C["grey"], linestyle=":", alpha=0.3, linewidth=2)
    ax.text(q_val, df_rate[target].max()*0.95, f"{q_val:.1f}%", color=C["orange"], fontsize=9, fontweight = "bold", ha="center")

for _, row in df_rate[df_rate["country_name"].isin(countries)].iterrows():
    ax.scatter(row[rate_col], row[target], color=Q_COLORS[row["rate_q"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[rate_col], row[target]), fontsize=9, color="white", fontweight="bold",
                xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("Real Interest Rate (%)", color="white")
ax.set_ylabel("Export Share (% world)", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8, title="Rate Quartile", title_fontsize=8)

print("Export Share by Rate Quartile :")
for q in ["Q1(Lowest)","Q2","Q3","Q4(Highest)"]:
    vals = df_rate[df_rate["rate_q"]==q][target]
    print(f"  {q.replace(chr(10),' '):<20} median={vals.median():.3f}  " f"mean={vals.mean():.3f}  n={len(vals)}")

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_6_Rate_QB.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question B :**

**Answer :** Yes — clear threshold around **3–5% real rate**.
Q1 (rate <2.9%) : median export share = 0.253 → 16x higher than Q4 (0.016).
Above ~5%, export share drops sharply and stays near zero.

**Key observation :** Q1 mean (0.968) >> Q1 median (0.253) → a few very
large exporters (China, USA, Germany) dominate Q1. Most low-rate countries
still export little — low rate is necessary but not sufficient.

**Contrast with CA threshold (QB) :** CA showed no hard threshold.
Real rate shows a clear one at ~3–5% → financial conditions matter more
as a threshold than trade balance position.

#### Question D — At same rate level, who exports more efficiently? v2
</div>

````python
# QD : boxplot Export Share by Rate Quartile — who exports most at each rate level?
from scipy.stats import kruskal

rate_col = "Real interest rate (%)"
target   = "export_share_structural"

df_qd = df_master[["country_name", target, rate_col, "size_profile"]].dropna().copy()
df_qd[rate_col] = df_qd[rate_col].clip(-20, 45)
df_qd["rate_q"] = pd.qcut(df_qd[rate_col], 4, labels=["Q1\n(Lowest)","Q2","Q3","Q4\n(Highest)"])

labels_q = ["Q1\n(Lowest)","Q2","Q3","Q4\n(Highest)"]
data_box  = [df_qd[df_qd["rate_q"]==q][target].dropna().values for q in labels_q]
stat, p   = kruskal(*data_box)

fig, ax = styled_figure(figsize=(12, 7))
ax.set_title(f" At same rate level, who exports most?\n"
             f"\n Export Share by Real Interest Rate Quartile  (KW H={stat:.1f}, p={p:.4f})", color="white")

#bp = ax.boxplot(data_box, patch_artist=True, medianprops=dict(color="white", linewidth=2))
bp = ax.boxplot(data_box, widths=width, patch_artist=True, 
                    medianprops  = dict(color="orange", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)
                   )

for patch, col in zip(bp["boxes"], [C["green"],C["blue"],C["orange"],C["red"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)

xlabels = [f"{q.split(chr(10))[0]}\n{q.split(chr(10))[1] if chr(10) in q else ''}\n(n={len(d)})"
           for q, d in zip(labels_q, data_box)]
ax.set_xticklabels(xlabels, color="white", fontsize=9)
ax.set_ylabel("Export Share (% world)", color="white")

for i, vals in enumerate(data_box):
    if len(vals) == 0: continue
    ax.text(i+1, np.median(vals), f"{np.median(vals):.3f}", ha="center", va="bottom", color="white", 
            fontsize=8, fontweight="bold")

# Annotate selected countries
for q_idx, q in enumerate(labels_q):
    sub = df_qd[(df_qd["rate_q"]==q) & (df_qd["country_name"].isin(countries))]
    for _, row in sub.iterrows():
        ax.scatter(q_idx+1, row[target], color=C["orange"], s=100, zorder=6,
                   edgecolors="white", linewidths=1.2)
        ax.annotate(row["country_name"], (q_idx+1, row[target]), fontsize=7, color="white", 
                    fontweight="bold", xytext=(10, 0), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

plt.tight_layout()
plt.savefig("../Figures/Link/G_1_6_Rate_QD_boxplot.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question D :**

**Answer :** KW H=19.4 (p=0.0002) → significant difference between groups.
Q1 (lowest rates) has dramatically higher export share — median 0.253 vs
0.016–0.024 for Q2/Q3/Q4.

**Key observation :** Q1 contains all major exporters — USA, Canada, Japan
are all in the lowest rate group. This confirms QB finding : large exporters
operate with low real rates. The separation is not gradual — it's a
step change between Q1 and the rest.

**Anomaly :** China is in Q2 (rate ~3%) not Q1 — yet dominates with 12.5%
export share. Brazil in Q4 (33% rate) still maintains 1% share →
both are GDP-driven anomalies, not rate-driven.

**Contrast with CA QD :** CA QD showed all countries as over-performers
(GDP dominates). Here the KW is significant — rate groups genuinely
discriminate between exporters. Low real rate is a structural feature
of large exporting economies, not just a coincidence.

````python
df_ga = df_master[["country_name","cagr_export","cagr_gdp","size_profile"]].dropna()
r_sp, p_sp = spearmanr(df_ga["cagr_gdp"], df_ga["cagr_export"])

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Does GDP growth explain export growth?"
             "\n CAGR Export vs CAGR GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_ga[df_ga["size_profile"]==sp]
    ax.scatter(sub["cagr_gdp"]*100, sub["cagr_export"]*100, color=col, alpha=0.5, s=40, label=sp)

lim = (df_ga[["cagr_export","cagr_gdp"]].abs().max().max())*100*1.1
ax.plot([-lim,lim],[-lim,lim], color="white", linestyle="--", alpha=0.4, label="Elasticity = 1")
ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)
ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_ga[df_ga["country_name"].isin(countries)].iterrows():
    ax.scatter(row["cagr_gdp"]*100, row["cagr_export"]*100, color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row["cagr_gdp"]*100, row["cagr_export"]*100), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("CAGR GDP (%)", color="white")
ax.set_ylabel("CAGR Export (%)", color="white")
ax.set_xlim (-15, 15)
ax.set_ylim (-20, 40)
ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/G_2_QA_cagr_export_vs_gdp.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question A :**

of their export CAGR — high dispersion at any GDP growth level.

**Anomaly :** China has positive GDP growth but negative export CAGR
→ decoupling confirmed (economy grows but exports decline).
Japan near zero on both axes → stagnant economy and stagnant exports.

**Answer :** GDP growth is a necessary but far from sufficient condition
for export growth. The relationship is loose and non-linear.

**Insight — Question C :**

**Key observation :** Leader group has the **lowest variance** (tight boxplot)

## G.4 — Export Stability × Economic Indicators
</div>

**Variables :**
- `volatility_export` : standard deviation of annual export growth (continuous)
- `stability_risk` : identical Spearman profile → treated together
- `stability_profile` : Stable / Moderate / Volatile / Very Volatile

**Note :** Volatility Export and Stability Risk have identical RF and Spearman results
→ analyzed together as one dimension.

**Selected indicators :**
| Indicator | Flag | RF imp | Spearman ρ | Signal |
|---|---|---|---|---|
| GDP (US$) | [S] | 0.112 | -0.283 | Size → stability |
| Real Interest Rate % | [E] | 0.049 | +0.198 | High rates → more volatile |
| GDP per Capita | [E] | 0.038 | -0.149 | Wealth → stability |

**Questions :**

### Question A — Do larger/richer economies have more stable exports?
</div>

````python
from scipy.stats import linregress

gdp_col  = "GDP (current US$)"
gdp_pc   = "GDP per capita (current US$)"
vol_col  = "volatility_export"

df_ga = df_master[["country_name", vol_col, gdp_col, gdp_pc, "size_profile"]].dropna().copy()

# Two scatter side by side
fig, axes = styled_subplots(2, 1, figsize=(12, 16))
fig.suptitle("Do larger/richer economies have more stable exports?", color="white", fontsize=13)

for ax, x_col, x_label, log_x in [(axes[0], gdp_col,  "GDP (US$) — log scale", True),
                                  (axes[1], gdp_pc,   "GDP per Capita (US$) — log scale", True)]:
    x_vals = np.log10(df_ga[x_col].clip(lower=1))
    r_sp, p_sp = spearmanr(x_vals, df_ga[vol_col])

    for sp, col in SIZE_COLORS.items():
        sub = df_ga[df_ga["size_profile"]==sp]
        ax.scatter(sub[x_col], sub[vol_col], color=col, alpha=0.5, s=35, label=sp)

    slope, intercept, _, _, _ = linregress(x_vals, df_ga[vol_col])
    x_line = np.linspace(x_vals.min(), x_vals.max(), 100)
    ax.plot(10**x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.6)

    for _, row in df_ga[df_ga["country_name"].isin(countries)].iterrows():
        ax.scatter(row[x_col], row[vol_col], color=SIZE_COLORS[row["size_profile"]], s=120, zorder=6, edgecolors="white", linewidths=1.5)
        ax.annotate(row["country_name"], (row[x_col], row[vol_col]), fontsize=8, color="white", fontweight="bold",
                    xytext=(8,3), textcoords="offset points", 
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    ax.set_xscale("log")
    ax.set_xlabel(x_label, color="white")
    ax.set_ylabel("Export Volatility", color="white")
    ax.set_ylim (0, 120)
    ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=7)

plt.tight_layout()
plt.savefig("../Figures/Link/G_4_QA_volatility_vs_gdp.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question A :**

**Answer :** Yes, weakly — larger (ρ=-0.28) and richer (ρ=-0.15) economies
tend to have more stable exports. Both relationships are significant but modest.

**Observation :** All selected Leaders (Japan, USA, Germany, China) cluster
in the low-volatility zone (<40). Small countries dominate the high-volatility
zone (>100) — commodity exporters and micro-states.

**Limit :** Size effect — larger economies have more diversified export
baskets by construction, not necessarily due to better economic management.

### Question B — Is there a GDP threshold for export stability?

**Insight — Question B :**

**Limit :** High variance within each quartile — size alone is insufficient

### Question C — How does Real Interest Rate differ across stability profiles?
</div>

````python
from scipy.stats import kruskal

rate_col = "Real interest rate (%)"
df_gc = df_master[["country_name","stability_profile", rate_col]].dropna().copy()
df_gc[rate_col] = df_gc[rate_col].clip(-10, 20)

groups_ordered = ["Stable","Moderate","Volatile","Very Volatile"]
group_colors   = [C["green"],C["blue"],C["orange"],C["red"]]
data_box = [df_gc[df_gc["stability_profile"]==g][rate_col].dropna().values for g in groups_ordered]
available = [(g,d,c) for g,d,c in zip(groups_ordered,data_box,group_colors) if len(d) >= 3]
groups_av, data_av, colors_av = zip(*available) if available else ([],[],[])
stat, p = kruskal(*list(data_av))

fig, ax = styled_figure(figsize=(12, 7))
ax.set_title(f" Real Interest Rate by Stability Profile"  f"\n KW H={stat:.1f} (p={p:.4f})", color="white")

bp = ax.boxplot(list(data_av), widths=width, patch_artist=True, 
                    medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)
                   )
for patch, col in zip(bp["boxes"], colors_av): patch.set_facecolor(col); patch.set_alpha(0.7)

xlabels = [f"{g}(n={len(d)})" for g,d in zip(groups_av, data_av)]
ax.set_xticklabels(xlabels, color="white", fontsize=9)
ax.set_ylabel("Real Interest Rate (%)", color="white")
ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)

for i, vals in enumerate(data_av):
    if len(vals) == 0: continue
    ax.text(i+1, np.median(vals), f"{np.median(vals):.1f}%", ha="center", va="bottom", color="white", fontsize=8, 
            fontweight="bold")

for _, row in df_gc[df_gc["country_name"].isin(countries)].iterrows():
    if row["stability_profile"] not in groups_av: continue
    g_idx = list(groups_av).index(row["stability_profile"]) + 1
    ax.scatter(g_idx, row[rate_col], color=C["orange"], s=100, zorder=6, edgecolors="white", linewidths=1.2)
    ax.annotate(row["country_name"], (g_idx, row[rate_col]), fontsize=7, color="white", fontweight="bold",
                xytext=(10,0), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

plt.tight_layout()
plt.savefig("../Figures/Link/G_4_QC_rate_by_stability.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — Question C :**

**Answer :** Borderline — KW H=7.4 (p=0.059), just above significance threshold.
Stable countries have lower real rates (median 4.3%) vs Very Volatile (9.6%).
The trend is economically coherent but statistically weak.

**Brazil anomaly :** Classified as Stable despite 33% real rate → GDP size
overrides the rate signal. Clipping to 20% would improve visual clarity.

**Limit :** p=0.059 → not significant at 5% level. Interpret as tendency,
not confirmed relationship. Real Interest Rate is a partial signal for
stability — necessary but not sufficient.

### Question D — At same GDP level, who has more stable exports?
</div>

````python
# Stability efficiency = 1/volatility relative to GDP quartile median
df_gd = df_ga.copy()
df_gd["gdp_q"] = pd.qcut(df_gd[gdp_col], 4, labels=["Q1","Q2","Q3","Q4"])

# Within each GDP quartile : who is below median volatility (more stable)?
for q in ["Q1","Q2","Q3","Q4"]:
    med = df_gd[df_gd["gdp_q"]==q][vol_col].median()
    df_gd.loc[df_gd["gdp_q"]==q, "vol_relative"] = df_gd.loc[df_gd["gdp_q"]==q, vol_col] - med

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" At same GDP level, who has more stable exports?"
             "\n Volatility relative to GDP quartile median (below 0 = more stable than peers)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_gd[df_gd["size_profile"]==sp]
    ax.scatter(sub[gdp_col], sub["vol_relative"], color=col, alpha=0.5, s=35, label=sp)

ax.axhline(0, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Quartile median (benchmark)")

ymax = df_gd["vol_relative"].abs().quantile(0.95)
ax.axhspan(-ymax, 0,    alpha=0.05, color=C["green"])
ax.axhspan(0, ymax,     alpha=0.05, color=C["red"])
ax.text(df_gd[gdp_col].min()*1.5,  ymax*0.8, "More volatile than peers", color=C["red"], fontsize=9, bbox = dict(facecolor ="white"))
ax.text(df_gd[gdp_col].min()*1.5, -ymax*0.9, "More stable than peers", color=C["green"], fontsize=9, bbox = dict(facecolor ="white"))

for _, row in df_gd[df_gd["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row["vol_relative"], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, 
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row["vol_relative"]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xscale("log")
ax.set_xlabel("GDP (US$) — log scale", color="white")
ax.set_ylabel("Volatility − quartile median", color="white")
ax.set_ylim(-ymax, ymax)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G_4_QD_stability_vs_peers.png", dpi=150, bbox_inches="tight")
plt.show()

print("Selected countries — stability vs peers :")
print(df_gd[df_gd["country_name"].isin(countries)][
      ["country_name","vol_relative","volatility_export","gdp_q","size_profile"]
      ].sort_values("vol_relative").round(3).to_string(index=False))
````

**Insight — Question D :**

**Answer :** Within Q4 (largest economies), stability varies dramatically.
Japan (-5.9) and USA (-3.1) are most stable relative to peers.
France (+41.5) is the most volatile Leader — surprising given its
developed economy status.

**Japan and USA stability :** Both have large domestic markets that
buffer export volatility — even when exports fluctuate, the domestic
base provides stability.

**France anomaly :** High volatility despite high GDP — possibly driven
by concentration in luxury goods, aerospace and agri-exports that are
more price-sensitive and cyclical.

**India and Canada** : more volatile than their Q4 peers → commodity
exposure (India: IT services cycles, Canada: energy/commodities).

### Question E — Does government spending (Expense % GDP) explain export stability?
</div>

**Rationale :** Expense % GDP is the top [E] indicator in RF (importance=0.081).
Higher government spending may proxy for economic diversification → more stable exports.
Also : public expenditure can cushion export sectors during downturns → reduce volatility.


````python
from scipy.stats import linregress

exp_col = "Expense (% of GDP)"
vol_col = "volatility_export"

df_qe = df_master[["country_name", vol_col, exp_col, "size_profile"]].dropna().copy()
df_qe[exp_col] = df_qe[exp_col].clip(0, 80)

r_sp, p_sp = spearmanr(df_qe[exp_col], df_qe[vol_col])
slope, intercept, _, _, _ = linregress(df_qe[exp_col], df_qe[vol_col])
x_line = np.linspace(df_qe[exp_col].min(), df_qe[exp_col].max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Does government spending explain export stability?"
             "\n Export Volatility vs Expense % GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_qe[df_qe["size_profile"]==sp]
    ax.scatter(sub[exp_col], sub[vol_col], color=col, alpha=0.5, s=40, label=sp)

ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--",
        alpha=0.6, label=f"Trend (ρ={r_sp:.2f})")

for _, row in df_qe[df_qe["country_name"].isin(countries)].iterrows():
    ax.scatter(row[exp_col], row[vol_col], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[exp_col], row[vol_col]), fontsize=9, color="white", fontweight="bold",
                xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("Expense % GDP", color="white")
ax.set_ylabel("Export Volatility", color="white")
ax.set_ylim (-5, 100)
ax.set_xlim (0, 60)
ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)

plt.tight_layout()
plt.savefig("../Figures/Link/G_4_QE_volatility_vs_expense.png", dpi=150, bbox_inches="tight")
plt.show()

print(f"Spearman ρ = {r_sp:.3f} (p={p_sp:.4f})")
print(f"n = {len(df_qe)}")
print(50*"=")
print("Selected countries :")
df_qe[df_qe["country_name"].isin(countries)][["country_name", exp_col, vol_col, "size_profile"]].\
      sort_values(exp_col).round(2)
````

**Insight — Question E :**

**Answer :** No — Expense % GDP shows no relationship with export volatility
(ρ=0.01, p=0.908). Government spending does not stabilize exports.

**Why RF showed importance=0.081 :** RF captures complex non-linear
interactions. Expense % GDP may interact with other variables (e.g. high
spending + high GDP = stability) but has no independent monotone effect.

**Implication :** Export stability cannot be achieved through fiscal policy
alone. The stabilizing factors (GDP size, moderate real rates) are structural,
not policy-driven in the short term.

### G.5.3 — Finding

**G.5 — Finding : Export Resilience is Not Explained by Economic Indicators**

**Objective :** Identify which economic indicators explain export resilience.

**Result :** The available macro indicators do not explain export resilience.
- Only size [S] indicators correlate weakly (GDP ρ=+0.20, FDI ρ=+0.23)
- No [E] economic indicator is significant

**What the figure shows :** Larger economies tend to be slightly more resilient

**Structural pattern — G.2 / G.3 / G.5 :**
Three export dimensions remain unexplained by macro fundamentals :
- G.2 : Export Growth → idiosyncratic
- G.3 : Export Consistency → idiosyncratic
- G.5 : Export Resilience → idiosyncratic

Only **export level** (G.1 — share, volume, per capita) is explained
by economic fundamentals. Export *dynamics* require sector-level,
institutional or policy data to explain — beyond the scope of this dataset.

Expense % GDP [E] ρ=+0.280 | Real Interest Rate [E] ρ=-0.181

**RF R²=0.930  OLS R²=0.379  Gap=0.551** 🔴 — strong non-linearity
The high gap confirms that trade openness follows threshold logic —
linear models massively underfit. RF captures country-size thresholds
and regime breaks that OLS cannot model.

## G.7 — General Observation : Export Analysis × Economic Indicators
</div>

**Structural finding — Two distinct worlds :**

**1. Export Level is explained by macro fundamentals**
G.1 (Export Share, Size Profile), G.6 (Export/GDP, Export per Capita)
→ GDP size, Current Account, Real Rate, GDP_pc, Expense% GDP all
  contribute to explain *how much* a country exports
→ ρ ranges from 0.18 to 0.94 — significant and actionable signals

**2. Export Dynamics are idiosyncratic**
G.2 (CAGR Export), G.3 (Consistency), G.4 (Stability), G.5 (Resilience)
→ No macro indicator explains export growth rate, consistency, or resilience
→ These dimensions follow country-specific structural logic —
  trade policy, geographic position, partner diversification —
  not captured by the 16 macro indicators available here.

**Implication for downstream sections :**
Export level variables (G.1, G.6) are the most tractable for H/I/J analysis.
Export dynamic variables (G.2–G.5) should be treated as outcome variables
to explain, not predictors — their drivers lie outside this indicator set.

**Key indicators across all G sections :**
| Indicator | Flag | Most relevant for |
|---|---|---|
| GDP (US$) | [S] | Export Share, Size Profile, Stability |
| GDP per Capita | [E] | Export_pc, Export/GDP, Stability |
| Current Account % GDP | [E] | Export Share, Export/GDP, Coverage |
| Real Interest Rate % | [E] | Export Share, Export/GDP, Export_pc, Stability |
| Expense % GDP | [E] | Export_pc, Export/GDP |

**Recurring anomalies across sections :**
- **Germany** : over-performer in every export dimension
  → most commercially efficient large economy at all scales
- **USA** : under-performer relative to size in G.1, G.6
  → service economy consistently reduces relative export performance
- **China** : over-performer in absolute volume (G.1 P1)
  but near-median in per-capita (G.1 P2) → collective volume model
- **France** : most volatile Leader (G.4), least resilient (G.5)
  → structural fragility despite large export volumes
- **Japan** : most stable (G.4) but least resilient (G.5)
  → stability ≠ resilience confirmed

**Methodological note :**
RF R² is systematically high (0.85–0.98) while OLS R² is low
for dynamic variables → relationships are non-linear and complex.
Macro fundamentals set the structural conditions for export level
but cannot predict export behavior over time.

````python
# ── H.1 : Import Share × GDP — Questions A/B/C/D ────────────────────────────
from scipy.stats import linregress, kruskal

gdp_col    = "GDP (current US$)"
target_imp = "import_share_structural"
df_h1 = df_master[["country_name", target_imp, gdp_col, "size_profile"]].dropna().copy()

# Question A : scatter log-log
log_gdp = np.log10(df_h1[gdp_col].clip(lower=1))
r_sp, p_sp = spearmanr(log_gdp, df_h1[target_imp])
slope, intercept, _, _, _ = linregress(log_gdp, df_h1[target_imp])
x_line = np.linspace(log_gdp.min(), log_gdp.max(), 100)

fig, axes = styled_subplots(2, 2, figsize=(18, 14))
fig.suptitle("H.1 — Import Share × GDP", color="white", fontsize=14)

# QA
ax = axes[0,0]
ax.set_title(f"Do larger economies dominate world imports?\nρ={r_sp:.2f}", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_h1[df_h1["size_profile"]==sp]
    ax.scatter(sub[gdp_col], sub[target_imp], color=col, alpha=0.5, s=35, label=sp)
ax.plot(10**x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.6)

for _, row in df_h1[df_h1["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row[target_imp], color=SIZE_COLORS[row["size_profile"]], s=120, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row[target_imp]), fontsize=8, color="white", fontweight="bold",
                xytext=(8,3), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))
ax.set_xscale("log"); ax.set_xlabel("GDP (US$) — log", color="white")
ax.set_ylabel("Import Share (% world)", color="white")
ax.text(0.05, 0.95, f"ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=7)

# QB — scatter colored by GDP quartile
df_h1["gdp_q"] = pd.qcut(df_h1[gdp_col], 4, labels=["Q1","Q2","Q3","Q4"])
Q_COLORS = {"Q1":C["red"],"Q2":C["orange"],"Q3":C["blue"],"Q4":C["green"]}
ax = axes[0,1]
ax.set_title("Is there a GDP threshold?", color="white")
for q, col in Q_COLORS.items():
    sub = df_h1[df_h1["gdp_q"]==q]
    ax.scatter(sub[gdp_col], sub[target_imp], color=col, alpha=0.6, s=40, label=q)
ax.plot(10**x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.5)
q_bounds = df_h1[gdp_col].quantile([0.25,0.50,0.75])
for q_val in q_bounds:
    ax.axvline(q_val, color=C["grey"], linestyle=":", linewidth =2, alpha=0.5)
    ax.text(q_val, df_h1[target_imp].max()*0.9, f"${q_val/1e9:.0f}B", color=C["orange"], fontsize=8, fontweight= "bold",ha="center")
for _, row in df_h1[df_h1["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row[target_imp], color=Q_COLORS[row["gdp_q"]], s=120, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row[target_imp]), fontsize=8, color="white", fontweight="bold",
                xytext=(8,3), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))
ax.set_xscale("log"); ax.set_xlabel("GDP (US$) — log", color="white")
ax.set_ylabel("Import Share (% world)", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=7)

# QC — KW boxplot GDP by size profile
groups_ordered = ["Leader","Major","Mid","Small"]
data_box = [df_h1[df_h1["size_profile"]==g][gdp_col].dropna().values for g in groups_ordered]
stat_kw, p_kw = kruskal(*data_box)
ax = axes[1,0]
ax.set_title(f"GDP by Size Profile (KW H={stat_kw:.1f}, p={p_kw:.4f})", color="white")
# bp = ax.boxplot(data_box, patch_artist=True, medianprops=dict(color="white", linewidth=2))

bp = ax.boxplot(data_box, widths=width, patch_artist=True, 
                    medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)
                   )

for patch, col in zip(bp["boxes"], [C["green"],C["blue"],C["orange"],C["red"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)
ax.set_xticklabels([f"{g}\n(n={len(d)})" for g,d in zip(groups_ordered,data_box)], color="white", fontsize=8)
ax.set_ylabel("GDP (US$)", color="white"); ax.set_yscale("log")
for i, vals in enumerate(data_box):
    if len(vals)==0: continue
    med = np.median(vals)
    ax.text(i+1, med, f"${med/1e12:.1f}T" if med>1e12 else f"${med/1e9:.0f}B",
            ha="center", va="bottom", color="white", fontsize=9, fontweight="bold")

# QD — Trade efficiency ratio
world_imp = df_h1[target_imp].sum()
df_h1["gdp_share"] = df_h1[gdp_col] / df_h1[gdp_col].sum() * 100
df_h1["imp_eff"]   = df_h1[target_imp] / df_h1["gdp_share"].replace(0, np.nan)
df_h1 = df_h1[df_h1[gdp_col] > 1e9]
ax = axes[1,1]
ax.set_title("Among same-GDP, who imports more than economic weight?\n"
             "Import Share / GDP Share", color="white")
for sp, col in SIZE_COLORS.items():
    sub = df_h1[df_h1["size_profile"]==sp]
    ax.scatter(sub[gdp_col], sub["imp_eff"], color=col, alpha=0.5, s=35, label=sp)
ax.axhline(1, color="white", linewidth=1.5, linestyle="--", alpha=0.6, label="Ratio=1")
ymax = df_h1["imp_eff"].quantile(0.9)
ax.axhspan(1, ymax*1.05, alpha=0.05, color=C["green"])
ax.axhspan(0, 1,         alpha=0.05, color=C["red"])
for _, row in df_h1[df_h1["country_name"].isin(countries)].iterrows():
    ax.scatter(row[gdp_col], row["imp_eff"], color=SIZE_COLORS[row["size_profile"]], s=120, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[gdp_col], row["imp_eff"]), fontsize=8, color="white", fontweight="bold",
                xytext=(8,3), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))
ax.set_xscale("log"); ax.set_xlabel("GDP (US$) — log", color="white")
ax.set_ylabel("Import Share / GDP Share", color="white")
ax.set_ylim(0, ymax*1.1)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=7)

plt.tight_layout()
plt.savefig("../Figures/Link/H_1_import_share_vs_gdp.png", dpi=150, bbox_inches="tight")
plt.show()

print("Selected countries — Import efficiency :")
df_h1[df_h1["country_name"].isin(countries)][["country_name","imp_eff","import_share_structural",
            "size_profile"]].sort_values("imp_eff", ascending=False).round(2)
````

**Insight — H.1 :**

**A :** Yes — ρ=0.96 (p<0.001). Strongest relationship in all of H.

USA has the highest import share (10.7%) — world's largest importer —

**Insight — H.4 :**

**A — CAGR Import vs CAGR GDP (ρ=0.41) :**
Stronger than Export equivalent (ρ=0.29). GDP growth predicts import growth
better than export growth → imports are more directly linked to domestic
economic activity. When GDP grows, domestic demand pulls imports.
China : negative import CAGR despite positive GDP → decoupling pattern
(domestic production substituting imports).

**B — Unemployment (ρ=-0.02, p=0.829) :** Not significant — 
the hypothesis that higher unemployment predicts import growth
is not confirmed. No linear relationship between labor market
conditions and import growth dynamics.

**Conclusion :** Import growth, like Export growth, is idiosyncratic —
no macro indicator explains it robustly.
GDP growth (ρ=0.41) remains the only meaningful predictor,
and even that explains less than 20% of the variance.

**Limit :** Gap 🔴 for both variables — dynamics remain idiosyncratic.
CAGR GDP (ρ=0.41) is the strongest available signal but still explains
only a portion of import growth variance.

## H.5 — Import Consistency (Negfreq Import)
</div>

**Key stats :**
[S] GDP ρ=+0.207 | [E] GDP_pc ρ=+0.183 | RF top [E] : Unemployment % (0.107)
⚠️ GDP sign is positive → larger countries have MORE negative import years
(counter-intuitive vs exports where large = more consistent)

**Specific question :** *"Does unemployment predict import decline frequency?"*
Unemployed → reduced domestic demand → more frequent import declines


````python
unemp_col = "Unemployment, total (% of total labor force) (modeled ILO estimate)"
gdp_col   = "GDP (current US$)"
negf_col  = "negfreq_import"

df_h5 = df_master[["country_name", negf_col, unemp_col, gdp_col, "size_profile"]].dropna().copy()
df_h5[unemp_col] = df_h5[unemp_col].clip(0, 30)

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Negfreq Import × Unemployment & GDP", color="white", fontsize=13)

for ax, x_col, x_label, do_log in [
    (axes[0], unemp_col, "Unemployment %", False), (axes[1], gdp_col,   "GDP (US$) — log", True)]:
    
    x_vals = np.log10(df_h5[x_col].clip(lower=1)) if do_log else df_h5[x_col]
    r_sp, p_sp = spearmanr(x_vals, df_h5[negf_col])
    slope, intercept, _, _, _ = linregress(x_vals, df_h5[negf_col])
    x_line = np.linspace(x_vals.min(), x_vals.max(), 100)

    for sp, col in SIZE_COLORS.items():
        sub = df_h5[df_h5["size_profile"]==sp]
        ax.scatter(sub[x_col], sub[negf_col], color=col, alpha=0.5, s=35, label=sp)

    ax.plot(10**x_line if do_log else x_line, intercept + slope * x_line, color="white", linewidth=1.5, 
            linestyle="--", alpha=0.6)

    for _, row in df_h5[df_h5["country_name"].isin(countries)].iterrows():
        ax.scatter(row[x_col], row[negf_col], color=SIZE_COLORS[row["size_profile"]], s=120, zorder=6,
                   edgecolors="white", linewidths=1.5)
        ax.annotate(row["country_name"], (row[x_col], row[negf_col]), fontsize=8, color="white", fontweight="bold",
                    xytext=(8,3), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    if do_log: ax.set_xscale("log")
    ax.set_xlabel(x_label, color="white")
    ax.set_ylabel("Negfreq Import", color="white")
    ax.text(0.05,0.95,f"ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=7)

plt.tight_layout()
plt.savefig("../Figures/Link/H_5_negfreq_import.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — H.5 :**

**Unemployment (ρ=+0.08, p=0.329) :** Not significant — the hypothesis

**GDP (ρ=+0.21, p=0.007) :** Significant but counter-intuitive.
Larger economies have MORE frequent negative import years.
This contrasts with Export consistency where large = more consistent.
Explanation : large economies have more cyclical import patterns — they
absorb global shocks (financial crises, supply disruptions) more intensely
because their import volumes are large and diverse.

**France (negfreq≈0.6)** is the clear outlier — most frequent negative
import years among all highlighted countries. Not explained by unemployment
or GDP size — structural import fragility specific to France.

**Germany (negfreq≈0.3) and India (negfreq≈0.3)** at similar levels
despite very different economic profiles — confirming that import
consistency follows country-specific patterns, not macro fundamentals.

**Explanation :** large economies have large and volatile domestic demand —
domestic cycles dominate over import smoothing, creating more frequent
negative import years despite overall economic stability.

- Inflation signal appears in H.2, H.3, H.6 — absent from equivalent G sections

  purchasing power erosion. It does not affect the supply side (exports)

## I.1 — Coverage Ratio & Balance Risk
</div>

**Variables :** `coverage_ratio_structural` | `balance_risk`

**Key stats :** CA% GDP [E] ρ=+0.528 | GDP [S] ρ=+0.574 | GDP_pc [E] ρ=+0.234

**Questions :**
1. Do surplus countries cover their imports with exports? → CA%
2. Do larger/wealthier countries have better coverage? → GDP + GDP_pc
3. Among same CA position, who has better coverage? → KW by CA quartile


````python
from scipy.stats import linregress

ca_col  = "Current account balance (% of GDP)"
gdp_col = "GDP (current US$)"
gdp_pc  = "GDP per capita (current US$)"
cov_col = "coverage_ratio_structural"  

df_i1 = df_master[["country_name", cov_col, ca_col, gdp_col, gdp_pc, "size_profile"]].dropna().copy()
df_i1[ca_col]  = df_i1[ca_col].clip(-30, 30)
df_i1[cov_col] = df_i1[cov_col].clip(0, 5)

# ── 3 separate figures
for x_col, x_label, do_log, fname, ref_line in [
    (ca_col,  "Current Account % GDP",  False, "I_1_cov_vs_ca",    True),
    (gdp_col, "GDP (US$) — log scale",  True,  "I_1_cov_vs_gdp",   False),
    (gdp_pc,  "GDP per Capita — log",   True,  "I_1_cov_vs_gdppc", False)]:
    
    x_vals = np.log10(df_i1[x_col].clip(lower=1)) if do_log else df_i1[x_col]
    r_sp, p_sp = spearmanr(x_vals.values, df_i1[cov_col].values)
    slope, intercept, _, _, _ = linregress(x_vals, df_i1[cov_col])
    x_line = np.linspace(x_vals.min(), x_vals.max(), 100)

    fig, ax = styled_figure(figsize=(13, 8))
    ax.set_title(f"Coverage Ratio × {x_label}",  color="white")

    for sp, col in SIZE_COLORS.items():
        sub = df_i1[df_i1["size_profile"]==sp]
        ax.scatter(sub[x_col], sub[cov_col], color=col, alpha=0.5, s=40, label=sp)

    ax.plot(10**x_line if do_log else x_line, intercept + slope * x_line, color="white", 
            linewidth=1.5, linestyle="--", alpha=0.6, label=f"Trend (ρ={r_sp:.2f})")

    if ref_line:
        ax.axhline(1, color=C["orange"], linestyle="--", alpha=0.5, label="Coverage=1 (balanced)")
        ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

    for _, row in df_i1[df_i1["country_name"].isin(countries)].iterrows():
        ax.scatter(row[x_col], row[cov_col], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", 
                   linewidths=1.5)
        ax.annotate(row["country_name"], (row[x_col], row[cov_col]), fontsize=9, color="white", 
                    fontweight="bold", xytext=(8,4), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    if do_log: ax.set_xscale("log")
    ax.set_xlabel(x_label, color="white")
    ax.set_ylabel("Coverage Ratio (Export/Import)", color="white")
    ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()

# ── Q3 : Coverage efficiency ratio vs CA quartile (replace residuals) ─────────
df_i1["ca_q"] = pd.qcut(df_i1[ca_col], 4, labels=["Q1\n(Deficit)","Q2","Q3","Q4\n(Surplus)"])
labels_q  = ["Q1\n(Deficit)","Q2","Q3","Q4\n(Surplus)"]
data_box  = [df_i1[df_i1["ca_q"]==q][cov_col].dropna().values for q in labels_q]

from scipy.stats import kruskal
stat, p = kruskal(*data_box)

fig, ax = styled_figure(figsize=(12, 7))
ax.set_title(f"Coverage Ratio by CA Quartile\n" f"KW H={stat:.1f} (p={p:.4f})", color="white")

bp = ax.boxplot(data_box, widths=width, patch_artist=True, 
                    medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)
                   )

for patch, col in zip(bp["boxes"], [C["red"],C["orange"],C["blue"],C["green"]]):
    patch.set_facecolor(col); patch.set_alpha(0.7)

ax.axhline(1, color=C["orange"], linestyle="--", alpha=0.5, label="Coverage=1")
xlabels = [f"{q.replace(chr(10),' ')}\n(n={len(d)})" for q,d in zip(labels_q, data_box)]
ax.set_xticklabels(xlabels, color="white", fontsize=9)
ax.set_ylabel("Coverage Ratio (Export/Import)", color="white")

for i, vals in enumerate(data_box):
    if len(vals)==0: continue
    ax.text(i+1, np.median(vals), f"{np.median(vals):.2f}", ha="center", va="bottom", color="white", fontsize=8, fontweight="bold")

for q_idx, q in enumerate(labels_q):
    sub = df_i1[(df_i1["ca_q"]==q) & (df_i1["country_name"].isin(countries))]
    for _, row in sub.iterrows():
        ax.scatter(q_idx+1.05, row[cov_col], color="orange", s=100, zorder=6, edgecolors="white", linewidths=1.2)
        ax.annotate(row["country_name"], (q_idx+1.05, row[cov_col]), fontsize=7, color="white", 
                    fontweight="bold", xytext=(10,0), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/I_1_cov_by_ca_quartile.png", dpi=150, bbox_inches="tight")
plt.show()

print("Coverage Ratio by CA Quartile :")
for q, vals in zip(labels_q, data_box):
    print(f"  {q.replace(chr(10),' '):<20} median={np.median(vals):.2f}  n={len(vals)}")
````

**Insight — I.1 :**

**Q1 — CA% → Coverage (ρ=0.53) :**
Surplus countries export more than they import → higher coverage ratio.
Causal and direct. China (surplus +1.7%) and Germany (surplus +7.4%)
both above ratio=1. USA and France (deficit) both below 1.
India below 1 despite near-zero CA → structural import dependence.

**Q2 — GDP → Coverage (ρ=0.58) :**
Larger economies tend to have higher coverage — size enables
export volume to match or exceed import volume.
But this is a [S] result — size explains coverage through scale,
not through economic structure.

**Q3 — GDP_pc → Coverage (ρ=0.23) :**
Weaker signal — wealth alone doesn't guarantee export coverage.
Germany (high GDP_pc, high coverage) vs USA (high GDP_pc, low coverage)
shows that the relationship depends on economic model, not just wealth.

**Q3 — Residuals :**
Brazil (0.49) and China (0.45) over-perform their CA position →
export more than their CA surplus would predict.
USA (-0.08) and India (-0.09) under-perform → imports exceed what
their CA position explains. GDP size does not compensate.

## I.2 — Trade Intensity
</div>

**Recall D1 :** Trade Intensity vs GDP_pc analyzed — ρ=0.29, scatter + boxplot.

**New questions (indicators not analyzed in D1) :**

- "Does CA surplus drive trade openness?" → CA% [E] ρ=+0.232

````python
from scipy.stats import kruskal

ca_col   = "Current account balance (% of GDP)"
rate_col = "Real interest rate (%)"

df_i5 = df_master[["country_name","balance_profile", ca_col, rate_col]].dropna().copy()
df_i5[ca_col]   = df_i5[ca_col].clip(-30, 30)
df_i5[rate_col] = df_i5[rate_col].clip(-20, 20)

groups_ordered = ["Strong Surplus","Slight Surplus","Slight Deficit","Structural Deficit"]
group_colors   = [C["green"],C["blue"],C["orange"],C["red"]]

fig, axes = styled_subplots(1, 2, figsize=(18, 7))
fig.suptitle("Balance Profile × CA% GDP & Real Interest Rate (KW)", color="white", fontsize=13)

for ax, ind_col, ind_label in [
    (axes[0], ca_col,   "Current Account % GDP"), (axes[1], rate_col, "Real Interest Rate %")]:
    
    data_box = [df_i5[df_i5["balance_profile"]==g][ind_col].dropna().values  for g in groups_ordered]
    available = [(g,d,c) for g,d,c in zip(groups_ordered,data_box,group_colors) if len(d) >= 3]
    if not available: continue
    g_av, d_av, c_av = zip(*available)

    stat, p = kruskal(*list(d_av))

    #bp = ax.boxplot(list(d_av), patch_artist=True, medianprops=dict(color="white", linewidth=2))
    bp = ax.boxplot(list(d_av), widths=width, patch_artist=True, 
                    medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)
                   )
    
    for patch, col in zip(bp["boxes"], c_av): patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{g}(n={len(d)})" for g,d in zip(g_av, d_av)]
    ax.set_xticklabels(xlabels, color="white", fontsize=7)
    ax.set_ylabel(ind_label, color="white")
    ax.set_title(f"KW H={stat:.1f} (p={p:.4f})", color="white")
    ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)

    for i, vals in enumerate(d_av):
        if len(vals) == 0: continue
        ax.text(i+1, np.median(vals), f"{np.median(vals):.1f}", ha="center", va="bottom",
                color="white", fontsize=8, fontweight="bold")

    # Annotate selected countries
    for _, row in df_i5[df_i5["country_name"].isin(countries)].iterrows():
        if row["balance_profile"] not in g_av: continue
        g_idx = list(g_av).index(row["balance_profile"]) + 1
        ax.scatter(g_idx+0.07, row[ind_col], color=C["orange"], s=100, zorder=6, edgecolors="white", linewidths=1.2)
        ax.annotate(row["country_name"], (g_idx+0.07, row[ind_col]), fontsize=7, color="white", 
                    fontweight="bold", xytext=(10,0), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

plt.tight_layout()
plt.savefig("../Figures/Link/I_5_balance_profile.png", dpi=150, bbox_inches="tight")
plt.show()
````

→ Progressive and monotone — KW and Spearman agree (robust relationship).
China and Brazil in Strong Surplus despite near-zero individual CA values
→ wide variance within groups.

**Real Interest Rate (KW H=11.0, p=0.012) :**
Significant but counter-intuitive pattern — Strong Surplus countries

**Key finding :** CA position (I.5 left) is the most discriminating

a secondary but non-monotone signal — high rates can coexist with

The structural_score is dominated by z_export_share (weight 0.60) —
a country can achieve "Strong" through size alone, regardless of dynamic
performance.

**Key contradiction — Germany :**
Germany holds the highest structural score (+4.6) yet is classified
Declining in J.3 (negative z_CAGR) and above-median volatile in G.4
(vol_relative=+21.4 among Leaders). A country can simultaneously be
the best structural performer and show deteriorating export dynamics.
This is not a paradox — it reveals that the structural score measures
*accumulated position* while the dynamic quadrant measures *current
trajectory*. The two dimensions are independent, and their divergence
is analytically meaningful : Germany is living on structural capital
while its growth engine slows.

**General signal :** structural score and dynamic quadrant are
complementary, not redundant. High score + Declining dynamic = mature
leader at risk. Low score + Star dynamic = emerging challenger with
fragile base. Both combinations exist in the data and tell different stories.

## J.3 — Dynamic Quadrant

## K.1 — Bilateral Score × Economic Indicators
</div>

**Variable :** `g_norm_rank` (higher = more resilient / less concentrated)

**Question :** Do countries with stronger macro fundamentals show less concentrated export partnerships?

Hypothesis : better CA position, higher GDP_pc → more diversified trade partners
→ lower bilateral dependency → higher g_norm_rank

**Key stats :** to be computed below


````python
from scipy.stats import spearmanr, linregress

g_col   = "g_norm_rank"
gdp_pc  = "GDP per capita (current US$)"
ca_col  = "Current account balance (% of GDP)"
rate_col = "Real interest rate (%)"
exp_col  = "Expense (% of GDP)"

df_k1 = df_master[["country_name", g_col, gdp_pc, ca_col, rate_col, exp_col, "size_profile"]].dropna().copy()
df_k1[ca_col]   = df_k1[ca_col].clip(-30, 30)
df_k1[rate_col] = df_k1[rate_col].clip(-20, 30)

print("Spearman correlations — g_norm_rank × macro indicators :")
print("="*55)
for ind_col, ind_label in [(gdp_pc,  "GDP per Capita"), (ca_col,  "Current Account % GDP"),
                           (rate_col,"Real Interest Rate %"), (exp_col, "Expense % GDP")]:
    
    x_vals = np.log10(df_k1[ind_col].clip(lower=1)) if ind_col==gdp_pc else df_k1[ind_col]
    r_sp, p_sp = spearmanr(x_vals.values, df_k1[g_col].values)
    sig = "✅" if p_sp < 0.05 else "❌"
    print(f"  {sig} ρ={r_sp:+.3f} (p={p_sp:.3f})  {ind_label}")

print("\nSelected countries — bilateral resilience rank :")
df_k1[df_k1["country_name"].isin(countries)][["country_name", g_col, "size_profile"] \
                                        ].sort_values(g_col, ascending=False).round(3)
````

````python
# Figure : scatter g_norm_rank vs best significant indicator
# Try CA% first — hypothesis : surplus → better diversification
r_ca, p_ca = spearmanr(df_k1[ca_col].values, df_k1[g_col].values)
r_pc, p_pc = spearmanr(np.log10(df_k1[gdp_pc].clip(lower=1)).values, df_k1[g_col].values)

# Use stronger signal
if abs(r_ca) >= abs(r_pc):   x_col, x_label, do_log = ca_col, "Current Account % GDP", False
else:    x_col, x_label, do_log = gdp_pc, "GDP per Capita (log)", True

x_vals = np.log10(df_k1[x_col].clip(lower=1)) if do_log else df_k1[x_col]
slope, intercept, _, _, _ = linregress(x_vals, df_k1[g_col])
x_line = np.linspace(x_vals.min(), x_vals.max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title(" Bilateral Resilience Rank × Economic Indicator \n"
             "Do better macro fundamentals reduce export partnership concentration?",  color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_k1[df_k1["size_profile"]==sp]
    ax.scatter(sub[x_col], sub[g_col], color=col, alpha=0.5, s=40, label=sp)

ax.plot(10**x_line if do_log else x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.6)

if not do_log: ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_k1[df_k1["country_name"].isin(countries)].iterrows():
    ax.scatter(row[x_col], row[g_col], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6, edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[x_col], row[g_col]), fontsize=9, color="white", 
                fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

if do_log: ax.set_xscale("log")
ax.set_xlabel(x_label, color="white")
ax.set_ylabel("Bilateral Resilience Rank (higher = less concentrated)", color="white")
r_plot, p_plot = spearmanr(x_vals.values, df_k1[g_col].values)
ax.text(0.05, 0.95, f"Spearman ρ={r_plot:.2f} (p={p_plot:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/K_1_bilateral_vs_macro.png", dpi=150, bbox_inches="tight")
plt.show()
````

## K.2 — Combined Score × Economic Indicators
</div>

**Variable :** `combined_score` = 0.70 × f_norm_rank + 0.30 × g_norm_rank

**Question :** Does the combined score (structural + bilateral resilience) confirm or enrich the macro relationships found in J.1 (structural only)?

If combined ≈ structural → bilateral adds little information
If combined diverges → bilateral dimension changes the macro relationships

````python
from scipy.stats import spearmanr

comb_col = "combined_score"
f_col    = "f_norm_rank"
g_col    = "g_norm_rank"

df_k2 = df_master[["country_name", comb_col, f_col, g_col, gdp_pc, ca_col, "size_profile"]].dropna().copy()
df_k2[ca_col] = df_k2[ca_col].clip(-30, 30)

# Compare ρ : structural vs combined for same indicators
print("Comparison — Spearman ρ : structural (f_norm) vs combined score")
print("="*60)
for ind_col, ind_label in [(gdp_pc, "GDP per Capita"), (ca_col, "Current Account % GDP")]:
    x_vals = np.log10(df_k2[ind_col].clip(lower=1)) if ind_col==gdp_pc else df_k2[ind_col]
    r_f, p_f = spearmanr(x_vals.values, df_k2[f_col].values)
    r_c, p_c = spearmanr(x_vals.values, df_k2[comb_col].values)
    print(f"  {ind_label}")
    print(f"    f_norm_rank  : ρ={r_f:+.3f} (p={p_f:.3f})")
    print(f"    combined     : ρ={r_c:+.3f} (p={p_c:.3f})")
    print(f"    Δρ = {r_c-r_f:+.3f}")

# Scatter f vs g colored by combined_profile
prof_col = "combined_profile"
df_k2b = df_master[["country_name", f_col, g_col, comb_col, prof_col]].dropna().copy()

PROF_COLORS = {
    "Global Champion": C["green"], "Competitive": C["blue"], "Vulnerable": C["orange"], "Fragile": C["red"] }

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Structural Rank × Bilateral Resilience Rank \n"
             "Tension between structural strength and geographic concentration", color="white")

for prof, col in PROF_COLORS.items():
    sub = df_k2b[df_k2b[prof_col]==prof]
    if len(sub)==0: continue
    ax.scatter(sub[f_col], sub[g_col], color=col, alpha=0.6, s=40, label=f"{prof} (n={len(sub)})")

ax.axvline(0.5, color="white", linestyle="--", alpha=0.4, linewidth=1)
ax.axhline(0.5, color="white", linestyle="--", alpha=0.4, linewidth=1)

ax.text(0.7, 0.85, "Strong & Resilient", color=C["green"], bbox= dict(facecolor = "white"),
        fontsize=8, alpha=0.7, fontweight="bold", transform=ax.transAxes)
ax.text(0.7, 0.2, "Strong but Exposed", color=C["orange"], bbox= dict(facecolor = "white"),
        fontsize=8, alpha=0.7, fontweight="bold", transform=ax.transAxes)
ax.text(0.15, 0.85, "Diversified Minor",  color=C["blue"], bbox= dict(facecolor = "white"),
        fontsize=8, alpha=0.7, fontweight="bold", transform=ax.transAxes)
ax.text(0.15, 0.2, "Weak & Vulnerable",  color=C["red"], bbox= dict(facecolor = "white"),
        fontsize=8, alpha=0.7, fontweight="bold", transform=ax.transAxes)

for _, row in df_k2b[df_k2b["country_name"].isin(countries)].iterrows():
    ax.scatter(row[f_col], row[g_col], color=PROF_COLORS[row[prof_col]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[f_col], row[g_col]),
                fontsize=9, color="white", fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("F rank — Structural Score (higher = stronger)", color="white")
ax.set_ylabel("G rank — Bilateral Resilience (higher = less concentrated)", color="white")
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/K_2_f_vs_g_combined.png", dpi=150, bbox_inches="tight")
plt.show()

print("\nSelected countries :")
df_k2b[df_k2b["country_name"].isin(countries)][["country_name", f_col, g_col, comb_col, prof_col]] \
.sort_values(comb_col, ascending=False).round(3)
````

## K.3 — Combined Profile × Economic Indicators
</div>

**Variable :** `combined_profile`
(Global Champion ≥0.80 / Competitive ≥0.55 / Vulnerable ≥0.30 / Fragile)

**Question :** What macro conditions characterize Global Champions vs Fragile countries?

Combining structural strength + bilateral resilience → does the combined profile
reveal macro patterns not visible in structural profile alone (J.1)?


````python
from scipy.stats import kruskal

comb_col = "combined_profile"
groups_cp = ["Global Champion","Competitive","Vulnerable","Fragile"]
colors_cp = [C["green"],C["blue"],C["orange"],C["red"]]

df_k3 = df_master[["country_name", comb_col, gdp_pc, ca_col]].dropna().copy()
df_k3[ca_col] = df_k3[ca_col].clip(-30, 30)

for ind_col, ind_label, do_log, fname in [
    (gdp_pc, "GDP per Capita", True,  "K_3_combined_profile_gdppc"),
    (ca_col, "CA % GDP",       False, "K_3_combined_profile_ca")]:
    df_tmp = df_k3.dropna(subset=[ind_col])
    data_kw = [df_tmp[df_tmp[comb_col]==g][ind_col].values for g in groups_cp]
    avail = [(g,d,c) for g,d,c in zip(groups_cp,data_kw,colors_cp) if len(d)>=3]
    if not avail: continue
    g_av, d_av, c_av = zip(*avail)
    stat, p = kruskal(*list(d_av))

    fig, ax = styled_figure(figsize=(12, 7))
    ax.set_title(f"{ind_label} by Combined Profile KW H={stat:.1f} (p={p:.4f})", color="white")

    bp = ax.boxplot(list(d_av) , patch_artist=True, 
                    medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2))
    
    for patch, col in zip(bp["boxes"], c_av):   patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{g}\n(n={len(d)})" for g, d in zip(g_av, d_av)]
    
    ax.set_xticklabels(xlabels, color="white", fontsize=8)
    ax.set_ylabel(ind_label, color="white")
    if do_log: ax.set_yscale("log")
    if not do_log: ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)

    for i, vals in enumerate(d_av):
        med = np.median(vals)
        label = f"${med/1e3:.0f}k" if do_log else f"{med:.1f}%"
        ax.text(i+1, med, label, ha="center", va="bottom", color="black", fontsize=10, fontweight="bold")

    for _, row in df_k3[df_k3["country_name"].isin(countries)].iterrows():
        if row[comb_col] not in g_av or pd.isna(row[ind_col]): continue
        g_idx = list(g_av).index(row[comb_col]) + 1
        ax.scatter(g_idx+0.07, row[ind_col], color=C["orange"], s=100, zorder=6, edgecolors="white", linewidths=1.2)
        ax.annotate(row["country_name"], (g_idx+0.07, row[ind_col]), fontsize=7, color="white", 
                    fontweight="bold", xytext=(10,0), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()
````

**Insight — K.3 Combined Profile × Macro :**

**GDP_pc (H=19.3, p=0.0002) :**
Global Champion ($34k median) is 4× above Competitive/Vulnerable/Fragile
($5–8k). The combined profile is a reliable proxy for economic wealth level.

**CA% (H=14.1, p=0.003) :**
Global Champion (+2.3%) is the only group with positive median CA.
Consistent with J.2 static quadrant (H=47.0) and I.5 balance profile
(η²=0.241) — CA position remains the most reliable [E] discriminator
across every trade classification in the notebook.

**Fragile group heterogeneity :** very wide CA dispersion — mixes deeply
deficit economies with some exceptions. The combined profile conflates
structurally weak AND bilaterally concentrated countries into one group,
creating internal heterogeneity not present in structural profile alone.

**Vs J.1 structural (H=27.3) :** combined profile discriminates less strongly
than structural alone — the bilateral dimension adds noise to the wealth signal
by including bilaterally concentrated rich countries (Canada, Japan) in
lower combined profiles.

**India anomaly confirmed :** classified Global Champion (combined) despite
$2k GDP_pc — the only Global Champion in the low-income range. Size achievement
(structural score) dominates the combined classification.

## K.4 — Strategic Quadrant Combined × Economic Indicators
</div>

**Variable :** `strategic_quadrant_combined`
(Strong & Resilient / Strong but Exposed / Diversified Minor / Weak & Vulnerable)

**Question :** Do Strong & Resilient countries (high structural + high bilateral resilience) show consistently better macro fundamentals than Weak & Vulnerable?

Key distinction from K.3 : this quadrant separates the **structural vs bilateral** tension
— a country can be structurally strong but geographically concentrated (Strong but Exposed).


````python
sqc_col    = "strategic_quadrant_combined"
groups_sqc = ["Strong & Resilient","Strong but Exposed", "Diversified Minor","Weak & Vulnerable"]
colors_sqc = [C["green"],C["orange"],C["blue"],C["red"]]

df_k4 = df_master[["country_name", sqc_col, gdp_pc, ca_col]].dropna().copy()
df_k4[ca_col] = df_k4[ca_col].clip(-30, 30)

for ind_col, ind_label, do_log, fname in [
    (gdp_pc, "GDP per Capita", True,  "K_4_sqc_gdppc"),  (ca_col, "CA % GDP", False, "K_4_sqc_ca")]:
    
    df_tmp = df_k4.dropna(subset=[ind_col])
    data_kw = [df_tmp[df_tmp[sqc_col]==g][ind_col].values for g in groups_sqc]
    avail = [(g,d,c) for g,d,c in zip(groups_sqc,data_kw,colors_sqc) if len(d)>=3]
    if not avail: continue
    g_av, d_av, c_av = zip(*avail)
    stat, p = kruskal(*list(d_av))

    fig, ax = styled_figure(figsize=(12, 7))
    ax.set_title(f" {ind_label} by Strategic Quadrant Combined"  f"\n KW H={stat:.1f} (p={p:.4f})", color="white")

    bp = ax.boxplot(list(d_av), patch_artist=True, medianprops  = dict(color="white", linewidth=2),
                    flierprops   = dict(marker="o", markerfacecolor="white", markeredgecolor="white", markersize=4, alpha=0.6),
                    whiskerprops = dict(color="white", linewidth=1, linestyle="--"),
                    capprops     = dict(color="white", linewidth=1.5),
                    boxprops     = dict(linewidth=1.2)                   )
    for patch, col in zip(bp["boxes"], c_av):     patch.set_facecolor(col); patch.set_alpha(0.7)

    xlabels = [f"{g}(n={len(d)})" for g,d in zip(g_av, d_av)]
    ax.set_xticklabels(xlabels, color="white", fontsize=7)
    ax.set_ylabel(ind_label, color="white")
    if do_log: ax.set_yscale("log")
    if not do_log: ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)

    for i, vals in enumerate(d_av):
        med = np.median(vals)
        label = f"${med/1e3:.0f}k" if do_log else f"{med:.1f}%"
        ax.text(i+1, med, label, ha="center", va="bottom", color="black", fontsize=10, fontweight="bold")

    for _, row in df_k4[df_k4["country_name"].isin(countries)].iterrows():
        if row[sqc_col] not in g_av or pd.isna(row[ind_col]): continue
        g_idx = list(g_av).index(row[sqc_col]) + 1
        ax.scatter(g_idx+0.07, row[ind_col], color=C["orange"], s=100, zorder=6, edgecolors="white", linewidths=1.2)
        ax.annotate(row["country_name"], (g_idx+0.07, row[ind_col]), fontsize=7, color="white", 
                    fontweight="bold", xytext=(10,0), textcoords="offset points",
                    bbox=dict(boxstyle="round, pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()

print("\nSelected countries — strategic quadrant combined :")
df_k4[df_k4["country_name"].isin(countries)][["country_name", sqc_col]]
````

## K.5 — General Observation : Composite Scores × Economic Indicators
</div>

**Two dimensions, two analytical logics :**

**F (structural score) :** explained by macro fundamentals —
GDP_pc (ρ=+0.32), CA% (ρ=+0.30). Consistent throughout G/H/I/J/K.
Combined score inherits these relationships (Δρ ≈ 0).

**G (bilateral resilience) :** not explained by macro — all indicators
non-significant. Geographic concentration of export partnerships is
driven by historical, political and geographic proximity factors
invisible in 16 World Bank indicators.

**Combined score :** inherits F relationships (ρ ≈ F alone) and adds
a bilateral adjustment that redistributes ranks without changing
macro signal strength. The bilateral dimension is analytically orthogonal
to macro fundamentals — it adds information that macro cannot provide.

**Practical implication :** to assess a country's trade position,
two independent analyses are needed — a macro-linked structural analysis
and a geography-driven bilateral analysis. Neither is redundant,
and neither can substitute for the other.

**Central finding for Direction 1 :**
Macro fundamentals explain WHAT countries trade (volume, share, openness),
HOW WELL they trade (structural score, coverage, balance) but NOT
WITH WHOM they trade (bilateral concentration, partner dependency).
The geographic dimension of trade requires partner-level data
beyond standard macro indicators.

**Rank adjustment finding :**
Canada's collapse (#5 → #48) and India's rise (#19 → #10) after bilateral
adjustment reveal that structural strength and geographic resilience are
independent — and that the most structurally impressive economies
(Canada, Japan) can simultaneously be the most geopolitically exposed.
This tension is the key insight of K and a natural entry point for L.

# L — Macro Fingerprint × Trade Profile
</div>

**Objective :** For each trade profile, what is the typical macro profile?
Does a specific combination of macro conditions correspond to a specific trade profile?

**Approach :**
1. Compute median + IQR of 5 macro indicators per trade profile group
2. Assess internal cohesion (low IQR = cohesive, high IQR = dispersed)
3. Choose visualization based on results :
   - ≥3 indicators distinct across profiles → **radar**
   - 1-2 indicators distinct → **grouped bar chart or heatmap**
   - Quasi-identical profiles → **finding only**

**Indicators :** GDP_pc | CA% | Real Rate | Expense% | Inflation

**Trade profiles analyzed :**
- `structural_profile` : Weak / Moderate / Good / Strong
- `profile_full_periode` : Robust / Stable but Concentrated / Vulnerable / Critical Dependency
- `meta_profile` : Global Leader / Established Power / Rising Challenger / Volatile Emerging / Transitional / Peripheral


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## L.1 — Median profiles & internal cohesion
</div>

````python
gdp_pc   = "GDP per capita (current US$)"
ca_col   = "Current account balance (% of GDP)"
rate_col = "Real interest rate (%)"
exp_col  = "Expense (% of GDP)"
infl_col = "Inflation, consumer prices (annual %)"

INDICATORS = {"GDP_pc": gdp_pc, "CA%": ca_col, "Real_Rate": rate_col, "Expense%" : exp_col, "Inflation": infl_col}

# Clip extremes
df_l = df_master.copy()
df_l[ca_col]   = df_l[ca_col].clip(-30, 30)
df_l[rate_col] = df_l[rate_col].clip(-20, 30)
df_l[infl_col] = df_l[infl_col].clip(-5, 30)

# Min-max normalize each indicator for radar comparability
scaler = MinMaxScaler()
ind_cols = list(INDICATORS.values())
df_l[[f"norm_{k}" for k in INDICATORS]] = scaler.fit_transform(df_l[ind_cols].fillna(df_l[ind_cols].median()))
norm_cols = {k: f"norm_{k}" for k in INDICATORS}

PROFILES = {
"structural_profile" : ["Weak","Moderate","Good","Strong"],
"profile_full_periode": ["Robust","Stable but Concentrated", "Vulnerable","Critical Dependency"],
"meta_profile": ["Global Leader","Established Power", "Rising Challenger", "Volatile Emerging", "Transitional","Peripheral"]
            }

for prof_col, groups in PROFILES.items():
    print(f"\n{'='*65}")
    print(f"  {prof_col}")
    print(f"{'='*65}")
    print(f"  {'Group':<30} {'GDP_pc':>8} {'CA%':>8} {'Rate':>8} {'Exp%':>8} {'Infl':>8}  {'Cohesion'}")
    print(f"  {'-'*30} {'-'*8} {'-'*8} {'-'*8} {'-'*8} {'-'*8}  {'-'*10}")

    profile_data = {}
    for g in groups:
        sub = df_l[df_l[prof_col]==g]
        if len(sub) < 3: continue
        medians = {k: sub[v].median() for k,v in norm_cols.items()}
        iqrs    = {k: iqr(sub[v].dropna()) for k,v in norm_cols.items()}
        avg_iqr = np.mean(list(iqrs.values()))
        cohesion = "High" if avg_iqr < 0.25 else "Medium" if avg_iqr < 0.40 else "Low"
        profile_data[g] = {"medians": medians, "iqrs": iqrs, "avg_iqr": avg_iqr, "n": len(sub)}
        print(f"  {g:<30} "
              f"{medians['GDP_pc']:>8.2f} {medians['CA%']:>8.2f} "
              f"{medians['Real_Rate']:>8.2f} {medians['Expense%']:>8.2f} "
              f"{medians['Inflation']:>8.2f}  {cohesion} ({avg_iqr:.2f})")

    # How many indicators discriminate across groups ?
    n_distinct = 0
    for k in INDICATORS:
        vals = [profile_data[g]["medians"][k] for g in profile_data if g in profile_data]
        if len(vals) == 0:
            print(f"  [SKIP] No data for this group")
            continue

        if max(vals) - min(vals) > 0.15:
            n_distinct += 1
        
    recommendation = ("→ RADAR" if n_distinct >= 3 else "→ BAR/HEATMAP" if n_distinct >= 1 else "→ FINDING ONLY")
    print(f"\n  Distinct indicators (range >0.15) : {n_distinct}/5  {recommendation}")
````

## L.2 — Visualization
</div>

````python
def plot_radar(df_plot, prof_col, groups, title, colors_dict, fname):
    """Radar chart for cohesive multi-indicator profiles"""
    labels = list(INDICATORS.keys())
    N = len(labels)
    angles = [n / float(N) * 2 * np.pi for n in range(N)]
    angles += angles[:1]

    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True), facecolor="#0f3460")
    ax.set_facecolor("#16213e")
    ax.spines["polar"].set_color("#ffffff30")
    ax.set_title(title, color="white", fontsize=13, pad=20)

    for g in groups:
        sub = df_plot[df_plot[prof_col]==g]
        if len(sub) < 3: continue
        values = [sub[v].median() for v in norm_cols.values()]
        values += values[:1]
        col = colors_dict.get(g, "white")
        ax.plot(angles, values, color=col, linewidth=2.5, label=f"{g} (n={len(sub)})")
        ax.fill(angles, values, color=col, alpha=0.12)

    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels, color="white", fontsize=10)
    ax.set_ylim(0, 1)
    ax.set_yticks([0.25, 0.50, 0.75])
    ax.set_yticklabels(["0.25","0.50","0.75"], color="white", fontsize=8)
    ax.tick_params(colors="white")
    ax.grid(color="white", alpha=0.2)
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=9, loc="upper right", bbox_to_anchor=(1.3, 1.1))
    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()


def plot_heatmap(df_plot, prof_col, groups, title, fname):
    """Heatmap for partially distinct profiles"""
    data = []
    for g in groups:
        sub = df_plot[df_plot[prof_col]==g]
        if len(sub) < 3: continue
        row = [sub[v].median() for v in norm_cols.values()]
        data.append(row)

    labels_g = [g for g in groups if len(df_plot[df_plot[prof_col]==g]) >= 3]
    df_heat = pd.DataFrame(data, index=labels_g, columns=list(INDICATORS.keys()))

    fig, ax = plt.subplots(figsize=(10, len(labels_g)*1.2 + 2), facecolor="#0f3460")
    ax.set_facecolor("#16213e")
    im = ax.imshow(df_heat.values, cmap="RdYlGn", vmin=0, vmax=1, aspect="auto")
    ax.set_xticks(range(len(INDICATORS)))
    ax.set_xticklabels(list(INDICATORS.keys()), color="white", fontsize=10)
    ax.set_yticks(range(len(labels_g)))
    ax.set_yticklabels(labels_g, color="white", fontsize=9)
    ax.set_title(title, color="white", fontsize=12, pad=15)

    for i in range(len(labels_g)):
        for j in range(len(INDICATORS)):
            ax.text(j, i, f"{df_heat.values[i,j]:.2f}", ha="center", va="center", color="black", fontsize=9, fontweight="bold")

    plt.colorbar(im, ax=ax, label="Normalized median")
    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()
````

## L.3 — Finding : No Multi-Indicator Macro Fingerprint
</div>

**Objective :** For each trade profile, identify a distinctive combination
of macro conditions — a "macro fingerprint" that characterizes the profile.

**Method :** Median of 5 normalized indicators (GDP_pc, CA%, Real Rate,
Expense%, Inflation) per trade profile group. Cohesion measured by average IQR.
Threshold : range >0.15 across groups = discriminating indicator.

**Results across all profiles tested :**

| Profile | Distinct indicators | Decision |
|---|---|---|
| structural_profile | 1/5 (GDP_pc only) | Heatmap |
| profile_full_periode | 0/5 | Finding only |
| meta_profile | 1/5 (GDP_pc only) | Heatmap |
| combined_profile | 1/5 (GDP_pc only) | Finding only |
| size_profile | 1/5 (GDP_pc only) | Heatmap |
| balance_profile | 0/5 | Finding only |
| growth_profile | 0/5 | Finding only |
| stability_profile | 0/5 | Finding only |
| consistency_profile | 0/5 | Finding only |

**Finding :** No trade profile shows a distinctive multi-indicator
macro fingerprint. GDP_pc is the only indicator that consistently
discriminates across profiles — and only for the top group
(Strong, Global Leader, Global Champion, Leader).

**Why the radar was not justified :**
All median profiles converge toward [0.44-0.55] on CA%, Real Rate,
Expense% and Inflation — the min-max normalization over the full
distribution compresses inter-group differences. Even indicators
known to be statistically significant in KW tests (CA% H=47.0 in J.2)
do not produce divergent medians at group level.

**Limits and perspective :**

1. **Normalization effect :** Min-max on full distribution may mask
   real group differences. A within-group standardization or
   a different scaling could reveal more distinct profiles.

2. **Missing indicators :** The 5 World Bank indicators used here
   may not capture the dimensions that truly differentiate trade profiles.
   Exchange rates, trade policy indices, institutional quality,
   infrastructure scores — absent from this dataset — could produce
   more distinctive fingerprints.

3. **Group heterogeneity :** All groups show High cohesion (IQR < 0.20)
   internally — but this means countries within the same trade profile
   are macro-homogeneous among themselves, not that profiles differ
   from each other. The macro space is dense — most countries occupy
   similar macro positions regardless of trade profile.

4. **Implication for Direction 1 :**
   Trade profiles are not determined by a combination of macro conditions.
   Individual indicators (GDP_pc, CA%) have predictive power in isolation
   (G/H/I/J/K findings) but do not combine into a distinctive multi-dimensional
   macro signature for any trade classification. The relationship between
   macro fundamentals and trade structure is simpler than expected —
   driven primarily by a single dimension (economic size/wealth)
   rather than by a complex interplay of macro conditions.

# M — Multi-indicator Prediction
</div>

**Objective :** Can macro fundamentals predict trade performance when combined?

**Input :** 6 macro indicators [E]
GDP_pc | CA% | Real Rate | Expense% | Inflation | Unemployment

**Targets :** 4 trade variables
structural_score | export_share | coverage_ratio | combined_score

**Models :** Random Forest Regression + OLS Multiple Regression

**Questions :**
1. Which indicator explains each target best? → RF feature importance
2. Which target is most predictable from macro fundamentals? → R² comparison
3. Who over/under-performs vs macro prediction? → Residual analysis (all 4 targets)


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

## M.1 — Model Training & Results Summary
</div>

````python
FEATURES = {
    "GDP_pc"      : "GDP per capita (current US$)",
    "CA%"         : "Current account balance (% of GDP)",
    "Real_Rate"   : "Real interest rate (%)",
    "Expense%"    : "Expense (% of GDP)",
    "Inflation"   : "Inflation, consumer prices (annual %)",
    "Unemployment": "Unemployment, total (% of total labor force) (modeled ILO estimate)",
}

TARGETS = {
    "coverage_ratio_structural": "Coverage Ratio",
    "export_share_structural"  : "Export Share",
    "structural_score"         : "Structural Score",
    "combined_score"           : "Combined Score",
}

feat_cols   = list(FEATURES.values())
feat_labels = list(FEATURES.keys())

results = {}

for t_col, t_label in TARGETS.items():
    df_m = df_master[["country_name","size_profile"] + feat_cols + [t_col]].dropna().copy()
    df_m["Current account balance (% of GDP)"]         = df_m["Current account balance (% of GDP)"].clip(-30,30)
    df_m["Real interest rate (%)"]                     = df_m["Real interest rate (%)"].clip(-20,30)
    df_m["Inflation, consumer prices (annual %)"]      = df_m["Inflation, consumer prices (annual %)"].clip(-5,30)

    X = df_m[feat_cols].values
    y = df_m[t_col].values

    scaler = StandardScaler()
    X_sc   = scaler.fit_transform(X)

    rf  = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
    rf.fit(X_sc, y)
    r2_rf  = r2_score(y, rf.predict(X_sc))
    cv_r2  = cross_val_score(rf, X_sc, y, cv=5, scoring="r2")

    ols = LinearRegression()
    ols.fit(X_sc, y)
    r2_ols = r2_score(y, ols.predict(X_sc))

    df_m["predicted"] = ols.predict(X_sc)
    df_m["residual"]  = df_m[t_col] - df_m["predicted"]

    results[t_col] = {
        "label"  : t_label,
        "r2_rf"  : r2_rf,
        "r2_ols" : r2_ols,
        "gap"    : r2_rf - r2_ols,
        "cv_mean": cv_r2.mean(),
        "cv_std" : cv_r2.std(),
        "fi"     : dict(zip(feat_labels, rf.feature_importances_)),
        "coef"   : dict(zip(feat_labels, ols.coef_)),
        "df"     : df_m,
        "n"      : len(df_m),
    }

print("M.1 — Model Results Summary :")
print("="*75)
print(f"  {'Target':<22} {'RF R²':>7} {'CV R²':>12} {'OLS R²':>8} {'Gap':>7} {'n':>5}")
print(f"  {'-'*22} {'-'*7} {'-'*12} {'-'*8} {'-'*7} {'-'*5}")
for t_col, res in results.items():
    gap_flag = "🔴" if res["gap"] > 0.20 else "🟡" if res["gap"] > 0.10 else "🟢"
    print(f"  {res['label']:<22} {res['r2_rf']:>7.3f} "
          f"{res['cv_mean']:>6.3f}±{res['cv_std']:.2f} "
          f"{res['r2_ols']:>8.3f} {res['gap']:>7.3f} {gap_flag} {res['n']:>5}")

print("\nFeature Importance (RF) — sorted by Coverage Ratio R² :")
print("="*75)
print(f"  {'Indicator':<14}", end="")
for t_col, res in results.items():
    print(f"  {res['label'][:13]:>13}", end="")
print()
for feat in feat_labels:
    print(f"  {feat:<14}", end="")
    for t_col, res in results.items():
        print(f"  {res['fi'][feat]:>13.3f}", end="")
    print()
````

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

## M.2 — Visualization : R² & Feature Importance
</div>

**Note :** Direction 2 relationships are often bidirectional — causal interpretation

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

## N.1 — Trade Structure → GDP Size
</div>

**Key stats :**
Import Share ρ=+0.959 | Export Share ρ=+0.937 | Combined Score ρ=+0.718
All significant — but largely tautological.

**Questions :**
1. Does trade structure determine economic size or is it the reverse?
2. Is the trade → GDP relationship symmetric with GDP → trade (G.1 ρ=0.94)?
3. Which trade variable has the strongest link to GDP size?


````python
from scipy.stats import spearmanr

gdp_col = "GDP (current US$)"
trade_vars = {
                "Export Share"   : "export_share_structural",
                "Import Share"   : "import_share_structural",
                "Structural Score": "structural_score",
                "Combined Score" : "combined_score",
                "Coverage Ratio" : "coverage_ratio_structural",
                "Trade Intensity": "trade_intensity",
              }

print("N.1 — Trade variables × GDP (Direction 2)")
print("="*55)
df_n1 = df_master[["country_name", gdp_col] +  list(trade_vars.values())].dropna().copy()

for label, col in trade_vars.items():
    r_sp, p_sp = spearmanr(np.log10(df_n1[gdp_col].clip(lower=1)), df_n1[col].values)
    sig = "✅" if p_sp < 0.05 else "❌"
    print(f"  {sig} ρ={r_sp:+.3f} (p={p_sp:.3f})  {label}")

print(f"\n  Comparison with Direction 1 :")
print(f"  G.1 : GDP → Export Share ρ=+0.94")
print(f"  N.1 : Export Share → GDP ρ=+0.94 (symmetric — same correlation)")
print(f"  → Spearman is non-directional. Cannot distinguish cause from effect.")
print(f"  → The relationship is bidirectional by construction.")
````

**Insight — N.1 :**

**Answer :** All trade variables correlate strongly with GDP size
(ρ=0.165 to 0.959). But this is mathematically symmetric with
Direction 1 — Spearman cannot distinguish cause from effect.

**Key finding :** The trade ↔ GDP relationship is bidirectional
by construction. A country is large because it trades, AND it
trades because it is large. Panel data with lagged variables
would be required to establish causal direction.

**Hierarchy :** Import Share (ρ=0.959) > Export Share (0.937) >
Combined Score (0.718) > Structural Score (0.650) → imports
slightly more correlated with GDP than exports, consistent with H.1.

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

## N.2 — Trade Structure → GDP per Capita
</div>

**Key stats :**
Export per Capita ρ=+0.823 | Import Share ρ=+0.447 | Export Share ρ=+0.441
Structural Score ρ=+0.330 | Trade Intensity ρ=+0.295 | Coverage Ratio ρ=+0.234

**Questions :**
1. Do countries that export more per person become wealthier?
2. Does trade openness (Trade Intensity) drive individual wealth?
3. Does structural trade performance predict wealth level?"*
4. Does better trade balance (Coverage Ratio) translate into higher wealth?
5. Is export_pc → GDP_pc stronger or weaker than GDP_pc → export_pc (G.1 P2 ρ=0.82)?


````python
from scipy.stats import linregress

gdp_pc_col = "GDP per capita (current US$)"
plot_configs = [
                ("export_pc",              "Export per Capita (log)",  True,  "N_2_exportpc_vs_gdppc"),
                ("trade_intensity",        "Trade Intensity",           False, "N_2_tradeintensity_vs_gdppc"),
                ("structural_score",       "Structural Score",          False, "N_2_structuralscore_vs_gdppc"),
                ]

for col, x_label, do_log, fname in plot_configs:
    df_n2 = df_master[["country_name", gdp_pc_col, col, "size_profile"]].dropna().copy()
    x_vals = np.log10(df_n2[col].clip(lower=1)) if do_log else df_n2[col]
    r_sp, p_sp = spearmanr(x_vals.values, np.log10(df_n2[gdp_pc_col].clip(lower=1)).values)
    slope, intercept, _, _, _ = linregress(x_vals, np.log10(df_n2[gdp_pc_col].clip(lower=1)))
    x_line = np.linspace(x_vals.min(), x_vals.max(), 100)

    fig, ax = styled_figure(figsize=(13, 8))
    ax.set_title(f"N.2 — {x_label} × GDP per Capita\n", color="white")

    for sp, c in SIZE_COLORS.items():
        sub = df_n2[df_n2["size_profile"]==sp]
        ax.scatter(sub[col], sub[gdp_pc_col], color=c, alpha=0.5, s=40, label=sp)

    ax.plot(10**x_line if do_log else x_line, 10**(intercept + slope * x_line),
            color="white", linewidth=1.5, linestyle="--", alpha=0.6, label=f"Trend (ρ={r_sp:.2f})")

    for _, row in df_n2[df_n2["country_name"].isin(countries)].iterrows():
        ax.scatter(row[col], row[gdp_pc_col], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
                   edgecolors="white", linewidths=1.5)
        ax.annotate(row["country_name"], (row[col], row[gdp_pc_col]),
                    fontsize=9, color="white", fontweight="bold", xytext=(8,4), textcoords="offset points",
                    bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

    if do_log: ax.set_xscale("log")
    ax.set_yscale("log")
    ax.set_xlabel(x_label, color="white")
    ax.set_ylabel("GDP per Capita (US$) — log", color="white")
    ax.text(0.5, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})",
            transform=ax.transAxes, color="white", fontsize=11, fontweight="bold")
    ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
    plt.tight_layout()
    plt.savefig(f"../Figures/Link/{fname}.png", dpi=150, bbox_inches="tight")
    plt.show()

print("Symmetric comparison :")
print("  Direction 1 : GDP_pc → Export_pc  ρ=+0.823 (G.1 P2)")
print("  Direction 2 : Export_pc → GDP_pc  ρ=+0.823 (N.2)")
print("  → Symmetric — same ρ by construction (Spearman non-directional)")
print("  → Both directions equally strong — true bidirectional relationship")
````

**Insight — N.2 :**

**Export_pc → GDP_pc (ρ=0.823) : strongest signal of Direction 2.**
Export per capita is the best trade proxy for individual wealth —
stronger than world share (biased by demographics) and structural
score (aggregates heterogeneous dimensions).

**China — most instructive case across all 3 panels :**
Systematically below the regression line — GDP_pc ($10k) lower than what
export_pc or structural score would predict. China achieves world trade
dominance through volume, not through individual wealth creation.
This is the clearest illustration that export power ≠ individual prosperity.

**Trade Intensity (ρ=0.29) — fan distribution :**
Very open economies can remain poor (small countries dependent on
one export sector) and large closed economies can be rich (USA, Japan).
Openness alone does not predict wealth.

**Symmetric finding :** Direction 1 (GDP_pc → export_pc ρ=0.823)
= Direction 2 (export_pc → GDP_pc ρ=0.823). The relationship is
equally strong in both directions — true bidirectional co-determination.

## N.3 — Trade Structure → Unemployment
</div>

**Key stats :**
Balance/GDP ρ=−0.191 ✅ | All other trade variables non-significant

**Questions :**
1. Does a positive trade balance reduce unemployment?
2. Does trade openness protect against unemployment?
3. Does export performance reduce unemployment pressure?


````python
unemp_col = "Unemployment, total (% of total labor force) (modeled ILO estimate)"
bal_col   = "balance_dep"
print("N.3 — Trade variables × Unemployment (Direction 2)")
print("="*55)
trade_vars_n3 = {
                "Balance / GDP"  : "balance_dep",
                "Export Share"   : "export_share_structural",
                "Trade Intensity": "trade_intensity",
                "Structural Score": "structural_score",
                "Coverage Ratio" : "coverage_ratio_structural"}

for label, col in trade_vars_n3.items():
    df_tmp = df_master[[unemp_col, col]].dropna()
    r_sp, p_sp = spearmanr(df_tmp[col].values, df_tmp[unemp_col].values)
    sig = "✅" if p_sp < 0.05 else "❌"
    print(f"  {sig} ρ={r_sp:+.3f} (p={p_sp:.3f})  {label}")

# Figure for Balance/GDP only (only significant signal)
df_n3 = df_master[["country_name", unemp_col, "balance_dep", "size_profile"]].dropna().copy()
df_n3[unemp_col] = df_n3[unemp_col].clip(0, 30)

from scipy.stats import linregress
r_sp, p_sp = spearmanr(df_n3["balance_dep"].values, df_n3[unemp_col].values)
slope, intercept, _, _, _ = linregress(df_n3["balance_dep"], df_n3[unemp_col])
x_line = np.linspace(df_n3["balance_dep"].min(), df_n3["balance_dep"].max(), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Does a positive trade balance reduce unemployment?" "\n Unemployment % vs Balance / GDP", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_n3[df_n3["size_profile"]==sp]
    ax.scatter(sub["balance_dep"], sub[unemp_col], color=col, alpha=0.5, s=40, label=sp)

ax.plot(x_line, intercept + slope * x_line, color="white", linewidth=1.5, linestyle="--", alpha=0.6)
ax.axhline(0, color=C["grey"], linestyle=":", alpha=0.4)
ax.axvline(0, color=C["grey"], linestyle=":", alpha=0.4)

for _, row in df_n3[df_n3["country_name"].isin(countries)].iterrows():
    ax.scatter(row["balance_dep"], row[unemp_col], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row["balance_dep"], row[unemp_col]),
                fontsize=9, color="white", fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xlabel("Balance / GDP", color="white")
ax.set_ylabel("Unemployment %", color="white")
ax.set_xlim (-7.5, 5)
ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/N_3_balance_vs_unemployment.png", dpi=150, bbox_inches="tight")
plt.show()
````

**Insight — N.3 :**

**Answer :** Weak but significant signal — Balance/GDP ρ=−0.191 (p=0.012).
A positive trade balance is associated with lower unemployment,
but the effect is modest and non-monotone.

**Surplus + low unemployment cluster :** Germany (`~3.5%`), Japan (`~2.8%`),
China (`~4.5%`) — all structural surplus countries. But causality is
likely reverse : high productivity simultaneously generates surplus
AND reduces unemployment. The trade → unemployment link is a
co-symptom, not a direct effect.

**Brazil anomaly :** slight surplus (+0.2%) but high unemployment (~12%).
A commodity surplus (agriculture, mining) creates little industrial
employment — breaks the expected correlation.

**Trade openness (Trade Intensity ρ=+0.084 non-sig) :**
Does not protect against unemployment. Being open to trade does not
mechanically reduce labor market pressure — consistent with H.4
finding that import growth is not driven by unemployment level.

## N.4 — Trade Structure → Inflation
</div>

**Key stats :**
Export per Capita ρ=−0.324 ✅ | Trade Intensity ρ=−0.146 ❌
All others non-significant

**Questions :**
1. Do higher per-capita exporters have lower inflation?
2. Does trade openness dampen price pressures?
3. Does structural trade strength correlate with price stability?
4. Is the trade → inflation link stronger than inflation → trade (H.3 ρ=−0.45)?


````python
infl_col  = "Inflation, consumer prices (annual %)"
exp_pc    = "export_pc"
ti_col    = "trade_intensity"

print("N.4 — Trade variables × Inflation (Direction 2)")
print("="*55)
trade_vars_n4 = {
                    "Export per Capita" : exp_pc,
                    "Trade Intensity"   : ti_col,
                    "Structural Score"  : "structural_score",
                    "Export Share"      : "export_share_structural",
                    "Coverage Ratio"    : "coverage_ratio_structural",
                }
for label, col in trade_vars_n4.items():
    df_tmp = df_master[[infl_col, col]].dropna()
    df_tmp[infl_col] = df_tmp[infl_col].clip(-5, 30)
    r_sp, p_sp = spearmanr(df_tmp[col].values, df_tmp[infl_col].values)
    sig = "✅" if p_sp < 0.05 else "❌"
    print(f"  {sig} ρ={r_sp:+.3f} (p={p_sp:.3f})  {label}")

# Figure : Export_pc × Inflation (strongest signal)
df_n4 = df_master[["country_name", infl_col, exp_pc, "size_profile"]].dropna().copy()
df_n4[infl_col] = df_n4[infl_col].clip(-5, 30)

from scipy.stats import linregress
r_sp, p_sp = spearmanr(np.log10(df_n4[exp_pc].clip(lower=1)).values, df_n4[infl_col].values)
slope, intercept, _, _, _ = linregress(np.log10(df_n4[exp_pc].clip(lower=1)), df_n4[infl_col])
x_log = np.linspace(np.log10(df_n4[exp_pc].min()), np.log10(df_n4[exp_pc].max()), 100)

fig, ax = styled_figure(figsize=(13, 8))
ax.set_title("Do higher per-capita exporters have lower inflation?" "\n Inflation % vs Export per Capita (log)", color="white")

for sp, col in SIZE_COLORS.items():
    sub = df_n4[df_n4["size_profile"]==sp]
    ax.scatter(sub[exp_pc], sub[infl_col], color=col, alpha=0.5, s=40, label=sp)

ax.plot(10**x_log, intercept + slope * x_log, color="white", linewidth=1.5, linestyle="--", alpha=0.6)

for _, row in df_n4[df_n4["country_name"].isin(countries)].iterrows():
    ax.scatter(row[exp_pc], row[infl_col], color=SIZE_COLORS[row["size_profile"]], s=150, zorder=6,
               edgecolors="white", linewidths=1.5)
    ax.annotate(row["country_name"], (row[exp_pc], row[infl_col]),
                fontsize=9, color="white", fontweight="bold", xytext=(8,4), textcoords="offset points",
                bbox=dict(boxstyle="round,pad=0.2", facecolor=C["dark"], alpha=0.7, edgecolor="white", linewidth=0.5))

ax.set_xscale("log")
ax.set_ylim(-5, 20)
ax.set_xlabel("Export per Capita (log)", color="white")
ax.set_ylabel("Inflation % (CPI)", color="white")
ax.text(0.05, 0.95, f"Spearman ρ={r_sp:.2f} (p={p_sp:.3f})", transform=ax.transAxes, color="white", fontsize=9)
ax.legend(facecolor="#16213e", labelcolor="white", fontsize=8)
plt.tight_layout()
plt.savefig("../Figures/Link/N_4_exportpc_vs_inflation.png", dpi=150, bbox_inches="tight")
plt.show()

print(f"\nSymmetric comparison :")
print(f"  Direction 1 : Inflation → Import_pc  ρ=−0.45 (H.3)")
print(f"  Direction 2 : Export_pc → Inflation  ρ=−0.32 (N.4)")
print(f"  → Direction 1 signal stronger : inflation constrains imports")
print(f"    more than exports predict inflation")
````

**Insight — N.4 :**

**Answer :** Moderate signal — Export_pc ρ=−0.323 (p<0.001).
Higher per-capita exporters tend to have lower inflation.
Two valid causal readings :

**Reading 1 (trade → inflation) :** Strong exporters have access
to diversified imported goods, exercising competitive pressure
on domestic prices. International competition disciplines inflation.

**Reading 2 (macro stability → export) :** Low-inflation countries
have credible monetary policy, stable exchange rates and long-term
industrial planning — conditions that favor competitive export development.

**Asymmetry vs Direction 1 :** H.3 showed inflation → import_pc ρ=−0.45
(stronger than N.4 ρ=−0.32). Inflation suppresses imports more directly
than it stimulates exports — the demand channel is stronger than the
supply channel.

**Cluster high inflation + low export_pc :** Mid countries with
inflation >15–30% — monetary instability erodes export competitiveness.
The relationship is direct here : volatile exchange rates destroy
exporter margins.

**Japan :** anti-thesis — highest export_pc in the group with near-zero
inflation. Chronic deflation + monetary discipline + export efficiency
coexist — but Japan's case shows this combination is not universally
replicable.


<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

## N.5 — Trade Profile → Macro Profile (L.1 Inverse)
</div>

**Objective :** For each macro indicator, what is the typical trade profile?
Mirror of L.1 : instead of macro → trade fingerprint, we test trade → macro fingerprint.

**Method :** Same Step 1 approach as L.1 :
- Compute median of macro indicators per trade profile group
- Assess if profiles diverge on ≥3 indicators → radar / heatmap / finding


````python
MACRO_INDICATORS = {
                    "GDP_pc"      : "GDP per capita (current US$)",
                    "CA%"         : "Current account balance (% of GDP)",
                    "Real_Rate"   : "Real interest rate (%)",
                    "Expense%"    : "Expense (% of GDP)",
                    "Inflation"   : "Inflation, consumer prices (annual %)",
                    }

df_n5 = df_master.copy()
df_n5["Current account balance (% of GDP)"]    = df_n5["Current account balance (% of GDP)"].clip(-30,30)
df_n5["Real interest rate (%)"]                = df_n5["Real interest rate (%)"].clip(-20,30)
df_n5["Inflation, consumer prices (annual %)"] = df_n5["Inflation, consumer prices (annual %)"].clip(-5,30)

ind_cols = list(MACRO_INDICATORS.values())
scaler   = MinMaxScaler()
df_n5[[f"norm_{k}" for k in MACRO_INDICATORS]] = scaler.fit_transform(
    df_n5[ind_cols].fillna(df_n5[ind_cols].median()))
norm_cols = {k: f"norm_{k}" for k in MACRO_INDICATORS}

TRADE_PROFILES = {
"meta_profile": ["Global Leader","Established Power","Rising Challenger", "Volatile Emerging","Transitional","Peripheral"],
"combined_profile": ["Global Champion","Competitive", "Vulnerable","Fragile"],
"structural_profile": ["Weak","Moderate","Good","Strong"],
                   }

for prof_col, groups in TRADE_PROFILES.items():
    print(f"\n{'='*65}")
    print(f"  Trade profile : {prof_col}")
    print(f"{'='*65}")
    print(f"  {'Group':<28} {'GDP_pc':>7} {'CA%':>7} {'Rate':>7} {'Exp%':>7} {'Infl':>7}  Cohesion")

    profile_data = {}
    for g in groups:
        sub = df_n5[df_n5[prof_col]==g]
        if len(sub) < 3: continue
        medians = {k: sub[v].median() for k,v in norm_cols.items()}
        avg_iqr = np.mean([iqr(sub[v].dropna()) for v in norm_cols.values()])
        cohesion = "High" if avg_iqr < 0.25 else "Medium" if avg_iqr < 0.40 else "Low"
        profile_data[g] = medians
        print(f"  {g:<28} "
              f"{medians['GDP_pc']:>7.2f} {medians['CA%']:>7.2f} "
              f"{medians['Real_Rate']:>7.2f} {medians['Expense%']:>7.2f} "
              f"{medians['Inflation']:>7.2f}  {cohesion} ({avg_iqr:.2f})")

    # Count distinct indicators
    n_distinct = sum(
        1 for k in MACRO_INDICATORS
        if max(v[k] for v in profile_data.values()) - min(v[k] for v in profile_data.values()) > 0.15)
    rec = ("→ RADAR" if n_distinct >= 3 else "→ HEATMAP" if n_distinct >= 1 else "→ FINDING ONLY")
    print(f"\n  Distinct indicators : {n_distinct}/5  {rec}")
````

**Insight — N.5 :**

Same result as L.1 — GDP_pc is the only indicator that discriminates
across trade profiles (1/5 distinct for all profiles).

**Inverse confirmation of L.1 :**
Whether we group by macro profile → look at trade (L.1)
or group by trade profile → look at macro (N.5),
the result is identical : only GDP_pc creates systematic divergence.

**Finding :** The macro fingerprint of trade profiles is uni-dimensional —
GDP_pc is the only systematic signal. Trade profiles cannot be characterized
by a multi-indicator macro combination. This confirms L.1 from the inverse direction.

<div style="background-color:#fdedec;padding-left:20px;border:1px solid;border-radius:5px;">

## N.6 — Prediction Inverse (RF) : Trade → Macro
</div>

**Objective :** Can trade structure predict macro fundamentals?

**Features (trade variables) :**
export_share | structural_score | coverage_ratio | trade_intensity | combined_score

**Targets (macro) :**
GDP per Capita | Unemployment % | Inflation %

**Same method as M — RF + OLS + CV 5-fold**


````python
TRADE_FEATURES = {
                    "Export_Share"   : "export_share_structural",
                    "Structural_Score": "structural_score",
                    "Coverage_Ratio" : "coverage_ratio_structural",
                    "Trade_Intensity": "trade_intensity",
                    "Combined_Score" : "combined_score",
                  }

MACRO_TARGETS = {
                    "GDP per capita (current US$)"                                        : "GDP per Capita",
                    "Unemployment, total (% of total labor force) (modeled ILO estimate)" : "Unemployment %",
                    "Inflation, consumer prices (annual %)"                               : "Inflation %",
                 }

feat_cols   = list(TRADE_FEATURES.values())
feat_labels = list(TRADE_FEATURES.keys())

results_n6 = {}

for t_col, t_label in MACRO_TARGETS.items():
    df_n6 = df_master[["country_name"] + feat_cols + [t_col]].dropna().copy()
    if t_col == "Inflation, consumer prices (annual %)":  df_n6[t_col] = df_n6[t_col].clip(-5, 30)

    X  = df_n6[feat_cols].values
    y  = np.log10(df_n6[t_col].clip(lower=0.1)) if "GDP" in t_col else df_n6[t_col].values

    scaler = StandardScaler()
    X_sc   = scaler.fit_transform(X)

    rf  = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
    rf.fit(X_sc, y)
    r2_rf = r2_score(y, rf.predict(X_sc))
    cv_r2 = cross_val_score(rf, X_sc, y, cv=5, scoring="r2")

    ols = LinearRegression()
    ols.fit(X_sc, y)
    r2_ols = r2_score(y, ols.predict(X_sc))

    results_n6[t_col] = {
                            "label"  : t_label,
                            "r2_rf"  : r2_rf,
                            "r2_ols" : r2_ols,
                            "gap"    : r2_rf - r2_ols,
                            "cv_mean": cv_r2.mean(),
                            "cv_std" : cv_r2.std(),
                            "fi"     : dict(zip(feat_labels, rf.feature_importances_)),
                            "n"      : len(df_n6),
                        }

print("N.6 — Prediction Inverse : Trade → Macro")
print("="*70)
print(f"  {'Target':<20} {'RF R²':>7} {'CV R²':>12} {'OLS R²':>8} {'Gap':>7} {'n':>5}")
print(f"  {'-'*20} {'-'*7} {'-'*12} {'-'*8} {'-'*7} {'-'*5}")
for t_col, res in results_n6.items():
    gap_flag = "🔴" if res["gap"]>0.20 else "🟡" if res["gap"]>0.10 else "🟢"
    print(f"  {res['label']:<20} {res['r2_rf']:>7.3f} "
          f"{res['cv_mean']:>6.3f}±{res['cv_std']:.2f} "
          f"{res['r2_ols']:>8.3f} {res['gap']:>7.3f} {gap_flag} {res['n']:>5}")

print("\nFeature Importance (RF) :")
print(f"  {'Feature':<18}", end="")
for res in results_n6.values():   print(f"  {res['label'][:12]:>12}", end="")
print()

for feat in feat_labels:
    print(f"  {feat:<18}", end="")
    for res in results_n6.values():
        print(f"  {res['fi'][feat]:>12.3f}", end="")
    print()

print("\nComparison with Direction 1 (M) :")
print("  Direction 1 : Macro → Trade  OLS R² = 0.12–0.28")
print("  Direction 2 : Trade → Macro  OLS R² = ??????  (see above)")
````

**Insight — N.6 :**

**CV R² reveals overfitting — key result :**
- GDP_pc : CV R²=0.027 (RF R²=0.888) → massive overfitting
- Unemployment : CV R²=−0.011 → RF cannot generalize at all
- Inflation : CV R²=−0.487 → worse than a constant predictor

**Conclusion :** Trade variables do NOT reliably predict macro outcomes
out-of-sample. The high RF R² values are artifacts of overfitting
on n=171–184 with only 5 features. CV R² is the reliable metric —
and it shows near-zero or negative generalization for all 3 targets.
Direction 2 prediction is not feasible with this dataset structure.

**Contrast with Direction 1 (M) :**
Direction 1 OLS R²=0.12–0.28 → weak but genuine linear signal
Direction 2 OLS R²=0.03–0.18 → weaker, near-zero for Unemployment and Inflation

**Direction 2 is structurally weaker than Direction 1 :**
Macro fundamentals predict trade structure better than trade structure
predicts macro fundamentals. The causal arrow, to the extent it can be
inferred from cross-sectional data, points more reliably from
macro → trade than from trade → macro.

## O.1 — What We Set Out to Answer
</div>

Two directions were investigated :

**Direction 1 :** Can macroeconomic fundamentals explain trade structure?
→ GDP, CA%, Real Rate, Expense%, Inflation, Unemployment as predictors
→ Trade variables as outcomes : export share, import share, coverage ratio,
  structural score, combined score, trade profiles and quadrants

**Direction 2 :** Does trade structure predict macroeconomic outcomes?
→ Trade variables as predictors
→ GDP_pc, Inflation, Unemployment as outcomes

The analysis covered 10 sections (G → N) and 40+ trade variables across
three levels : raw dimensions (G/H/I), classifications (J/K), and
multi-indicator synthesis (L/M/N).


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## O.2 — Core Findings
</div>

**Finding 1 — Trade level is explained by macro fundamentals**

Export share (G.1 ρ=0.94), import share (H.1 ρ=0.96), coverage ratio (I.1 ρ=0.53),
trade openness (G.6, H.2) — all respond to macro conditions, primarily GDP size
and current account position. The larger and more surplus-oriented an economy,
the more it dominates world trade.

**Finding 2 — Trade dynamics are idiosyncratic**

Export growth (G.2), consistency (G.3), resilience (G.5), import consistency (H.5),
import resilience (H.7) — none are explained by macro indicators. The 16 World Bank
variables cannot predict whether a country's exports will grow, stay stable, or
recover from shocks. These dimensions require sector-level or policy-level data.

**Finding 3 — A clear hierarchy of macro predictors**

Across all sections, the same ranking emerges :
1. **GDP_pc / GDP** — dominant predictor of trade level and structural quality
2. **Current Account % GDP** — strongest [E] signal for balance and coverage
3. **Real Interest Rate** — moderator, non-monotone, persistent across sections
4. **Expense % GDP** — proxy for economic development and openness
5. **Inflation** — hits imports harder than exports (H.3 vs G.6)
6. **Unemployment** — weak, specific only to trade balance (I.3, N.3)

**Finding 4 — Multi-indicator analysis confirms but does not extend**

The prediction model (M) confirms RF explains 85–90% of trade variance —
but this is non-linear and does not survive cross-validation on small samples.
The L.1 macro fingerprint analysis shows no multi-dimensional profile exists —
GDP_pc alone discriminates trade profiles, not a combination of macro conditions.

**Finding 5 — Direction 2 is weaker than Direction 1**

Trade structure predicts GDP_pc (Export_pc ρ=+0.823) but this is symmetric
with Direction 1 and cannot establish causal priority. Trade → Unemployment
and Trade → Inflation signals exist but are modest (ρ = −0.19 to −0.32).
Cross-validation confirms trade variables do not reliably predict macro outcomes
out-of-sample.

## O.5 — Limits
</div>

**1. Cross-sectional data — no causal direction**
Spearman correlations and KW tests cannot establish whether macro conditions
cause trade structure or vice versa. Both directions coexist.
Panel data with lagged variables (does GDP growth in t predict export growth in t+1?)
would be required for genuine causal inference.

**2. Period effect — COVID distortion**
The 2015–2024 window includes the COVID collapse (2020) and rebound (2021),
which dominate variance in growth and volatility variables. This likely explains
why export dynamics (G.2, G.3, G.5) appear idiosyncratic — a structural relationship
may exist over longer or more stable periods.

**3. Missing dimensions**
The 6 economic indicators capture macro conditions but not :
- Natural resource endowments (Canada, Russia residuals in M.3)
- Demographic scale (India, Brazil over-performance)
- Geographic proximity to markets
- Trade policy and agreements
- Institutional quality and rule of law

**4. Sample size constraints**
n=99 for complete-case prediction models. RF R² values (0.85–0.90)
are likely optimistic without proper cross-validation — confirmed
by CV R² collapse in N.6 (near-zero for Unemployment and Inflation).


<div style="background-color:#e9f7ef;padding-left:20px;border:1px solid;border-radius:5px;">

## O.6 — Perspectives
</div>

**1. Panel data analysis**
Testing whether macro conditions in year t predict trade structure in year t+k
would transform correlational findings into causal evidence. The 2015–2024
annual data available in the Gold Layer makes this feasible.

**2. Extended indicator set**
Adding population, natural resource index, regional trade integration score,
and institutional quality indicators would reduce the systematic residuals
identified in M.3 and strengthen the predictive model.

**3. Non-linear modeling**
The RF/OLS gap (0.60–0.72) shows the relationship is fundamentally non-linear.
Gradient boosting with proper CV, or neural network approaches, could
better capture the interaction effects between macro indicators.

**4. Direction 2 with panel data**
Does export growth Granger-cause GDP growth? Does trade balance improvement
reduce unemployment over 3–5 years? These questions require time-lagged
analysis not possible with the current cross-sectional structure.

**5. Sector-level decomposition**
The residual analysis (M.3) shows that natural resource and manufacturing
exporters behave differently. Decomposing trade by sector (commodities vs
manufactures vs services) would reveal heterogeneous macro relationships
within the same country groupings.
